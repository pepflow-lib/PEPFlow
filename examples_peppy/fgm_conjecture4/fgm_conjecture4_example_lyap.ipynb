# FGM Conjecture 4: an explicit two-square Lyapunov proof

For convex $L$-smooth $f$, a minimizer $x_{\star}$, and
$\|x_{0}-x_{\star}\|\le R$, consider the final gap $f(y_{N})-f(x_{\star})$.
Initialize $x_{0}=y_{0}=z_{0}$ and $\theta_{0}=1$. For $0\le k<N$, use
$$
\begin{aligned}
\theta_{k+1}&=\frac{1+\sqrt{1+4\theta_{k}^{2}}}{2},&
y_{k+1}&=x_{k}-\nabla f(x_{k})/L,\\
z_{k+1}&=z_k-\theta_k\nabla f(x_k)/L,&
x_{k+1}&=(1-1/\theta_{k+1})y_{k+1}+z_{k+1}/\theta_{k+1}.
\end{aligned}
$$
Two explicit squares per iteration prove the bound using only the current state.

## Proof Statement

### Theorem F.3 (FGM at $y_N$)

Let $f$ be convex and $L$-smooth, $L>0$, let $x_\star$ minimize $f$, and assume
$\|x_0-x_\star\|\le R$. Fix $N\ge1$ and run the displayed FGM iteration from
$x_0=y_0=z_0$. The displayed three-update scheme is equivalent to the momentum form
(Code Cell 10).

Set $\theta_{-1}=0$, $C_0=0$,
$C_{i+1}=\theta_i^{-2}\sum_{h=0}^i\theta_h^3$, and
$$D_j=2C_N+1-C_j,\qquad \tau_N=\frac1{2D_0},\qquad a_{-1}=0,\quad a_N=1.$$
The paper's interpolation weights are
$$
a_j=\frac{\theta_j^2(\theta_{N-1}^2-\theta_j^2+C_{j+1})}
{\theta_{N-1}^2D_{j+1}}\quad(0\le j<N),\qquad
b_j=a_j-a_{j-1}\quad(0\le j\le N),
$$
$$\sigma_j=\frac{a_j}{\theta_j^2}-\frac{a_{j-1}}{\theta_{j-1}^2}\quad(0\le j<N),$$
where the second term of $\sigma_0$ is zero. Thus $a_0=b_0=1/(D_0-1)$ and
$b_N=1/(C_N+1)$. The terminal gradient $\nabla f(y_N)$ is a proof witness.
For $1\le k<N$, put
$$
\boldsymbol w_k=\begin{bmatrix}z_{k+1}-x_\star\\y_{k+1}-z_{k+1}\end{bmatrix}\in\mathcal H^2,\qquad
\mathbf Q_k=\frac12\operatorname{diag}\left(
\frac{1-a_k}{D_0-\theta_k^2},
\frac{\theta_k^2/\theta_{N-1}^2-a_k}{\theta_k^2-C_{k+1}}\right).
$$
The initial/interior potential in (32) and the terminal potential in (33) are
$$
V_0=0,\qquad
V_k=a_k(f(x_k)-f(x_\star))-\tau_NL\|x_0-x_\star\|^2-\frac{a_k}{2L}\|\nabla f(x_k)\|^2+L\boldsymbol w_k^\top \mathbf Q_k\boldsymbol w_k
\quad(1\le k<N),\qquad
V_N=f(y_N)-f(x_\star)-\tau_NL\|x_0-x_\star\|^2.
$$
The exact decrements below make it nonincreasing, and therefore
$$f(y_N)-f(x_\star)\le\frac{LR^2}{2(2C_N+1)}.$$
The interior matrices are unused for $N=1$.
For $1\le k<N-1$, define $s_k,t_k\in\mathcal H$ by
$$s_k=z_{k+1}-x_\star-(D_0-\theta_k^2)\nabla f(x_{k+1})/L,\qquad
t_k=\nabla f(x_{k+1})/L-\frac{y_{k+1}-z_{k+1}}{\theta_k^2-C_{k+1}}.$$
For $N\ge2$, the terminal directions use $z_N,y_N,\nabla f(y_N)$ in these
formulas at $k=N-1$. Then $S_k=L\alpha_k\|s_k\|^2+L\beta_k\|t_k\|^2$
for $1\le k<N$. These directions are not evaluated for $N=1$.


### Proof outline and exact verification index

Smooth convex interpolation gives
$$I_f(x,y)=f(y)-f(x)+\langle\nabla f(y),x-y\rangle
+\frac1{2L}\|\nabla f(x)-\nabla f(y)\|^2\le0,$$
where $\nabla f(x_\star)=0$.
For $0\le k<N-1$ and $1\le k<N-1$, respectively, the square coefficients are
$$
\alpha_k=\frac{(D_0b_{k+1}-\theta_{k+1})(D_0-\theta_{k+1}^2)
+(D_0a_{k+1}-\theta_{k+1}^2)\theta_{k+1}}
{2D_0(D_0-\theta_{k+1}^2)(D_0-\theta_k^2)},
$$
$$
\beta_k=\frac{\theta_k^2(\theta_k^2-C_{k+1})}{2}
\left[\sigma_{k+1}+\frac{\theta_{k+1}-1}{\theta_{k+1}^2-C_{k+2}}
\left(\frac1{\theta_{N-1}^2}-\frac{a_{k+1}}{\theta_{k+1}^2}\right)\right].
$$
At $k=N-1$, use
$$\alpha_{N-1}=\frac{b_N}{2(D_0-\theta_{N-1}^2)},\qquad
\beta_{N-1}=\frac{b_N(\theta_{N-1}^2-C_N)}2.$$
Use $S_k=L\alpha_k\|s_k\|^2+L\beta_k\|t_k\|^2$ with the directions defined above. The base term for $N\ge2$ is
$$S_0=\frac{LD_0}{2(D_0-1)^2}\|\nabla f(x_0)/L-(x_0-x_\star)/D_0\|^2
+L\alpha_0\|x_0-x_\star-\nabla f(x_0)/L-(D_0-1)\nabla f(x_1)/L\|^2.$$
These give the decrement identities (34):
$$
\begin{aligned}
V_1-V_0&=b_0I_f(x_\star,x_0)+a_0I_f(x_0,x_1)+b_1I_f(x_\star,x_1)-S_0,\\
V_{k+1}-V_k&=a_kI_f(x_k,x_{k+1})+b_{k+1}I_f(x_\star,x_{k+1})-S_k
&& (1\le k<N-1),\\
V_N-V_{N-1}&=a_{N-1}I_f(x_{N-1},y_N)+b_NI_f(x_\star,y_N)-S_{N-1}.
\end{aligned}
$$
For $N=2$ the middle range is empty. For $N=1$ the first line alone is used,
with $x_1$ replaced by $y_1$, $D_0=3$, $a_0=b_0=b_1=1/2$, and $\alpha_0=1/8$;
no factor $(\theta_0^2-C_1)^{-1}$ is evaluated.
All used $a_j,b_j,\alpha_k,\beta_k$ are strictly positive. The terminal
formula gives $\beta_0=0$ when $N=1$, but that coefficient is unused.
Thus $S_k\ge0$, each decrement is nonpositive, and the terminal potential plus
the radius assumption yields the bound.

```python
from pathlib import Path
from itertools import combinations
from fractions import Fraction
import hashlib
import json
import sys
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp

# Locate the project from the Peppy root, project root, or a notebook subdirectory.
_candidates = [Path.cwd(), *Path.cwd().parents]
ROOT = next(q for q in _candidates if (q / "pepflow").is_dir())
sys.path.insert(0, str(ROOT))
import pepflow as pf  # noqa: E402
import pepflow.lyapunov_utils as lu  # noqa: E402

STUDY = ROOT / "examples_peppy" / "fgm_conjecture4"
STATE_STEM = "fgm_conjecture4"
%matplotlib inline
```

```python
L = pf.Parameter("L")
R = pf.Parameter("R")
f = pf.SmoothConvexFunction(is_basis=True, tags=["f"], L=L)
```

```python
def theta_sequence(N, digits=50):
    theta = [sp.Integer(1)]
    for _ in range(N):
        value = (1 + sp.sqrt(1 + 4 * theta[-1] ** 2)) / 2
        theta.append(value if digits is None else sp.N(value, digits))
    return theta


def coefficient_rows(N):
    theta, rows = theta_sequence(N), [[]]
    for k in range(N):
        rows.append(
            [
                (1 - 1 / theta[k + 1]) * rows[k][j] + theta[j] / theta[k + 1]
                for j in range(k)
            ]
            + [1 + (theta[k] - 1) / theta[k + 1]]
        )
    return rows


def tau_N_value(N):
    C_N = 1 + sum(coefficient_rows(N - 1)[-1], sp.Integer(0))
    return 1 / (2 * (2 * C_N + 1))


def make_ctx_fgm_c4_explicit_squares(ctx_name, N):
    theta = theta_sequence(N)
    ctx = pf.PEPContext(ctx_name).set_as_current()
    x = pf.Vector(is_basis=True, tags=["x_0", "y_0", "z_0"])
    z = x
    f.set_stationary_point("x_star")
    for k in range(N):
        g = f.grad(x)
        y = (x - g / L).add_tag(f"y_{k + 1}")
        z = (z - theta[k] * g / L).add_tag(f"z_{k + 1}")
        x = ((1 - 1 / theta[k + 1]) * y + z / theta[k + 1]).add_tag(f"x_{k + 1}")
    return ctx


def get_pep_setup(N, params):
    assert isinstance(N, int) and N >= 1
    assert params.get("L", 1) > 0 and params.get("R", 1) >= 0
    ctx = make_ctx_fgm_c4_explicit_squares(f"fgm_{N}", N)
    pb = pf.PEPBuilder(ctx)
    pb.add_initial_constraint(
        ((ctx["x_0"] - ctx["x_star"]) ** 2).le(R**2, name="initial_condition")
    )
    pb.set_performance_metric(f(ctx[f"y_{N}"]) - f(ctx["x_star"]))
    return ctx, pb, f
```

## Multiplier and slack verification (numerical diagnostics)

The numerical check reconstructs the paper's $a_i$ and $b_i$ from the saved
FGM gradient coefficient rows and compares them with the saved interpolation
multipliers. Finite numerical checks do not establish their signs for every
horizon.

The saved certificate is checked by reconstructing its weighted interpolation
residuals and LDL slack exactly in Gram, function, and constant coordinates.
The initial constraint is included. All Gram, function, and constant
coordinates are checked, and no small LDL coefficient is discarded.
The exact proof in Code Cells 10–15 is independent of these diagnostics.

```python
certificate_keys = (
    "N_verify",
    "params_json",
    "relaxed_constraints",
    "S_matrix",
    "S_row_names",
    "S_col_names",
    "basis_vectors",
    "basis_scalars",
    "opt_value",
    "tau_sol",
    "tau_name",
    "named_dual_values",
    "constraint_metadata",
    "normalization",
    "certificate_provenance",
    "numerical_checks",
    "numerical_decomposition",
)
payload = {key: certificate[key] for key in certificate_keys}
assert (
    hashlib.sha256(
        json.dumps(
            payload, sort_keys=True, separators=(",", ":"), allow_nan=False
        ).encode()
    ).hexdigest()
    == certificate["certificate_id"]
)
basis_num = list(ctx_num.basis_vectors())
assert [str(v) for v in basis_num] == certificate["basis_vectors"]
assert [str(v) for v in ctx_num.basis_scalars()] == certificate["basis_scalars"]
constraints_num = [
    *pb_num.init_conditions,
    *f_num.get_interpolation_constraints(ctx_num),
]
duals_num = certificate["named_dual_values"]
assert {c.name for c in constraints_num} == set(duals_num)
weighted_num = {}
for constraint in constraints_num:
    value = duals_num[constraint.name]
    if constraint.name in pb_num.relaxed_constraints:
        assert value is None
        continue
    assert value is not None and value >= -1e-6 and constraint.cmp.name == "LE"
    weighted_num[constraint.name] = float(value) * (constraint.lhs - constraint.rhs)
certificate_slack_matrix = np.asarray(certificate["S_matrix"], dtype=float)
LDL_D_num = np.asarray(certificate["numerical_decomposition"]["D"], dtype=float)
LDL_E_num = np.asarray(
    certificate["numerical_decomposition"]["ell_coords"], dtype=float
)
assert (
    np.max(abs(LDL_E_num.T @ LDL_D_num @ LDL_E_num - certificate_slack_matrix)) < 1e-8
)
assert np.linalg.eigvalsh(certificate_slack_matrix).min() >= -1e-6
certificate_slack = sum(
    (
        float(certificate_slack_matrix[i, j]) * basis_num[i] * basis_num[j]
        for i, j in np.ndindex(certificate_slack_matrix.shape)
    ),
    pf.Scalar.zero(),
)


def numeric_residual(expression):
    value = pm_num.eval_scalar(expression)
    return [
        float(np.max(np.abs(np.asarray(component, dtype=float)), initial=0))
        for component in (value.inner_prod_coords, value.func_coords, value.offset)
    ]


certificate_diff = (
    pb_num.performance_metric
    - float(certificate["opt_value"])
    - sum(weighted_num.values(), pf.Scalar.zero())
    + certificate_slack
)
certificate_errors = numeric_residual(certificate_diff)
assert max(certificate_errors) < 1e-5
print("Certificate residuals [Gram, function, constant]:", certificate_errors)
print(
    "LDL reconstruction error:",
    np.max(abs(LDL_E_num.T @ LDL_D_num @ LDL_E_num - certificate_slack_matrix)),
)
pf.pprint_labeled_matrix(
    certificate_slack_matrix,
    certificate["S_row_names"],
    certificate["S_col_names"],
    precision=5,
)

rows_num = coefficient_rows(N_num - 1)
h_rows_num = [row + [sp.Integer(0)] * (N_num - len(row)) for row in rows_num]
h_rows_num.append(rows_num[-1] + [sp.Integer(1)])
C_N_num = sum(h_rows_num[-1])
a_num = [
    sum(h_rows_num[-1][: i + 1]) / (2 * C_N_num - sum(h_rows_num[i]))
    for i in range(N_num)
]
b_num = (
    [a_num[0]] + [a_num[i] - a_num[i - 1] for i in range(1, N_num)] + [1 - a_num[-1]]
)


def paper_point_num(k):
    return ctx_num[f"y_{N_num}"] if k == N_num else ctx_num[f"x_{k}"]


candidate_num = {
    f"{f_num}:{ctx_num['x_star']},{point}": b_num[i]
    for i, point in ((i, paper_point_num(i)) for i in range(N_num + 1))
}
candidate_num.update(
    {
        f"{f_num}:{paper_point_num(i)},{paper_point_num(i + 1)}": a_num[i]
        for i in range(N_num)
    }
)
lambda_error = max(
    abs(float(value) - duals_num[name]) for name, value in candidate_num.items()
)
certificate_slack_from_weights = (
    float(tau_N_value(N_num)) * (ctx_num["x_0"] - ctx_num["x_star"]) ** 2
    - pb_num.performance_metric
)
for constraint in f_num.get_interpolation_constraints(ctx_num):
    certificate_slack_from_weights += float(candidate_num.get(constraint.name, 0)) * (
        constraint.lhs - constraint.rhs
    )
assert (
    lambda_error < 1e-7
    and max(numeric_residual(certificate_slack_from_weights - certificate_slack)) < 1e-7
)
print("Candidate multiplier error:", lambda_error)
print(
    "Candidate slack error:",
    numeric_residual(certificate_slack_from_weights - certificate_slack),
)
print(
    "Numerical certificate verified; global analytic certificate status: not inferred from these checks."
)
```

## Partial sums and rank profile (numerical diagnostics)

The saved LDL factors are grouped by their latest gradient. Their partial
certificate sums add the saved weights on $I_f(x_\star,x_k)$ and
$I_f(x_k,x_{k+1})$, using $y_N$ at the endpoint, then subtract the assigned
slack pieces. The rank check identifies the point and gradient span used by
the discovery calculation. The final paper potential $V_k$ is the explicit
expression in (32) and (33).


```python
ell_num = [
    sum((float(a) * v for a, v in zip(row, basis_num)), pf.Vector.zero())
    for row in LDL_E_num
]
gradient_columns = [
    int(np.flatnonzero(pm_num.eval_vector(f_num.grad(p)).coords)[0])
    for p in (paper_point_num(k) for k in range(N_num + 1))
]
pieces_num = {j: pf.Scalar.zero() for j in range(N_num + 1)}
block_start = 0
while block_start < len(LDL_D_num):
    block = (
        [block_start, block_start + 1]
        if block_start + 1 < len(LDL_D_num)
        and LDL_D_num[block_start, block_start + 1] != 0
        else [block_start]
    )
    support = np.max(abs(LDL_E_num[np.ix_(block, gradient_columns)]), axis=0)
    latest = (
        int(np.flatnonzero(support > 1e-12).max()) if np.any(support > 1e-12) else 0
    )
    pieces_num[latest] += sum(
        (
            float(LDL_D_num[i, j]) * ell_num[i] * ell_num[j]
            for i in block
            for j in block
        ),
        pf.Scalar.zero(),
    )
    block_start += len(block)
assert (
    max(
        numeric_residual(sum(pieces_num.values(), pf.Scalar.zero()) - certificate_slack)
    )
    < 1e-8
)
partial_certificate_num = [
    float(certificate["opt_value"]) + weighted_num[certificate["tau_name"]]
]
partial_sum = partial_certificate_num[0] - pieces_num[0]
for k in range(N_num):
    partial_sum += weighted_num[f"{f_num}:{ctx_num['x_star']},{paper_point_num(k)}"]
    partial_sum += weighted_num[
        f"{f_num}:{paper_point_num(k)},{paper_point_num(k + 1)}"
    ]
    partial_sum -= pieces_num[k + 1]
    if k == N_num - 1:
        partial_sum += weighted_num[
            f"{f_num}:{ctx_num['x_star']},{paper_point_num(N_num)}"
        ]
    partial_certificate_num.append(partial_sum)
assert (
    max(numeric_residual(partial_certificate_num[-1] - pb_num.performance_metric))
    < 1e-5
)
ranks = [
    int(
        np.sum(
            np.linalg.svd(
                np.asarray(
                    pm_num.eval_scalar(partial_certificate).inner_prod_coords,
                    dtype=float,
                ),
                compute_uv=False,
            )
            > 1e-4
        )
    )
    for partial_certificate in partial_certificate_num
]
assert ranks == certificate["rank_profile"]
print("Ranks of numerical partial sums (absolute tolerance 1e-4):", ranks)
print(
    "Terminal residual [Gram, function, constant]:",
    numeric_residual(partial_certificate_num[-1] - pb_num.performance_metric),
)
```

## Special vectors and coefficient extraction

The search uses tagged iterates, existing gradients, and their pairwise
differences. It filters by column space and then searches for the sparsest
coefficient matrix. The adopted interior order is
$(y_{k+1}-x_{\star},z_{k+1}-x_{\star},\nabla f(x_k)/L)$, equivalent to the selected span
because $z_{k+1}=\theta_{k+1}x_{k+1}-(\theta_{k+1}-1)y_{k+1}$.
At $k=1,N-1$ only the first and third vectors are needed; at $k=0$ the
basis is $(x_{0}-x_{\star})$, and the terminal quadratic basis is empty.

```python
candidates, candidate_labels, directions = [], [], []


def add_candidate(label, vector):
    coords = np.asarray(pm_num.eval_vector(vector).coords, dtype=float)
    norm = np.linalg.norm(coords)
    if norm <= 1e-12:
        return
    unit = coords / norm
    unit *= np.sign(unit[np.flatnonzero(abs(unit) > 1e-12)[0]])
    if any(np.max(abs(unit - old)) <= 1e-12 for old in directions):
        return
    candidates.append(vector)
    candidate_labels.append(label)
    directions.append(unit)


for tag in ["x_star"] + [
    f"{seq}_{i}" for seq in ["x", "y", "z"] for i in range(N_num + 1)
]:
    add_candidate(tag, ctx_num[tag])
for i in range(N_num + 1):
    point = paper_point_num(i)
    add_candidate(
        f"grad f(x_{i})/L" if i < N_num else f"grad f(y_{N_num})/L",
        f_num.grad(point) / params_num["L"],
    )
for vector in basis_num:
    add_candidate(str(vector), vector)
for (left_label, left), (right_label, right) in combinations(
    list(zip(candidate_labels, candidates)), 2
):
    add_candidate(f"({right_label})-({left_label})", right - left)
labels_by_id = {id(v): label for v, label in zip(candidates, candidate_labels)}
print("Candidate directions:", len(candidates))
for k, partial_certificate in enumerate(partial_certificate_num):
    if ranks[k]:
        filtered = lu.vectors_in_column_space(
            partial_certificate,
            candidates,
            pep_context=ctx_num,
            resolve_parameters=params_num,
            rtol=1e-4,
            atol=1e-4,
        )
        selected, coeff = lu.find_basis_with_sparsest_coefficients(
            partial_certificate,
            filtered,
            pep_context=ctx_num,
            resolve_parameters=params_num,
            zero_tol=1e-6,
            indep_tol=1e-7,
            span_tol=1e-5,
        )
    else:
        selected, coeff = [], np.zeros((0, 0))
    assert len(selected) == ranks[k]
    print(
        f"k={k}: rank={ranks[k]}, basis={[labels_by_id[id(v)] for v in selected]}, "
        f"near-zero coefficients={np.sum(abs(coeff) < 1e-6)}/{coeff.size}"
    )
```

## Closed-form Lyapunov coefficients

From this point onward the executable verification uses the paper's notation.
The code writes $\theta_k^2=\theta_{k+1}^2-\theta_{k+1}$ and uses the moment sequence $\rho_k$ from Lemma F.4 directly. Code Cell 11 substitutes the literal theorem definitions
of $a,b,\sigma,\mathbf Q,\alpha,\beta$ into independent expressions for both sides.
All potentials use the paper's normalization $V_0=0$ and
$V_N=f(y_N)-f(x_\star)-\tau_NL\|x_0-x_\star\|^2$.

### Positivity of the explicit coefficients

For $i\ge1$, use
$\rho_i=(1-1/\theta_i)(\rho_{i-1}+1)$.
The theta recurrence gives $\theta_i-\theta_{i-1}>1/2$.
At each induction step, $\rho_{i-1}+1\le(2\theta_{i-1}+1)/3<2\theta_i/3$, using $\theta_i-\theta_{i-1}>1/2$. Starting from $\rho_0=0$, induction uses
$$\rho_i-\rho_{i-1}-\tfrac13=\left(2\theta_i/3-(\rho_{i-1}+1)\right)/\theta_i,\qquad
\frac{2\theta_i+1}{3}-(\rho_i+1)=(1-1/\theta_i)\left(2\theta_i/3-(\rho_{i-1}+1)\right)>0.$$
Consequently $\rho_i+1\le(2\theta_i+1)/3$ and $C_i$ increases.
Thus $D_j\ge C_N+1>0$,
$D_0-\theta_i^2\ge\rho_{N-1}+2>0$ for $0\le i<N$, and, for $i\ge1$,
$$\theta_i^2-C_{i+1}\ge\frac{(\theta_i-1)(3\theta_i+1)}6>0.$$
The initial gap $\theta_0^2-C_1=0$ is never inverted.

For $1\le i<N$, use $\theta_{N-1}\ge\theta_i$ and the moment upper bound above.
Code Cell 10 checks the literal positive expansion
$$
\begin{aligned}
\theta_i\theta_{N-1}^2D_iD_{i+1}\sigma_i={}&\frac{\theta_i}{18}\bigl[(3\theta_i-4)(\theta_i-1)
+2(3\theta_i+1)(\theta_{N-1}-\theta_i)+6(\theta_{N-1}-\theta_i)^2\bigr]\\
&+\left(2\theta_i/3-(\rho_{i-1}+1)\right)\left[\theta_{N-1}^2-\theta_i^2+\frac{\theta_i}{2}+\frac{\theta_{N-1}}{3}+\frac23\right]+C_i\left((2\theta_{N-1}+1)/3-(\rho_{N-1}+1)\right)>0.
\end{aligned}
$$
Each term is nonnegative, and the first is positive since
$\theta_i\ge\theta_1>4/3$. Therefore $\sigma_i>0$. Since
$\sigma_0=a_0=1/(D_0-1)>1/D_0$, the ratios $a_i/\theta_i^2$ increase and
$$b_i=\theta_i\frac{a_i}{\theta_i^2}+\theta_{i-1}^2\sigma_i>0,\quad
D_0b_i-\theta_i>0,\quad D_0a_i-\theta_i^2>0.$$
Also $b_0=a_0>0$ and $b_N=1/(C_N+1)>0$.
The exact tail identity is
$$\frac1{\theta_{N-1}^2}-\frac{a_i}{\theta_i^2}
=\frac{\rho_{N-1}+2-(\rho_i+1)}{\theta_{N-1}^2D_{i+1}}>0.$$
The monotonicity of $\rho_i$ proves this last numerator positive. These signs
make every displayed interior $\alpha_k,\beta_k$ strictly positive.
At the terminal index $\alpha_{N-1}>0$ and $\beta_{N-1}>0$ for $N\ge2$;
for $N=1$, the unused terminal $\beta_0$ is zero.
This is an all-horizon analytic argument, not a finite-horizon test.

For the rate and multiplier numerators, Code Cell 10 checks the telescoping
increments and their initial values:
$$\sum_{h=0}^i\theta_h=\theta_i^2,\qquad
\sum_{h=0}^i\theta_h\theta_{h-1}^2
=\theta_i^2(\theta_i^2-C_{i+1}),\qquad
\sum_{h=0}^i\theta_h^3=\theta_i^2C_{i+1}.$$
The fixed-step recurrence gives
$y_N=x_0-\sum_{h=0}^{N-1}\theta_h(1-\theta_{h-1}^2/\theta_{N-1}^2)\nabla f(x_h)/L$.
Applying the two independently checked telescoping sums to these coefficients gives
$$\sum_{h=0}^{N-1}\theta_h\left(1-\frac{\theta_{h-1}^2}{\theta_{N-1}^2}\right)
=\sum_{h=0}^{N-1}\theta_h-\frac1{\theta_{N-1}^2}\sum_{h=0}^{N-1}\theta_h\theta_{h-1}^2
=\theta_{N-1}^2-(\theta_{N-1}^2-C_N)=C_N.$$
The same argument for a prefix proves the two formulas for $a_i$.

```python
# Code Cell 10: exact scalar identities and the analytic positivity argument.
# Standalone prerequisite: SymPy. No discovery or certificate cells are needed.
import sympy as sp

theta_i, theta_Nminus1, rho_iminus1, rho_Nminus1 = sp.symbols(
    "theta_i theta_Nminus1 rho_iminus1 rho_Nminus1", real=True
)
theta_iminus1_squared = theta_i**2 - theta_i
rho_i = (1 - 1 / theta_i) * (rho_iminus1 + 1)
C_i = (theta_iminus1_squared + rho_iminus1 + 1) / 2  # C_i
C_iplus1 = (theta_i**2 + rho_i + 1) / 2  # C_{i+1}
D_0 = theta_Nminus1**2 + rho_Nminus1 + 2
D_i, D_iplus1 = D_0 - C_i, D_0 - C_iplus1
a_iminus1 = (
    theta_iminus1_squared
    * (theta_Nminus1**2 - theta_iminus1_squared + C_i)
    / (theta_Nminus1**2 * D_i)
)
a_i = (
    theta_i**2
    * (theta_Nminus1**2 - theta_i**2 + C_iplus1)
    / (theta_Nminus1**2 * D_iplus1)
)
sigma_i = a_i / theta_i**2 - a_iminus1 / theta_iminus1_squared
b_i = a_i - a_iminus1
positive_expansion = (
    theta_i
    * (
        (3 * theta_i - 4) * (theta_i - 1)
        + 2 * (3 * theta_i + 1) * (theta_Nminus1 - theta_i)
        + 6 * (theta_Nminus1 - theta_i) ** 2
    )
    / 18
    + (2 * theta_i / 3 - (rho_iminus1 + 1))
    * (
        theta_Nminus1**2
        - theta_i**2
        + theta_i / 2
        + theta_Nminus1 / 3
        + sp.Rational(2, 3)
    )
    + C_i * ((2 * theta_Nminus1 + 1) / 3 - (rho_Nminus1 + 1))
)
scalar_residuals = {
    "theta telescoping": theta_i**2 - theta_iminus1_squared - theta_i,
    "C recurrence": C_iplus1 - theta_i - (1 - 1 / theta_i) * C_i,
    "cube prefix increment": theta_i**2 * C_iplus1
    - theta_iminus1_squared * C_i
    - theta_i**3,
    "fourth-power telescoping": theta_i**4
    - theta_iminus1_squared**2
    - (2 * theta_i**3 - theta_i**2),
    "weighted prefix increment": theta_i**2 * (theta_i**2 - C_iplus1)
    - theta_iminus1_squared * (theta_iminus1_squared - C_i)
    - theta_i * theta_iminus1_squared,
    "moment increment lower bound": rho_i
    - rho_iminus1
    - sp.Rational(1, 3)
    - (2 * theta_i / 3 - (rho_iminus1 + 1)) / theta_i,
    "moment upper bound induction": (2 * theta_i + 1) / 3
    - (rho_i + 1)
    - (1 - 1 / theta_i) * (2 * theta_i / 3 - (rho_iminus1 + 1)),
    "positive gap lower bound": (theta_i**2 - (2 * theta_i + 1) / 3) / 2
    - (theta_i - 1) * (3 * theta_i + 1) / 6,
    "positive sigma numerator": theta_i * theta_Nminus1**2 * D_i * D_iplus1 * sigma_i
    - positive_expansion,
    "positive tail": 1 / theta_Nminus1**2
    - a_i / theta_i**2
    - (rho_Nminus1 + 2 - (rho_i + 1)) / (theta_Nminus1**2 * D_iplus1),
    "star-weight identity": b_i
    - theta_i * a_i / theta_i**2
    - theta_iminus1_squared * sigma_i,
    "multiplier prefix formula": (
        theta_i**2 - theta_i**2 * (theta_i**2 - C_iplus1) / theta_Nminus1**2
    )
    / D_iplus1
    - a_i,
}
scalar_residuals = {name: sp.cancel(value) for name, value in scalar_residuals.items()}
assert all(value == 0 for value in scalar_residuals.values())
# Explicit initial summands initialize the three independently checked increments.
theta_0, theta_minus1, rho_0 = (
    sp.Integer(1),
    sp.Integer(0),
    sp.Integer(0),
)
C_1 = (theta_0**2 + rho_0 + 1) / 2
assert C_1 == 1
assert theta_0 == theta_0**2
assert theta_0 * theta_minus1**2 == theta_0**2 * (theta_0**2 - C_1)
assert theta_0**3 == theta_0**2 * C_1
print("Scalar identities:", scalar_residuals)

# Algorithm equivalence with the raw gradient and actual positions.
L = sp.Symbol("L", positive=True)
theta_k, theta_kplus1, x_k, y_k, grad_x_k = sp.symbols(
    "theta_k theta_kplus1 x_k y_k grad_x_k", real=True
)
y_kplus1 = x_k - grad_x_k / L
z_kplus1 = theta_k * x_k - (theta_k - 1) * y_k - theta_k * (grad_x_k / L)
x_kplus1 = (1 - 1 / theta_kplus1) * y_kplus1 + z_kplus1 / theta_kplus1
assert (
    sp.cancel(x_kplus1 - y_kplus1 - (theta_k - 1) / theta_kplus1 * (y_kplus1 - y_k))
    == 0
)
theta_j, theta_jminus1_squared = sp.symbols("theta_j theta_jminus1_squared")
h_k_j = theta_j * (1 - theta_jminus1_squared / (theta_kplus1**2 - theta_kplus1))
h_kplus1_j = theta_j * (1 - theta_jminus1_squared / theta_kplus1**2)
assert (
    sp.cancel((1 - 1 / theta_kplus1) * h_k_j + theta_j / theta_kplus1 - h_kplus1_j) == 0
)
new_row_residual = (
    1
    + (theta_k - 1) / theta_kplus1
    - theta_k * (1 - (theta_k**2 - theta_k) / theta_kplus1**2)
)
assert (
    sp.rem(
        sp.together(new_row_residual).as_numer_denom()[0],
        theta_k**2 - (theta_kplus1**2 - theta_kplus1),
        theta_k,
    )
    == 0
)
z_kplus1_state, y_kplus1_state = sp.symbols("z_kplus1_state y_kplus1_state", real=True)
x_kplus1_state = z_kplus1_state + (1 - 1 / theta_kplus1) * (
    y_kplus1_state - z_kplus1_state
)
assert (
    sp.cancel(
        x_kplus1_state
        - z_kplus1_state
        - (theta_kplus1**2 - theta_kplus1)
        * (y_kplus1_state - z_kplus1_state)
        / theta_kplus1**2
    )
    == 0
)
assert (
    sp.cancel(
        x_kplus1_state
        - grad_x_k / L
        - (z_kplus1_state - theta_kplus1 * (grad_x_k / L))
        - (
            (theta_kplus1**2 - theta_kplus1)
            * (y_kplus1_state - z_kplus1_state)
            / theta_kplus1**2
            + (theta_kplus1 - 1) * (grad_x_k / L)
        )
    )
    == 0
)
print("Momentum/mixing equivalence, row recurrence, and state update: exact zeros")
```

## Symbolic recursion context

Code Cell 11 provides literal coefficient definitions and an exact polynomial
checker. A scalar quadratic in formal coordinates represents the corresponding
inner products in arbitrary dimension: every diagonal coefficient and every
cross coefficient is checked independently. Function-value symbols and the
constant coefficient are also checked. No positivity of a Gram matrix or
function-value symbol is assumed in these algebraic identities.
The proof defines the $3\times3$ matrix $\mathbf S_k$ below and evaluates it on the column formed by $\boldsymbol w_k$ and $\nabla f(x_{k+1})/L$, without a separate expanded-state alias.


```python
# Code Cell 11: literal paper coefficients and exact coefficient-wise checking.
# Prerequisite: Code Cell 10 (SymPy import only).
def paper_coefficients(theta_kplus1, rho_k, theta_Nminus1, rho_Nminus1):
    theta_k_squared = theta_kplus1**2 - theta_kplus1
    rho_kplus1 = (1 - 1 / theta_kplus1) * (rho_k + 1)
    C_kplus1 = (theta_k_squared + rho_k + 1) / 2  # C_{k+1}
    C_kplus2 = (theta_kplus1**2 + rho_kplus1 + 1) / 2  # C_{k+2}
    D_0 = theta_Nminus1**2 + rho_Nminus1 + 2
    a_k = (
        theta_k_squared
        * (theta_Nminus1**2 - theta_k_squared + C_kplus1)
        / (theta_Nminus1**2 * (D_0 - C_kplus1))
    )
    a_kplus1 = (
        theta_kplus1**2
        * (theta_Nminus1**2 - theta_kplus1**2 + C_kplus2)
        / (theta_Nminus1**2 * (D_0 - C_kplus2))
    )
    b_kplus1, sigma_kplus1 = (
        a_kplus1 - a_k,
        a_kplus1 / theta_kplus1**2 - a_k / theta_k_squared,
    )
    alpha_k = (
        (D_0 * b_kplus1 - theta_kplus1) * (D_0 - theta_kplus1**2)
        + (D_0 * a_kplus1 - theta_kplus1**2) * theta_kplus1
    ) / (2 * D_0 * (D_0 - theta_kplus1**2) * (D_0 - theta_k_squared))
    beta_k = (
        theta_k_squared
        * (theta_k_squared - C_kplus1)
        / 2
        * (
            sigma_kplus1
            + (theta_kplus1 - 1)
            / (theta_kplus1**2 - C_kplus2)
            * (1 / theta_Nminus1**2 - a_kplus1 / theta_kplus1**2)
        )
    )
    return dict(
        theta_kplus1=theta_kplus1,
        theta_Nminus1=theta_Nminus1,
        theta_k_squared=theta_k_squared,
        C_kplus1=C_kplus1,
        C_kplus2=C_kplus2,
        D_0=D_0,
        a_k=a_k,
        a_kplus1=a_kplus1,
        b_kplus1=b_kplus1,
        sigma_kplus1=sigma_kplus1,
        Q_k_11=(1 - a_k) / (2 * (D_0 - theta_k_squared)),
        Q_k_22=(theta_k_squared / theta_Nminus1**2 - a_k)
        / (2 * (theta_k_squared - C_kplus1)),
        Q_kplus1_11=(1 - a_kplus1) / (2 * (D_0 - theta_kplus1**2)),
        Q_kplus1_22=(theta_kplus1**2 / theta_Nminus1**2 - a_kplus1)
        / (2 * (theta_kplus1**2 - C_kplus2)),
        alpha_k=alpha_k,
        beta_k=beta_k,
    )


def formal_coefficients(values):
    symbols = {name: sp.Symbol(name, real=True) for name in values}
    return symbols, {symbols[name]: sp.factor(value) for name, value in values.items()}


def exact_polynomial(
    name,
    expression,
    variables,
    substitutions=None,
    reducer=lambda value: sp.cancel(sp.together(value)),
    required_monomials=(),
):
    """Check every independent coordinate, function, and constant coefficient."""
    polynomial = sp.Poly(sp.expand(expression), *variables)
    monomials = set(polynomial.monoms()) | set(required_monomials)
    monomials.add((0,) * len(variables))
    residuals = {}
    for monomial in sorted(monomials, reverse=True):
        coefficient = polynomial.coeff_monomial(monomial)
        if substitutions:
            coefficient = coefficient.subs(substitutions, simultaneous=True)
        residuals[monomial] = reducer(coefficient)
    assert all(value == 0 for value in residuals.values()), (name, residuals)
    print(
        name
        + ": every coefficient is exactly zero ("
        + str(len(residuals))
        + " checked)"
    )
    return residuals


def I_f(f_left, f_right, x_left, x_right, grad_left, grad_right, L):
    return (
        f_right
        - f_left
        + grad_right * (x_left - x_right)
        + (grad_left - grad_right) ** 2 / (2 * L)
    )


# Set f(x_star)=0 by an objective shift; the symbols f_x_k are centered function values.
# Independent absolute scalar coordinates; L is a parameter, not a polynomial generator.
z_kplus1, y_kplus1, x_star, grad_x_k, grad_x_kplus1, x_0_minus_x_star = sp.symbols(
    "z_kplus1 y_kplus1 x_star grad_x_k grad_x_kplus1 x_0_minus_x_star", real=True
)
f_x_k, f_x_kplus1 = sp.symbols("f_x_k f_x_kplus1", real=True)
theta_kplus1_formal, rho_k_formal, theta_Nminus1_formal, rho_Nminus1_formal = (
    sp.symbols("theta_kplus1 rho_k theta_Nminus1 rho_Nminus1", real=True)
)
```

## Symbolic step recursion verification

For $1\le k<N-1$, the algorithm gives
$$x_k=y_{k+1}+\nabla f(x_k)/L,\quad
x_{k+1}-x_\star=z_{k+1}-x_\star+
\frac{\theta_k^2}{\theta_{k+1}^2}(y_{k+1}-z_{k+1}).$$
The same update gives
$$\boldsymbol w_{k+1}=\operatorname{diag}\left(1,\frac{\theta_k^2}{\theta_{k+1}^2}\right)\boldsymbol w_k
-\begin{bmatrix}\theta_{k+1}\\1-\theta_{k+1}\end{bmatrix}\frac{\nabla f(x_{k+1})}{L}.$$
Define the symmetric matrix $\mathbf S_k\in\mathbb R^{3\times3}$ directly:
$$\mathbf S_k=
\begin{pmatrix}\mathbf Q_k&-\frac12\begin{bmatrix}b_{k+1}\\\theta_k^2\sigma_{k+1}\end{bmatrix}\\
-\frac12\begin{bmatrix}b_{k+1}&\theta_k^2\sigma_{k+1}\end{bmatrix}&a_{k+1}\end{pmatrix}
-\begin{pmatrix}1&0&-\theta_{k+1}\\0&\theta_k^2/\theta_{k+1}^2&\theta_{k+1}-1\end{pmatrix}^{\!\top}
\mathbf Q_{k+1}\begin{pmatrix}1&0&-\theta_{k+1}\\0&\theta_k^2/\theta_{k+1}^2&\theta_{k+1}-1\end{pmatrix}.$$
Code Cell 12 checks all six independent entries of the rank-two identity
$$\mathbf S_k=\alpha_k\begin{bmatrix}1\\0\\-(D_0-\theta_k^2)\end{bmatrix}
\begin{bmatrix}1&0&-(D_0-\theta_k^2)\end{bmatrix}
+\beta_k\begin{bmatrix}0\\-1/(\theta_k^2-C_{k+1})\\1\end{bmatrix}
\begin{bmatrix}0&-1/(\theta_k^2-C_{k+1})&1\end{bmatrix}.$$
In particular, its mixed position entry vanishes. Using the defined directions,
$$L\begin{bmatrix}\boldsymbol w_k\\\nabla f(x_{k+1})/L\end{bmatrix}^{\!\top}
\mathbf S_k\begin{bmatrix}\boldsymbol w_k\\\nabla f(x_{k+1})/L\end{bmatrix}
=L\alpha_k\|s_k\|^2+L\beta_k\|t_k\|^2=S_k.$$
The independently checked interpolation expansion gives
$$a_k I_f(x_k,x_{k+1})+b_{k+1}I_f(x_\star,x_{k+1})-(V_{k+1}-V_k)
=L\begin{bmatrix}\boldsymbol w_k\\\nabla f(x_{k+1})/L\end{bmatrix}^{\!\top}
\mathbf S_k\begin{bmatrix}\boldsymbol w_k\\\nabla f(x_{k+1})/L\end{bmatrix}=S_k.$$
All coefficients in the actual point and raw-gradient coordinates, including
function values and constants, are checked separately.


```python
# Code Cell 12: exact interior matrix, position, raw-gradient, function and constant checks.
# The mixed state entry is checked in the state basis, not as an absolute-coordinate monomial.
step_values = paper_coefficients(
    theta_kplus1_formal, rho_k_formal, theta_Nminus1_formal, rho_Nminus1_formal
)
q, step_substitution = formal_coefficients(step_values)
Q_k = sp.diag(q["Q_k_11"], q["Q_k_22"])
Q_kplus1 = sp.diag(q["Q_kplus1_11"], q["Q_kplus1_22"])
state_map = sp.Matrix(
    [
        [1, 0, -q["theta_kplus1"]],
        [0, q["theta_k_squared"] / q["theta_kplus1"] ** 2, q["theta_kplus1"] - 1],
    ]
)
S_k_matrix = sp.diag(q["Q_k_11"], q["Q_k_22"], q["a_kplus1"])
S_k_matrix[0, 2] = S_k_matrix[2, 0] = -q["b_kplus1"] / 2
S_k_matrix[1, 2] = S_k_matrix[2, 1] = -q["theta_k_squared"] * q["sigma_kplus1"] / 2
S_k_matrix -= state_map.T * Q_kplus1 * state_map
w_k_Q_k_w_k = (
    q["Q_k_11"] * (z_kplus1 - x_star) ** 2 + q["Q_k_22"] * (y_kplus1 - z_kplus1) ** 2
)
w_kplus1_Q_kplus1_w_kplus1 = (
    q["Q_kplus1_11"]
    * (z_kplus1 - x_star - q["theta_kplus1"] * (grad_x_kplus1 / L)) ** 2
    + q["Q_kplus1_22"]
    * (
        q["theta_k_squared"] * (y_kplus1 - z_kplus1) / q["theta_kplus1"] ** 2
        + (q["theta_kplus1"] - 1) * (grad_x_kplus1 / L)
    )
    ** 2
)
s_k = z_kplus1 - x_star - (q["D_0"] - q["theta_k_squared"]) * (grad_x_kplus1 / L)
t_k = grad_x_kplus1 / L - (y_kplus1 - z_kplus1) / (q["theta_k_squared"] - q["C_kplus1"])
S_k_over_L = q["alpha_k"] * s_k**2 + q["beta_k"] * t_k**2
s_factor = sp.Matrix([1, 0, -(q["D_0"] - q["theta_k_squared"])])
t_factor = sp.Matrix([0, -1 / (q["theta_k_squared"] - q["C_kplus1"]), 1])
matrix_residuals = {}
for i in range(3):
    for j in range(i, 3):
        matrix_residuals[i, j] = sp.cancel(
            (
                S_k_matrix
                - q["alpha_k"] * s_factor * s_factor.T
                - q["beta_k"] * t_factor * t_factor.T
            )[i, j].subs(step_substitution, simultaneous=True)
        )
assert all((value == 0 for value in matrix_residuals.values()))
assert S_k_matrix[0, 1] == 0
print(
    "Six independent slack-matrix entries and zero mixed state coefficient: exact zeros"
)
quadratic_residuals = exact_polynomial(
    "Paper slack-matrix factorization in actual coordinates",
    (
        sp.Matrix([z_kplus1 - x_star, y_kplus1 - z_kplus1, grad_x_kplus1 / L]).T
        * S_k_matrix
        * sp.Matrix([z_kplus1 - x_star, y_kplus1 - z_kplus1, grad_x_kplus1 / L])
    )[0]
    - S_k_over_L,
    (z_kplus1, y_kplus1, x_star, grad_x_kplus1),
    step_substitution,
)
x_k = y_kplus1 + grad_x_k / L
x_kplus1 = (
    z_kplus1 + q["theta_k_squared"] * (y_kplus1 - z_kplus1) / q["theta_kplus1"] ** 2
)
weighted_I_f = q["a_k"] * I_f(
    f_x_k, f_x_kplus1, x_k, x_kplus1, grad_x_k, grad_x_kplus1, L
) + q["b_kplus1"] * I_f(0, f_x_kplus1, x_star, x_kplus1, 0, grad_x_kplus1, L)
expanded_interpolation = (
    q["a_kplus1"] * f_x_kplus1
    - q["a_k"] * f_x_k
    + L * q["a_k"] * (grad_x_k / L) ** 2 / 2
    + L * q["a_kplus1"] * (grad_x_kplus1 / L) ** 2 / 2
    - L
    * (grad_x_kplus1 / L)
    * (
        q["b_kplus1"] * (z_kplus1 - x_star)
        + q["theta_k_squared"] * q["sigma_kplus1"] * (y_kplus1 - z_kplus1)
    )
)
V_k = (
    q["a_k"] * f_x_k
    - L * x_0_minus_x_star**2 / (2 * q["D_0"])
    - L * q["a_k"] * (grad_x_k / L) ** 2 / 2
    + L * w_k_Q_k_w_k
)
V_kplus1 = (
    q["a_kplus1"] * f_x_kplus1
    - L * x_0_minus_x_star**2 / (2 * q["D_0"])
    - L * q["a_kplus1"] * (grad_x_kplus1 / L) ** 2 / 2
    + L * w_kplus1_Q_kplus1_w_kplus1
)
step_variables = (
    z_kplus1,
    y_kplus1,
    x_star,
    grad_x_k,
    grad_x_kplus1,
    x_0_minus_x_star,
    f_x_k,
    f_x_kplus1,
)
interpolation_residuals = exact_polynomial(
    "Interior interpolation expansion",
    weighted_I_f - expanded_interpolation,
    step_variables,
    step_substitution,
)
bridge_residuals = exact_polynomial(
    "Interpolation minus potential difference  = L [w_k; gradient/L]^T S_k [w_k; gradient/L]",
    weighted_I_f
    - (V_kplus1 - V_k)
    - L
    * (
        sp.Matrix([z_kplus1 - x_star, y_kplus1 - z_kplus1, grad_x_kplus1 / L]).T
        * S_k_matrix
        * sp.Matrix([z_kplus1 - x_star, y_kplus1 - z_kplus1, grad_x_kplus1 / L])
    )[0],
    step_variables,
    step_substitution,
)
complete_decrement = (V_kplus1 - V_k - weighted_I_f + L * S_k_over_L).subs(
    q["b_kplus1"], q["a_kplus1"] - q["a_k"]
)
decrement_residuals = exact_polynomial(
    "Complete interior decrement", complete_decrement, step_variables, step_substitution
)
```

## Base case and boundary symbolic verification

For $N\ge2$, $x_1=y_1=z_1=x_0-\nabla f(x_0)/L$.
Code Cell 13 verifies the first line of Eq. (34) using the
literal interior definition of $V_1$ and the base square $S_0$.
The only algebraic theta constraint is $\theta_1^2-\theta_1-1=0$;
polynomial remainders enforce it exactly. The undefined matrix $\mathbf Q_0$ is never evaluated. In particular, this proof does not divide by the
zero gap $\theta_0^2-C_1$.

```python
# Raw gradients; centered coordinates x_star=0. In N=1 grad_1 is at y_1;
# in N=2 grad_1 is at x_1 and grad_2 is at y_2.
theta_1 = sp.Symbol("theta_1", real=True)
first_relation = theta_1**2 - theta_1 - 1


def reduce_first(value):
    numerator, denominator = sp.fraction(sp.cancel(sp.together(value)))
    denominator_remainder = sp.rem(denominator, first_relation, theta_1)
    assert denominator_remainder != 0, "A forbidden zero denominator was evaluated"
    return sp.cancel(sp.rem(numerator, first_relation, theta_1) / denominator_remainder)


first_keys = (
    "theta_kplus1",
    "D_0",
    "a_k",
    "a_kplus1",
    "b_kplus1",
    "Q_kplus1_11",
    "Q_kplus1_22",
    "alpha_k",
)
first_generic = paper_coefficients(
    theta_1, sp.Integer(0), theta_Nminus1_formal, rho_Nminus1_formal
)
first_values = {name: reduce_first(first_generic[name]) for name in first_keys}
q, first_substitution = formal_coefficients(first_values)
grad_x_0, grad_x_1, f_x_0, f_x_1 = sp.symbols(
    "grad_x_0 grad_x_1 f_x_0 f_x_1", real=True
)
x_0, x_1 = (x_0_minus_x_star, x_0_minus_x_star - grad_x_0 / L)
V_0 = sp.Integer(0)
V_1 = (
    q["a_kplus1"] * f_x_1
    - L * x_0_minus_x_star**2 / (2 * q["D_0"])
    - L * q["a_kplus1"] * (grad_x_1 / L) ** 2 / 2
    + L
    * (
        q["Q_kplus1_11"] * (x_1 - q["theta_kplus1"] * (grad_x_1 / L)) ** 2
        + q["Q_kplus1_22"] * ((q["theta_kplus1"] - 1) * (grad_x_1 / L)) ** 2
    )
)
a_0 = 1 / (q["D_0"] - 1)
weighted_base = (
    a_0 * I_f(0, f_x_0, 0, x_0, 0, grad_x_0, L)
    + a_0 * I_f(f_x_0, f_x_1, x_0, x_1, grad_x_0, grad_x_1, L)
    + q["b_kplus1"] * I_f(0, f_x_1, 0, x_1, 0, grad_x_1, L)
)
S_0 = (
    L
    * q["D_0"]
    / (2 * (q["D_0"] - 1) ** 2)
    * (grad_x_0 / L - x_0_minus_x_star / q["D_0"]) ** 2
    + L * q["alpha_k"] * (x_1 - (q["D_0"] - 1) * (grad_x_1 / L)) ** 2
)
assert reduce_first(first_values["a_k"] - 1 / (first_values["D_0"] - 1)) == 0
base_residuals = exact_polynomial(
    "Initial decrement (N >= 2)",
    V_1 - V_0 - weighted_base + S_0,
    (x_0_minus_x_star, grad_x_0, grad_x_1, f_x_0, f_x_1),
    first_substitution,
    reduce_first,
)
```

### Terminal identity symbolic verification

For $N\ge2$, $x_{N-1}=y_N+\nabla f(x_{N-1})/L$.
Code Cell 14 checks the terminal specialization of $\mathbf Q_k,a_k$ and every coefficient of
$$\boldsymbol w_{N-1}^{\top}\mathbf Q_{N-1}\boldsymbol w_{N-1}
+\frac{\|\nabla f(y_N)\|^2}{2L^2}
-\frac{b_N}{L}\langle\nabla f(y_N),y_N-x_\star\rangle
=\alpha_{N-1}\|s_{N-1}\|^2+\beta_{N-1}\|t_{N-1}\|^2=S_{N-1}/L.$$
It then checks the interpolation bridge and complete terminal decrement.
The terminal potential is $f(y_N)-f(x_\star)-\tau_NL\|x_0-x_\star\|^2$; the positive gap
$\theta_{N-1}^2-C_N$ is used only for $N\ge2$.


```python
# Code Cell 14: terminal gradients are at x_(N-1) and y_N; L remains a positive parameter.
theta_Nminus1_squared, rho_Nminus1 = sp.symbols(
    "theta_Nminus1_squared rho_Nminus1", real=True
)
C_N = (theta_Nminus1_squared + rho_Nminus1 + 1) / 2
D_0 = 2 * C_N + 1
b_N = 1 / (C_N + 1)
a_Nminus1 = (
    theta_Nminus1_squared
    * (theta_Nminus1_squared - theta_Nminus1_squared + C_N)
    / (theta_Nminus1_squared * (D_0 - C_N))
)
Q_Nminus1_11 = (1 - a_Nminus1) / (2 * (D_0 - theta_Nminus1_squared))
Q_Nminus1_22 = (theta_Nminus1_squared / theta_Nminus1_squared - a_Nminus1) / (
    2 * (theta_Nminus1_squared - C_N)
)
alpha_Nminus1, beta_Nminus1 = (
    b_N / (2 * (D_0 - theta_Nminus1_squared)),
    b_N * (theta_Nminus1_squared - C_N) / 2,
)
assert sp.cancel(a_Nminus1 - (1 - b_N)) == 0
assert sp.cancel(Q_Nminus1_11 - alpha_Nminus1) == 0
assert sp.cancel(Q_Nminus1_22 - b_N / (2 * (theta_Nminus1_squared - C_N))) == 0
assert sp.cancel(b_N * (C_N + 1) - 1) == 0
z_N, y_N, grad_x_Nminus1, grad_y_N = sp.symbols(
    "z_N y_N grad_x_Nminus1 grad_y_N", real=True
)
f_x_Nminus1, f_y_N = sp.symbols("f_x_Nminus1 f_y_N", real=True)
w_Nminus1_Q_Nminus1_w_Nminus1 = (
    Q_Nminus1_11 * (z_N - x_star) ** 2 + Q_Nminus1_22 * (y_N - z_N) ** 2
)
terminal_quadratic = (
    w_Nminus1_Q_Nminus1_w_Nminus1
    + (grad_y_N / L) ** 2 / 2
    - b_N * (grad_y_N / L) * (y_N - x_star)
)
s_Nminus1 = z_N - x_star - (D_0 - theta_Nminus1_squared) * (grad_y_N / L)
t_Nminus1 = grad_y_N / L - (y_N - z_N) / (theta_Nminus1_squared - C_N)
S_Nminus1_over_L = alpha_Nminus1 * s_Nminus1**2 + beta_Nminus1 * t_Nminus1**2
terminal_factorization_residuals = exact_polynomial(
    "Terminal quadratic factorization",
    terminal_quadratic - S_Nminus1_over_L,
    (z_N, y_N, x_star, grad_y_N),
)
x_Nminus1 = y_N + grad_x_Nminus1 / L
weighted_I_f_Nminus1 = a_Nminus1 * I_f(
    f_x_Nminus1, f_y_N, x_Nminus1, y_N, grad_x_Nminus1, grad_y_N, L
) + b_N * I_f(0, f_y_N, x_star, y_N, 0, grad_y_N, L)
V_Nminus1 = (
    a_Nminus1 * f_x_Nminus1
    - L * x_0_minus_x_star**2 / (2 * D_0)
    - L * a_Nminus1 * (grad_x_Nminus1 / L) ** 2 / 2
    + L * w_Nminus1_Q_Nminus1_w_Nminus1
)
V_N = f_y_N - L * x_0_minus_x_star**2 / (2 * D_0)
terminal_variables = (
    z_N,
    y_N,
    x_star,
    grad_x_Nminus1,
    grad_y_N,
    x_0_minus_x_star,
    f_x_Nminus1,
    f_y_N,
)
terminal_bridge_residuals = exact_polynomial(
    "Terminal interpolation-to-quadratic-form bridge",
    weighted_I_f_Nminus1 - (V_N - V_Nminus1) - L * terminal_quadratic,
    terminal_variables,
)
terminal_decrement_residuals = exact_polynomial(
    "Complete terminal decrement (N >= 2)",
    V_N - V_Nminus1 - weighted_I_f_Nminus1 + L * S_Nminus1_over_L,
    terminal_variables,
)
```

### Single-iteration endpoint and two-iteration boundary

Code Cell 15 verifies the literal exceptional identity
$$f(y_1)-f(x_\star)-\frac L6\|x_0-x_\star\|^2
=\frac12(I_f(x_\star,x_0)+I_f(x_0,y_1)+I_f(x_\star,y_1))
-\frac{3L}{8}\|\nabla f(x_0)/L-(x_0-x_\star)/3\|^2-\frac L8\|x_0-x_\star-\nabla f(x_0)/L-2\nabla f(y_1)/L\|^2.$$
For $N=2$ it builds the iterates and coefficients directly and checks both
actual transitions. This additionally checks compatibility of the base and
terminal formulas when no generic interior transition exists.
Finally it checks the endpoint normalization
$f(y_N)-f(x_\star)-\tau_NLR^2=V_N+\tau_NL(\|x_0-x_\star\|^2-R^2)$.
The short-horizon checks supplement the symbolic generic proofs and do not
replace the all-horizon positivity argument.

```python
# Raw gradients; centered coordinates x_star=0. In N=1 grad_1 is at y_1;
# in N=2 grad_1 is at x_1 and grad_2 is at y_2.
grad_x_0, grad_y_1, grad_y_2, f_x_0, f_y_1, f_y_2 = sp.symbols(
    "grad_x_0 grad_y_1 grad_y_2 f_x_0 f_y_1 f_y_2", real=True
)
x_0, y_1 = (x_0_minus_x_star, x_0_minus_x_star - grad_x_0 / L)
I_f_xstar_x0 = I_f(0, f_x_0, 0, x_0, 0, grad_x_0, L)
I_f_x0_y1 = I_f(f_x_0, f_y_1, x_0, y_1, grad_x_0, grad_y_1, L)
I_f_xstar_y1 = I_f(0, f_y_1, 0, y_1, 0, grad_y_1, L)
V_1_N1 = f_y_1 - L * x_0_minus_x_star**2 / 6
S_0_N1 = (
    3 * L * (grad_x_0 / L - x_0_minus_x_star / 3) ** 2 / 8
    + L * (x_0_minus_x_star - grad_x_0 / L - 2 * (grad_y_1 / L)) ** 2 / 8
)
one_residuals = exact_polynomial(
    "N=1 combined base-terminal identity",
    V_1_N1 - (I_f_xstar_x0 + I_f_x0_y1 + I_f_xstar_y1) / 2 + S_0_N1,
    (x_0_minus_x_star, grad_x_0, grad_y_1, f_x_0, f_y_1),
)
theta_1_N2 = theta_1
theta_1_squared_N2 = theta_1_N2**2
C_2_N2 = (1 + theta_1_N2**3) / theta_1_squared_N2
D_0_N2 = 2 * C_2_N2 + 1
a_0_N2 = 1 / (D_0_N2 - 1)
a_1_N2 = C_2_N2 / (C_2_N2 + 1)
b_1_N2, b_2_N2 = (a_1_N2 - a_0_N2, 1 - a_1_N2)
alpha_0_N2 = (
    (D_0_N2 * b_1_N2 - theta_1_N2) * (D_0_N2 - theta_1_squared_N2)
    + (D_0_N2 * a_1_N2 - theta_1_squared_N2) * theta_1_N2
) / (2 * D_0_N2 * (D_0_N2 - theta_1_squared_N2) * (D_0_N2 - 1))
Q_1_11_N2 = (1 - a_1_N2) / (2 * (D_0_N2 - theta_1_squared_N2))
Q_1_22_N2 = (theta_1_squared_N2 / theta_1_squared_N2 - a_1_N2) / (
    2 * (theta_1_squared_N2 - C_2_N2)
)
alpha_1_N2 = b_2_N2 / (2 * (D_0_N2 - theta_1_squared_N2))
beta_1_N2 = b_2_N2 * (theta_1_squared_N2 - C_2_N2) / 2
x_1 = y_1
z_2, y_2 = (y_1 - theta_1_N2 * (grad_y_1 / L), x_1 - grad_y_1 / L)
V_0_N2 = sp.Integer(0)
V_1_N2 = (
    a_1_N2 * f_y_1
    - L * x_0_minus_x_star**2 / (2 * D_0_N2)
    - L * a_1_N2 * (grad_y_1 / L) ** 2 / 2
    + L * (Q_1_11_N2 * z_2**2 + Q_1_22_N2 * (y_2 - z_2) ** 2)
)
V_2_N2 = f_y_2 - L * x_0_minus_x_star**2 / (2 * D_0_N2)
S_0_N2 = (
    L
    * D_0_N2
    / (2 * (D_0_N2 - 1) ** 2)
    * (grad_x_0 / L - x_0_minus_x_star / D_0_N2) ** 2
    + L * alpha_0_N2 * (y_1 - (D_0_N2 - 1) * (grad_y_1 / L)) ** 2
)
S_1_N2 = (
    L * alpha_1_N2 * (z_2 - (D_0_N2 - theta_1_squared_N2) * (grad_y_2 / L)) ** 2
    + L * beta_1_N2 * (grad_y_2 / L - (y_2 - z_2) / (theta_1_squared_N2 - C_2_N2)) ** 2
)
weighted_I_f_0_N2 = a_0_N2 * (I_f_xstar_x0 + I_f_x0_y1) + b_1_N2 * I_f_xstar_y1
weighted_I_f_1_N2 = a_1_N2 * I_f(
    f_y_1, f_y_2, x_1, y_2, grad_y_1, grad_y_2, L
) + b_2_N2 * I_f(0, f_y_2, 0, y_2, 0, grad_y_2, L)
two_variables = (x_0_minus_x_star, grad_x_0, grad_y_1, grad_y_2, f_x_0, f_y_1, f_y_2)
two_base_residuals = exact_polynomial(
    "N=2 initial transition",
    V_1_N2 - V_0_N2 - weighted_I_f_0_N2 + S_0_N2,
    two_variables,
    reducer=reduce_first,
)
two_terminal_residuals = exact_polynomial(
    "N=2 terminal transition",
    V_2_N2 - V_1_N2 - weighted_I_f_1_N2 + S_1_N2,
    two_variables,
    reducer=reduce_first,
)
R, tau_N, f_y_N = sp.symbols("R tau_N f_y_N", real=True)
V_N = f_y_N - tau_N * L * x_0_minus_x_star**2
assert (
    sp.expand(f_y_N - tau_N * L * R**2 - V_N - tau_N * L * (x_0_minus_x_star**2 - R**2))
    == 0
)
print("Endpoint/radius normalization: exact zero; radius inequality is analytic")
```

Every exact residual above is checked coefficient by coefficient, including
function values and constants. The analytic domain argument proves the
interpolation weights and used square weights positive and every denominator
well-defined. Therefore $V_N\le\cdots\le V_0=0$, and
$$f(y_N)-f(x_\star)=V_N+\tau_NL\|x_0-x_\star\|^2
\le\tau_NLR^2=\frac{LR^2}{2(2C_N+1)}.$$
Code Cells 10–15 constitute the selected exact proof run; the saved numerical
experiments in Code Cells 1–9 are separate diagnostics.
