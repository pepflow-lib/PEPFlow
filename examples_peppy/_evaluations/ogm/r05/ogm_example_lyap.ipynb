# OGM: numerical discovery and Lyapunov workflow

Let $f$ be convex and $L$-smooth on a finite-dimensional real Euclidean space, with $L>0$, $R\geq0$, $N\geq1$, $\nabla f(x_\star)=0$ and $\|x_0-x_\star\|^2\leq R^2$. Set $z_0=x_0$. The target is $f(x_N)-f(x_\star)$.

With $\theta_{-1}=0$, define
$$\theta_k=\frac{1+\sqrt{1+4\theta_{k-1}^2}}2\quad(0\leq k<N),\qquad
\theta_N=\frac{1+\sqrt{1+8\theta_{N-1}^2}}2.$$
For $k=0,\ldots,N-1$, perform
$$y_{k+1}=x_k-\frac1L\nabla f(x_k),\qquad
z_{k+1}=z_k-\frac{2\theta_k}{L}\nabla f(x_k),$$
$$x_{k+1}=\left(1-\frac1{\theta_{k+1}}\right)y_{k+1}
+\frac1{\theta_{k+1}}z_{k+1}.$$
The terminal coefficient uses the factor $8$, including when $N=1$.

## Proof Statement

### Theorem
Under the stated smooth convex assumptions and OGM updates, let $g_{k}=\nabla f(x_{k})$, $d=x_{0}-x_{\star}$ and $T=\theta_{N}$. Define $V_{0}=0$, and, for $1\leq k<N$,
$$V_{k}=\frac{1}{T^2}\left[2\theta_{k}^2(f(x_{k})-f(x_{\star}))
-\frac{\theta_{k}^2}{L}\|g_{k}\|^2+\frac L2\|z_{k+1}-x_{\star}\|^2-\frac L2\|d\|^2\right],$$
with the separate terminal value
$$V_{N}=f(x_{N})-f(x_{\star})-\frac{L}{2T^2}\|d\|^2.$$
For every integer $N\geq1$, this sequence is nonincreasing. Consequently
$$\boxed{f(x_{N})-f(x_{\star})\leq\frac{LR^2}{2\theta_{N}^2}.}$$
The constant is sharp: the one-dimensional function $f(x)=Lx^2/2$, with $x_{\star}=0$ and $x_{0}=R$, attains equality. For $N=1$, the interior range is empty and $\theta_{1}=2$; the bound is $LR^2/8$.

### Proof outline
For any evaluated points $a,b$, smooth convex interpolation gives
$$I_{a,b}=f(b)-f(a)-\langle g_b,b-a\rangle+\frac1{2L}\|g_b-g_a\|^2\leq0,$$
where $g_{x_{\star}}=0$. Let $B=L\|d\|^2/2$ and, for $0\leq k<N$,
$$H_{k}=2\theta_{k}^2(f(x_{k})-f(x_{\star}))-\frac{\theta_{k}^2}{L}\|g_{k}\|^2+\frac L2\|z_{k+1}-x_{\star}\|^2.$$
Thus $V_{k}=(H_{k}-B)/T^2$ for $1\leq k<N$, $V_{0}=0$, and $V_{N}=f(x_{N})-f(x_{\star})-B/T^2$.

Since $\theta_{0}=1$ and $z_{1}=x_{0}-2g_{0}/L$, direct expansion gives
$$H_{0}-B=2I_{x_{\star},x_{0}}.$$
For $0\leq k\leq N-2$, the ordinary recurrence $\theta_{k+1}^2-\theta_{k+1}=\theta_{k}^2$ and the algorithm updates give the exact identity
$$H_{k+1}-H_{k}=2\theta_{k+1}I_{x_{\star},x_{k+1}}+2\theta_{k}^2I_{x_{k},x_{k+1}}.$$
At the terminal step, the distinct relation $T^2-T=2\theta_{N-1}^2$ gives
$$T^2(f(x_{N})-f(x_{\star}))-H_{N-1}
=T I_{x_{\star},x_{N}}+2\theta_{N-1}^2 I_{x_{N-1},x_{N}}
-\frac L2\left\|z_{N}-x_{\star}-\frac TL g_{N}\right\|^2.$$
These are polynomial identities in inner products and function values, so their exact coordinate verifications below apply in any finite dimension.

For $N\geq2$, the corresponding actual sequence starts with
$$V_{1}-V_{0}=\frac{2I_{x_{\star},x_{0}}+2\theta_{1}I_{x_{\star},x_{1}}+2I_{x_{0},x_{1}}}{T^2}\leq0.$$
For $1\leq k\leq N-2$ its increment is
$$V_{k+1}-V_{k}=\frac{2\theta_{k+1}I_{x_{\star},x_{k+1}}+2\theta_{k}^2I_{x_{k},x_{k+1}}}{T^2}\leq0,$$
and its last increment is
$$V_{N}-V_{N-1}=\frac{T I_{x_{\star},x_{N}}+2\theta_{N-1}^2 I_{x_{N-1},x_{N}}
-\frac L2\|z_{N}-x_{\star}-(T/L)g_{N}\|^2}{T^2}\leq0.$$
For $N=1$ the direct identity is
$$V_{1}=\tfrac12(I_{x_{\star},x_{0}}+I_{x_{\star},x_{1}}+I_{x_{0},x_{1}})
-\frac L8\|z_{1}-x_{\star}-(2/L)g_{1}\|^2\leq0.$$
Every theta is positive by its positive-root definition, $L>0$, and all squared norms are nonnegative. Hence every displayed increment has the asserted sign. Now $V_{N}\leq V_{0}=0$ implies
$$f(x_{N})-f(x_{\star})\leq\frac{L\|d\|^2}{2T^2}\leq\frac{LR^2}{2T^2}.$$
No division by $R$ is used, so the argument includes $R=0$.

For sharpness, $f(x)=Lx^2/2$ has $y_{k+1}=0$. Induction from $\theta_{0}=1$, $x_{0}=z_{0}=R$ gives $z_{k}=(-1)^kR$ and $x_{k}=(-1)^kR/\theta_{k}$, using the special terminal coefficient in the last update. Thus $f(x_{N})=LR^2/(2T^2)$.

## Function and algorithm definitions
The finite-horizon solver uses floating-point evaluations of the supplied theta recurrence. The initial condition uses $R^2$, with $R$ denoting the radius.

```python
L = pf.Parameter("L")
R = pf.Parameter("R")
f = pf.SmoothConvexFunction(is_basis=True, tags=["f"], L=L)


def theta_sequence(N):
    """Numerical coefficients; preserve the special terminal recurrence."""
    N = int(N)
    if N < 1:
        raise ValueError("N must be positive")
    previous = 0.0
    theta = []
    for _ in range(N):
        previous = (1.0 + math.sqrt(1.0 + 4.0 * previous**2)) / 2.0
        theta.append(previous)
    theta.append((1.0 + math.sqrt(1.0 + 8.0 * previous**2)) / 2.0)
    return theta
```

```python
def make_ctx_ogm(ctx_name, N):
    """Encode all iterates, with z_0=x_0 and grad f(x_star)=0."""
    N = int(N)
    theta = theta_sequence(N)
    ctx = pf.PEPContext(ctx_name).set_as_current()
    x = pf.Vector(is_basis=True, tags=["x_0"])
    z = x
    z.add_tag("z_0")
    f.set_stationary_point("x_star")
    for k in range(N):
        g = f.grad(x)
        g.add_tag(f"g_{k}")
        y = x - g / L
        y.add_tag(f"y_{k + 1}")
        z = z - (2 * theta[k] / L) * g
        z.add_tag(f"z_{k + 1}")
        x = (1 - 1 / theta[k + 1]) * y + z / theta[k + 1]
        x.add_tag(f"x_{k + 1}")
    f.grad(x).add_tag(f"g_{N}")
    return ctx


def get_pep_setup(N, params):
    """Build the normalized PEP; params resolve L and R at solve time."""
    N = int(N)
    ctx = make_ctx_ogm(f"ctx_{N}", N)
    pb = pf.PEPBuilder(ctx)
    pb.add_initial_constraint(
        ((ctx["x_0"] - ctx["x_star"]) ** 2).le(R**2, name="initial_condition")
    )
    pb.set_performance_metric(f(ctx[f"x_{N}"]) - f(ctx["x_star"]))
    return ctx, pb, f
```

### Initial closed-form candidates
The stationary-point multipliers divided by $\theta_i$, and consecutive multipliers divided by $\theta_i^2$, have the same numerical normalization. This suggests
$$\lambda_{\star,i}=\frac{2\theta_i}{\theta_N^2}\ (i<N),\quad
\lambda_{\star,N}=\frac1{\theta_N},\quad
\lambda_{i,i+1}=\frac{2\theta_i^2}{\theta_N^2},$$
with other interpolation multipliers zero. The nearly rank-one Gram slack suggests
$$S(G)=\frac{L}{2\theta_N^2}\left\|z_N-x_\star-\frac{\theta_N}{L}g_N\right\|^2.$$
These formulas are numerical candidates. Their all-horizon identities and domain conditions have not yet been established. No target-specific reference source was consulted.

```python
def lamb(ri, ci, N, theta):
    def index(tag):
        if tag == "x_star":
            return -1
        if tag in ("z_0", "x_0"):
            return 0
        return int(tag.split("_")[1])

    i, j = index(ri), index(ci)
    if i == -1 and 0 <= j < N:
        return 2 * theta[j] / theta[N] ** 2
    if i == -1 and j == N:
        return 1 / theta[N]
    if 0 <= i < N and j == i + 1:
        return 2 * theta[i] ** 2 / theta[N] ** 2
    return 0


def S_guess(ctx, obj, N, theta, L):
    w = ctx[f"z_{N}"] - ctx["x_star"] - (theta[N] / L) * obj.grad(ctx[f"x_{N}"])
    return (L / (2 * theta[N] ** 2)) * (w**2)
```

```python
theta = theta_sequence(N)
labels = b2["lambda_row_names"]
candidate_lambda = np.array(
    [[lamb(ri, ci, N, theta) for ci in b2["lambda_col_names"]] for ri in labels]
)
observed_lambda = np.asarray(b2["lambda_matrix"], dtype=float)
# Dropped interpolation entries have display NaNs; the complete named map records None.
active = np.isfinite(observed_lambda)
assert np.allclose(candidate_lambda[active], observed_lambda[active], atol=1e-4, rtol=0)
for i, ri in enumerate(labels):
    for j, ci in enumerate(b2["lambda_col_names"]):
        if f"f:{ri},{ci}" in b2["relaxed_constraints"]:
            assert candidate_lambda[i, j] == 0
print(
    "Maximum active lambda formula discrepancy:",
    np.max(np.abs(candidate_lambda[active] - observed_lambda[active])),
)
pf.pprint_labeled_matrix(candidate_lambda, labels, b2["lambda_col_names"], precision=6)
```

```python
def inventory(ctx, pb):
    entries = [(c, "initial", None, None) for c in pb.init_conditions]
    assert not ctx.oper_to_duplets, "This OGM case has no operators"
    for owner in ctx.func_to_triplets:
        groups = ctx.get_constraint_data(owner)
        assert not groups.psd_dict, "This OGM case has only scalar interpolation"
        entries += [
            (c, "interpolation", str(owner), str(group))
            for group, cs in groups.sc_dict.items()
            for c in cs
        ]
    assert len({c.name for c, *_ in entries}) == len(entries)
    return entries


def scalar_coordinates(pm, scalar):
    value = pm.eval_scalar(pf.Scalar.zero() + scalar)
    return {
        "gram": np.asarray(value.inner_prod_coords, dtype=float).tolist(),
        "func": np.asarray(value.func_coords, dtype=float).tolist(),
        "offset": float(value.offset),
    }


def verify_certificate(ctx, pb, data, dense_value):
    """Check all coordinates and cone signs without another solve."""
    pm = pf.ExpressionManager(ctx, resolve_parameters=json.loads(data["params_json"]))
    basis = ctx.basis_vectors()
    assert [str(v) for v in basis] == data["basis_vectors"]
    assert [str(v) for v in ctx.basis_scalars()] == data["basis_scalars"]
    entries = inventory(ctx, pb)
    assert (
        {c.name for c, *_ in entries}
        == set(data["named_dual_values"])
        == set(data["constraint_metadata"])
    )
    assert set(pb.relaxed_constraints) == set(data["relaxed_constraints"])
    weighted = pf.Scalar.zero()
    duals = []
    for c, origin, owner, group in entries:
        meta = data["constraint_metadata"][c.name]
        assert meta == dict(
            origin=origin,
            owner=owner,
            group=group,
            kind="scalar",
            comparator=c.cmp.value,
            shape=[],
            relaxed=c.name in pb.relaxed_constraints,
        )
        value = data["named_dual_values"][c.name]
        if meta["relaxed"]:
            assert value is None
            continue
        assert value is not None and np.isfinite(value)
        residual = c.lhs - c.rhs
        if c.cmp in (pf.Comparator.GE, pf.Comparator.SEQ):
            residual = -residual
        weighted += value * (pf.Scalar.zero() + residual)
        if c.cmp != pf.Comparator.EQ:
            duals.append(value)
    S = np.asarray(data["S_matrix"], dtype=float)
    D = np.asarray(data["numerical_decomposition"]["D"], dtype=float)
    E = np.asarray(data["numerical_decomposition"]["ell_coords"], dtype=float)
    S_numeric = sum(
        (float(S[i, j]) * basis[i] * basis[j] for i, j in np.ndindex(S.shape)),
        pf.Scalar.zero(),
    )
    residual = scalar_coordinates(
        pm, pb.performance_metric - data["opt_value"] - weighted + S_numeric
    )

    def max_abs(a):
        return float(np.max(np.abs(a), initial=0))

    checks = dict(
        identity_gram_max_abs=max_abs(residual["gram"]),
        identity_func_max_abs=max_abs(residual["func"]),
        identity_offset_abs=abs(residual["offset"]),
        S_reconstruction_max_abs=max_abs(E.T @ D @ E - S),
        dense_relaxed_gap=abs(dense_value - data["opt_value"]),
        minimum_scalar_inequality_dual=min(duals, default=0),
        minimum_S_eigenvalue=float(np.linalg.eigvalsh(S).min()),
        minimum_matrix_inequality_eigenvalue=0.0,
    )
    tolerances = dict(
        identity_atol=1e-5,
        identity_rtol=0.0,
        reconstruction_atol=1e-4,
        reconstruction_rtol=1e-5,
        dual_tol=1e-6,
        psd_tol=1e-6,
        dense_opt_atol=1e-5,
    )
    passed = (
        data["certificate_provenance"]["solver_status"] == "optimal"
        and np.all(np.isfinite(S))
        and np.allclose(S, S.T, atol=1e-6, rtol=0)
        and all(
            checks[k] <= 1e-5
            for k in (
                "identity_gram_max_abs",
                "identity_func_max_abs",
                "identity_offset_abs",
            )
        )
        and np.allclose(E.T @ D @ E, S, atol=1e-4, rtol=1e-5)
        and checks["dense_relaxed_gap"] < 1e-5
        and checks["minimum_scalar_inequality_dual"] >= -1e-6
        and checks["minimum_S_eigenvalue"] >= -1e-6
    )
    return {
        "tolerances": tolerances,
        "residuals": checks,
        "passed": bool(passed),
    }, residual
```

### B2 conclusion
All selected inequality multipliers and $S$ satisfy the numerical cone checks. Normalizing interpolation and initial constraints to $R_c\leq0$, the verified identity is
$$f(x_N)-f(x_\star)-\operatorname{opt}-\sum_c d_c R_c+S(G)=0.$$
The identity is checked in Gram, function-value and constant coordinates. Since $d_c\geq0$ and $S(G)\geq0$, this gives the bound numerically at the selected sample. The candidate formulas match this certificate, while the complete numerical LDL factors remain authoritative for subsequent replay. Numerical certificate: verified. ## Lyapunov partial sums and rank profile
Write $I_{i,j}\leq0$ for the smooth convex interpolation residual, and use the selected numerical multipliers. Set $V_0=0$ and, for $1\leq k\leq N$,
$$V_k=\sum_{i=0}^{k}\lambda_{\star,i}I_{\star,i}
+\sum_{i=0}^{k-1}\lambda_{i,i+1}I_{i,i+1}-\mathbf1_{k=N}S(G).$$
Thus $V_1-V_0$ includes the initial $I_{\star,0}$ contribution; later increments add the next stationary-point and consecutive interpolation terms. The final increment additionally subtracts the complete PSD slack. Every increment is nonpositive under the numerical cone checks. The numerical rank-one candidate for $S$ remains informative, but the grouping retains the entire saved PSD matrix, including its tiny components.

The interior Gram rank is three at both $N=4$ and $N=6$, excluding exactly $V_0,V_N$. These exclusions affect the rank comparison only. The endpoint terms and every constraint remain included in the reconstruction and coverage obligations.

```python
from runtime import NOTEBOOK_DIR  # ty: ignore[unresolved-import]  # Notebook bootstrap adds the helper directory to sys.path.


def rebuild(certificate):
    from ogm_setup import get_pep_setup  # ty: ignore[unresolved-import]  # Imported from the notebook working directory.

    assert certificate["closed_form_search"]["completed"]
    assert digest_state(certificate) == certificate["certificate_id"]
    for relative, expected in certificate["certificate_provenance"][
        "source_hashes"
    ].items():
        assert (
            hashlib.sha256((NOTEBOOK_DIR / relative).read_bytes()).hexdigest()
            == expected
        )
    N = certificate["N_verify"]
    params = json.loads(certificate["params_json"])
    ctx, pb, obj = get_pep_setup(N, params)
    pb.set_relaxed_constraints(certificate["relaxed_constraints"])
    checks, _ = verify_certificate(ctx, pb, certificate, certificate["opt_value"])
    assert checks["passed"]
    assert (
        certificate["S_row_names"]
        == certificate["S_col_names"]
        == certificate["basis_vectors"]
    )
    assert certificate["normalization"]["bound_value"] == certificate["opt_value"]
    assert (
        certificate["named_dual_values"]["initial_condition"] == certificate["tau_sol"]
    )
    return ctx, pb, obj, pf.ExpressionManager(ctx, resolve_parameters=params)


def construct(certificate, verbose=True):
    ctx, pb, obj, pm = rebuild(certificate)
    N = certificate["N_verify"]
    basis = ctx.basis_vectors()
    group_terms, term_assignment = {}, {}
    for c, origin, owner, group in inventory(ctx, pb):
        value = certificate["named_dual_values"][c.name]
        if origin == "initial" or value is None:
            continue
        residual = c.lhs - c.rhs
        if c.cmp in (pf.Comparator.GE, pf.Comparator.SEQ):
            residual = -residual
        group_terms[c.name] = value * (pf.Scalar.zero() + residual)
        left, right = c.name.split(":")[1].split(",")
        j = 0 if right == "z_0" else int(right.split("_")[1])
        assert left == "x_star" or left == ("z_0" if j == 1 else f"x_{j - 1}")
        term_assignment[c.name] = j - 1
    S = np.asarray(certificate["S_matrix"], dtype=float)
    S_scalar = sum(
        (float(S[i, j]) * basis[i] * basis[j] for i, j in np.ndindex(S.shape)),
        pf.Scalar.zero(),
    )
    # The candidate rank-one square is accurate to 1e-4. Retain the complete
    # saved PSD block, including small numerical components, at the last step.
    S_piece_terms = {"terminal_full_S": S_scalar}
    S_assignment = {"terminal_full_S": N - 1}
    assert np.linalg.eigvalsh(S).min() >= -1e-6
    lyap = [pf.Scalar.zero()]
    partial_sum = sum(
        (term for name, term in group_terms.items() if term_assignment[name] == -1),
        pf.Scalar.zero(),
    )
    for k in range(N):
        partial_sum += sum(
            (term for name, term in group_terms.items() if term_assignment[name] == k),
            pf.Scalar.zero(),
        )
        partial_sum -= sum(
            (piece for name, piece in S_piece_terms.items() if S_assignment[name] == k),
            pf.Scalar.zero(),
        )
        lyap.append(partial_sum)
    coords = [scalar_coordinates(pm, v) for v in lyap]
    singular_values = [
        np.linalg.svd(np.asarray(c["gram"]), compute_uv=False).tolist() for c in coords
    ]
    ranks = [int(np.sum(np.array(s) > 1e-4)) for s in singular_values]
    if verbose:
        for k, rank in enumerate(ranks):
            print(f"rank V_{k}: {rank}")
            if k == 0:
                print()
        print(
            "Interior rank is constant at this N:",
            len(ranks[1:-1]) >= 3 and len(set(ranks[1:-1])) == 1,
            "boundary exclusions:",
            (1, 1),
        )
    coverage = scalar_coordinates(
        pm, lyap[N] - sum(group_terms.values(), pf.Scalar.zero()) + S_scalar
    )

    def maximum(c):
        return max(
            float(np.max(np.abs(c["gram"]), initial=0)),
            float(np.max(np.abs(c["func"]), initial=0)),
            abs(c["offset"]),
        )

    assert maximum(coverage) <= 1e-5
    increment_errors = []
    for k in range(N):
        assigned = {-1, 0} if k == 0 else {k}
        increment = sum(
            (v for name, v in group_terms.items() if term_assignment[name] in assigned),
            pf.Scalar.zero(),
        )
        increment -= sum(
            (v for name, v in S_piece_terms.items() if S_assignment[name] in assigned),
            pf.Scalar.zero(),
        )
        error = maximum(scalar_coordinates(pm, lyap[k + 1] - lyap[k] - increment))
        assert error <= 1e-5
        increment_errors.append(error)
    result = {
        "N": N,
        "certificate_id": certificate["certificate_id"],
        "recipe_id": "chronological_star_and_forward_pairs_terminal_full_S",
        "rank_profile": ranks,
        "singular_values": singular_values,
        "rank_convention": {
            "tolerance": 1e-4,
            "scaling": "none, raw Gram coordinate matrix",
            "coordinate_order": "x0, xstar, gradients chronologically",
        },
        "rank_credible": all(
            not any(5e-5 <= s <= 2e-4 for s in values) for values in singular_values
        ),
        "certificate_checked": True,
        "grouping_checked": True,
        "coverage_coordinates": coverage,
        "coverage_max_abs": maximum(coverage),
        "increment_errors": increment_errors,
        "lyap_coordinates": coords,
        "constraint_assignment": term_assignment,
        "S_assignment": S_assignment,
        "boundary_constraints": [],
        "boundary_S": [],
        "omitted_constraints": [],
        "sign_convention": "Residuals <=0 with nonnegative duals; subtract PSD S only at terminal. Hence V_(k+1)<=V_k, including combined initial contributions.",
    }
    return result
```

### Candidate-vector scan

For $1\leq k<N$, use the ordered basis
$$\mathcal B_k=(x_0-x_\star,\ z_{k+1}-x_\star,\ g_k).$$
At $k=0$ the quadratic part is zero and the basis is empty. At $k=N$ use only $x_0-x_\star$. All of these cases are checked below.

```python
def V_k_basis(k, ctx=ctx4, obj=obj4, N=N4):
    if k == 0:
        return []
    anchor = ctx["x_0"] - ctx["x_star"]
    if k == N:
        return [anchor]
    return [anchor, ctx[f"z_{k + 1}"] - ctx["x_star"], obj.grad(ctx[f"x_{k}"])]


def V_k_basis_labels(k, N=N4):
    if k == 0:
        return []
    if k == N:
        return ["x_0-x_star"]
    return ["x_0-x_star", f"z_{k + 1}-x_star", f"g_{k}"]


coeff4, quadratic_residuals = {}, {}
for k in range(N4 + 1):
    vectors = V_k_basis(k)
    matrix = (
        np.column_stack([pm4.eval_vector(v).coords for v in vectors]).astype(float)
        if vectors
        else np.zeros((len(basis4), 0))
    )
    rank = int(np.linalg.matrix_rank(matrix, tol=1e-7)) if vectors else 0
    assert rank == len(vectors) == observation["rank_profile"][k]
    C = find_symmetric_coefficient_matrix(
        lyap4[k],
        vectors,
        pep_context=ctx4,
        resolve_parameters=params4,
        indep_tol=1e-7,
        span_tol=1e-5,
    )
    target = np.asarray(observation["lyap_coordinates"][k]["gram"])
    error = float(np.linalg.norm(matrix @ C @ matrix.T - target, ord="fro"))
    assert error <= 1e-5 * max(1.0, np.linalg.norm(target, ord="fro"))
    coeff4[k] = C
    quadratic_residuals[str(k)] = error
    print(f"k={k}: rank {rank} basis {V_k_basis_labels(k)}; reconstruction {error:.3g}")
```

### Coefficient matrices
Matrices use the stated basis order, with symmetric off-diagonal entries counted twice. Function-gap coefficients and offsets are retained separately. The full structure is assembled explicitly and checked against the original partial sums.

```python
function_coefficients = {}
component_residuals = {}
for k in range(N4 + 1):
    target_func = np.asarray(observation["lyap_coordinates"][k]["func"])
    f_gap = scalar_coordinates(pm4, obj4(ctx4[f"x_{k}"]) - obj4(ctx4["x_star"]))
    gap_coords = np.asarray(f_gap["func"])
    a = float(target_func @ gap_coords / (gap_coords @ gap_coords)) if k else 0.0
    function_coefficients[k] = a
    assert np.allclose(a * gap_coords, target_func, atol=1e-4, rtol=1e-5)
    assert abs(observation["lyap_coordinates"][k]["offset"]) <= 1e-5
    print(f"k={k}: coefficient of f(x_{k})-f_star: {a:.12g}; offset 0")
    pf.pprint_labeled_matrix(coeff4[k], V_k_basis_labels(k), precision=7)


def V_k_structure(k):
    vectors = V_k_basis(k)
    C = coeff4[k]
    quadratic = sum(
        (float(C[i, j]) * vectors[i] * vectors[j] for i, j in np.ndindex(C.shape)),
        pf.Scalar.zero(),
    )
    return quadratic + function_coefficients[k] * (
        obj4(ctx4[f"x_{k}"]) - obj4(ctx4["x_star"])
    )


for k in range(N4 + 1):
    residual = scalar_coordinates(pm4, V_k_structure(k) - lyap4[k])
    errors = {
        name: float(np.max(np.abs(residual[name]), initial=0))
        for name in ("gram", "func", "offset")
    }
    assert max(errors.values()) <= 1e-4
    component_residuals[str(k)] = errors
print("Full-expression reconstruction:", component_residuals)
```

The numerical diagonal matrices suggest
$$C_k=\operatorname{diag}\left(-\frac{L}{2\theta_N^2},\frac{L}{2\theta_N^2},-\frac{\theta_k^2}{L\theta_N^2}\right),\qquad
a_k=\frac{2\theta_k^2}{\theta_N^2}\quad(1\leq k<N).$$
The terminal candidates are $C_N=-L/(2\theta_N^2)$ and $a_N=1$. These are candidate formulas, not yet analytic identities.

```python
def coeff_pattern(k, N, theta, L):
    if k == 0:
        return np.zeros((0, 0)), 0.0
    if k == N:
        return np.array([[-L / (2 * theta[N] ** 2)]]), 1.0
    return np.diag(
        [
            -L / (2 * theta[N] ** 2),
            L / (2 * theta[N] ** 2),
            -(theta[k] ** 2) / (L * theta[N] ** 2),
        ]
    ), 2 * theta[k] ** 2 / theta[N] ** 2


formula_errors = {}
for k in range(N4 + 1):
    C_candidate, a_candidate = coeff_pattern(k, N4, theta_sequence(N4), params4["L"])
    C_error = float(np.max(np.abs(C_candidate - coeff4[k]), initial=0))
    a_error = abs(a_candidate - function_coefficients[k])
    assert max(C_error, a_error) < 1e-4
    formula_errors[str(k)] = {"quadratic": C_error, "function": a_error}
    print(
        f"Candidate formula k={k}: quadratic error={C_error:.3g}; function error={a_error:.3g}"
    )
```

## Closed-form Lyapunov function and exact proof
The diagonal coefficients suggested by B4 are now verified analytically, using the minimal contexts below. It is convenient to work with $H_{k}=T^2V_{k}+B$ in the interior. The base identity and ordinary-step recurrence show inductively that
$$H_{k}-B=2\sum_{i=0}^{k}\theta_{i}I_{x_{\star},x_i}
+2\sum_{i=0}^{k-1}\theta_{i}^2I_{x_i,x_{i+1}}\quad(0\leq k<N).$$
This establishes the exact coefficient formula and its agreement with the partial-sum construction, rather than relying on numerical fitting. The terminal identity supplies the separately normalized endpoint. The complete formulas for $V_{k}$, including function gaps, appear in the theorem above.

### Symbolic recursion context and coefficient formulas
Parameters $L,t,u$ are positive. The ordinary relation is $t^2=u^2-u$; the terminal relation is $2t^2=u^2-u$. Exact polynomial remainders impose only these proved recurrence identities. Denominators are powers of positive $L,u$.

```python
import json

import numpy as np
import pepflow as pf
import sympy as sp


b4_proof = json.loads((NOTEBOOK_DIR / "state/ogm_b4.json").read_text())
assert (
    b4_proof["numerical_reconstruction"]["certificate_id"] == b4_proof["certificate_id"]
)
assert b4_proof["numerical_reconstruction"]["passed"]
k_symbol, N_symbol = sp.symbols("k N", integer=True, nonnegative=True)
L_symbol, t_symbol, u_symbol = sp.symbols("L t u", positive=True)
R_symbol = sp.symbols("R", nonnegative=True)
Lp, tp, up = pf.Parameter("L"), pf.Parameter("t"), pf.Parameter("u")
half = sp.Rational(1, 2)
symbolic_parameters: dict = {"L": L_symbol, "t": t_symbol, "u": u_symbol}
proof_results = {}


def interpolation(function, a, b):
    """Smooth convex interpolation residual I_(a,b) <= 0."""
    return (
        function(b)
        - function(a)
        - function.grad(b) * (b - a)
        + (function.grad(b) - function.grad(a)) ** 2 / (2 * Lp)
    )


def H(theta, function, x, z_next, star):
    """Unnormalized interior expression, including its function gap."""
    return (
        2 * theta**2 * (function(x) - function(star))
        - theta**2 / Lp * (function.grad(x) ** 2)
        + Lp / 2 * ((z_next - star) ** 2)
    )


def exact_components(expression, manager, relation=None):
    value = manager.eval_scalar(pf.Scalar.zero() + expression, sympy_mode=True)

    def reduce(entry):
        entry = sp.cancel(sp.sympify(entry))
        numerator, denominator = sp.fraction(entry)
        if relation is not None:
            numerator = sp.rem(
                sp.Poly(numerator, t_symbol), sp.Poly(relation, t_symbol)
            ).as_expr()
        return sp.simplify(numerator / denominator)

    return (
        sp.Matrix(value.inner_prod_coords).applyfunc(reduce),
        sp.Matrix(list(value.func_coords)).applyfunc(reduce),
        reduce(value.offset),
    )


def check_exact(name, difference, manager, context, relation=None):
    gram, funcs, offset = exact_components(difference, manager, relation)
    assert all(value == 0 for value in [*gram, *funcs, offset]), (
        name,
        gram,
        funcs,
        offset,
    )
    pf.pprint_labeled_matrix(
        np.array(gram.tolist(), dtype=object),
        [str(v) for v in context.basis_vectors()],
        precision=6,
    )
    print(name, "Gram residual exactly zero:", gram == sp.zeros(*gram.shape))
    print(name, "function residual:", funcs.T, "; offset:", offset)
    proof_results[name] = {
        "gram_zero": True,
        "function_zero": True,
        "offset_zero": True,
        "relation": str(relation) if relation is not None else None,
        "gram_dimension": list(gram.shape),
        "function_dimension": list(funcs.shape),
    }
    return True
```

```python
T_symbol = sp.symbols("T", positive=True)
C_closed = sp.diag(
    -L_symbol / (2 * T_symbol**2),
    L_symbol / (2 * T_symbol**2),
    -(t_symbol**2) / (L_symbol * T_symbol**2),
)
pf.pprint_labeled_matrix(
    np.array(C_closed.tolist(), dtype=object), ["d", "q_k", "g_k"], precision=0
)
gap_symbol, d_squared, q_squared, g_squared = sp.symbols(
    "gap d_squared q_squared g_squared", real=True
)
V_symbolic = (
    2 * t_symbol**2 * gap_symbol
    - t_symbol**2 * g_squared / L_symbol
    + L_symbol * q_squared / 2
    - L_symbol * d_squared / 2
) / T_symbol**2
print("Interior V_k:", V_symbolic)
print("LaTeX:", sp.latex(V_symbolic))
```

## Symbolic Step Recursion Verification

For $0\leq k\leq N-2$, set $t=\theta_{k}$, $u=\theta_{k+1}$. Verify
$$H_{k+1}-H_{k}=2uI_{x_{\star},x_{k+1}}+2t^2I_{x_k,x_{k+1}},\qquad t^2=u^2-u.$$
The residual $\mathrm{LHS}-\mathrm{RHS}$ must be zero in Gram, function-value and constant coordinates.

```python
ctx_step = pf.PEPContext("ogm_symbolic_step").set_as_current()
f_step = pf.SmoothConvexFunction(is_basis=True, tags=["f_{step}"], L=Lp)
star_step = f_step.set_stationary_point("x_{star}")
x_step = pf.Vector(is_basis=True, tags=["x_{k}"])
z_step = pf.Vector(is_basis=True, tags=["z_{k+1}"])
g_step = f_step.grad(x_step)
x_next_step = (1 - 1 / up) * (x_step - g_step / Lp) + z_step / up
x_next_step.add_tag("x_{k+1}")
g_next_step = f_step.grad(x_next_step)
z_next_step = z_step - (2 * up / Lp) * g_next_step
z_next_step.add_tag("z_{k+2}")
step_LHS = H(up, f_step, x_next_step, z_next_step, star_step) - H(
    tp, f_step, x_step, z_step, star_step
)
step_RHS = 2 * up * interpolation(
    f_step, star_step, x_next_step
) + 2 * tp**2 * interpolation(f_step, x_step, x_next_step)
step_diff = step_LHS - step_RHS
pm_step = pf.ExpressionManager(ctx_step, resolve_parameters=symbolic_parameters)
step_verified = check_exact(
    "step", step_diff, pm_step, ctx_step, relation=t_symbol**2 - u_symbol**2 + u_symbol
)
```

## Base Case and Boundary Symbolic Verification

Using $\theta_{0}=1$ and $z_{1}=x_{0}-2g_{0}/L$, verify
$$H_{0}-\frac L2\|x_{0}-x_{\star}\|^2=2I_{x_{\star},x_{0}}.$$
The residual $\mathrm{LHS}-\mathrm{RHS}$ must be zero in every scalar component. The right side is nonpositive, giving the base bound $H_{0}\leq B$.

```python
ctx_base = pf.PEPContext("ogm_symbolic_base").set_as_current()
f_base = pf.SmoothConvexFunction(is_basis=True, tags=["f_{base}"], L=Lp)
star_base = f_base.set_stationary_point("x_{star}")
x_base = pf.Vector(is_basis=True, tags=["x_{0}"])
g_base = f_base.grad(x_base)
z1_base = x_base - (2 / Lp) * g_base
initial_energy = (Lp / 2) * ((x_base - star_base) ** 2)
base_LHS = H(1, f_base, x_base, z1_base, star_base) - initial_energy
base_RHS = 2 * interpolation(f_base, star_base, x_base)
base_diff = base_LHS - base_RHS
pm_base = pf.ExpressionManager(ctx_base, resolve_parameters=symbolic_parameters)
base_verified = check_exact("base", base_diff, pm_base, ctx_base)
```

### Boundary Identity Symbolic Verification

Let $t=\theta_{N-1}$ and $u=\theta_{N}$, with $2t^2=u^2-u$. Verify
$$u^2(f(x_{N})-f(x_{\star}))-H_{N-1}
=uI_{x_{\star},x_{N}}+2t^2I_{x_{N-1},x_{N}}
-\frac L2\|z_{N}-x_{\star}-(u/L)g_{N}\|^2.$$
The residual $\mathrm{LHS}-\mathrm{RHS}$ must be zero in every scalar component.

```python
ctx_terminal = pf.PEPContext("ogm_symbolic_terminal").set_as_current()
f_terminal = pf.SmoothConvexFunction(is_basis=True, tags=["f_{terminal}"], L=Lp)
star_terminal = f_terminal.set_stationary_point("x_{star}")
x_terminal = pf.Vector(is_basis=True, tags=["x_{N-1}"])
z_terminal = pf.Vector(is_basis=True, tags=["z_{N}"])
g_terminal = f_terminal.grad(x_terminal)
x_final = (1 - 1 / up) * (x_terminal - g_terminal / Lp) + z_terminal / up
x_final.add_tag("x_{N}")
g_final = f_terminal.grad(x_final)
terminal_square = (Lp / 2) * ((z_terminal - star_terminal - (up / Lp) * g_final) ** 2)
terminal_LHS = up**2 * (f_terminal(x_final) - f_terminal(star_terminal)) - H(
    tp, f_terminal, x_terminal, z_terminal, star_terminal
)
terminal_RHS = (
    up * interpolation(f_terminal, star_terminal, x_final)
    + 2 * tp**2 * interpolation(f_terminal, x_terminal, x_final)
    - terminal_square
)
terminal_diff = terminal_LHS - terminal_RHS
pm_terminal = pf.ExpressionManager(ctx_terminal, resolve_parameters=symbolic_parameters)
boundary_verified = check_exact(
    "boundary",
    terminal_diff,
    pm_terminal,
    ctx_terminal,
    relation=2 * t_symbol**2 - u_symbol**2 + u_symbol,
)
```

### Exceptional horizon $N=1$

There is no ordinary interior transition when $N=1$. Here $\theta_{0}=1$, $\theta_{1}=2$, and $x_{1}=x_{0}-3g_{0}/(2L)$. Verify the direct identity
$$f(x_{1})-f(x_{\star})-\frac L8\|x_{0}-x_{\star}\|^2
=\frac12(I_{x_{\star},x_{0}}+I_{x_{\star},x_{1}}+I_{x_{0},x_{1}})
-\frac L8\|z_{1}-x_{\star}-(2/L)g_{1}\|^2.$$
Again $\mathrm{LHS}-\mathrm{RHS}=0$.

```python
ctx_one = pf.PEPContext("ogm_symbolic_N1").set_as_current()
f_one = pf.SmoothConvexFunction(is_basis=True, tags=["f_{one}"], L=Lp)
star_one = f_one.set_stationary_point("x_{star}")
x0_one = pf.Vector(is_basis=True, tags=["x_{0}"])
g0_one = f_one.grad(x0_one)
z1_one = x0_one - (2 / Lp) * g0_one
x1_one = half * (x0_one - g0_one / Lp) + half * z1_one
g1_one = f_one.grad(x1_one)
one_LHS = f_one(x1_one) - f_one(star_one) - (Lp / 8) * ((x0_one - star_one) ** 2)
one_RHS = (
    half * interpolation(f_one, star_one, x0_one)
    + half * interpolation(f_one, star_one, x1_one)
    + half * interpolation(f_one, x0_one, x1_one)
    - (Lp / 8) * ((z1_one - star_one - (2 / Lp) * g1_one) ** 2)
)
one_diff = one_LHS - one_RHS
pm_one = pf.ExpressionManager(ctx_one, resolve_parameters=symbolic_parameters)
one_verified = check_exact("N1", one_diff, pm_one, ctx_one)
```

### Signs, domain, and normalization
Starting at $\theta_{0}=1$, both positive-root recurrences produce positive real values. Their polynomial identities justify the reductions used above. Therefore the interpolation multipliers and square coefficient are positive. Every interpolation residual is nonpositive for convex $L$-smooth $f$, so all increments decrease. The initial-distance assumption then gives the theorem exactly, including $R=0$. These domain and sign facts hold for all integer horizons, not only the numerical samples.

```python
# Polynomial consequences of the positive-root recurrences are checked exactly.
root_regular = (1 + sp.sqrt(1 + 4 * t_symbol**2)) / 2
root_terminal = (1 + sp.sqrt(1 + 8 * t_symbol**2)) / 2
assert sp.simplify(root_regular**2 - root_regular - t_symbol**2) == 0
assert sp.simplify(root_terminal**2 - root_terminal - 2 * t_symbol**2) == 0
assert root_regular.is_positive and root_terminal.is_positive
thetaN_symbol = sp.symbols("theta_N", positive=True)
assert (2 * u_symbol / thetaN_symbol**2).is_positive
assert (2 * t_symbol**2 / thetaN_symbol**2).is_positive
assert (1 / thetaN_symbol).is_positive
assert (L_symbol / (2 * thetaN_symbol**2)).is_positive
print(
    "Positive roots, recurrence polynomial identities, and all multiplier/square signs verified."
)
print(
    "Normalization: V_N<=0 implies gap_N<=L*||x_0-x_star||^2/(2*theta_N^2)<=L*R^2/(2*theta_N^2)."
)
```

### Telescoped full certificate
Adding the exact base, ordinary steps, and terminal identity cancels every $H_k$, giving
$$f(x_{N})-f(x_{\star})-\frac{L}{2T^2}\|d\|^2
=\sum_{i=0}^{N-1}\frac{2\theta_i}{T^2}I_{x_{\star},x_i}
+\frac1T I_{x_{\star},x_N}
+\sum_{i=0}^{N-1}\frac{2\theta_i^2}{T^2}I_{x_i,x_{i+1}}
-\frac{L}{2T^2}\|z_N-x_{\star}-(T/L)g_N\|^2.$$
The same cancellation works when the ordinary-step sum is empty at $N=1$. Hence the global lambda and slack candidates of B2 are analytically verified as well. Adding $(L/(2T^2))(\|d\|^2-R^2)$ to the right gives the full-PEP identity normalized by $LR^2/(2T^2)$. All interpolation multipliers and the initial multiplier are nonnegative, and the final slack is a nonnegative square.

### Sharpness
For $f(x)=Lx^2/2$, write $x_k=\sigma R/t$ and $z_k=\sigma R$. The exact scalar update below flips the sign and divides the new iterate by the new coefficient. Taking $\sigma=(-1)^k$ proves the equality example by induction.

```python
# For f(x)=L*x^2/2, x_star=0, one exact scalar recurrence proves attainability.
sigma = sp.symbols("sigma", real=True)
x_quad = sigma * R_symbol / t_symbol
z_quad = sigma * R_symbol
g_quad = L_symbol * x_quad
y_quad = sp.simplify(x_quad - g_quad / L_symbol)
z_next_quad = sp.simplify(z_quad - 2 * t_symbol * g_quad / L_symbol)
x_next_quad = sp.simplify((1 - 1 / u_symbol) * y_quad + z_next_quad / u_symbol)
assert y_quad == 0
assert z_next_quad == -sigma * R_symbol
assert x_next_quad == -sigma * R_symbol / u_symbol
print(
    "Quadratic example: z_k=(-1)^k R and x_k=(-1)^k R/theta_k; bound attained at every horizon."
)
```

The exact symbolic step, base, terminal, and exceptional-horizon identities all vanish in Gram, function and constant coordinates. Their signs prove the nonincreasing Lyapunov sequence and the sharp guarantee
$$f(x_N)-f(x_{\star})\leq\frac{LR^2}{2\theta_N^2}.$$
