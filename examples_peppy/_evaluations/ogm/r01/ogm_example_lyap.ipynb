# OGM: smooth convex objective-gap discovery
Let $f$ be convex with $L$-Lipschitz gradient, $L>0$, and let $\nabla f(x_{\star})=0$. Assume $R\geq0$, $\|x_{0}-x_{\star}\|^{2}\leq R^{2}$, $z_{0}=x_{0}$, and integer $N\geq1$. The performance metric is $f(x_{N})-f(x_{\star})$.

Set $\theta_{-1}=0$ and
$$\theta_{k}=\frac{1+\sqrt{1+4\theta_{k-1}^{2}}}2\quad(0\leq k<N),\qquad \theta_{N}=\frac{1+\sqrt{1+8\theta_{N-1}^{2}}}2.$$
For $k=0,\ldots,N-1$,
$$y_{k+1}=x_{k}-\frac1L\nabla f(x_{k}),\qquad z_{k+1}=z_{k}-\frac{2\theta_{k}}{L}\nabla f(x_{k}),$$
$$x_{k+1}=\left(1-\frac1{\theta_{k+1}}\right)y_{k+1}+\frac{z_{k+1}}{\theta_{k+1}}.$$
The terminal theta recurrence is distinct and is retained in every numerical instance.

## Proof Statement

### Theorem
Let $f$ be convex and $L$-smooth on a finite-dimensional Euclidean space, with $L>0$, $\nabla f(x_{\star})=0$, and $\|x_{0}-x_{\star}\|^{2}\leq R^{2}$, where $R\geq0$. For every integer $N\geq1$, run the algorithm above with its special terminal recurrence. Write $g_{k}=\nabla f(x_{k})$, $f_{\star}=f(x_{\star})$, $T=\theta_{N}$, and $a=x_{0}-x_{\star}$.

A nonincreasing certificate is $V_{0}=0$, with
$$V_{k}=\frac{2\theta_{k}^{2}}{T^{2}}(f(x_{k})-f_{\star})-\frac{\theta_{k}^{2}}{LT^{2}}\|g_{k}\|^{2}+\frac{L}{2T^{2}}\left(\|z_{k+1}-x_{\star}\|^{2}-\|a\|^{2}\right),\quad 1\leq k<N,$$
and
$$V_{N}=f(x_{N})-f_{\star}+\frac{L}{2T^{2}}\left(\left\|z_{N}-x_{\star}-\frac{T}{L}g_{N}\right\|^{2}-\|a\|^{2}\right).$$
The interior range is empty when $N=1$. These expressions satisfy $V_{N}\leq\cdots\leq V_{0}=0$, and therefore
$$\boxed{f(x_{N})-f(x_{\star})\leq\frac{LR^{2}}{2\theta_{N}^{2}}.}$$


### Proof outline
For evaluated points $u,v$, write $g(u)=\nabla f(u)$ and define the nonpositive smooth convex interpolation residual
$$\mathcal I(u,v)=f(v)-f(u)+\langle g(v),u-v\rangle+\frac{1}{2L}\|g(u)-g(v)\|^{2}\leq0.$$
Use $\mathcal I_{i,j}=\mathcal I(x_{i},x_{j})$ and $\mathcal I_{\star,j}=\mathcal I(x_{\star},x_{j})$.

To verify the displayed $V_{k}$, define auxiliary expressions
$$\Phi_{k}=2\theta_{k}^{2}(f(x_{k})-f_{\star})-\frac{\theta_{k}^{2}}{L}\|g_{k}\|^{2}+\frac{L}{2}\|z_{k+1}-x_{\star}\|^{2},\quad 0\leq k<N,$$
$$\Phi_{N}=T^{2}(f(x_{N})-f_{\star})+\frac{L}{2}\left\|z_{N}-x_{\star}-\frac{T}{L}g_{N}\right\|^{2},\qquad A=\frac{L}{2}\|a\|^{2}.$$
Thus $V_{k}=(\Phi_{k}-A)/T^{2}$ for $1\leq k\leq N$, while $V_{0}=0$.

The exact base, ordinary-step, and terminal-step identities are
$$\Phi_{0}-A=2\mathcal I_{\star,0},$$
$$\Phi_{j+1}-\Phi_{j}=2\theta_{j+1}\mathcal I_{\star,j+1}+2\theta_{j}^{2}\mathcal I_{j,j+1},\quad 0\leq j\leq N-2,$$
$$\Phi_{N}-\Phi_{N-1}=T\mathcal I_{\star,N}+2\theta_{N-1}^{2}\mathcal I_{N-1,N}.$$
The ordinary identity uses $\theta_{j+1}^{2}-\theta_{j+1}=\theta_{j}^{2}$; the terminal identity uses $T^{2}-T=2\theta_{N-1}^{2}$. These are the two different recurrences in the input.

In particular, for $N\geq2$,
$$V_{1}-V_{0}=\frac{2\mathcal I_{\star,0}+2\theta_{1}\mathcal I_{\star,1}+2\mathcal I_{0,1}}{T^{2}};$$
for $1\leq k\leq N-2$,
$$V_{k+1}-V_{k}=\frac{2\theta_{k+1}\mathcal I_{\star,k+1}+2\theta_{k}^{2}\mathcal I_{k,k+1}}{T^{2}};$$
and, when $N\geq2$,
$$V_{N}-V_{N-1}=\frac{T\mathcal I_{\star,N}+2\theta_{N-1}^{2}\mathcal I_{N-1,N}}{T^{2}}.$$
For $N=1$, $T=2$ and the direct identity is
$$V_{1}=\frac{2\mathcal I_{\star,0}+2\mathcal I_{\star,1}+2\mathcal I_{0,1}}{4}.$$
Every multiplier is nonnegative: $\theta_{0}=1$, the ordinary recurrence preserves $\theta_{j}\geq1$, and the terminal recurrence gives $T\geq2$. All divisions are valid since $L>0$ and $T>0$. Thus all displayed increments are nonpositive.

Finally, the terminal formula gives
$$f(x_{N})-f_{\star}=V_{N}+\frac{L}{2T^{2}}\|a\|^{2}-\frac{L}{2T^{2}}\left\|z_{N}-x_{\star}-\frac{T}{L}g_{N}\right\|^{2}\leq\frac{LR^{2}}{2T^{2}}.$$
This uses the nonnegative terminal square and the initial radius bound, and includes $R=0$. The notebook below verifies each identity in all symbolic scalar components. The proof is local and valid for arbitrary horizon; a separate global analytic multiplier-table verification is not claimed.

```python
import math

L = pf.Parameter("L")
R = pf.Parameter("R")
f = pf.SmoothConvexFunction(is_basis=True, tags=["f"], L=L)
```

## Algorithm context and PEP setup

```python
def theta_sequence(N):
    """Numerical coefficients; the terminal recurrence has factor eight."""
    theta = []
    previous = 0.0
    for k in range(int(N) + 1):
        factor = 8 if k == N else 4
        previous = (1 + math.sqrt(1 + factor * previous**2)) / 2
        theta.append(previous)
    return theta


def make_ctx_ogm(ctx_name, N, **kwargs):
    ctx = pf.PEPContext(ctx_name).set_as_current()
    x = pf.Vector(is_basis=True, tags=["x_0", "z_0"])
    z = x
    f.set_stationary_point("x_star")
    theta = theta_sequence(N)
    for k in range(int(N)):
        g = f.grad(x)
        g.add_tag(f"g_{k}")
        y = x - g / L
        y.add_tag(f"y_{k + 1}")
        z = z - (2 * theta[k] / L) * g
        z.add_tag(f"z_{k + 1}")
        x = (1 - 1 / theta[k + 1]) * y + z / theta[k + 1]
        x.add_tag(f"x_{k + 1}")
    f.grad(x).add_tag(f"g_{N}")
    return ctx


def get_pep_setup(N, params):
    ctx = make_ctx_ogm(f"ctx_{N}", N)
    pb = pf.PEPBuilder(ctx)
    pb.add_initial_constraint(
        ((ctx["x_0"] - ctx["x_star"]) ** 2).le(R**2, name="initial_condition")
    )
    pb.set_performance_metric(f(ctx[f"x_{N}"]) - f(ctx["x_star"]))
    return ctx, pb, f
```

## Lambda and S verification
The initial search suggests, for $i<N$,
$$\lambda_{\star,i}=\frac{2\theta_{i}}{\theta_{N}^{2}},\qquad
\lambda_{i,i+1}=\frac{2\theta_{i}^{2}}{\theta_{N}^{2}},\qquad
\lambda_{\star,N}=\frac1{\theta_{N}}.$$
The candidate remainder is
$$S=\frac{L}{2\theta_{N}^{2}}\left\|z_{N}-x_{\star}-\frac{\theta_{N}}{L}g_{N}\right\|^{2}.$$
These are candidates supported by the selected numerical sample, not yet an arbitrary-horizon proof. At $L=R=1$, the following code reassembles every active scalar constraint and verifies the identity
$$f(x_{N})-f(x_{\star})-B-\sum_{c} d_{c}R_{c}+S=0,$$
where each $R_{c}\leq0$. The code checks the Gram, function-value, and constant components, multiplier signs, and the full saved matrix reconstruction.

```python
import hashlib
import json

import numpy as np
import pepflow as pf
import sympy as sp
from pepflow.utils import Comparator

from runtime import ROOT


def verify_certificate(state):
    for name, expected in state['certificate_provenance']['source_hashes'].items():
        assert hashlib.sha256((ROOT / name).read_bytes()).hexdigest() == expected
    N = state['N_verify']
    ctx, pb, obj = get_pep_setup(N, {'L': 1, 'R': 1})
    interpolation = obj.get_interpolation_constraints(ctx)
    entries = list(pb.init_conditions) + list(interpolation)
    assert set(c.name for c in entries) == set(state['constraint_metadata']) == set(state['named_dual_values'])
    assert state['basis_vectors'] == [str(v) for v in ctx.basis_vectors()]
    assert state['basis_scalars'] == [str(v) for v in ctx.basis_scalars()]
    assert state['S_row_names'] == state['basis_vectors'] == state['S_col_names']
    pm = pf.ExpressionManager(ctx, resolve_parameters={'L': sp.Integer(1), 'R': sp.Integer(1)})
    theta = theta_sequence(N)
    thetaN = theta[-1]
    # Candidate single square in the original basis, retaining its entire matrix.
    w = ctx[f'z_{N}'] - ctx['x_star'] - thetaN * ctx[f'g_{N}']
    S_numeric = w**2 / (2 * thetaN**2)
    E = np.asarray(pm.eval_vector(w).coords, dtype=float).reshape(1, -1)
    D = np.array([[1 / (2 * thetaN**2)]])
    S_reconstructed = E.T @ D @ E
    S_array = np.asarray(state['S_matrix'], dtype=float)
    weights = pf.Scalar.zero()
    inequality_values = []
    lambda_errors = []
    candidate_lambdas = {}
    def index(tag):
        return 0 if tag == 'z_0' else -1 if tag == 'x_star' else int(tag.split('_')[1])
    for con in entries:
        meta = state['constraint_metadata'][con.name]
        value = state['named_dual_values'][con.name]
        assert meta['kind'] == 'scalar' and meta['comparator'] == con.cmp.value
        assert meta['relaxed'] == (con.name in state['relaxed_constraints'])
        if meta['relaxed']:
            assert value is None
            continue
        assert value is not None and np.isfinite(value)
        residual = con.lhs - con.rhs
        if con.cmp in (Comparator.GE, Comparator.SEQ):
            residual = -residual
        weights += float(value) * (pf.Scalar.zero() + residual)
        if con.cmp != Comparator.EQ:
            inequality_values.append(float(value))
        if meta['origin'] == 'interpolation':
            left, right = con.name.split(':')[1].split(',')
            i, j = index(left), index(right)
            if i == -1:
                candidate = 1 / thetaN if j == N else 2 * theta[j] / thetaN**2
            else:
                assert j == i + 1
                candidate = 2 * theta[i]**2 / thetaN**2
            candidate_lambdas[con.name] = candidate
            lambda_errors.append(abs(value - candidate))
    diff = pb.performance_metric - state['opt_value'] - weights + S_numeric
    checked = pm.eval_scalar(diff)
    def max_abs(a):
        return float(np.max(np.abs(np.asarray(a, dtype=float)), initial=0.0))
    residuals = dict(identity_gram_max_abs=max_abs(checked.inner_prod_coords), identity_func_max_abs=max_abs(checked.func_coords), identity_offset_abs=max_abs(checked.offset), S_reconstruction_max_abs=max_abs(S_reconstructed-S_array), minimum_scalar_inequality_dual=min(inequality_values), minimum_S_eigenvalue=float(np.linalg.eigvalsh(S_array).min()), minimum_matrix_inequality_eigenvalue=0.0, lambda_candidate_max_abs=max(lambda_errors), conjecture_gap=abs(state['opt_value'] - 1/(2*thetaN**2)))
    tolerances = dict(identity_atol=1e-5, identity_rtol=0.0, reconstruction_atol=1e-4, reconstruction_rtol=1e-5, dual_tol=1e-6, psd_tol=1e-6, dense_opt_atol=1e-5)
    assert state['certificate_provenance']['solver_status'] == 'optimal'
    assert all(residuals[k] <= tolerances['identity_atol'] for k in ('identity_gram_max_abs','identity_func_max_abs','identity_offset_abs'))
    assert np.allclose(S_array, S_array.T, atol=1e-6, rtol=0)
    assert np.allclose(S_reconstructed, S_array, atol=1e-4, rtol=1e-5)
    assert residuals['minimum_scalar_inequality_dual'] >= -1e-6
    assert residuals['minimum_S_eigenvalue'] >= -1e-6
    assert residuals['lambda_candidate_max_abs'] <= 1e-4
    assert residuals['conjecture_gap'] <= 1e-5
    decomposition = dict(method='direct', D=D.tolist(), ell_coords=E.tolist(), basis_labels=state['basis_vectors'], discarded_matrix_max_abs=residuals['S_reconstruction_max_abs'], candidate_source='w = z_N - x_star - theta_N*g_N; S_numeric = w**2/(2*theta_N**2), L=1')
    return ctx, pb, pm, checked, S_reconstructed, candidate_lambdas, decomposition, {'tolerances': tolerances, 'residuals': residuals, 'passed': True}


def certificate_digest(state):
    keys = ('N_verify', 'params_json', 'relaxed_constraints', 'S_matrix', 'S_row_names', 'S_col_names', 'basis_vectors', 'basis_scalars', 'opt_value', 'tau_sol', 'tau_name', 'named_dual_values', 'constraint_metadata', 'normalization', 'certificate_provenance', 'numerical_checks', 'numerical_decomposition')
    return hashlib.sha256(json.dumps({key: state[key] for key in keys}, sort_keys=True, separators=(',', ':'), allow_nan=False).encode()).hexdigest()
```

## Partial-sum Lyapunov construction and rank profile
Normalize each interpolation inequality as $R_{i,j}\leq0$. Set $V_{0}=0$ and form $V_{k}$ from the weighted star inequalities through index $k$ and consecutive inequalities through $(k-1,k)$. Thus the first increment includes the initial star term, and subsequent increments obey
$$V_{k+1}-V_{k}=\lambda_{\star,k+1}R_{\star,k+1}+\lambda_{k,k+1}R_{k,k+1}\leq0.$$
The single nonnegative remainder square is reserved for the terminal boundary. All active interpolation terms are assigned; no terms are omitted. The independently reconstructed $N=4$ and $N=6$ trials have interior rank three with fixed endpoint exclusions $(1,1)$ at absolute tolerance $10^{-4}$ in unscaled Gram coordinates. At least three indices remain at each horizon, and their singular values are well separated from the threshold. This finite-sample observation supports basis discovery; it is not an all-horizon proof.

```python
import numpy as np
import pepflow as pf
from pepflow.constraint import ScalarConstraint

assert certificate_digest(certificate) == certificate['certificate_id']
assert certificate['closed_form_search']['completed']
assert certificate['certificate_provenance']['pep_type'] == 'primal'
ctx, pb, pm, checked, _, _, _, fresh_checks = verify_certificate(certificate)
N_int = certificate['N_verify']
pb.set_relaxed_constraints(certificate['relaxed_constraints'])
assert certificate['params_json'] == '{"L": 1, "R": 1}'
assert certificate['normalization']['initial_condition'] == '||x_0-x_star||^2 <= R^2'
assert certificate['normalization']['performance_metric'] == 'f(x_N) - f(x_star)'
assert certificate['normalization']['tau_name'] == 'initial_condition'
assert abs(certificate['normalization']['bound_value']-certificate['opt_value']) < 1e-5
assert abs(certificate['named_dual_values']['initial_condition']-certificate['tau_sol']) < 1e-6
basis = list(ctx.basis_vectors())
S_array = np.asarray(certificate['S_matrix'], dtype=float)
S_scalar = sum((float(S_array[i,j])*basis[i]*basis[j] for i,j in np.ndindex(S_array.shape)), pf.Scalar.zero())
records = [(c,'initial',None,None) for c in pb.init_conditions]
assert not ctx.oper_to_duplets
for owner in ctx.func_to_triplets:
    data = ctx.get_constraint_data(owner)
    assert not data.psd_dict
    records.extend((c,'interpolation',str(owner),str(group)) for group, family in data.sc_dict.items() for c in family)
weighted_terms, group_terms, interpolation_pairs = {}, {}, {}
for con, origin, owner_name, group in records:
    assert isinstance(con, ScalarConstraint)
    meta = dict(origin=origin,owner=owner_name,group=group,kind='scalar',comparator=con.cmp.value,shape=[],relaxed=con.name in pb.relaxed_constraints)
    assert meta == certificate['constraint_metadata'][con.name]
    val = certificate['named_dual_values'][con.name]
    if meta['relaxed']:
        assert val is None
        continue
    assert val is not None
    r = con.lhs - con.rhs
    if con.cmp in (Comparator.GE, Comparator.SEQ):
        r = -r
    weighted_terms[con.name] = float(val)*r
    if origin == 'interpolation':
        group_terms[con.name] = weighted_terms[con.name]
        interpolation_pairs[con.name] = tuple(con.name.split(':')[1].split(','))
assert set(weighted_terms) - set(group_terms) == {'initial_condition'}
raw_diff = pb.performance_metric-certificate['opt_value']-sum(weighted_terms.values(),pf.Scalar.zero())+S_scalar

def coords_max(expression):
    value = pm.eval_scalar(expression)
    return [float(np.max(np.abs(np.asarray(part,dtype=float)),initial=0)) for part in (value.inner_prod_coords,value.func_coords,value.offset)]

assert max(coords_max(raw_diff)) < 1e-5
certificate_checked = True
extra_duals = {}
```

```python
# Each assigned weighted interpolation residual is <= 0, so V is nonincreasing.
# The single nonnegative S square is left for the final boundary argument.
term_assignment = {}
for name,(left,right) in interpolation_pairs.items():
    j = 0 if right == 'z_0' else int(right.split('_')[1])
    term_assignment[name] = -1 if j == 0 else j-1
S_assignment = {'terminal_square': 'boundary'}
assert set(term_assignment) == set(group_terms)
assert set(S_assignment) == set(S_piece_terms)
lyap = [pf.Scalar.zero()]
partial_sum = sum((group_terms[name] for name,target in term_assignment.items() if target == -1),pf.Scalar.zero())
for step in range(N_int):
    partial_sum += sum((group_terms[name] for name,target in term_assignment.items() if target == step),pf.Scalar.zero())
    lyap.append(partial_sum)
rank_tolerance = 1e-4
rank_left,rank_right = 1,1
ranks, singular_values = [],[]
for k,Vk in enumerate(lyap):
    matrix = np.asarray(pm.eval_scalar(Vk).inner_prod_coords,dtype=float)
    singular_values.append(np.linalg.svd(matrix,compute_uv=False).tolist())
    ranks.append(int(np.linalg.matrix_rank(matrix,tol=rank_tolerance)))
    print(f'rank V_{k}: {ranks[-1]}')
    if k == 0:
        print()
retained_ranks = ranks[rank_left:N_int+1-rank_right]
print('Interior rank is constant at this N:',len(retained_ranks)>=3 and len(set(retained_ranks))==1,'boundary exclusions:',(rank_left,rank_right))
boundary_names = []
boundary_S = ['terminal_square']
leftover = -S_piece_terms['terminal_square']
coverage = lyap[N_int]+leftover-(sum(group_terms.values(),pf.Scalar.zero())-S_scalar)
coverage_residual = coords_max(coverage)
assert max(coverage_residual) < 1e-5
for k in range(N_int):
    targets = {-1,0} if k == 0 else {k}
    increment = sum((group_terms[name] for name,target in term_assignment.items() if target in targets),pf.Scalar.zero())
    assert max(coords_max(lyap[k+1]-lyap[k]-increment)) < 1e-5
rank_credible = all(not any(rank_tolerance/10 <= s <= rank_tolerance*10 for s in values) for values in singular_values)
grouping_checked = True
```

Candidates include tagged iterates, gradients, auxiliary displacements, the initial displacement, the terminal square vector, and pairwise differences. Zero vectors and coordinate duplicates are removed.

```python
from itertools import combinations
from pepflow.lyapunov_utils import vectors_in_column_space, find_basis_with_sparsest_coefficients

params_sp={'L':sp.Integer(1),'R':sp.Integer(1)}
assert ranks == certificate['rank_profile']
assert certificate['rank_search']['status']=='confirmed'
for k,Vk in enumerate(lyap):
    evaluated=pm.eval_scalar(Vk)
    for actual,saved in ((evaluated.inner_prod_coords,certificate['lyap_inner_prod_coords'][k]),(evaluated.func_coords,certificate['lyap_func_coords'][k]),(evaluated.offset,certificate['lyap_offsets'][k])):
        assert np.allclose(np.asarray(actual,dtype=float),np.asarray(saved,dtype=float),atol=1e-4,rtol=1e-5)
theta=theta_sequence(N_int)
candidate_pairs=[('anchor',ctx['x_0']-ctx['x_star']),('terminal_q',ctx[f'z_{N_int}']-ctx['x_star']-theta[-1]*ctx[f'g_{N_int}'])]
for k in range(N_int+1):
    for family in ('x','z','g','y'):
        if family=='y' and k==0:
            continue
        v=ctx[f'{family}_{k}']
        candidate_pairs.append((f'{family}_{k}',v))
        if family!='g':
            candidate_pairs.append((f'{family}_{k}-x_star',v-ctx['x_star']))
base_pairs=list(candidate_pairs)
for (label_a,v_a),(label_b,v_b) in combinations(base_pairs,2):
    candidate_pairs.append((f'({label_a})-({label_b})',v_a-v_b))
unique_pairs=[]
unique_coords=[]
for label,v in candidate_pairs:
    coords=np.asarray(pm.eval_vector(v).coords,dtype=float)
    if np.linalg.norm(coords)<1e-6:
        continue
    if any(np.allclose(coords,old,atol=1e-6,rtol=0) or np.allclose(coords,-old,atol=1e-6,rtol=0) for old in unique_coords):
        continue
    unique_pairs.append((label,v))
    unique_coords.append(coords)
candidates=[v for _,v in unique_pairs]
label_by_id={id(v):label for label,v in unique_pairs}
```

### Selected basis pattern
The following exhaustive sparse-basis search is executed for every nonzero-rank index, including the terminal case.

```python
sparse_basis_by_k={}
sparse_coeff_by_k={}
for k in range(1,N_int+1):
    found,C=find_basis_with_sparsest_coefficients(lyap[k],filtered[k],pep_context=ctx,resolve_parameters={'L':sp.Integer(1),'R':sp.Integer(1)},zero_tol=1e-6,indep_tol=1e-7,span_tol=1e-5)
    sparse_basis_by_k[k]=found
    sparse_coeff_by_k[k]=C
    assert len(found)==ranks[k]
    print(f'k={k}, rank={ranks[k]} sparse basis:',[label_by_id[id(v)] for v in found])
    print('zero coefficients:',int(np.sum(np.abs(C)<1e-6)),'/',C.size)

```

The search selects the interior basis
$$[x_{0}-x_{\star},\ g_{k},\ z_{k+1}-x_{\star}],\qquad 1\leq k<N.$$
At $k=N$ it selects $[x_{0}-x_{\star},\ z_{N}-x_{\star}-\theta_{N}g_{N}]$ for the normalized sample $L=1$. At $k=0$ the quadratic part is zero and uses the empty basis. Independence and full reconstruction are checked below.

```python
def V_k_basis(k):
    if k == 0:
        return []
    anchor = ctx['x_0']-ctx['x_star']
    if k == N_int:
        return [anchor,ctx[f'z_{N_int}']-ctx['x_star']-theta_sequence(N_int)[-1]*ctx[f'g_{N_int}']]
    return [anchor,ctx[f'g_{k}'],ctx[f'z_{k+1}']-ctx['x_star']]


def V_k_basis_labels(k):
    if k == 0:
        return []
    if k == N_int:
        return ['x_0-x_star','z_N-x_star-theta_N*g_N']
    return ['x_0-x_star',f'g_{k}',f'z_{k+1}-x_star']
```

### Coefficient matrices
The basis order is the displayed order above. Each symmetric matrix uses the convention that off-diagonal entries contribute twice. We also extract the function-gap coefficient and retain the constant term.

```python
from pepflow.lyapunov_utils import find_symmetric_coefficient_matrix

params_sp={'L':sp.Integer(1),'R':sp.Integer(1)}
required_indices=list(range(N_int+1))
coeff_by_k={}
coefficient_residuals={}
function_sample_coefficients={}
function=next(iter(ctx.func_to_triplets))
for k in required_indices:
    selected_basis=V_k_basis(k)
    evaluated=pm.eval_scalar(lyap[k])
    target=np.asarray(evaluated.inner_prod_coords,dtype=float)
    if selected_basis:
        matrix=np.column_stack([pm.eval_vector(v).coords for v in selected_basis]).astype(float)
        assert np.linalg.matrix_rank(matrix,tol=1e-7)==len(selected_basis)==ranks[k]
        C=find_symmetric_coefficient_matrix(lyap[k],selected_basis,pep_context=ctx,resolve_parameters={'L':sp.Integer(1),'R':sp.Integer(1)},indep_tol=1e-7,span_tol=1e-5)
        assert np.array_equal(C,C.T)
        difference=matrix@C@matrix.T-target
    else:
        assert ranks[k]==0
        C=np.zeros((0,0))
        difference=-target
    coefficient_residuals[k]=float(np.linalg.norm(difference,ord='fro'))
    assert coefficient_residuals[k] <= 1e-5*max(1.0,float(np.linalg.norm(target,ord='fro')))
    coeff_by_k[k]=C
    gap=pm.eval_scalar(function(ctx[f'x_{k}'])-function(ctx['x_star']))
    gap_coords=np.asarray(gap.func_coords,dtype=float)
    target_function=np.asarray(evaluated.func_coords,dtype=float)
    alpha=float(np.dot(gap_coords,target_function)/np.dot(gap_coords,gap_coords))
    function_sample_coefficients[k]=alpha
    assert np.allclose(alpha*gap_coords,target_function,atol=1e-4,rtol=1e-5)
    assert abs(float(evaluated.offset)) < 1e-5
    print(f'k={k}: rank {ranks[k]} basis {V_k_basis_labels(k)}')
    print('Quadratic reconstruction residual:',coefficient_residuals[k])
    if C.size:
        pf.pprint_labeled_matrix(C,V_k_basis_labels(k),precision=8)
    print('Function term coefficient:',alpha,'multiplying f(x_k)-f_star; offset:',float(evaluated.offset))

```

```python
def V_k_structure(k):
    vectors=V_k_basis(k)
    C=coeff_by_k[k]
    quadratic=sum((float(C[i,j])*vectors[i]*vectors[j] for i,j in np.ndindex(C.shape)),pf.Scalar.zero())
    function=next(iter(ctx.func_to_triplets))
    return quadratic+function_sample_coefficients[k]*(function(ctx[f'x_{k}'])-function(ctx['x_star']))


component_residuals={}
for k in range(N_int+1):
    diff=pm.eval_scalar(V_k_structure(k)-lyap[k])
    target=pm.eval_scalar(lyap[k])
    component_residuals[str(k)]={}
    for name,residual,target_component in (('gram',diff.inner_prod_coords,target.inner_prod_coords),('function',diff.func_coords,target.func_coords),('offset',diff.offset,target.offset)):
        residual=np.asarray(residual,dtype=float)
        scale=float(np.max(np.abs(np.asarray(target_component,dtype=float)),initial=0))
        error=float(np.max(np.abs(residual),initial=0))
        assert error <= 1e-4+1e-5*scale
        component_residuals[str(k)][name]=error
print('All-index full-component reconstruction:',component_residuals)
```

At the B4 discovery stage, the diagonal coefficient pattern below was a numerical candidate checked against every extracted matrix and function coefficient. The final symbolic section establishes the formulas analytically.

```python
def coeff_pattern(k,N):
    theta=theta_sequence(N)
    tau=1/(2*theta[-1]**2)
    if k==0:
        return np.zeros((0,0)),0.0
    if k==N:
        return np.diag([-tau,tau]),1.0
    return np.diag([-tau,-theta[k]**2/theta[-1]**2,tau]),2*theta[k]**2/theta[-1]**2


formula_residuals={}
for k in range(N_int+1):
    C_guess,alpha_guess=coeff_pattern(k,N_int)
    error=float(np.max(np.abs(C_guess-coeff_by_k[k]),initial=0))
    alpha_error=abs(alpha_guess-function_sample_coefficients[k])
    assert error<=1e-4 and alpha_error<=1e-4
    formula_residuals[str(k)]=dict(quadratic=error,function=alpha_error)
    print('Candidate formula residual',k,formula_residuals[str(k)])

for k in range(N_int+1):
    C_guess, alpha_guess = coeff_pattern(k, N_int)
    if C_guess.size:
        print("Candidate closed-form coefficient matrix at k=", k)
        pf.pprint_labeled_matrix(C_guess, V_k_basis_labels(k), precision=8)
    print("Candidate function coefficient:", alpha_guess)
```

### Block 4 conclusion
For the normalized sample, the supported interior expression is
$$V_{k}=\frac{2\theta_{k}^{2}}{\theta_{N}^{2}}(f(x_{k})-f_{\star})-\frac{\theta_{k}^{2}}{\theta_{N}^{2}}\|g_{k}\|^{2}+\frac{\|z_{k+1}-x_{\star}\|^{2}-\|x_{0}-x_{\star}\|^{2}}{2\theta_{N}^{2}},\quad 1\leq k<N.$$
The endpoint expressions are $V_{0}=0$ and
$$V_{N}=f(x_{N})-f_{\star}+\frac{\|z_{N}-x_{\star}-\theta_{N}g_{N}\|^{2}-\|x_{0}-x_{\star}\|^{2}}{2\theta_{N}^{2}}.$$
Every quadratic, function, and constant component is numerically reconstructed, including endpoints. These formulas are candidates at this stage. B5 must restore general scaling and prove the step, base, and terminal identities with the necessary sign conditions.

## Exact coefficient derivation and proof domain

The numerical matrices suggested the coefficients, but the following calculation establishes them over the full domain. Let $\Phi_{k}=A_{k}(f(x_{k})-f_{\star})+B_{k}\|g_{k}\|^{2}+C\|z_{k+1}-x_{\star}\|^{2}$. The selected formulas are $A_{k}=2\theta_{k}^{2}$, $B_{k}=-\theta_{k}^{2}/L$, and $C=L/2$. Matching function-value coefficients in the step identity forces the consecutive multiplier $A_{k}=2\theta_{k}^{2}$ and star multiplier $A_{k+1}-A_{k}=2\theta_{k+1}$. The last equality follows from the ordinary recurrence. At the terminal step the coefficient becomes $A_{N}=T^{2}$, so the star multiplier is $T^{2}-2\theta_{N-1}^{2}=T$. The checks below verify all remaining quadratic and constant coefficients, rather than relying on the fit.

The resulting closed form is $V_{0}=0$ and $V_{k}=(\Phi_{k}-L\|x_{0}-x_{\star}\|^{2}/2)/T^{2}$ for $1\leq k\leq N$, with the full interior and terminal $\Phi_{k}$ given in the proof outline. The local symbolic parameters $t,s$ stand for consecutive theta values; polynomial reduction uses exactly $t^{2}=s^{2}-s$ or $2t^{2}=s^{2}-s$. Denominators are nonzero because $L>0$ and all theta values used here are positive.

```python
with (notebook_dir / "state/ogm_b4.json").open() as stream:
    b4 = json.load(stream)
assert certificate_digest(b4) == b4["certificate_id"]
assert b4["numerical_reconstruction"]["certificate_id"] == b4["certificate_id"]
assert set(component_residuals) == {str(k) for k in range(N_int+1)}
assert all(max(values.values()) < 1e-4 for values in component_residuals.values())
print("Same-certificate B4 structure rechecked above; now proving the symbolic identities.")
```

```python

ell_sym,t_sym,s_sym,T_sym=sp.symbols('L t s T',positive=True)
R_sym=sp.symbols('R',nonnegative=True)
L_proof=pf.Parameter('L_proof')
t_proof=pf.Parameter('t_proof')
s_proof=pf.Parameter('s_proof')


def interpolation_residual(fun,u,v):
    return fun(v)-fun(u)+fun.grad(v)*(u-v)+(fun.grad(u)-fun.grad(v))**2/(2*L_proof)


def exact_components(expression,manager,relation=None):
    value=manager.eval_scalar(pf.Scalar.zero()+expression,sympy_mode=True)
    def reduce_entry(entry):
        entry=sp.cancel(sp.sympify(entry))
        if relation is not None:
            numerator,denominator=sp.fraction(entry)
            entry=sp.cancel(sp.rem(numerator,relation,t_sym)/denominator)
        return sp.simplify(entry)
    gram=sp.Matrix(value.inner_prod_coords).applyfunc(reduce_entry)
    functions=sp.Matrix(list(value.func_coords)).applyfunc(reduce_entry)
    offset=reduce_entry(value.offset)
    return gram,functions,offset


def assert_exact(name,expression,manager,context,relation=None):
    gram,functions,offset=exact_components(expression,manager,relation)
    assert all(entry==0 for entry in [*gram,*functions,offset]),(name,gram,functions,offset)
    print(name,'Gram residual:')
    pf.pprint_labeled_matrix(np.array(gram.tolist(),dtype=object),[str(v) for v in context.basis_vectors()],precision=0)
    print(name,'function residual:',functions.T)
    print(name,'offset residual:',offset)
    return {'gram_zero':True,'function_zero':True,'offset_zero':True}


symbolic_results={}

```

## Symbolic Step Recursion Verification

For $0\leq j\leq N-2$, the identity is
$$\Phi_{j+1}-\Phi_{j}=2s\mathcal I_{\star,j+1}+2t^{2}\mathcal I_{j,j+1},\qquad t=\theta_{j},\ s=\theta_{j+1},\ t^{2}=s^{2}-s.$$
The residual $\mathrm{LHS}-\mathrm{RHS}$ must vanish in every Gram, function, and constant coordinate.

```python
# Nonterminal transition: t=theta_j, s=theta_{j+1}, 0 <= j <= N-2.
step_ctx=pf.PEPContext('ogm_symbolic_step').set_as_current()
step_f=pf.SmoothConvexFunction(is_basis=True,tags=['f_{step}'],L=L_proof)
step_x=pf.Vector(is_basis=True,tags=['x_{j}'])
step_z=pf.Vector(is_basis=True,tags=['z_{j}'])
step_f.set_stationary_point('x_{star}')
step_star=step_ctx['x_{star}']
step_g=step_f.grad(step_x)
step_p=step_z-2*t_proof/L_proof*step_g
step_next=(1-1/s_proof)*(step_x-step_g/L_proof)+step_p/s_proof
step_gnext=step_f.grad(step_next)
step_q=step_p-2*s_proof/L_proof*step_gnext
Phi_old=2*t_proof**2*(step_f(step_x)-step_f(step_star))-t_proof**2/L_proof*step_g**2+L_proof/2*(step_p-step_star)**2
Phi_new=2*s_proof**2*(step_f(step_next)-step_f(step_star))-s_proof**2/L_proof*step_gnext**2+L_proof/2*(step_q-step_star)**2
step_LHS=Phi_new-Phi_old
step_RHS=2*s_proof*interpolation_residual(step_f,step_star,step_next)+2*t_proof**2*interpolation_residual(step_f,step_x,step_next)
step_diff=step_LHS-step_RHS
step_pm=pf.ExpressionManager(step_ctx,resolve_parameters={'L_proof':ell_sym,'t_proof':t_sym,'s_proof':s_sym})
symbolic_results['step']=assert_exact('Nonterminal step',step_diff,step_pm,step_ctx,t_sym**2-(s_sym**2-s_sym))
```

## Base Case and Boundary Symbolic Verification

With $\theta_{0}=1$ and $z_{0}=x_{0}$,
$$\Phi_{0}-\frac{L}{2}\|x_{0}-x_{\star}\|^{2}=2\mathcal I_{\star,0}.$$
The residual $\mathrm{LHS}-\mathrm{RHS}$ must vanish. Since the right side is nonpositive, this initializes the decrease argument.

```python
# Initial auxiliary potential Phi_0, with theta_0=1 and z_0=x_0.
base_ctx=pf.PEPContext('ogm_symbolic_base').set_as_current()
base_f=pf.SmoothConvexFunction(is_basis=True,tags=['f_{base}'],L=L_proof)
base_x=pf.Vector(is_basis=True,tags=['x_{0}'])
base_f.set_stationary_point('x_{star}')
base_star=base_ctx['x_{star}']
base_g=base_f.grad(base_x)
base_z1=base_x-2/L_proof*base_g
base_initial=L_proof/2*(base_x-base_star)**2
base_Phi0=2*(base_f(base_x)-base_f(base_star))-base_g**2/L_proof+L_proof/2*(base_z1-base_star)**2
base_LHS=base_Phi0-base_initial
base_RHS=2*interpolation_residual(base_f,base_star,base_x)
base_diff=base_LHS-base_RHS
base_pm=pf.ExpressionManager(base_ctx,resolve_parameters={'L_proof':ell_sym})
symbolic_results['base']=assert_exact('Initial identity',base_diff,base_pm,base_ctx)
```

### Boundary Identity Symbolic Verification

The special terminal recurrence gives $2t^{2}=s^{2}-s$ and
$$\Phi_{N}-\Phi_{N-1}=s\mathcal I_{\star,N}+2t^{2}\mathcal I_{N-1,N},\qquad t=\theta_{N-1},\ s=\theta_{N}.$$
The terminal metric identity is
$$\Phi_{N}-s^{2}(f(x_{N})-f_{\star})=\frac{L}{2}\left\|z_{N}-x_{\star}-\frac{s}{L}g_{N}\right\|^{2}\geq0.$$
Each residual $\mathrm{LHS}-\mathrm{RHS}$ must vanish. The nonnegative square is subtracted when solving for the objective gap.

```python
# Special terminal transition: t=theta_{N-1}, s=theta_N, s^2-s=2t^2.
terminal_ctx=pf.PEPContext('ogm_symbolic_terminal').set_as_current()
terminal_f=pf.SmoothConvexFunction(is_basis=True,tags=['f_{terminal}'],L=L_proof)
terminal_x=pf.Vector(is_basis=True,tags=['x_{N-1}'])
terminal_z=pf.Vector(is_basis=True,tags=['z_{N-1}'])
terminal_f.set_stationary_point('x_{star}')
terminal_star=terminal_ctx['x_{star}']
terminal_g=terminal_f.grad(terminal_x)
terminal_p=terminal_z-2*t_proof/L_proof*terminal_g
terminal_next=(1-1/s_proof)*(terminal_x-terminal_g/L_proof)+terminal_p/s_proof
terminal_gnext=terminal_f.grad(terminal_next)
terminal_q=terminal_p-terminal_star-s_proof/L_proof*terminal_gnext
terminal_old=2*t_proof**2*(terminal_f(terminal_x)-terminal_f(terminal_star))-t_proof**2/L_proof*terminal_g**2+L_proof/2*(terminal_p-terminal_star)**2
terminal_Phi=s_proof**2*(terminal_f(terminal_next)-terminal_f(terminal_star))+L_proof/2*terminal_q**2
terminal_LHS=terminal_Phi-terminal_old
terminal_RHS=s_proof*interpolation_residual(terminal_f,terminal_star,terminal_next)+2*t_proof**2*interpolation_residual(terminal_f,terminal_x,terminal_next)
terminal_diff=terminal_LHS-terminal_RHS
terminal_pm=pf.ExpressionManager(terminal_ctx,resolve_parameters={'L_proof':ell_sym,'t_proof':t_sym,'s_proof':s_sym})
symbolic_results['terminal_step']=assert_exact('Terminal step',terminal_diff,terminal_pm,terminal_ctx,t_sym**2-(s_sym**2-s_sym)/2)
# The final potential controls the requested metric, without dropping a negative term.
boundary_LHS=terminal_Phi-s_proof**2*(terminal_f(terminal_next)-terminal_f(terminal_star))
boundary_RHS=L_proof/2*terminal_q**2
boundary_diff=boundary_LHS-boundary_RHS
symbolic_results['boundary']=assert_exact('Terminal metric identity',boundary_diff,terminal_pm,terminal_ctx)
```

### Small horizon and radius normalization

For $N=1$, $\theta_{0}=1$, $\theta_{1}=2$, and $x_{1}=x_{0}-3g_{0}/(2L)$. The direct identity is
$$\Phi_{1}-\frac{L}{2}\|x_{0}-x_{\star}\|^{2}=2\mathcal I_{\star,0}+2\mathcal I_{\star,1}+2\mathcal I_{0,1}.$$
The normalization identity is
$$\frac{L}{2T^{2}}\|x_{0}-x_{\star}\|^{2}-\frac{LR^{2}}{2T^{2}}=\frac{L}{2T^{2}}\left(\|x_{0}-x_{\star}\|^{2}-R^{2}\right)\leq0.$$
Both residuals must vanish; no division by $R$ is used, so $R=0$ is included.

```python
# N=1 is covered directly: theta_0=1, theta_1=2, x_1=x_0-3g_0/(2L).
small_ctx=pf.PEPContext('ogm_symbolic_N1').set_as_current()
small_f=pf.SmoothConvexFunction(is_basis=True,tags=['f_{N=1}'],L=L_proof)
small_x=pf.Vector(is_basis=True,tags=['x_{0}'])
small_f.set_stationary_point('x_{star}')
small_star=small_ctx['x_{star}']
small_g=small_f.grad(small_x)
small_z1=small_x-2/L_proof*small_g
small_x1=small_x-sp.Rational(3,2)/L_proof*small_g
small_g1=small_f.grad(small_x1)
small_final=4*(small_f(small_x1)-small_f(small_star))+L_proof/2*(small_z1-small_star-2/L_proof*small_g1)**2
small_LHS=small_final-L_proof/2*(small_x-small_star)**2
small_RHS=2*interpolation_residual(small_f,small_star,small_x)+2*interpolation_residual(small_f,small_star,small_x1)+2*interpolation_residual(small_f,small_x,small_x1)
small_diff=small_LHS-small_RHS
small_pm=pf.ExpressionManager(small_ctx,resolve_parameters={'L_proof':ell_sym})
symbolic_results['N1']=assert_exact('N=1 direct certificate',small_diff,small_pm,small_ctx)
# Radius normalization is an identity, and the initial residual is nonpositive.
R_proof=pf.Parameter('R_proof')
normalization_LHS=L_proof/(2*s_proof**2)*(small_x-small_star)**2-L_proof*R_proof**2/(2*s_proof**2)
normalization_RHS=L_proof/(2*s_proof**2)*((small_x-small_star)**2-R_proof**2)
normalization_diff=normalization_LHS-normalization_RHS
normalization_pm=pf.ExpressionManager(small_ctx,resolve_parameters={'L_proof':ell_sym,'s_proof':T_sym,'R_proof':R_sym})
symbolic_results['normalization']=assert_exact('Radius normalization',normalization_diff,normalization_pm,small_ctx)
print('Signs: L>0, theta_j>=1, theta_N>=2, so every local multiplier and square coefficient is nonnegative.')
print('Each interpolation residual is nonpositive by convex L-smooth interpolation. Hence the potentials decrease.')
```

### Signs, domain, and telescoping conclusion

By induction from $\theta_{0}=1$, every ordinary theta is at least one. Since $\theta_{N-1}\geq1$, the terminal root is at least two. Thus all local multipliers $2\theta_{j+1}$, $2\theta_{j}^{2}$, and $\theta_{N}$ are positive; $L/2$ and $1/T^{2}$ are positive. The smooth convex interpolation residuals are nonpositive. Consequently $\Phi_{N}\leq\Phi_{N-1}\leq\cdots\leq\Phi_{0}\leq L\|x_{0}-x_{\star}\|^{2}/2$. The explicitly defined $V_{0},\ldots,V_{N}$ are therefore nonincreasing. The terminal square and initial radius yield
$$f(x_{N})-f(x_{\star})\leq\frac{LR^{2}}{2\theta_{N}^{2}}.$$
All local identities were checked exactly for arbitrary admissible theta pairs and $L>0$.
