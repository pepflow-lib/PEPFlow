# Proximal gradient: Lyapunov discovery

Let $f$ be convex and $L$-smooth with $L>0$, and let $g$ be closed, proper and convex on a finite-dimensional Euclidean space. Put $h=f+g$. Suppose $x_\star$ minimizes $h$, $R\geq0$, and $\|x_0-x_\star\|^2\leq R^2$. For integer $N\geq1$, the updates are
$$x_{k+1}=\operatorname{prox}_{g/L}\left(x_k-\frac1L\nabla f(x_k)\right),\qquad k=0,\ldots,N-1.$$
The performance metric is $h(x_N)-h(x_\star)$. The associated proximal subgradient is
$$s_{k+1}=L(x_k-x_{k+1})-\nabla f(x_k)\in\partial g(x_{k+1}).$$


## Proof Statement

### Theorem

Under the assumptions above, for every integer $N\geq1$,
$$h(x_{N})-h(x_{\star})\leq\frac{LR^2}{4N}.$$
Let $r_{k}=x_{k}-x_{\star}$, $a_{k}=\nabla f(x_{k})$, $F_{k}=f(x_{k})-f(x_{\star})$, and $G_{k}=g(x_{k})-g(x_{\star})$. An explicit nonincreasing certificate is $V_{0}=0$,
$$\boxed{V_{k}=-\frac{L}{4N}\|r_{0}\|^2
+\frac{L(N-k)}{(2N-k)^2}\|r_{k}\|^2
+\frac{\|a_{k}-a_{\star}\|^2}{2L(2N-k)}
-\frac{\langle r_{k},a_{k}\rangle}{2N-k}
+\frac{k+1}{2N-k}F_{k}+\frac{k}{2N-k}G_{k},\quad1\leq k<N,}$$
and
$$V_{N}=F_{N}+G_{N}-\frac{L}{4N}\|r_{0}\|^2.$$
The interior range is empty when $N=1$. All identities below are exact for symbolic $N,L$ and the stated index ranges; finite-sample numerical evidence is kept separately.


### Proof outline

Write $s_{i}=L(x_{i-1}-x_{i})-a_{i-1}\in\partial g(x_{i})$, and choose $s_{\star}=-a_{\star}$. For evaluated points define the nonpositive interpolation residuals
$$I^{f}(u,v)=f(v)-f(u)+\langle\nabla f(v),u-v\rangle+\frac{\|\nabla f(u)-\nabla f(v)\|^2}{2L},$$
$$I^{g}(u,v)=g(v)-g(u)+\langle s_{v},u-v\rangle.$$
The potential has quadratic part
$$-\frac{L}{4N}\|r_{0}\|^2+\frac{L(N-k)}{(2N-k)^2}\|r_{k}\|^2
+\frac{\|a_{k}-a_{\star}\|^2}{2L(2N-k)}-\frac{\langle r_{k},a_{k}\rangle}{2N-k},$$
and function part $((k+1)F_{k}+kG_{k})/(2N-k)$ for $1\leq k<N$; its endpoints are $V_{0}=0$ and $V_{N}=F_{N}+G_{N}-L\|r_{0}\|^2/(4N)$.

The supplied SOS aid, now verified in the exact local identities, is $S_{i}=\sum_{j=1}^{3}c_{j,i}\|u_{j,i}\|^2$, where
$$c_{1,i}=\frac{N}{(2N+1-i)(2N-i)L},\qquad
u_{1,i}=\frac{i}{2N}a_{i}+\frac{2N-i}{2N}a_{i-1}-a_{\star},$$
$$c_{2,i}=\frac{1}{2L}\left(\frac{i-1}{2N-i}\frac{2N+2}{2N+1}
+\frac{2N}{(2N-i)^2(2N+1)}\right),\qquad
u_{2,i}=a_{i-1}+s_{i}-\frac{Lr_{i-1}}{2N+1-i},$$
$$c_{3,i}=\frac{i}{2N+1-i}\frac{2N}{2N+1}\frac{1}{2L},\qquad
u_{3,i}=\frac{2N+1}{2N}(a_{i}+s_{i})-\frac{a_{i-1}+s_{i}}{2N}-\frac{Lr_{i}}{2N-i}.$$
All these coefficients are nonnegative for $1\leq i\leq N$ and $L>0$.

For $1\leq i<N$, put
$$\eta^{f}_{i}=\frac{2N+1}{(2N-i)(2N+1-i)},\quad
\eta^{g}_{i}=\frac{2N}{(2N-i)(2N+1-i)},\quad
\theta^{f}_{i}=\frac{i}{2N+1-i},\quad\theta^{g}_{i}=\frac{i-1}{2N+1-i}.$$
For $N\geq2$, the exact base identity is
$$V_{1}=\frac{I^{f}(x_{\star},x_{0})}{2N}
+\eta^{f}_{1}I^{f}(x_{\star},x_{1})+\frac{I^{f}(x_{0},x_{1})}{2N}
+\eta^{g}_{1}I^{g}(x_{\star},x_{1})-S_{1}\leq0.$$
For $2\leq i<N$, the exact step identity is
$$V_{i}-V_{i-1}=\eta^{f}_{i}I^{f}(x_{\star},x_{i})
+\theta^{f}_{i}I^{f}(x_{i-1},x_{i})+\eta^{g}_{i}I^{g}(x_{\star},x_{i})
+\theta^{g}_{i}I^{g}(x_{i-1},x_{i})-S_{i}\leq0.$$
The distinct terminal identity is
$$V_{N}-V_{N-1}=\frac{I^{f}(x_{\star},x_{N})}{N+1}
+\frac{N}{N+1}I^{f}(x_{N-1},x_{N})+\frac{2}{N+1}I^{g}(x_{\star},x_{N})
+\frac{N-1}{N+1}I^{g}(x_{N-1},x_{N})-S_{N}\leq0.$$
No $g(x_{0})$ evaluation is required. When $N=1$, use the separately verified identity
$$V_{1}=\tfrac12 I^{f}(x_{\star},x_{0})+\tfrac12 I^{f}(x_{\star},x_{1})
+\tfrac12 I^{f}(x_{0},x_{1})+I^{g}(x_{\star},x_{1})-S_{1}\leq0.$$
Every displayed decrease follows from nonpositive residuals and nonnegative coefficients and squares. Thus $V_{N}\leq V_{0}=0$ and
$$h(x_{N})-h(x_{\star})\leq\frac{L}{4N}\|r_{0}\|^2\leq\frac{LR^2}{4N}.$$
Summing these same exact local identities gives the full-PEP certificate with initial multiplier $L/(4N)$ and total slack $\sum_{i=1}^{N}S_{i}$. The numerical polar correction used for stable basis discovery is not part of this analytic identity.

```python
import pepflow as pf
import numpy as np
import matplotlib.pyplot as plt

assert Path(pf.__file__).resolve().is_relative_to(repo_root)
print("Approved framework:", pf.__file__)
```

```python
L = pf.Parameter("L")
R = pf.Parameter("R")
f = pf.SmoothConvexFunction(is_basis=True, tags=["f"], L=L)
g = pf.ConvexFunction(is_basis=True, tags=["g"])
h = f + g
```

```python
def make_ctx_pgm(ctx_name: str, N, **kwargs) -> pf.PEPContext:
    ctx = pf.PEPContext(ctx_name).set_as_current()
    x = pf.Vector(is_basis=True, tags=["x_0"])
    h.set_stationary_point("x_star")
    for i in range(int(N)):
        x = g.prox(x - f.grad(x) / L, 1 / L, tag=f"x_{i + 1}")
        f.grad(x)
    return ctx


def get_pep_setup(N, params):
    ctx = make_ctx_pgm(f"pgm_{N}", N)
    pb = pf.PEPBuilder(ctx)
    pb.add_initial_constraint(
        ((ctx["x_0"] - ctx["x_star"]) ** 2).le(R**2, name="initial_condition")
    )
    pb.set_performance_metric(h(ctx[f"x_{N}"]) - h(ctx["x_star"]))
    return ctx, pb, f
```

## Full-PEP certificate

At $N=4$, both interpolation families enter the primal PEP. The dense certificate is preserved in `state/pgm_dense.json`. We tested a sparse pattern retaining only $(x_\star,x_i)$ and $(x_i,x_{i+1})$ inequalities, retaining the initial constraint. Each selected multiplier and the PSD slack come from the same CLARABEL solve. The following checks reconstruct the context without solving again.


```python
import hashlib
import sympy as sp

b1 = json.loads((notebook_dir / "state" / "pgm_b1.json").read_text())
dense = json.loads((notebook_dir / "state" / "pgm_dense.json").read_text())
selected = json.loads(
    (notebook_dir / "supplementary" / "b2-sparse-candidate.json").read_text()
)
N_int = selected["N_verify"]
N = sp.S(N_int)
ctx, pb, obj = get_pep_setup(N_int, {"L": 1, "R": 1})
entries = [(c, "initial", None, None) for c in pb.init_conditions]
for owner in ctx.func_to_triplets:
    owner.get_interpolation_constraints(ctx)
    groups = ctx.get_constraint_data(owner)
    entries.extend(
        (c, "interpolation", str(owner), str(group))
        for group, constraints in groups.sc_dict.items()
        for c in constraints
    )
assert {c.name for c, *_ in entries} == set(selected["constraint_metadata"])
assert selected["basis_vectors"] == [str(v) for v in ctx.basis_vectors()]
assert selected["basis_scalars"] == [str(v) for v in ctx.basis_scalars()]
relaxed = []
for c, origin, *_ in entries:
    if origin == "initial":
        continue
    left, right = c.name.split(":")[1].split(",")
    keep = left == "x_star"
    if left != "x_star" and right != "x_star":
        keep = int(right.split("_")[1]) == int(left.split("_")[1]) + 1
    if not keep:
        relaxed.append(c.name)
assert relaxed == selected["relaxed_constraints"]
pb.set_relaxed_constraints(relaxed)
pm = pf.ExpressionManager(ctx, resolve_parameters={"L": sp.S(1), "R": sp.S(1)})
S_array = np.asarray(selected["S_matrix"], dtype=float)
basis = ctx.basis_vectors()
print("Dense optimum:", dense["opt_value"])
print("Sparse optimum:", selected["opt_value"])
print("Dense/sparse difference:", abs(dense["opt_value"] - selected["opt_value"]))
print("Basis:", selected["basis_vectors"])
for sample_name, sample in [("dense", dense), ("sparse", selected)]:
    for owner_name in ("f", "g"):
        active = [
            (name, value)
            for name, value in sample["named_dual_values"].items()
            if name.startswith(owner_name + ":") and value is not None and value > 1e-6
        ]
        print(sample_name, owner_name, "active multipliers:", len(active))
    eig = np.linalg.eigvalsh(np.asarray(sample["S_matrix"]))
    print(
        sample_name, "S rank at 1e-4:", np.sum(np.abs(eig) > 1e-4), "eigenvalues:", eig
    )
```

### Initial multiplier formula search

Rationalizing the sparse entries suggests consecutive multipliers
$$\lambda^f_{i,i+1}=\frac{i+1}{2N-i}\quad(0\leq i<N),\qquad
\lambda^g_{i,i+1}=\frac{i}{2N-i}\quad(1\leq i<N).$$
Conservation of function-value coefficients then gives the minimizer multipliers as consecutive differences, with a final coefficient closing the sum to one. We compare every active coefficient, including the initial multiplier $L/(4N)$. These are candidates, not an all-horizon proof.


```python
def candidate_multiplier(name, horizon):
    horizon = sp.S(horizon)
    if name == "initial_condition":
        return sp.S(1) / (4 * horizon)
    owner, pair = name.split(":")
    left, right = pair.split(",")
    if left == "x_star":
        j = int(right.split("_")[1])
        if j == horizon:
            return sp.S(1 if owner == "f" else 2) / (horizon + 1)
        return (2 * horizon + (1 if owner == "f" else 0)) / (
            (2 * horizon - j) * (2 * horizon + 1 - j)
        )
    if right == "x_star":
        return sp.S(0)
    i, j = int(left.split("_")[1]), int(right.split("_")[1])
    return (
        (i + (1 if owner == "f" else 0)) / (2 * horizon - i) if j == i + 1 else sp.S(0)
    )


lambda_errors = []
for name, value in selected["named_dual_values"].items():
    candidate = candidate_multiplier(name, N)
    if value is not None:
        error = abs(float(candidate) - value)
        lambda_errors.append(error)
        print(name, "numerical=", round(value, 9), "candidate=", candidate)
    else:
        assert candidate == 0
print("Maximum multiplier comparison error:", max(lambda_errors))
assert max(lambda_errors) < 1e-4
for owner_name, indices in [
    ("f", ["star", *range(N_int + 1)]),
    ("g", ["star", *range(1, N_int + 1)]),
]:
    labels = [f"x_{i}" for i in indices]
    candidate_matrix = sp.Matrix(
        [
            [
                0 if a == b else candidate_multiplier(f"{owner_name}:{a},{b}", N)
                for b in labels
            ]
            for a in labels
        ]
    )
    pf.pprint_labeled_matrix(np.asarray(candidate_matrix, dtype=object), labels, labels)
```

### Supplied SOS candidate and normalization

The three square families below are the **supplied assistance**. They are tested against this attempt's selected certificate. Here $a_i=\nabla f(x_i)$ and $s_i=L(x_{i-1}-x_i)-a_{i-1}\in\partial g(x_i)$. The certificate is normalized as
$$h(x_N)-h(x_\star)-\mathrm{opt}-\sum_c d_c R_c+S=0,$$
where every $R_c\leq0$. The candidate uses normalization factor 1. All its coefficients are nonnegative for $L>0$, $N\geq1$, $1\leq i\leq N$: denominators are positive and numerators are sums of nonnegative terms. Numerical agreement here does not establish the complete identity for arbitrary $N$.


```python
S_candidate = pf.Scalar.zero()
square_vectors, square_coefficients, pieces = [], [], []
for ii in range(1, N_int + 1):
    i = sp.S(ii)
    a = f.grad(ctx[f"x_{ii}"])
    ap = f.grad(ctx[f"x_{ii - 1}"])
    ast = f.grad(ctx["x_star"])
    s = g.grad(ctx[f"x_{ii}"])
    c1 = N / ((2 * N + 1 - i) * (2 * N - i) * L)
    u1 = i / (2 * N) * a + (2 * N - i) / (2 * N) * ap - ast
    c2 = (
        (i - 1) / (2 * N - i) * (2 * N + 2) / (2 * N + 1)
        + 2 * N / ((2 * N - i) ** 2 * (2 * N + 1))
    ) / (2 * L)
    u2 = ap + s - L / (2 * N + 1 - i) * (ctx[f"x_{ii - 1}"] - ctx["x_star"])
    c3 = i / (2 * N + 1 - i) * (2 * N) / (2 * N + 1) / (2 * L)
    u3 = (
        (2 * N + 1) / (2 * N) * (a + s)
        - (ap + s) / (2 * N)
        - L / (2 * N - i) * (ctx[f"x_{ii}"] - ctx["x_star"])
    )
    piece = c1 * u1**2 + c2 * u2**2 + c3 * u3**2
    S_candidate += piece
    pieces.append(piece)
    square_vectors.extend([u1, u2, u3])
    square_coefficients.extend([c1, c2, c3])
E = np.asarray([pm.eval_vector(v).coords for v in square_vectors], dtype=float)
D = np.diag([float(c.get_value({"L": sp.S(1)})) for c in square_coefficients])
candidate_eval = pm.eval_scalar(S_candidate)
S_reconstructed = E.T @ D @ E


def max_abs(values):
    return float(np.max(np.abs(np.asarray(values, dtype=float)), initial=0.0))


print("Supplied SOS reconstruction residual:", max_abs(S_reconstructed - S_array))
assert np.allclose(S_reconstructed, S_array, atol=1e-4, rtol=1e-5)
assert max_abs(candidate_eval.func_coords) == 0 and max_abs(candidate_eval.offset) == 0
pf.pprint_labeled_matrix(
    S_reconstructed, selected["basis_vectors"], selected["basis_vectors"]
)
```

### Complete fixed-horizon certificate verification
The saved multipliers and slack are checked in all coordinates, with no active interpolation family omitted. The supplied SOS is also checked as a decomposition of this same slack within the frozen tolerances.

```python
weighted = pf.Scalar.zero()
exact_weighted = pf.Scalar.zero()
inequality_duals = []
for c, *_ in entries:
    value = selected["named_dual_values"][c.name]
    residual = c.lhs - c.rhs
    if c.cmp in (pf.Comparator.GE, pf.Comparator.SEQ):
        residual = -residual
    meta = selected["constraint_metadata"][c.name]
    assert meta["comparator"] == c.cmp.value and meta["kind"] == "scalar"
    if meta["relaxed"]:
        assert value is None
        continue
    assert value is not None
    weighted += value * residual
    exact_weighted += candidate_multiplier(c.name, N) * residual
    if c.cmp != pf.Comparator.EQ:
        inequality_duals.append(value)
S_saved = sum(
    (float(S_array[i, j]) * basis[i] * basis[j] for i, j in np.ndindex(S_array.shape)),
    pf.Scalar.zero(),
)
assert pb.performance_metric is not None
checked = pm.eval_scalar(
    pb.performance_metric - selected["opt_value"] - weighted + S_saved
)
candidate_checked = pm.eval_scalar(
    pb.performance_metric - selected["opt_value"] - weighted + S_candidate
)
exact_checked = pm.eval_scalar(
    pb.performance_metric - sp.S(1) / (4 * N) - exact_weighted + S_candidate,
    sympy_mode=True,
)
exact_fixed_horizon = all(
    sp.simplify(v) == 0
    for values in (
        exact_checked.inner_prod_coords,
        exact_checked.func_coords,
        [exact_checked.offset],
    )
    for v in np.asarray(values, dtype=object).flat
)
residuals = dict(
    identity_gram_max_abs=max_abs(checked.inner_prod_coords),
    identity_func_max_abs=max_abs(checked.func_coords),
    identity_offset_abs=max_abs(checked.offset),
    S_reconstruction_max_abs=max_abs(S_reconstructed - S_array),
    candidate_identity_gram_max_abs=max_abs(candidate_checked.inner_prod_coords),
    dense_relaxed_gap=abs(dense["opt_value"] - selected["opt_value"]),
    conjecture_gap=abs(selected["opt_value"] - 1 / (4 * N_int)),
    minimum_scalar_inequality_dual=min(inequality_duals),
    minimum_S_eigenvalue=float(np.linalg.eigvalsh(S_array).min()),
    minimum_matrix_inequality_eigenvalue=0.0,
)
tolerances = json.loads((repo_root / "settings.json").read_text())["tolerances"]
assert selected["certificate_provenance"]["solver_status"] == "optimal"
assert np.all(np.isfinite(S_array)) and np.allclose(
    S_array, S_array.T, atol=1e-6, rtol=0
)
assert all(
    residuals[k] <= 1e-5
    for k in (
        "identity_gram_max_abs",
        "identity_func_max_abs",
        "identity_offset_abs",
        "candidate_identity_gram_max_abs",
    )
)
assert np.allclose(S_reconstructed, S_array, atol=1e-4, rtol=1e-5)
assert residuals["dense_relaxed_gap"] < 1e-5 and residuals["conjecture_gap"] < 1e-5
assert residuals["minimum_scalar_inequality_dual"] >= -1e-6
assert residuals["minimum_S_eigenvalue"] >= -1e-6
assert exact_fixed_horizon
print("Numerical certificate verified:", True)
print("Residuals:", residuals)
print("Exact candidate identity at N=4:", exact_fixed_horizon)
print("Analytic full-PEP proof for arbitrary N:", False)
pf.pprint_labeled_matrix(
    np.asarray(checked.inner_prod_coords),
    selected["basis_vectors"],
    selected["basis_vectors"],
)
```

## Lyapunov partial sums

Let $I^f_{a,b}\leq0$ and $I^g_{a,b}\leq0$ denote normalized interpolation residuals, and let $S_i\geq0$ be the supplied three-square block. With $V_0=0$, assign each inequality to its destination index and define
$$V_{k+1}-V_k=\sum_{c:\,\mathrm{destination}(c)=k+1}d_c I_c-S_{k+1},$$
including $d_{\star,0}^fI^f_{\star,0}$ in the first increment. Thus the sequence is nonincreasing under the certificate's sign convention. The initial distance constraint is reserved for the boundary argument.

At the first horizon $N=4$, all three interior ranks are 4. All interpolation terms and all SOS pieces are assigned, with no boundary omissions. The full-coordinate coverage residual is within $10^{-5}$. The same recipe is confirmed at $N=6$ using that horizon’s own complete certificate; the comparison below replays the visible construction.


Endpoint reconstruction exposed accumulated error in the supplied numerical SOS approximation. We use polar alignment of its factor rows to the same saved PSD slack, discarding only slack eigenvalues at or below the frozen PSD tolerance and checking the full reconstruction. Every revised step piece is a sum of three squares. This is a numerical decomposition for the selected certificate; it does not alter any certificate multiplier or claim an analytic decomposition.

```python
# Visible/replayable selected-certificate and partial-sum checks; no solves.
import hashlib
import json
from pathlib import Path

import numpy as np
import pepflow as pf
import sympy as sp

# Caller supplies notebook_dir, certificate and get_pep_setup.
N_int = certificate["N_verify"]
params_sp = {
    name: sp.S(value) for name, value in json.loads(certificate["params_json"]).items()
}
ctx, pb, obj = get_pep_setup(N_int, params_sp)
ctx.set_as_current()
pb.set_relaxed_constraints(certificate["relaxed_constraints"])
certificate_keys = (
    "N_verify",
    "params_json",
    "relaxed_constraints",
    "S_matrix",
    "S_row_names",
    "S_col_names",
    "basis_vectors",
    "basis_scalars",
    "opt_value",
    "tau_sol",
    "tau_name",
    "named_dual_values",
    "constraint_metadata",
    "normalization",
    "certificate_provenance",
    "numerical_checks",
    "numerical_decomposition",
)
digest = hashlib.sha256(
    json.dumps(
        {key: certificate[key] for key in certificate_keys},
        sort_keys=True,
        separators=(",", ":"),
        allow_nan=False,
    ).encode()
).hexdigest()
assert digest == certificate["certificate_id"]
for path, expected in certificate["certificate_provenance"]["source_hashes"].items():
    source_path = Path(path).resolve()
    assert source_path.is_relative_to(notebook_dir)
    assert hashlib.sha256(source_path.read_bytes()).hexdigest() == expected
assert (
    Path(pf.__file__)
    .resolve()
    .is_relative_to(notebook_dir / "supplementary" / "approved")
)
assert certificate["certificate_provenance"]["pep_type"] == "primal"
assert certificate["certificate_provenance"]["solver_status"] == "optimal"
assert certificate["closed_form_search"]["completed"]
assert certificate["normalization"]["tau_name"] == "initial_condition"
assert certificate["normalization"]["initial_condition"] == "||x_0-x_star||^2 <= R^2"
assert certificate["normalization"]["performance_metric"] == "h(x_N)-h(x_star)"
assert certificate["normalization"]["bound_formula"] == "L*R**2/(4*N)"
assert certificate["normalization"]["bound_value"] == certificate["opt_value"]
records = [(c, "initial", None, None) for c in pb.init_conditions]
assert not ctx.oper_to_duplets
for owner in ctx.func_to_triplets:
    owner.get_interpolation_constraints(ctx)
    data = ctx.get_constraint_data(owner)
    assert not data.psd_dict
    records.extend(
        (c, "interpolation", str(owner), str(group))
        for group, family in data.sc_dict.items()
        for c in family
    )
assert len({c.name for c, *_ in records}) == len(records)
assert (
    {c.name for c, *_ in records}
    == set(certificate["named_dual_values"])
    == set(certificate["constraint_metadata"])
)
basis = ctx.basis_vectors()
basis_labels = [str(v) for v in basis]
assert (
    basis_labels
    == certificate["basis_vectors"]
    == certificate["S_row_names"]
    == certificate["S_col_names"]
)
assert [str(v) for v in ctx.basis_scalars()] == certificate["basis_scalars"]
pm = pf.ExpressionManager(ctx, resolve_parameters=params_sp)
tol = certificate["numerical_checks"]["tolerances"]
S_array = np.asarray(certificate["S_matrix"], dtype=float)
assert S_array.shape == (len(basis), len(basis))
assert np.all(np.isfinite(S_array))
assert np.allclose(S_array, S_array.T, atol=tol["psd_tol"], rtol=0)
assert np.linalg.eigvalsh(S_array).min() >= -tol["psd_tol"]
weighted_terms, group_terms, extra_duals = {}, {}, {}
for c, origin, owner, group in records:
    assert isinstance(c, pf.ScalarConstraint)
    dropped = c.name in certificate["relaxed_constraints"]
    meta = dict(
        origin=origin,
        owner=owner,
        group=group,
        kind="scalar",
        comparator=c.cmp.value,
        shape=[],
        relaxed=dropped,
    )
    assert meta == certificate["constraint_metadata"][c.name]
    value = certificate["named_dual_values"][c.name]
    if dropped:
        assert value is None
        continue
    assert value is not None and np.isfinite(value)
    if c.cmp != pf.Comparator.EQ:
        assert value >= -tol["dual_tol"]
    residual = (
        c.rhs - c.lhs
        if c.cmp in (pf.Comparator.GE, pf.Comparator.SEQ)
        else c.lhs - c.rhs
    )
    weighted_terms[c.name] = value * residual
    if origin == "interpolation":
        group_terms[c.name] = weighted_terms[c.name]
    elif c.name != certificate["tau_name"]:
        group_terms[c.name] = weighted_terms[c.name]
        extra_duals[c.name] = value
assert set(certificate["relaxed_constraints"]).issubset(
    certificate["constraint_metadata"]
)
assert certificate["named_dual_values"]["initial_condition"] == certificate["tau_sol"]
S_scalar = sum(
    (float(S_array[i, j]) * basis[i] * basis[j] for i, j in np.ndindex(S_array.shape)),
    pf.Scalar.zero(),
)


def scalar_residual(expression):
    evaluated = pm.eval_scalar(expression)
    return [
        float(np.max(np.abs(np.asarray(values, dtype=float)), initial=0.0))
        for values in (
            evaluated.inner_prod_coords,
            evaluated.func_coords,
            evaluated.offset,
        )
    ]


assert pb.performance_metric is not None
certificate_residuals = scalar_residual(
    pb.performance_metric
    - certificate["opt_value"]
    - sum(weighted_terms.values(), pf.Scalar.zero())
    + S_scalar
)
assert max(certificate_residuals) <= tol["identity_atol"]
certificate_checked = True
# Decomposition: align supplied square factors to the same saved PSD slack.
D = np.asarray(certificate["numerical_decomposition"]["D"], dtype=float)
E = np.asarray(certificate["numerical_decomposition"]["ell_coords"], dtype=float)
assert E.shape == (3 * N_int, len(basis)) and D.shape == (3 * N_int, 3 * N_int)
assert np.allclose(D, np.diag(np.diag(D)), atol=0, rtol=0)
assert np.min(np.diag(D)) >= 0
assert certificate["numerical_decomposition"]["basis_labels"] == basis_labels
assert np.allclose(
    E.T @ D @ E,
    S_array,
    atol=tol["reconstruction_atol"],
    rtol=tol["reconstruction_rtol"],
)
A_supplied = np.sqrt(D) @ E
polar_U, polar_singular, polar_Vt = np.linalg.svd(A_supplied, full_matrices=False)
slack_eigenvalues, slack_vectors = np.linalg.eigh(S_array)
slack_sqrt = (
    slack_vectors
    @ np.diag(
        np.sqrt(np.where(slack_eigenvalues > tol["psd_tol"], slack_eigenvalues, 0))
    )
    @ slack_vectors.T
)
E = polar_U @ polar_Vt @ slack_sqrt
D = np.eye(3 * N_int)
polar_reconstruction = float(np.max(np.abs(E.T @ E - S_array)))
assert polar_reconstruction <= tol["identity_atol"]
decomposition_trial = dict(
    method="polar_aligned_supplied_squares",
    eigenvalue_cutoff=tol["psd_tol"],
    discarded_slack_eigenvalues=slack_eigenvalues[
        slack_eigenvalues <= tol["psd_tol"]
    ].tolist(),
    reconstruction_max_abs=polar_reconstruction,
    D=D.tolist(),
    ell_coords=E.tolist(),
)
ell = [sum((float(c) * v for c, v in zip(row, basis)), pf.Vector.zero()) for row in E]
S_piece_terms = {
    f"S_{i + 1}": sum(
        (float(D[j, j]) * ell[j] ** 2 for j in range(3 * i, 3 * i + 3)),
        pf.Scalar.zero(),
    )
    for i in range(N_int)
}
piece_residuals = scalar_residual(
    sum(S_piece_terms.values(), pf.Scalar.zero()) - S_scalar
)
assert max(piece_residuals) <= tol["identity_atol"]
for piece in S_piece_terms.values():
    evaluated = pm.eval_scalar(piece)
    assert (
        np.max(np.abs(evaluated.func_coords), initial=0.0) == 0
        and evaluated.offset == 0
    )
    assert (
        np.linalg.eigvalsh(np.asarray(evaluated.inner_prod_coords, dtype=float)).min()
        >= -tol["psd_tol"]
    )
# Grouping convention: each increment is a sum of nonpositive residuals minus PSD pieces.
term_assignment = {}
for name in group_terms:
    left, right = name.split(":")[1].split(",")
    assert right != "x_star"
    j = int(right.split("_")[1])
    term_assignment[name] = j - 1
S_assignment = {f"S_{i + 1}": i for i in range(N_int)}
lyap = [pf.Scalar.zero()]
partial_sum = sum(
    (group_terms[name] for name, step in term_assignment.items() if step == -1),
    pf.Scalar.zero(),
)
for k in range(N_int):
    partial_sum += sum(
        (group_terms[name] for name, step in term_assignment.items() if step == k),
        pf.Scalar.zero(),
    )
    partial_sum -= S_piece_terms[f"S_{k + 1}"]
    lyap.append(partial_sum)
rank_tolerance, rank_left, rank_right = 1e-4, 1, 1
ranks, singular_values = [], []
for k, value in enumerate(lyap):
    matrix = np.asarray(pm.eval_scalar(value).inner_prod_coords, dtype=float)
    singular = np.linalg.svd(matrix, compute_uv=False)
    singular_values.append(singular.tolist())
    rank = int(np.sum(singular > rank_tolerance))
    ranks.append(rank)
    print(f"rank V_{k}: {rank}")
    if k == 0:
        print()
retained_ranks = ranks[rank_left : N_int + 1 - rank_right]
print(
    "Interior rank is constant at this N:",
    len(retained_ranks) >= 3 and len(set(retained_ranks)) == 1,
    "boundary exclusions:",
    (rank_left, rank_right),
)
coverage_residuals = scalar_residual(
    lyap[-1] - (sum(group_terms.values(), pf.Scalar.zero()) - S_scalar)
)
assert max(coverage_residuals) <= tol["identity_atol"]
increment_residuals = []
for k in range(N_int):
    steps = {-1, 0} if k == 0 else {k}
    increment = (
        sum(
            (
                group_terms[name]
                for name, step in term_assignment.items()
                if step in steps
            ),
            pf.Scalar.zero(),
        )
        - S_piece_terms[f"S_{k + 1}"]
    )
    residual = scalar_residual(lyap[k + 1] - lyap[k] - increment)
    increment_residuals.append(residual)
    assert max(residual) <= tol["identity_atol"]
grouping_checked = True
rank_credible = all(
    not any(rank_tolerance / 2 < v < rank_tolerance * 2 for v in values)
    for values in singular_values
)
print("Coverage residuals:", coverage_residuals, "rank credible:", rank_credible)
```

```python
def V_k_basis(k):
    if k == 0:
        return []
    r0 = ctx["x_0"] - ctx["x_star"]
    if k == N_int:
        return [r0]
    xk = ctx[f"x_{k}"]
    return [r0, xk - ctx["x_star"], f.grad(xk) - f.grad(ctx["x_star"]), f.grad(xk)]


def V_k_basis_labels(k):
    if k == 0:
        return []
    if k == N_int:
        return ["r_0"]
    return ["r_0", f"r_{k}", f"a_{k}-a_star", f"a_{k}"]
```

### Coefficient matrices
We use $V_k^{\mathrm{quad}}=\sum_{i,j}C_{ij}(k)\langle b_{k,i},b_{k,j}\rangle$, so off-diagonal coefficients count twice. Every numerical entry is retained, and function gaps and offsets are checked separately.

```python
coeff_by_k, coefficient_residuals, independence = {}, {}, {}
for k in range(N_int + 1):
    vectors = V_k_basis(k)
    B = (
        np.column_stack([pm.eval_vector(v).coords for v in vectors]).astype(float)
        if vectors
        else np.zeros((len(basis), 0))
    )
    rank = int(np.linalg.matrix_rank(B, tol=1e-7)) if vectors else 0
    assert rank == len(vectors) == ranks[k]
    independence[k] = rank
    C = find_symmetric_coefficient_matrix(
        lyap[k],
        vectors,
        pep_context=ctx,
        resolve_parameters=params_sp,
        indep_tol=1e-7,
        span_tol=1e-5,
    )
    target = np.asarray(pm.eval_scalar(lyap[k]).inner_prod_coords, dtype=float)
    residual = float(np.linalg.norm(B @ C @ B.T - target, ord="fro"))
    assert residual <= 1e-5 * max(1.0, np.linalg.norm(target, ord="fro"))
    coeff_by_k[k], coefficient_residuals[k] = C, residual
    print(
        f"k={k}: rank {rank} basis",
        V_k_basis_labels(k),
        "reconstruction residual:",
        residual,
    )
    if vectors:
        pf.pprint_labeled_matrix(C, V_k_basis_labels(k), V_k_basis_labels(k))
    ev = pm.eval_scalar(lyap[k])
    print(
        "Function coordinates:",
        [
            (str(v), float(c))
            for v, c in zip(ctx.basis_scalars(), ev.func_coords)
            if abs(c) > 1e-7
        ],
        "offset:",
        ev.offset,
    )
```

```python
# Identify meaningful function gaps without discarding the saved residual.
function_coeff_by_k = {0: [0.0, 0.0]}
for k in range(1, N_int + 1):
    Fk = f(ctx[f"x_{k}"]) - f(ctx["x_star"])
    Gk = g(ctx[f"x_{k}"]) - g(ctx["x_star"])
    columns = np.column_stack(
        [pm.eval_scalar(Fk).func_coords, pm.eval_scalar(Gk).func_coords]
    ).astype(float)
    target = np.asarray(pm.eval_scalar(lyap[k]).func_coords, dtype=float)
    weights = np.linalg.lstsq(columns, target, rcond=None)[0]
    assert np.max(np.abs(columns @ weights - target), initial=0.0) <= 1e-5
    function_coeff_by_k[k] = weights.tolist()


def V_k_structure(k):
    vectors = V_k_basis(k)
    C = coeff_by_k[k]
    value = sum(
        (float(C[i, j]) * vectors[i] * vectors[j] for i, j in np.ndindex(C.shape)),
        pf.Scalar.zero(),
    )
    if k > 0:
        alpha, beta = function_coeff_by_k[k]
        value += alpha * (f(ctx[f"x_{k}"]) - f(ctx["x_star"]))
        value += beta * (g(ctx[f"x_{k}"]) - g(ctx["x_star"]))
    return value


component_residuals = {}
for k in range(N_int + 1):
    diff = pm.eval_scalar(V_k_structure(k) - lyap[k])
    values = [diff.inner_prod_coords, diff.func_coords, diff.offset]
    residual = [
        float(np.max(np.abs(np.asarray(v, dtype=float)), initial=0.0)) for v in values
    ]
    assert max(residual) <= 1e-5
    component_residuals[str(k)] = dict(zip(["gram", "function", "offset"], residual))
    print(
        f"k={k}: full-expression residuals",
        component_residuals[str(k)],
        "function-gap coefficients",
        function_coeff_by_k[k],
    )
print("All required indices 0 through N reconstructed with no constant terms.")
```

### Block 4 conclusion

The numerically reconstructed structure is
$$V_k=\sum_{i,j=0}^{3}c_{ij}(k,N,L)\langle b_{k,i},b_{k,j}\rangle
+\alpha(k,N)\,[f(x_k)-f(x_\star)]+\beta(k,N)\,[g(x_k)-g(x_\star)],\quad 1\leq k<N.$$
Here $c_{ij}=c_{ji}$, and the ten independent entries and the two function coefficients remain named analytic unknowns. At $k=0$, $V_0=0$. At $k=N$, the corresponding structure is
$$V_N=c_T(N,L)\|r_0\|^2+\alpha_T(N)[f(x_N)-f(x_\star)]+\beta_T(N)[g(x_N)-g(x_\star)].$$
There are no constant terms. All these sample expressions reconstruct every Gram, function-value and offset coordinate within the frozen tolerance. Numerical reconstruction does not establish an arbitrary-horizon theorem; Block 5 must derive coefficients and prove the exact step, base and terminal identities.


## Closed form and exact proof
The B4 numerical structure is now resolved analytically. The sparse pattern suggests a reduced coefficient ansatz; the following full-coordinate equations determine its coefficients, and the subsequent exact identities prove the selected solution without relying on numerical fitting.

### Symbolic context and residual definitions
All auxiliary contexts below encode the actual proximal update and stationarity of $f+g$ at the minimizer. The supplied square families are carried into the local equations explicitly.

```python
# Exact local PGM identities; this source is included visibly in the notebook.
from typing import Any

import numpy as np
import pepflow as pf
import sympy as sp

k_sym, N_sym = sp.symbols("k N", integer=True)
L_sym = sp.Symbol("L", positive=True)
KP, NP, LP = pf.Parameter("k_proof"), pf.Parameter("N_proof"), pf.Parameter("L_proof")
symbolic_parameters: dict[str, Any] = {
    "k_proof": k_sym,
    "N_proof": N_sym,
    "L_proof": L_sym,
}


def residual_f(fun, u, v):
    return (
        fun(v)
        - fun(u)
        + fun.grad(v) * (u - v)
        + (fun.grad(u) - fun.grad(v)) ** 2 / (2 * LP)
    )


def residual_g(fun, u, v):
    return fun(v) - fun(u) + fun.grad(v) * (u - v)


def supplied_piece(fun, convex, xprev, xnext, xstar, i):
    ap, ai, ast = fun.grad(xprev), fun.grad(xnext), fun.grad(xstar)
    si = convex.grad(xnext)
    c1 = NP / ((2 * NP + 1 - i) * (2 * NP - i) * LP)
    u1 = i / (2 * NP) * ai + (2 * NP - i) / (2 * NP) * ap - ast
    c2 = (
        (i - 1) / (2 * NP - i) * (2 * NP + 2) / (2 * NP + 1)
        + 2 * NP / ((2 * NP - i) ** 2 * (2 * NP + 1))
    ) / (2 * LP)
    u2 = ap + si - LP / (2 * NP + 1 - i) * (xprev - xstar)
    c3 = i / (2 * NP + 1 - i) * (2 * NP) / (2 * NP + 1) / (2 * LP)
    u3 = (
        (2 * NP + 1) / (2 * NP) * (ai + si)
        - (ap + si) / (2 * NP)
        - LP / (2 * NP - i) * (xnext - xstar)
    )
    return c1 * u1**2 + c2 * u2**2 + c3 * u3**2


def analytic_potential(fun, convex, x, xstar, r0, index):
    r = x - xstar
    a, ast = fun.grad(x), fun.grad(xstar)
    return (
        -LP / (4 * NP) * r0**2
        + LP * (NP - index) / (2 * NP - index) ** 2 * r**2
        + (a - ast) ** 2 / (2 * LP * (2 * NP - index))
        - r * a / (2 * NP - index)
        + (index + 1) / (2 * NP - index) * (fun(x) - fun(xstar))
        + index / (2 * NP - index) * (convex(x) - convex(xstar))
    )


def symbolic_components(expression, manager):
    evaluated = manager.eval_scalar(expression, sympy_mode=True)
    gram = sp.Matrix(evaluated.inner_prod_coords).applyfunc(sp.cancel)
    functions = sp.Matrix(list(evaluated.func_coords)).applyfunc(sp.cancel)
    offset = sp.cancel(evaluated.offset)
    return gram, functions, offset


def verify_identity(label, expression, context, parameters):
    manager = pf.ExpressionManager(context, resolve_parameters=parameters)
    gram, functions, offset = symbolic_components(expression, manager)
    passed = all(value == 0 for value in [*gram, *functions, offset])
    print(
        label,
        "exact zero:",
        passed,
        "function residual:",
        functions.T,
        "offset:",
        offset,
    )
    pf.pprint_labeled_matrix(
        np.asarray(gram, dtype=object),
        [str(v) for v in context.basis_vectors()],
        [str(v) for v in context.basis_vectors()],
    )
    assert passed, (label, gram, functions, offset)
    return True
```

### Derive the coefficients
Use the ansatz $A_{k}\|r_{k}\|^2+B_{k}\|a_{k}-a_{\star}\|^2+C_{k}\langle r_{k},a_{k}\rangle+\alpha_{k}F_{k}+\beta_{k}G_{k}$ plus a constant multiple of $\|r_{0}\|^2$. Equate all Gram, function-value and constant coordinates of its difference with the supported local interpolation/SOS combination. The base identity fixes the remaining constant.


```python
# Solve the exact coefficient equations in the sparsity pattern suggested by B4.
ctx_derive = pf.PEPContext("pgm_coefficient_derivation").set_as_current()
f_derive = pf.SmoothConvexFunction(is_basis=True, tags=["f_{derive}"], L=LP)
g_derive = pf.ConvexFunction(is_basis=True, tags=["g_{derive}"])
xs_derive = (f_derive + g_derive).set_stationary_point("x_{star,derive}")
x_derive = pf.Vector(is_basis=True, tags=["x_{k,derive}"])
f_derive.grad(x_derive)
g_derive(x_derive)
xn_derive = g_derive.prox(
    x_derive - f_derive.grad(x_derive) / LP, 1 / LP, tag="x_{k+1,derive}"
)
f_derive.grad(xn_derive)
unknown_names = [
    "A_k",
    "B_k",
    "C_k",
    "alpha_k",
    "beta_k",
    "A_next",
    "B_next",
    "C_next",
    "alpha_next",
    "beta_next",
]
unknown_symbols = sp.symbols(" ".join(unknown_names), real=True)
unknown_parameters = [pf.Parameter(name) for name in unknown_names]


def coefficient_ansatz(fun, convex, x, xstar, coefficients):
    aa, bb, cc, ff, gg = coefficients
    r = x - xstar
    a, ast = fun.grad(x), fun.grad(xstar)
    return (
        aa * r**2
        + bb * (a - ast) ** 2
        + cc * r * a
        + ff * (fun(x) - fun(xstar))
        + gg * (convex(x) - convex(xstar))
    )


derive_LHS = coefficient_ansatz(
    f_derive, g_derive, xn_derive, xs_derive, unknown_parameters[5:]
) - coefficient_ansatz(f_derive, g_derive, x_derive, xs_derive, unknown_parameters[:5])
derive_RHS = (
    (2 * NP + 1)
    / ((2 * NP - KP - 1) * (2 * NP - KP))
    * residual_f(f_derive, xs_derive, xn_derive)
    + (KP + 1) / (2 * NP - KP) * residual_f(f_derive, x_derive, xn_derive)
    + 2
    * NP
    / ((2 * NP - KP - 1) * (2 * NP - KP))
    * residual_g(g_derive, xs_derive, xn_derive)
    + KP / (2 * NP - KP) * residual_g(g_derive, x_derive, xn_derive)
    - supplied_piece(f_derive, g_derive, x_derive, xn_derive, xs_derive, KP + 1)
)
derive_parameters: dict[str, Any] = {
    **symbolic_parameters,
    **dict(zip(unknown_names, unknown_symbols)),
}
derive_manager = pf.ExpressionManager(ctx_derive, resolve_parameters=derive_parameters)
derive_gram, derive_functions, derive_offset = symbolic_components(
    derive_LHS - derive_RHS, derive_manager
)
coefficient_equations = list(
    dict.fromkeys(
        value
        for value in [*derive_gram, *derive_functions, derive_offset]
        if value != 0
    )
)
coefficient_solutions = sp.solve(
    coefficient_equations, unknown_symbols, dict=True, simplify=False
)
assert len(coefficient_solutions) == 1
coefficient_solution = {
    str(symbol): sp.factor(value) for symbol, value in coefficient_solutions[0].items()
}
print("Derived coefficients from full-coordinate step equations:")
for name, value in coefficient_solution.items():
    print(name, "=", value)
expected_current = [
    L_sym * (N_sym - k_sym) / (2 * N_sym - k_sym) ** 2,
    1 / (2 * L_sym * (2 * N_sym - k_sym)),
    -1 / (2 * N_sym - k_sym),
    (k_sym + 1) / (2 * N_sym - k_sym),
    k_sym / (2 * N_sym - k_sym),
]
expected_next = [sp.cancel(value.subs(k_sym, k_sym + 1)) for value in expected_current]
assert all(
    sp.cancel(coefficient_solutions[0][symbol] - expected) == 0
    for symbol, expected in zip(unknown_symbols, expected_current + expected_next)
)
assert all(
    sp.cancel(eq.subs(coefficient_solutions[0])) == 0 for eq in coefficient_equations
)
# The recurrence leaves an additive constant multiple of ||r_0||^2.
# Determine it from the initial identity, then verify the terminal normalization.
ctx_anchor = pf.PEPContext("pgm_anchor_derivation").set_as_current()
f_anchor = pf.SmoothConvexFunction(is_basis=True, tags=["f_{anchor}"], L=LP)
g_anchor = pf.ConvexFunction(is_basis=True, tags=["g_{anchor}"])
xs_anchor = (f_anchor + g_anchor).set_stationary_point("x_{star,anchor}")
x0_anchor = pf.Vector(is_basis=True, tags=["x_{0,anchor}"])
f_anchor.grad(x0_anchor)
x1_anchor = g_anchor.prox(
    x0_anchor - f_anchor.grad(x0_anchor) / LP, 1 / LP, tag="x_{1,anchor}"
)
f_anchor.grad(x1_anchor)
t_anchor = pf.Parameter("t_anchor")
t_symbol = sp.Symbol("t_anchor", real=True)
anchor_coefficients = [
    LP * (NP - 1) / (2 * NP - 1) ** 2,
    1 / (2 * LP * (2 * NP - 1)),
    -1 / (2 * NP - 1),
    2 / (2 * NP - 1),
    1 / (2 * NP - 1),
]
anchor_LHS = (
    coefficient_ansatz(f_anchor, g_anchor, x1_anchor, xs_anchor, anchor_coefficients)
    + t_anchor * (x0_anchor - xs_anchor) ** 2
)
anchor_RHS = (
    residual_f(f_anchor, xs_anchor, x0_anchor) / (2 * NP)
    + (2 * NP + 1)
    / ((2 * NP - 1) * 2 * NP)
    * residual_f(f_anchor, xs_anchor, x1_anchor)
    + residual_f(f_anchor, x0_anchor, x1_anchor) / (2 * NP)
    + residual_g(g_anchor, xs_anchor, x1_anchor) / (2 * NP - 1)
    - supplied_piece(f_anchor, g_anchor, x0_anchor, x1_anchor, xs_anchor, sp.S(1))
)
anchor_parameters: dict[str, Any] = {**symbolic_parameters, "t_anchor": t_symbol}
anchor_manager = pf.ExpressionManager(ctx_anchor, resolve_parameters=anchor_parameters)
anchor_gram, anchor_functions, anchor_offset = symbolic_components(
    anchor_LHS - anchor_RHS, anchor_manager
)
anchor_equations = [
    value for value in [*anchor_gram, *anchor_functions, anchor_offset] if value != 0
]
anchor_solution = sp.solve(anchor_equations, t_symbol, dict=True)
assert anchor_solution == [{t_symbol: -L_sym / (4 * N_sym)}]
print("Initial-distance coefficient:", anchor_solution)
```

### Closed-form conclusion
For $1\leq k<N$,
$$V_{k}=-\frac{L\|r_{0}\|^2}{4N}+\frac{L(N-k)\|r_{k}\|^2}{(2N-k)^2}
+\frac{\|a_{k}-a_{\star}\|^2}{2L(2N-k)}-\frac{\langle r_{k},a_{k}\rangle}{2N-k}
+\frac{(k+1)F_{k}+kG_{k}}{2N-k}.$$
Set $V_{0}=0$ and $V_{N}=F_{N}+G_{N}-L\|r_{0}\|^2/(4N)$. The next sections verify the exact identities for these expressions; they establish the formulas over arbitrary horizons, not merely the observed samples.


## Symbolic Step Recursion Verification

For $1\leq k\leq N-2$, with $i=k+1$, verify
$$\mathrm{LHS}=V_{k+1}-V_{k},\quad
\mathrm{RHS}=\eta^{f}_{i}I^{f}(x_{\star},x_{i})+\theta^{f}_{i}I^{f}(x_{i-1},x_{i})
+\eta^{g}_{i}I^{g}(x_{\star},x_{i})+\theta^{g}_{i}I^{g}(x_{i-1},x_{i})-S_{i}.$$
The residual $\mathrm{LHS}-\mathrm{RHS}$ must vanish in every coordinate.


```python
ctx_step = pf.PEPContext("pgm_symbolic_step").set_as_current()
f_step = pf.SmoothConvexFunction(is_basis=True, tags=["f_{step}"], L=LP)
g_step = pf.ConvexFunction(is_basis=True, tags=["g_{step}"])
xstar_step = (f_step + g_step).set_stationary_point("x_{star,step}")
r0_step = pf.Vector(is_basis=True, tags=["r_{0,step}"])
x_step = pf.Vector(is_basis=True, tags=["x_{k}"])
f_step.grad(x_step)
g_step(x_step)
xn_step = g_step.prox(x_step - f_step.grad(x_step) / LP, 1 / LP, tag="x_{k+1}")
f_step.grad(xn_step)
step_LHS = analytic_potential(
    f_step, g_step, xn_step, xstar_step, r0_step, KP + 1
) - analytic_potential(f_step, g_step, x_step, xstar_step, r0_step, KP)
step_RHS = (
    (2 * NP + 1)
    / ((2 * NP - KP - 1) * (2 * NP - KP))
    * residual_f(f_step, xstar_step, xn_step)
    + (KP + 1) / (2 * NP - KP) * residual_f(f_step, x_step, xn_step)
    + 2
    * NP
    / ((2 * NP - KP - 1) * (2 * NP - KP))
    * residual_g(g_step, xstar_step, xn_step)
    + KP / (2 * NP - KP) * residual_g(g_step, x_step, xn_step)
    - supplied_piece(f_step, g_step, x_step, xn_step, xstar_step, KP + 1)
)
step_diff = step_LHS - step_RHS
step_verified = verify_identity(
    "Interior step, 1<=k<=N-2", step_diff, ctx_step, symbolic_parameters
)
```

## Base Case and Boundary Symbolic Verification

For $N\geq2$, verify
$$V_{1}=\frac{I^{f}(x_{\star},x_{0})}{2N}+\frac{2N+1}{(2N-1)2N}I^{f}(x_{\star},x_{1})
+\frac{I^{f}(x_{0},x_{1})}{2N}+\frac{I^{g}(x_{\star},x_{1})}{2N-1}-S_{1}.$$
The residual of the left-hand side minus the right-hand side must vanish in every coordinate. This uses the actual starting point and $V_{0}=0$.


```python
ctx_base = pf.PEPContext("pgm_symbolic_base").set_as_current()
f_base = pf.SmoothConvexFunction(is_basis=True, tags=["f_{base}"], L=LP)
g_base = pf.ConvexFunction(is_basis=True, tags=["g_{base}"])
xstar_base = (f_base + g_base).set_stationary_point("x_{star,base}")
x0_base = pf.Vector(is_basis=True, tags=["x_{0,base}"])
f_base.grad(x0_base)
x1_base = g_base.prox(x0_base - f_base.grad(x0_base) / LP, 1 / LP, tag="x_{1,base}")
f_base.grad(x1_base)
base_LHS = analytic_potential(
    f_base, g_base, x1_base, xstar_base, x0_base - xstar_base, sp.S(1)
)
base_RHS = (
    residual_f(f_base, xstar_base, x0_base) / (2 * NP)
    + (2 * NP + 1) / ((2 * NP - 1) * 2 * NP) * residual_f(f_base, xstar_base, x1_base)
    + residual_f(f_base, x0_base, x1_base) / (2 * NP)
    + residual_g(g_base, xstar_base, x1_base) / (2 * NP - 1)
    - supplied_piece(f_base, g_base, x0_base, x1_base, xstar_base, sp.S(1))
)
base_diff = base_LHS - base_RHS
base_verified = verify_identity("Base, N>=2", base_diff, ctx_base, symbolic_parameters)
```

### Boundary Identity Symbolic Verification

For $N\geq2$, verify
$$V_{N}-V_{N-1}=\frac{I^{f}(x_{\star},x_{N})}{N+1}+\frac{N I^{f}(x_{N-1},x_{N})}{N+1}
+\frac{2I^{g}(x_{\star},x_{N})}{N+1}+\frac{(N-1)I^{g}(x_{N-1},x_{N})}{N+1}-S_{N}.$$
Here $V_{N}=F_{N}+G_{N}-L\|r_{0}\|^2/(4N)$; the residual $\mathrm{LHS}-\mathrm{RHS}$ must vanish in every coordinate.


```python
ctx_terminal = pf.PEPContext("pgm_symbolic_terminal").set_as_current()
f_terminal = pf.SmoothConvexFunction(is_basis=True, tags=["f_{terminal}"], L=LP)
g_terminal = pf.ConvexFunction(is_basis=True, tags=["g_{terminal}"])
xstar_terminal = (f_terminal + g_terminal).set_stationary_point("x_{star,terminal}")
r0_terminal = pf.Vector(is_basis=True, tags=["r_{0,terminal}"])
xp_terminal = pf.Vector(is_basis=True, tags=["x_{N-1}"])
f_terminal.grad(xp_terminal)
g_terminal(xp_terminal)
xn_terminal = g_terminal.prox(
    xp_terminal - f_terminal.grad(xp_terminal) / LP, 1 / LP, tag="x_{N}"
)
f_terminal.grad(xn_terminal)
V_terminal = (
    f_terminal(xn_terminal)
    - f_terminal(xstar_terminal)
    + g_terminal(xn_terminal)
    - g_terminal(xstar_terminal)
    - LP / (4 * NP) * r0_terminal**2
)
terminal_LHS = V_terminal - analytic_potential(
    f_terminal, g_terminal, xp_terminal, xstar_terminal, r0_terminal, NP - 1
)
terminal_RHS = (
    residual_f(f_terminal, xstar_terminal, xn_terminal) / (NP + 1)
    + NP / (NP + 1) * residual_f(f_terminal, xp_terminal, xn_terminal)
    + 2 / (NP + 1) * residual_g(g_terminal, xstar_terminal, xn_terminal)
    + (NP - 1) / (NP + 1) * residual_g(g_terminal, xp_terminal, xn_terminal)
    - supplied_piece(
        f_terminal, g_terminal, xp_terminal, xn_terminal, xstar_terminal, NP
    )
)
terminal_diff = terminal_LHS - terminal_RHS
boundary_verified = verify_identity(
    "Terminal, N>=2", terminal_diff, ctx_terminal, symbolic_parameters
)
```

### Single-iteration boundary

For $N=1$, verify directly
$$F_{1}+G_{1}-\frac{L\|r_{0}\|^2}{4}
=\frac{I^{f}(x_{\star},x_{0})+I^{f}(x_{\star},x_{1})+I^{f}(x_{0},x_{1})}{2}
+I^{g}(x_{\star},x_{1})-S_{1}.$$
The residual must vanish; no interior formula is evaluated outside its domain.


```python
ctx_single = pf.PEPContext("pgm_symbolic_single").set_as_current()
f_single = pf.SmoothConvexFunction(is_basis=True, tags=["f_{single}"], L=LP)
g_single = pf.ConvexFunction(is_basis=True, tags=["g_{single}"])
xstar_single = (f_single + g_single).set_stationary_point("x_{star,single}")
x0_single = pf.Vector(is_basis=True, tags=["x_{0,single}"])
f_single.grad(x0_single)
x1_single = g_single.prox(
    x0_single - f_single.grad(x0_single) / LP, 1 / LP, tag="x_{1,single}"
)
f_single.grad(x1_single)
single_LHS = (
    f_single(x1_single)
    - f_single(xstar_single)
    + g_single(x1_single)
    - g_single(xstar_single)
    - LP / sp.S(4) * (x0_single - xstar_single) ** 2
)
single_RHS = (
    sp.Rational(1, 2) * residual_f(f_single, xstar_single, x0_single)
    + sp.Rational(1, 2) * residual_f(f_single, xstar_single, x1_single)
    + sp.Rational(1, 2) * residual_f(f_single, x0_single, x1_single)
    + residual_g(g_single, xstar_single, x1_single)
    - supplied_piece(f_single, g_single, x0_single, x1_single, xstar_single, sp.S(1))
)
single_diff = single_LHS - single_RHS
single_verified = verify_identity(
    "Single iteration, N=1",
    single_diff,
    ctx_single,
    {**symbolic_parameters, "N_proof": sp.S(1)},
)
```

### Signs, domain, and normalization

For $1\leq i\leq N$, write $i=1+u$ and $N=1+u+v$ with nonnegative integers $u,v$. All residual multipliers and square coefficients become nonnegative rational functions with positive denominators when $L>0$. Also $2N-k\geq N+1>0$ on the interior potential's domain. The terminal convention yields the exact normalization identity
$$h(x_{N})-h(x_{\star})-\frac{LR^2}{4N}
=V_{N}+\frac{L}{4N}(\|r_{0}\|^2-R^2).$$
Its residual is checked below. Both terms on the right are nonpositive by the verified decreases and the initial constraint. This also covers $R=0$ without dividing by $R$.

The consecutive multipliers telescope into the star multipliers:
$$\frac{j+1}{2N-j}-\frac{j}{2N+1-j}=\frac{2N+1}{(2N-j)(2N+1-j)},$$
$$\frac{j}{2N-j}-\frac{j-1}{2N+1-j}=\frac{2N}{(2N-j)(2N+1-j)}.$$
Together with the explicit endpoint coefficients, this verifies that summing the local identities gives the complete analytic full-PEP certificate, with no omitted interpolation family, constant or boundary term.
