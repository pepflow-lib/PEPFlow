# Proximal gradient: numerical discovery and Lyapunov analysis
Let $f$ be convex and $L$-smooth, $g$ closed, proper and convex, and $h=f+g$ on a finite-dimensional real Euclidean space. Assume $L>0$, $R\geq0$, $N\geq1$, and $x_{\star}\in\arg\min h$ with $\|x_{0}-x_{\star}\|^{2}\leq R^{2}$. Proximal steps are well-defined.
$$x_{k+1}=\operatorname{prox}_{g/L}\left(x_{k}-\frac{1}{L}\nabla f(x_{k})\right),\qquad k=0,\ldots,N-1.$$
The metric is $h(x_{N})-h(x_{\star})$. The equivalent proximal relation is $s_{k+1}=L(x_{k}-x_{k+1})-\nabla f(x_{k})\in\partial g(x_{k+1})$.
The fixed PGM SOS expression in the supplied input is authorized assistance; its certificate role is verified in the exact proof below.

## Proof Statement

### Theorem
Under the assumptions stated above, proximal gradient with step $1/L$ satisfies
$$0\leq h(x_{N})-h(x_{\star})\leq \frac{LR^{2}}{4N}.$$
For $N\geq2$, write $f_{i}=f(x_{i})$, $g_{i}=g(x_{i})$, $h_{i}=f_{i}+g_{i}$, $a_{i}=\nabla f(x_{i})$ and $d_{i}=x_{i}-x_{\star}$. Set $V_{0}=0$, and for $1\leq k<N$ define
$$V_{k}=-\frac{L}{4N}\|d_{0}\|^{2}+\frac{L(N-k)}{(2N-k)^{2}}\|d_{k}\|^{2}+\frac{\|a_{k}-a_{\star}\|^{2}}{2L(2N-k)}-\frac{\langle d_{k},a_{k}\rangle}{2N-k}+\frac{k+1}{2N-k}(f_{k}-f_{\star})+\frac{k}{2N-k}(g_{k}-g_{\star}).$$
At the terminal index define
$$V_{N}=h(x_{N})-h(x_{\star})-\frac{L}{4N}\|d_{0}\|^{2}.$$
For $N=1$, use only the two exceptional values $V_{0}=0$ and the displayed terminal formula.
The sequence is nonincreasing. The bound is sharp.

### Proof outline
Use the full potential
$$V_{k}=-\frac{L}{4N}\|d_{0}\|^{2}+\frac{L(N-k)}{(2N-k)^{2}}\|d_{k}\|^{2}+\frac{\|a_{k}-a_{\star}\|^{2}}{2L(2N-k)}-\frac{\langle d_{k},a_{k}\rangle}{2N-k}+\frac{k+1}{2N-k}(f_{k}-f_{\star})+\frac{k}{2N-k}(g_{k}-g_{\star}),\quad1\leq k<N,$$
with $V_{0}=0$ and $V_{N}=h_{N}-h_{\star}-L\|d_{0}\|^{2}/(4N)$.
Define the nonpositive interpolation residuals
$$I_{f}(u,v)=f(v)-f(u)+\langle\nabla f(v),u-v\rangle+\frac{\|\nabla f(u)-\nabla f(v)\|^{2}}{2L}\leq0,$$
$$I_{g}(u,v)=g(v)-g(u)+\langle s_{v},u-v\rangle\leq0,$$
where $s_{i}=L(x_{i-1}-x_{i})-a_{i-1}$ and $s_{\star}=-a_{\star}$. Write $I_{f}(\star,i)=I_{f}(x_{\star},x_{i})$ and likewise for other pairs. Let
$$\alpha_{i}=\frac{i}{2N+1-i},\qquad\beta_{i}=\frac{i}{2N-i},\quad\beta_{0}=0.$$
The supplied SOS aid, now verified as part of this proof, is $S_{i}=c_{1,i}\|u_{1,i}\|^{2}+c_{2,i}\|u_{2,i}\|^{2}+c_{3,i}\|u_{3,i}\|^{2}$, with
$$c_{1,i}=\frac{N}{(2N+1-i)(2N-i)L},\quad u_{1,i}=\frac{i}{2N}a_{i}+\frac{2N-i}{2N}a_{i-1}-a_{\star},$$
$$c_{2,i}=\frac1{2L}\left(\frac{i-1}{2N-i}\frac{2N+2}{2N+1}+\frac{2N}{(2N-i)^{2}(2N+1)}\right),\quad u_{2,i}=a_{i-1}+s_{i}-\frac{L}{2N+1-i}d_{i-1},$$
$$c_{3,i}=\frac{i}{2N+1-i}\frac{2N}{2N+1}\frac1{2L},\quad u_{3,i}=\frac{2N+1}{2N}(a_{i}+s_{i})-\frac{a_{i-1}+s_{i}}{2N}-\frac{L}{2N-i}d_{i}.$$
For $N\geq2$, the base identity is
$$V_{1}=\alpha_{1} I_{f}(\star,0)+(\alpha_{2}-\alpha_{1})I_{f}(\star,1)+\alpha_{1} I_{f}(0,1)+\beta_{1} I_{g}(\star,1)-S_{1}.$$
For $2\leq i\leq N-1$, the interior identity is
$$V_{i}-V_{i-1}=(\alpha_{i+1}-\alpha_{i})I_{f}(\star,i)+\alpha_{i} I_{f}(i-1,i)+(\beta_{i}-\beta_{i-1})I_{g}(\star,i)+\beta_{i-1}I_{g}(i-1,i)-S_{i}.$$
The final identity is
$$V_{N}-V_{N-1}=(1-\alpha_{N})I_{f}(\star,N)+\alpha_{N} I_{f}(N-1,N)+(1-\beta_{N-1})I_{g}(\star,N)+\beta_{N-1}I_{g}(N-1,N)-S_{N}.$$
All coefficients are nonnegative: $2N-i\geq N>0$, $2N+1-i\geq N+1>0$;
$$\alpha_{i+1}-\alpha_{i}=\frac{2N+1}{(2N-i)(2N+1-i)},\qquad\beta_{i}-\beta_{i-1}=\frac{2N}{(2N-i)(2N+1-i)},$$
and $1-\alpha_{N}=1/(N+1)$, $1-\beta_{N-1}=2/(N+1)$. The three SOS coefficients are nonnegative term by term. Thus $V_{N}\leq\cdots\leq V_{1}\leq V_{0}=0$ and
$$h_{N}-h_{\star}\leq\frac{L}{4N}\|d_{0}\|^{2}\leq\frac{LR^{2}}{4N}.$$
For $N=1$, the separate identity is
$$V_{1}=\tfrac12 I_{f}(\star,0)+\tfrac12 I_{f}(\star,1)+\tfrac12 I_{f}(0,1)+I_{g}(\star,1)-S_{1}\leq0.$$
This also covers $R=0$; the proof never divides by $R$. For sharpness with $R>0$, take $f=0$, $g(x)=c\max(x,0)$, $c=LR/(2N)$, $x_{0}=R$, $x_{\star}=0$. Then $x_{k}=R-kR/(2N)>0$ for $0\leq k\leq N$, so $h_{N}-h_{\star}=LR^{2}/(4N)$.

```python
import json
import matplotlib.pyplot as plt
import pepflow as pf
```

```python
L = pf.Parameter("L")
R = pf.Parameter("R")
f = pf.SmoothConvexFunction(is_basis=True, tags=["f"], L=L)
g = pf.ConvexFunction(is_basis=True, tags=["g"])
h = f + g

```

```python
def make_ctx_pgm(ctx_name, N, **kwargs):
    ctx = pf.PEPContext(ctx_name).set_as_current()
    x = pf.Vector(is_basis=True, tags=["x_0"])
    h.set_stationary_point("x_star")
    for i in range(N):
        x = g.prox(x - f.grad(x) / L, 1 / L, tag=f"x_{i + 1}")
    f(x)
    return ctx


def get_pep_setup(N, params):
    ctx = make_ctx_pgm(f"ctx_{N}", N)
    pb = pf.PEPBuilder(ctx)
    pb.add_initial_constraint(
        ((ctx["x_0"] - ctx["x_star"]) ** 2).le(R ** 2, name="initial_condition")
    )
    pb.set_performance_metric(h(ctx[f"x_{N}"]) - h(ctx["x_star"]))
    return ctx, pb, f
```

## Dense and relaxed proof solves
A dense CLARABEL solve was followed by dropping entries below $10^{-6}$, then the consecutive/optimizer relaxation shown below. After terminal reconstruction exposed numerical ambiguity, the final certificate was captured from a dual solve with the derived interpolation-multiplier candidates imposed as equalities and S left as a free PSD variable. CLARABEL defaults were used. All original residuals are LE, so the certificate sign is performance minus weighted residuals plus S; there are no original equality or matrix residuals. Saved values are replayed and checked without a new solve.

```python
import hashlib
import json
import numpy as np
import sympy as sp
import pepflow as pf
from pepflow.utils import NUMERICAL_TYPE

b1 = json.loads((notebook_dir / 'state/pgm_b1.json').read_text())
dense = json.loads((notebook_dir / 'state/pgm_dense.json').read_text())
raw = json.loads((notebook_dir / 'state/pgm_selected_raw.json').read_text())
N_int = raw['N']
N = sp.Integer(N_int)
params_sp: dict[str, NUMERICAL_TYPE] = {'L': sp.Integer(1), 'R': sp.Integer(1)}
ctx, pb, obj = get_pep_setup(N_int, params_sp)
pm = pf.ExpressionManager(ctx, resolve_parameters=params_sp)
basis = ctx.basis_vectors()
assert raw['basis_vectors'] == [str(v) for v in basis]
assert raw['basis_scalars'] == [str(v) for v in ctx.basis_scalars()]
for name, digest in raw['source_hashes'].items():
    assert hashlib.sha256((notebook_dir / name).read_bytes()).hexdigest() == digest
print('Dense objective:', dense['opt_value'])
print('Basis:', raw['basis_vectors'])
for owner in ('f', 'g'):
    print('Dense active entries', owner, sum(v > 1e-6 for k,v in dense['named_dual_values'].items() if k.startswith(owner+':')))
print('Dense S rank (tol=1e-4):', np.linalg.matrix_rank(np.array(dense['S_matrix']), tol=1e-4))
# Relaxation recipe: retain only optimizer-to-iterate and forward consecutive pairs.
entries = list(pb.init_conditions)
for owner in (f, g):
    entries.extend(owner.get_interpolation_constraints(ctx))
def retained(name):
    if ':' not in name:
        return True
    left, right = name.split(':')[1].split(',')
    return left == 'x_star' or (right != 'x_star' and int(right[2:]) == int(left[2:]) + 1)
assert set(raw['relaxed_constraints']) == {c.name for c in entries if not retained(c.name)}
assert set(raw['named_dual_values']) == {c.name for c in entries}
print('Relaxed constraints dropped:', len(raw['relaxed_constraints']))
print('Selected objective:', raw['opt_value'], 'status:', raw['status'])
print('Dense/relaxed gap:', abs(dense['opt_value'] - raw['opt_value']))

```

## Lambda and S verification
The candidate consecutive multipliers are $\alpha_{i}=i/(2N+1-i)$ for $f$ and $\beta_{i}=i/(2N-i)$ for $g$. Optimizer-row multipliers are differences of consecutive coefficients, with the terminal $f$ weight $1-\alpha_{N}$ and $\beta_{N}=1$. They are only numerical candidates here.
The supplied SOS is declared assistance. Its scale relative to this objective normalization is one. All three squared vectors per iteration are retained. The numerical certificate, candidate comparison, and full analytic proof have separate statuses.

```python
# Candidate multipliers inferred from the selected consecutive-pair sample.
def lambda_candidate(owner, left, right, n):
    j = int(right[2:])
    def alpha(t):
        return sp.Rational(t, 2*n+1-t)
    def beta(t):
        return sp.Rational(t, 2*n-t)
    if owner == 'f':
        return alpha(j) if left != 'x_star' else (alpha(j+1)-alpha(j) if j<n else 1-alpha(n))
    return beta(j-1) if left != 'x_star' else beta(j)-beta(j-1)

lambda_errors = []
for owner in ('f', 'g'):
    labels = ['x_star'] + [f'x_{i}' for i in range(0 if owner == 'f' else 1,N_int+1)]
    candidate_matrix = np.zeros((len(labels), len(labels)))
    for row,left in enumerate(labels):
        for col,right in enumerate(labels):
            name = f'{owner}:{left},{right}'
            value = raw['named_dual_values'].get(name)
            if value is not None:
                candidate = lambda_candidate(owner,left,right,N_int)
                candidate_matrix[row,col] = float(candidate)
                lambda_errors.append(abs(float(candidate)-value))
                print(name, value, 'candidate', candidate)
    print('Candidate lambda', owner)
    pf.pprint_labeled_matrix(candidate_matrix, labels)
assert max(lambda_errors) <= 1e-4

# Supplied SOS aid, implemented in exact rational arithmetic.
def supplied_sos(ctx, n, smoothness):
    n = sp.Integer(n)
    smoothness = sp.sympify(smoothness)
    star = ctx['x_star']
    astar = f.grad(star)
    pieces, coefficients, vectors = [], [], []
    for i in range(1,int(n)+1):
        xp, x = ctx[f'x_{i-1}'], ctx[f'x_{i}']
        ap, a, s = f.grad(xp), f.grad(x), g.grad(x)
        c1 = n/((2*n+1-i)*(2*n-i)*smoothness)
        u1 = i/(2*n)*a+(2*n-i)/(2*n)*ap-astar
        c2 = ((i-1)/(2*n-i)*(2*n+2)/(2*n+1)+2*n/((2*n-i)**2*(2*n+1)))/(2*smoothness)
        u2 = ap+s-smoothness/(2*n+1-i)*(xp-star)
        c3 = i/(2*n+1-i)*(2*n)/(2*n+1)/(2*smoothness)
        u3 = (2*n+1)/(2*n)*(a+s)-(ap+s)/(2*n)-smoothness/(2*n-i)*(x-star)
        coefficients.extend([c1,c2,c3])
        vectors.extend([u1,u2,u3])
        pieces.append(c1*u1**2+c2*u2**2+c3*u3**2)
    return pieces, coefficients, vectors

pieces, coefficients, vectors = supplied_sos(ctx, N_int, 1)
S_candidate = sum(pieces, pf.Scalar.zero())
S_eval = pm.eval_scalar(S_candidate)
S_array = np.array(raw['S_matrix'])
D = np.diag(np.array(coefficients,dtype=float))
E = np.array([pm.eval_vector(v).coords for v in vectors],dtype=float)
S_reconstructed = E.T @ D @ E
assert np.all(np.array(coefficients,dtype=float) >= 0)
assert np.allclose(S_reconstructed,S_array,atol=1e-4,rtol=1e-5)
assert np.max(np.abs(np.array(S_eval.func_coords,dtype=float)),initial=0) == 0
assert float(S_eval.offset) == 0
print('Supplied aid normalization factor: 1')
print('SOS reconstruction residual:', np.max(np.abs(S_reconstructed-S_array)))
pf.pprint_labeled_matrix(S_reconstructed,raw['basis_vectors'])

# Complete numerical identity including initial constraint and both functions.
weighted = pf.Scalar.zero()
inequality_duals = []
for c in entries:
    value = raw['named_dual_values'][c.name]
    if raw['constraint_metadata'][c.name]['relaxed']:
        assert value is None
        continue
    assert value is not None
    assert c.cmp.value == 'LE'
    weighted += value * (c.lhs-c.rhs)
    inequality_duals.append(value)
diff = pm.eval_scalar(pb.performance_metric-raw['opt_value']-weighted+S_candidate)
def maxabs(x):
    return float(np.max(np.abs(np.asarray(x,dtype=float)),initial=0))
residuals = dict(identity_gram_max_abs=maxabs(diff.inner_prod_coords),identity_func_max_abs=maxabs(diff.func_coords),identity_offset_abs=maxabs(diff.offset),S_reconstruction_max_abs=maxabs(S_reconstructed-S_array),dense_relaxed_gap=abs(dense['opt_value']-raw['opt_value']),minimum_scalar_inequality_dual=min(inequality_duals),minimum_S_eigenvalue=float(np.linalg.eigvalsh(S_array).min()),minimum_matrix_inequality_eigenvalue=0.0)
assert raw['status'] == 'optimal'
assert all(residuals[k] <= 1e-5 for k in ('identity_gram_max_abs','identity_func_max_abs','identity_offset_abs'))
assert residuals['dense_relaxed_gap'] < 1e-5
assert residuals['minimum_scalar_inequality_dual'] >= -1e-6
assert residuals['minimum_S_eigenvalue'] >= -1e-6
assert np.allclose(S_array,S_array.T,atol=1e-6,rtol=0)
print('Complete numerical certificate checks:',residuals)
pf.pprint_labeled_matrix(np.asarray(diff.inner_prod_coords,dtype=float),raw['basis_vectors'])
# Independent candidate identity at this sample only; no all-horizon proof claim.
exact_weighted = sp.Rational(1,4*N_int)*(pb.init_conditions[0].lhs-pb.init_conditions[0].rhs)
for c in entries[1:]:
    if not raw['constraint_metadata'][c.name]['relaxed']:
        owner,pair = c.name.split(':')
        left,right = pair.split(',')
        exact_weighted += lambda_candidate(owner,left,right,N_int)*(c.lhs-c.rhs)
candidate_diff = pm.eval_scalar(pb.performance_metric-sp.Rational(1,4*N_int)-exact_weighted+S_candidate)
print('Candidate fixed-sample Gram residual:',maxabs(candidate_diff.inner_prod_coords))
print('Candidate fixed-sample function residual:',maxabs(candidate_diff.func_coords))
print('Candidate fixed-sample offset:',candidate_diff.offset)

```

The complete fixed-horizon numerical certificate passes. The initial closed-form search found candidate lambda formulas and confirmed the supplied SOS numerically. At this discovery stage the arbitrary-horizon identity and Lyapunov theorem remained open; the final symbolic sections below resolve them.

## Partial-sum Lyapunov construction and rank profile
Let $I^f_{u,v}\leq0$ and $I^g_{u,v}\leq0$ denote the normalized interpolation residuals, and $S_{i}\geq0$ the three-square piece supplied at iteration $i$. Set $V_{0}=0$, add the $f$ optimizer-to-zero term in the first increment, and thereafter accumulate
$$V_{k+1}-V_{k}=\sum_{c\text{ assigned to }k}\lambda_{c} I_{c}-S_{k+1}\leq0.$$
Each interpolation residual appears exactly once in the partial sums, and all square pieces are subtracted. The complete original grouping is restored after refining the coherent certificates. The numerical profile is rank four for every interior index at $N=4$ and $N=6$, with fixed exclusions of $V_{0},V_{N}$. The terminal rank is one because the initial squared-distance normalization remains. Full-coordinate coverage is checked below. These finite-sample observations do not establish the analytic theorem.

```python
def reconstruct_partial_sums(certificate):
    import json
    N_int = certificate['N_verify']
    params_sp = json.loads(certificate['params_json'])
    ctx, pb, obj = get_pep_setup(N_int, params_sp)
    pb.set_relaxed_constraints(certificate['relaxed_constraints'])
    REPO_ROOT = notebook_dir

    import hashlib
    import json
    from pathlib import Path
    import numpy as np
    import sympy as sp
    import pepflow as pf
    from pepflow.constraint import ScalarConstraint
    from pepflow.operator import LinearOperatorTranspose

    from pepflow.utils import name_to_vector_tuple

    certificate_checked = False
    if not certificate.get("numerical_certificate_verified", False):
        raise ValueError("Recover or verify the complete numerical certificate in Block 2")
    if not certificate.get("closed_form_search", {}).get("completed", False):
        raise ValueError("Complete or recover Block 2's initial closed-form attempt")
    certificate_keys = (
        "N_verify", "params_json", "relaxed_constraints",
        "S_matrix", "S_row_names", "S_col_names", "basis_vectors", "basis_scalars",
        "opt_value", "tau_sol", "tau_name", "named_dual_values", "constraint_metadata",
        "normalization", "certificate_provenance", "numerical_checks",
        "numerical_decomposition",
    )
    payload = {key: certificate[key] for key in certificate_keys}
    digest = hashlib.sha256(json.dumps(
        payload, sort_keys=True, separators=(",", ":"), allow_nan=False
    ).encode("utf-8")).hexdigest()
    if digest != certificate["certificate_id"]:
        raise ValueError("Certificate components changed; recover one coherent certificate")
    provenance = certificate["certificate_provenance"]
    if provenance["pep_type"] not in ("primal", "dual"):
        raise ValueError("Recover and verify the certificate's actual dual sign convention")
    for relative_path, expected_hash in provenance["source_hashes"].items():
        source_path = (REPO_ROOT / relative_path).resolve()
        if not source_path.is_relative_to(REPO_ROOT):
            raise ValueError("Certificate source path must stay in the selected repository")
        if hashlib.sha256(source_path.read_bytes()).hexdigest() != expected_hash:
            raise ValueError(f"Certificate source changed: {relative_path}")
    if not Path(pf.__file__).resolve().is_relative_to(REPO_ROOT):
        raise ValueError("The imported PEPFlow source is not the selected repository")
    if N_int != certificate["N_verify"]:
        raise ValueError("Certificate horizon differs from the rebuilt setup")
    saved_params = json.loads(certificate["params_json"])
    expected_params = {
        key: (sp.Rational(value) if isinstance(value, str) else
              sp.nsimplify(value, rational=True, tolerance=1e-9)
              if isinstance(value, float) else sp.S(value))
        for key, value in saved_params.items()
    }
    if params_sp != expected_params:
        raise ValueError("Certificate parameters differ from the rebuilt setup")
    if not certificate["numerical_checks"]["passed"]:
        raise ValueError("Block 2 numerical checks did not pass")
    if provenance['pep_type'] == 'dual':
        # This sample contains only scalar LE primal residuals, whose signs agree.
        assert all(meta['kind'] == 'scalar' and meta['comparator'] == 'LE' for meta in certificate['constraint_metadata'].values())
        assert certificate.get('sign_convention')
    tol = certificate["numerical_checks" ]["tolerances"]

    records = [(c, "initial", None, None) for c in pb.init_conditions]
    owners = list(ctx.func_to_triplets) + [
        owner for owner in ctx.oper_to_duplets
        if not isinstance(owner, LinearOperatorTranspose)
    ]
    for owner in owners:
        data = ctx.get_constraint_data(owner)
        for group, family in data.sc_dict.items():
            records.extend((c, "interpolation", str(owner), str(group)) for c in family)
        records.extend((c, "interpolation", str(owner), str(group))
                       for group, c in data.psd_dict.items())
    constraints_by_name = {c.name: c for c, _, _, _ in records}
    if len(constraints_by_name) != len(records):
        raise ValueError("Constraint names must identify one unique contribution")
    saved_meta = certificate["constraint_metadata"]
    saved_duals = certificate["named_dual_values"]
    if set(constraints_by_name) != set(saved_meta) or set(saved_duals) != set(saved_meta):
        raise ValueError("The saved certificate does not cover the rebuilt constraints")
    relaxed = set(certificate["relaxed_constraints"])
    if set(pb.relaxed_constraints) != relaxed or not relaxed.issubset(constraints_by_name):
        raise ValueError("The selected relaxation differs from the saved certificate")
    if pb.ctx is not ctx:
        raise ValueError("The builder and expression context differ")
    basis = list(ctx.basis_vectors())
    basis_labels = [repr(v) for v in basis]
    scalar_labels = [repr(v) for v in ctx.basis_scalars()]
    if basis_labels != certificate["basis_vectors"] or scalar_labels != certificate["basis_scalars"]:
        raise ValueError("Certificate coordinate order differs from the rebuilt context")
    if certificate["S_row_names"] != basis_labels or certificate["S_col_names"] != basis_labels:
        raise ValueError("S labels do not match the rebuilt basis")
    S_array = np.asarray(certificate["S_matrix"], dtype=float)
    if S_array.shape != (len(basis), len(basis)) or not np.all(np.isfinite(S_array)):
        raise ValueError("Invalid S shape or entries")
    if not np.allclose(S_array, S_array.T, atol=tol["psd_tol"], rtol=0):
        raise ValueError("S is not numerically symmetric")
    if len(basis) and np.linalg.eigvalsh((S_array + S_array.T) / 2).min() < -tol["psd_tol"]:
        raise ValueError("S failed the recorded PSD tolerance")

    pm = pf.ExpressionManager(ctx, resolve_parameters=params_sp)
    S_scalar = sum((float(S_array[i, j]) * basis[i] * basis[j]
                    for i, j in np.ndindex(S_array.shape)), pf.Scalar.zero())

    weighted_terms = {}
    interpolation_terms = {}
    interpolation_pairs = {}
    extra_terms = {}
    extra_duals = {}
    for constraint, origin, owner, group in records:
        name = constraint.name
        kind = "scalar" if isinstance(constraint, ScalarConstraint) else "psd"
        residual = constraint.lhs - constraint.rhs
        shape = [] if kind == "scalar" else list(np.shape(residual))
        expected_meta = dict(origin=origin, owner=owner, group=group, kind=kind,
                             comparator=constraint.cmp.value, shape=shape,
                             relaxed=name in relaxed)
        if any(saved_meta[name].get(key) != value for key, value in expected_meta.items()):
            raise ValueError(f"Constraint metadata changed: {name}")
        value = saved_duals[name]
        if name in relaxed:
            if value is not None:
                raise ValueError(f"A relaxed constraint must have no retained multiplier: {name}")
            continue
        if value is None:
            raise ValueError(f"Missing unrelaxed dual: {name}")
        dual = np.asarray(value, dtype=float)
        if list(dual.shape) != shape or not np.all(np.isfinite(dual)):
            raise ValueError(f"Invalid saved dual: {name}")
        # Normalize inequalities to residual <= 0; primal equality duals keep their sign.
        if constraint.cmp.name in ("GE", "SEQ"):
            residual = -residual
        if kind == "scalar":
            if constraint.cmp.name != "EQ" and float(dual) < -tol["dual_tol"]:
                raise ValueError(f"Negative inequality multiplier: {name}")
            contribution = pf.Scalar.zero() + float(dual) * residual
        else:
            if constraint.cmp.name != "EQ" and not np.allclose(dual, dual.T, atol=tol["psd_tol"], rtol=0):
                raise ValueError(f"Nonsymmetric matrix dual: {name}")
            if constraint.cmp.name != "EQ" and dual.size and np.linalg.eigvalsh((dual + dual.T) / 2).min() < -tol["psd_tol"]:
                raise ValueError(f"Matrix dual is not PSD: {name}")
            contribution = sum((float(dual[i, j]) * residual[j, i]
                                for i, j in np.ndindex(dual.shape)), pf.Scalar.zero())
        weighted_terms[name] = contribution
        if origin == "interpolation" and kind == "scalar":
            interpolation_terms[name] = contribution
            try:
                interpolation_pairs[name] = name_to_vector_tuple(name)
            except (ValueError, IndexError):
                interpolation_pairs[name] = None  # Assign this term explicitly when grouping.
        elif name != certificate["tau_name"]:
            extra_terms[name] = contribution
            extra_duals[name] = value

    tau_name = certificate["tau_name"]
    if (tau_name not in weighted_terms or saved_meta[tau_name]["origin"] != "initial"
            or saved_meta[tau_name]["kind"] != "scalar" or tau_name in relaxed
            or certificate["normalization"]["tau_name"] != tau_name):
        raise ValueError("Missing or inconsistent bound normalization constraint")
    if not np.allclose(saved_duals[tau_name], certificate["tau_sol"], atol=tol["dual_tol"], rtol=0):
        raise ValueError("The saved normalization dual is inconsistent")
    if not np.isfinite(certificate["opt_value"]) or not np.allclose(
        certificate["normalization"]["bound_value"], certificate["opt_value"],
        atol=tol["identity_atol"], rtol=tol["identity_rtol"]
    ):
        raise ValueError("The saved bound/normalization is inconsistent")
    certificate_diff = (pb.performance_metric - float(certificate["opt_value"])
                        - sum(weighted_terms.values(), pf.Scalar.zero()) + S_scalar)
    evaluated_diff = pm.eval_scalar(certificate_diff)
    for label, coords in (
        ("Gram", evaluated_diff.inner_prod_coords),
        ("function", evaluated_diff.func_coords),
        ("offset", np.asarray(evaluated_diff.offset)),
    ):
        if not np.allclose(np.asarray(coords, dtype=float), 0,
                           atol=tol["identity_atol"], rtol=tol["identity_rtol"]):
            raise ValueError(f"Numerical certificate {label} identity failed")
    decomposition = certificate["numerical_decomposition"]
    D_saved = np.asarray(decomposition["D"], dtype=float)
    E_saved = np.asarray(decomposition["ell_coords"], dtype=float)
    if decomposition["basis_labels"] != basis_labels or E_saved.ndim != 2:
        raise ValueError("Numerical S decomposition basis differs")
    if E_saved.shape[1] != len(basis) or D_saved.shape != (E_saved.shape[0], E_saved.shape[0]):
        raise ValueError("Numerical S decomposition shape differs")
    if not np.all(np.isfinite(D_saved)) or not np.all(np.isfinite(E_saved)):
        raise ValueError("Numerical S decomposition contains nonfinite values")
    if not np.allclose(E_saved.T @ D_saved @ E_saved, S_array,
                       atol=tol["reconstruction_atol"], rtol=tol["reconstruction_rtol"]):
        raise ValueError("Numerical S decomposition does not reconstruct this certificate")
    certificate_checked = True
    print("Numerical certificate checked for this N and parameter sample.")


    # Direct pieces from the B2 aid, each a sum of three nonnegative squares.
    pieces, coefficients, vectors = supplied_sos(ctx, N_int, 1)
    S_piece_terms = {f'S_{i+1}':piece for i,piece in enumerate(pieces)}
    S_assignment = {name:i for i,name in enumerate(S_piece_terms)}
    # With interpolation residuals <= 0, every assigned contribution minus its
    # nonnegative S piece gives a nonpositive increment.
    group_terms = {name:term for name,term in weighted_terms.items() if name != tau_name}
    term_assignment = {}
    for name in group_terms:
        left, right = name.split(':')[1].split(',')
        term_assignment[name] = int(right[2:])-1
    assert set(term_assignment) == set(group_terms)
    assert set(S_assignment) == set(S_piece_terms)
    lyap = [pf.Scalar.zero()]
    partial_sum = sum((term for name,term in group_terms.items() if term_assignment[name] == -1), pf.Scalar.zero())
    for step in range(N_int):
        partial_sum += sum((term for name,term in group_terms.items() if term_assignment[name] == step), pf.Scalar.zero())
        partial_sum -= S_piece_terms[f'S_{step+1}']
        lyap.append(partial_sum)
    rank_tolerance = 1e-4
    rank_left, rank_right = 1, 1
    ranks, singular_values = [], []
    for k, value in enumerate(lyap):
        matrix = pm.eval_scalar(value).inner_prod_coords.astype(float)
        singular_values.append(np.linalg.svd(matrix,compute_uv=False).tolist())
        rank = int(np.linalg.matrix_rank(matrix,tol=rank_tolerance))
        ranks.append(rank)
        print(f'rank V_{k}: {rank}')
        if k == 0:
            print()
    retained_ranks = ranks[rank_left:N_int+1-rank_right]
    print('Interior rank is constant at this N:', len(retained_ranks)>=3 and len(set(retained_ranks))==1,'boundary exclusions:',(rank_left,rank_right))
    assert len(retained_ranks)>=3 and len(set(retained_ranks))==1
    # Credibility check records all singular values, including numerical residuals.
    # No retained singular value lies in [tolerance/2, 2*tolerance].
    rank_credible = all(not any(5e-5 <= v <= 2e-4 for v in singular_values[k]) for k in range(1,N_int))
    assert rank_credible




    boundary_names = [name for name, step in term_assignment.items() if step == "boundary"]
    boundary_S = [name for name, step in S_assignment.items() if step == "boundary"]
    if sum(name in interpolation_terms for name in boundary_names) > 1 or len(boundary_S) > 1:
        raise ValueError("Revisit the permitted terminal interpolation/S leftovers")
    omitted = sum((group_terms[name] for name, step in term_assignment.items()
                   if step == "omitted"), pf.Scalar.zero())
    leftover = sum((group_terms[name] for name in boundary_names), pf.Scalar.zero())
    leftover -= sum((S_piece_terms[name] for name in boundary_S), pf.Scalar.zero())
    coverage_diff = (lyap[N_int] + leftover + omitted
                     - (sum(group_terms.values(), pf.Scalar.zero()) - S_scalar))
    piece_diff = sum(S_piece_terms.values(), pf.Scalar.zero()) - S_scalar
    for label, expression in (("coverage", coverage_diff), ("S pieces", piece_diff), ("omitted", omitted)):
        value = pm.eval_scalar(expression)
        for coords in (value.inner_prod_coords, value.func_coords, np.asarray(value.offset)):
            if not np.allclose(np.asarray(coords, dtype=float), 0,
                               atol=tol["identity_atol"], rtol=tol["identity_rtol"]):
                raise ValueError(f"Full-coordinate {label} check failed")
    # Every subtracted S piece must have the nonnegative numerical sign used here.
    for name, piece in S_piece_terms.items():
        value = pm.eval_scalar(piece)
        matrix = np.asarray(value.inner_prod_coords, dtype=float)
        if not np.allclose(value.func_coords, 0, atol=tol["identity_atol"], rtol=0) or not np.allclose(value.offset, 0, atol=tol["identity_atol"], rtol=0):
            raise ValueError(f"S piece has unexpected function/offset terms: {name}")
        if matrix.size and np.linalg.eigvalsh((matrix + matrix.T) / 2).min() < -tol["psd_tol"]:
            raise ValueError(f"S piece has the wrong numerical sign: {name}")
    # Step identities follow the chosen assignments; check them, including the initial terms.
    for k in range(N_int):
        assigned_steps = {-1, 0} if k == 0 else {k}
        increment = sum((group_terms[name] for name, step in term_assignment.items()
                         if step in assigned_steps), pf.Scalar.zero())
        increment -= sum((S_piece_terms[name] for name, step in S_assignment.items()
                          if step in assigned_steps), pf.Scalar.zero())
        value = pm.eval_scalar(lyap[k + 1] - lyap[k] - increment)
        for coords in (value.inner_prod_coords, value.func_coords, np.asarray(value.offset)):
            if not np.allclose(np.asarray(coords, dtype=float), 0,
                               atol=tol["identity_atol"], rtol=tol["identity_rtol"]):
                raise ValueError(f"Partial-sum increment identity failed at k={k}")
    grouping_checked = True
    M_final = pm.eval_scalar(lyap[N_int]).inner_prod_coords.astype(float)
    rank_final = int(np.linalg.matrix_rank(M_final, tol=rank_tolerance))
    print(f"lyap[{N_int}] rank:", rank_final)
    print("Coverage checked with all scalar components and explicit boundary leftovers.")

    evaluated = [pm.eval_scalar(value) for value in lyap]
    observation = dict(N=N_int,certificate_id=certificate['certificate_id'],recipe_id='forward_star_consecutive_three_sos_refined_dual',rank_profile=ranks,rank_convention={'tolerance':1e-4,'scaling':'unscaled Gram coordinates','order':'setup basis order'},rank_credible=rank_credible,certificate_checked=certificate_checked,grouping_checked=grouping_checked,singular_values=singular_values)
    return dict(observation=observation,lyap_inner_prod_coords=[np.array(v.inner_prod_coords,dtype=float).tolist() for v in evaluated],lyap_func_coords=[np.array(v.func_coords,dtype=float).tolist() for v in evaluated],lyap_offsets=[float(v.offset) for v in evaluated],grouping_coverage=dict(constraint_assignment=term_assignment,S_assignment=S_assignment,normalization_constraint=tau_name,boundary_constraints=boundary_names,boundary_S=boundary_S,omitted_constraints=[],checked=True),extra_duals=extra_duals)
```

### Candidate-vector scan
Scan tagged iterate gaps, gradients, proximal displacements, basis vectors and pairwise differences; remove zero and duplicate coordinates.

```python
candidate_pairs = [('x0-star',ctx['x_0']-ctx['x_star'])]
for i in range(N_int+1):
    candidate_pairs += [(f'x{i}-star',ctx[f'x_{i}']-ctx['x_star']),
                        (f'a{i}-astar',f.grad(ctx[f'x_{i}'])-f.grad(ctx['x_star']))]
    if i:
        candidate_pairs += [(f'hgrad{i}',f.grad(ctx[f'x_{i}'])+g.grad(ctx[f'x_{i}'])),
                            (f'proxstep{i}',ctx[f'x_{i-1}']-ctx[f'x_{i}'])]
candidate_pairs += [(str(v),v) for v in ctx.basis_vectors()]
base = list(candidate_pairs)
for (left,u),(right,v) in combinations(base,2):
    candidate_pairs.append((f'({left})-({right})',u-v))
unique_pairs, seen_coords = [], []
for label,vector in candidate_pairs:
    coords = np.array(pm.eval_vector(vector).coords,dtype=float)
    if np.linalg.norm(coords)<1e-7 or any(np.allclose(coords,previous,atol=1e-7,rtol=0) for previous in seen_coords):
        continue
    unique_pairs.append((label,vector))
    seen_coords.append(coords)
labels_by_id = {id(vector):label for label,vector in unique_pairs}
in_column = {}
for k in range(1,N_int+1):
    in_column[k] = vectors_in_column_space(lyap[k],[v for _,v in unique_pairs],pep_context=ctx,resolve_parameters=params_sp,atol=1e-4,rtol=1e-4)
    print(f'V_{k} column-space candidates:')
    for vector in in_column[k]:
        print('  ',labels_by_id[id(vector)])

```

### Selected basis pattern

For $1\leq k<N$, use $[x_{0}-x_{\star},\ x_{k}-x_{\star},\ a_{k}-a_{\star},\ a_{k}]$. At $k=0$ the basis is empty; at $k=N$ it is $[x_{0}-x_{\star}]$.

```python
def V_k_basis(k):
    if k == 0:
        return []
    if k == N_int:
        return [ctx['x_0']-ctx['x_star']]
    return [ctx['x_0']-ctx['x_star'],ctx[f'x_{k}']-ctx['x_star'],f.grad(ctx[f'x_{k}'])-f.grad(ctx['x_star']),f.grad(ctx[f'x_{k}'])]

def V_k_basis_labels(k):
    if k == 0:
        return []
    if k == N_int:
        return ['x_0-x_star']
    return ['x_0-x_star',f'x_{k}-x_star',f'a_{k}-a_star',f'a_{k}']

for k in range(N_int+1):
    vectors = V_k_basis(k)
    basis_rank = int(np.linalg.matrix_rank(np.column_stack([pm.eval_vector(v).coords for v in vectors]).astype(float),tol=1e-7)) if vectors else 0
    assert basis_rank == len(vectors) == ranks[k]
    print(f'k={k}: rank {basis_rank} basis {V_k_basis_labels(k)}')

```

### Coefficient matrices
All indices, including initial and terminal cases, are reconstructed in the stated basis order. Off-diagonal entries contribute twice.

```python
coeff_by_k, coefficient_residuals = {}, {}
for k in range(N_int+1):
    vectors = V_k_basis(k)
    target = pm.eval_scalar(lyap[k]).inner_prod_coords.astype(float)
    if vectors:
        C = find_symmetric_coefficient_matrix(lyap[k],vectors,pep_context=ctx,resolve_parameters=params_sp,indep_tol=1e-7,span_tol=1e-5)
        B = np.column_stack([pm.eval_vector(v).coords for v in vectors]).astype(float)
        residual = float(np.linalg.norm(B@C@B.T-target,ord='fro'))
    else:
        C = np.zeros((0,0))
        residual = float(np.linalg.norm(target,ord='fro'))
    assert np.array_equal(C,C.T)
    assert residual <= 1e-5*max(1.0,np.linalg.norm(target,ord='fro'))
    coeff_by_k[k], coefficient_residuals[k] = C, residual
    print(f'k={k}: reconstruction residual {residual}')
    if vectors:
        pf.pprint_labeled_matrix(C,V_k_basis_labels(k))
    evaluated = pm.eval_scalar(lyap[k])
    print('Function terms:',[(label,float(c)) for label,c in zip(b3['basis_scalars'],evaluated.func_coords) if abs(float(c))>1e-6], 'offset:',evaluated.offset)

```

Full-expression reconstruction retains the observed function gaps and scalar offsets. The subsequent coefficient formulas are numerical candidates only.

```python
def V_k_structure(k):
    vectors = V_k_basis(k)
    C = coeff_by_k[k]
    quadratic = sum((float(C[i,j])*vectors[i]*vectors[j] for i,j in np.ndindex(C.shape)),pf.Scalar.zero())
    if k == 0:
        return quadratic
    f_weight = 1 if k == N_int else (k+1)/(2*N_int-k)
    g_weight = 1 if k == N_int else k/(2*N_int-k)
    return quadratic + f_weight*(f(ctx[f'x_{k}'])-f(ctx['x_star'])) + g_weight*(g(ctx[f'x_{k}'])-g(ctx['x_star']))

component_residuals = {}
for k in range(N_int+1):
    diff = pm.eval_scalar(V_k_structure(k)-lyap[k])
    component_residuals[str(k)] = {'gram':maxabs(diff.inner_prod_coords),'function':maxabs(diff.func_coords),'offset':maxabs(diff.offset)}
    assert all(value <= 1e-4 for value in component_residuals[str(k)].values())
print('Full-component reconstruction:',component_residuals)

# Candidate coefficient formulas, inferred from the observed rational matrices.
def coeff_pattern(k,n,smoothness=1):
    ell = sp.sympify(smoothness)
    if k == 0:
        return sp.zeros(0)
    if k == n:
        return sp.Matrix([[-ell/(4*n)]])
    C = sp.zeros(4)
    C[0,0] = -ell/(4*n)
    C[1,1] = ell*(n-k)/(2*n-k)**2
    C[2,2] = 1/(2*ell*(2*n-k))
    C[1,3] = C[3,1] = -sp.Rational(1,2*(2*n-k))
    return C

for k in range(N_int+1):
    predicted = np.array(coeff_pattern(k,N_int),dtype=float).reshape(coeff_by_k[k].shape)
    assert np.allclose(predicted,coeff_by_k[k],atol=1e-4,rtol=1e-5)
    print(f'k={k}: coefficient candidate residual {maxabs(predicted-coeff_by_k[k])}')
```

### Block 4 conclusion
The reconstructed interior structure is
$$V_{k}=-\frac{L}{4N}\|x_{0}-x_{\star}\|^{2}+\frac{L(N-k)}{(2N-k)^{2}}\|x_{k}-x_{\star}\|^{2}+\frac{\|a_{k}-a_{\star}\|^{2}}{2L(2N-k)}-\frac{\langle x_{k}-x_{\star},a_{k}\rangle}{2N-k}+\frac{k+1}{2N-k}(f_{k}-f_{\star})+\frac{k}{2N-k}(g_{k}-g_{\star}).$$
The exceptional cases are $V_{0}=0$ and $V_{N}=h_{N}-h_{\star}-\frac{L}{4N}\|x_{0}-x_{\star}\|^{2}$. Every numerical component reconstructs. The displayed coefficients were inferred numerically in Block 4. The exact symbolic identities and domain arguments below now establish them for arbitrary horizons.

## Closed-form conclusion and symbolic recursion context
For $N\geq2$, write $a_{i}=\nabla f(x_{i})$ and $d_{i}=x_{i}-x_{\star}$. Set $V_{0}=0$, and for $1\leq k<N$ define
$$V_{k}=-\frac{L}{4N}\|d_{0}\|^{2}+\frac{L(N-k)}{(2N-k)^{2}}\|d_{k}\|^{2}+\frac{\|a_{k}-a_{\star}\|^{2}}{2L(2N-k)}-\frac{\langle d_{k},a_{k}\rangle}{2N-k}+\frac{k+1}{2N-k}(f_{k}-f_{\star})+\frac{k}{2N-k}(g_{k}-g_{\star}).$$
At the terminal index define
$$V_{N}=h(x_{N})-h(x_{\star})-\frac{L}{4N}\|d_{0}\|^{2}.$$
For $N=1$, use only the two exceptional values $V_{0}=0$ and the displayed terminal formula.

```python
b4 = json.loads((notebook_dir/'state/pgm_b4.json').read_text())
assert b4['numerical_reconstruction']['certificate_id'] == b4['certificate_id'] == b3['certificate_id']
print('Verified numerical structure loaded; exact proof starts below.')
# Symbolic local definitions. All coefficients remain exact.
k_sym, n_sym = sp.symbols('k N', integer=True, positive=True)
ell_sym = sp.Symbol('L', positive=True)
kp, np_, lp = pf.Parameter('k_symbolic'), pf.Parameter('N_symbolic'), pf.Parameter('L_symbolic')
symbolic_mapping: dict = {'k_symbolic':k_sym,'N_symbolic':n_sym,'L_symbolic':ell_sym}

def interior_potential(x, anchor, star, ff, gg, k, n, ell):
    a, astar = ff.grad(x), ff.grad(star)
    return (-ell/(4*n)*anchor**2 + ell*(n-k)/(2*n-k)**2*(x-star)**2
            + (a-astar)**2/(2*ell*(2*n-k)) - (x-star)*a/(2*n-k)
            + (k+1)/(2*n-k)*(ff(x)-ff(star)) + k/(2*n-k)*(gg(x)-gg(star)))

def If(ff,u,v,ell):
    return ff(v)-ff(u)+ff.grad(v)*(u-v)+(ff.grad(u)-ff.grad(v))**2/(2*ell)

def Ig(gg,u,v):
    return gg(v)-gg(u)+gg.grad(v)*(u-v)

def alpha(i,n):
    return i/(2*n+1-i)

def beta(i,n):
    return i/(2*n-i)

def local_square_piece(x,y,star,ff,gg,i,n,ell):
    ap,a,astar,s = ff.grad(x),ff.grad(y),ff.grad(star),gg.grad(y)
    c1 = n/((2*n+1-i)*(2*n-i)*ell)
    c2 = ((i-1)/(2*n-i)*(2*n+2)/(2*n+1)+2*n/((2*n-i)**2*(2*n+1)))/(2*ell)
    c3 = i/(2*n+1-i)*(2*n)/(2*n+1)/(2*ell)
    u1 = i/(2*n)*a+(2*n-i)/(2*n)*ap-astar
    u2 = ap+s-ell/(2*n+1-i)*(x-star)
    u3 = (2*n+1)/(2*n)*(a+s)-(ap+s)/(2*n)-ell/(2*n-i)*(y-star)
    return c1*u1**2+c2*u2**2+c3*u3**2

def exact_components(expression,manager):
    value = manager.eval_scalar(expression,sympy_mode=True)
    gram = sp.Matrix(value.inner_prod_coords).applyfunc(lambda z:sp.factor(sp.cancel(z)))
    funcs = sp.Matrix(list(value.func_coords)).applyfunc(lambda z:sp.factor(sp.cancel(z)))
    offset = sp.factor(sp.cancel(sp.sympify(value.offset)))
    return gram,funcs,offset

def check_exact(name,expression,manager,context):
    gram,funcs,offset = exact_components(expression,manager)
    assert all(z == 0 for z in [*gram,*funcs,offset]), (name,gram,funcs,offset)
    print(name, 'all Gram/function/offset residuals exactly zero')
    pf.pprint_labeled_matrix(np.array(gram.tolist(),dtype=object),[str(v) for v in context.basis_vectors()])
    pf.pprint_labeled_matrix(np.array(funcs.tolist(),dtype=object),[str(v) for v in context.basis_scalars()],['residual'])
    print('constant residual:',offset)
    return {'gram_shape':list(gram.shape),'function_shape':list(funcs.shape),'offset':str(offset),'exact_zero':True}

```

## Symbolic Step Recursion Verification

For $i=k+1$ and $1\leq k\leq N-2$, verify $$V_{i}-V_{i-1}=(\alpha_{i+1}-\alpha_{i})I_{f}(\star,i)+\alpha_{i} I_{f}(i-1,i)+(\beta_{i}-\beta_{i-1})I_{g}(\star,i)+\beta_{i-1}I_{g}(i-1,i)-S_{i}.$$ The residual $\mathrm{LHS}-\mathrm{RHS}$ must be exactly zero in all components.

```python
ctx_step = pf.PEPContext('symbolic_step').set_as_current()
f_step = pf.SmoothConvexFunction(is_basis=True,tags=['f_{step}'],L=lp)
g_step = pf.ConvexFunction(is_basis=True,tags=['g_{step}'])
star_step = (f_step+g_step).set_stationary_point('x_{star,step}')
x_step = pf.Vector(is_basis=True,tags=['x_{k}'])
anchor_step = pf.Vector(is_basis=True,tags=['d_{0}'])
y_step = g_step.prox(x_step-f_step.grad(x_step)/lp,1/lp,tag='x_{k+1}')
f_step(y_step)
i_step = kp+1
LHS_step = interior_potential(y_step,anchor_step,star_step,f_step,g_step,kp+1,np_,lp)-interior_potential(x_step,anchor_step,star_step,f_step,g_step,kp,np_,lp)
RHS_step = ((alpha(i_step+1,np_)-alpha(i_step,np_))*If(f_step,star_step,y_step,lp)
            + alpha(i_step,np_)*If(f_step,x_step,y_step,lp)
            + (beta(i_step,np_)-beta(i_step-1,np_))*Ig(g_step,star_step,y_step)
            + beta(i_step-1,np_)*Ig(g_step,x_step,y_step)
            - local_square_piece(x_step,y_step,star_step,f_step,g_step,i_step,np_,lp))
diff_step = LHS_step-RHS_step
pm_step = pf.ExpressionManager(ctx_step,resolve_parameters=symbolic_mapping)
step_evidence = check_exact('Interior step: 1 <= k <= N-2',diff_step,pm_step,ctx_step)

```

## Base Case and Boundary Symbolic Verification

For $N\geq2$, verify $$V_{1}=\alpha_{1}I_{f}(\star,0)+(\alpha_{2}-\alpha_{1})I_{f}(\star,1)+\alpha_{1}I_{f}(0,1)+\beta_{1}I_{g}(\star,1)-S_{1}.$$ The residual $\mathrm{LHS}-\mathrm{RHS}$ must be exactly zero. This uses the actual starting displacement $d_{0}=x_{0}-x_{\star}$.

```python
ctx_base = pf.PEPContext('symbolic_base').set_as_current()
f_base = pf.SmoothConvexFunction(is_basis=True,tags=['f_{base}'],L=lp)
g_base = pf.ConvexFunction(is_basis=True,tags=['g_{base}'])
star_base = (f_base+g_base).set_stationary_point('x_{star,base}')
x_base = pf.Vector(is_basis=True,tags=['x_{0,base}'])
y_base = g_base.prox(x_base-f_base.grad(x_base)/lp,1/lp,tag='x_{1,base}')
f_base(y_base)
LHS_base = interior_potential(y_base,x_base-star_base,star_base,f_base,g_base,sp.Integer(1),np_,lp)
RHS_base = (alpha(sp.Integer(1),np_)*If(f_base,star_base,x_base,lp)
            + (alpha(sp.Integer(2),np_)-alpha(sp.Integer(1),np_))*If(f_base,star_base,y_base,lp)
            + alpha(sp.Integer(1),np_)*If(f_base,x_base,y_base,lp)
            + beta(sp.Integer(1),np_)*Ig(g_base,star_base,y_base)
            - local_square_piece(x_base,y_base,star_base,f_base,g_base,sp.Integer(1),np_,lp))
diff_base = LHS_base-RHS_base
pm_base = pf.ExpressionManager(ctx_base,resolve_parameters=symbolic_mapping)
base_evidence = check_exact('Base: N >= 2',diff_base,pm_base,ctx_base)

```

### Boundary Identity Symbolic Verification

For $N\geq2$, verify $$V_{N}-V_{N-1}=(1-\alpha_{N})I_{f}(\star,N)+\alpha_{N}I_{f}(N-1,N)+(1-\beta_{N-1})I_{g}(\star,N)+\beta_{N-1}I_{g}(N-1,N)-S_{N},$$ where $V_{N}=h_{N}-h_{\star}-L\|d_{0}\|^{2}/(4N)$. The residual $\mathrm{LHS}-\mathrm{RHS}$ must be exactly zero.

```python
ctx_terminal = pf.PEPContext('symbolic_terminal').set_as_current()
f_terminal = pf.SmoothConvexFunction(is_basis=True,tags=['f_{terminal}'],L=lp)
g_terminal = pf.ConvexFunction(is_basis=True,tags=['g_{terminal}'])
star_terminal = (f_terminal+g_terminal).set_stationary_point('x_{star,terminal}')
x_terminal = pf.Vector(is_basis=True,tags=['x_{N-1}'])
anchor_terminal = pf.Vector(is_basis=True,tags=['d_{0,terminal}'])
y_terminal = g_terminal.prox(x_terminal-f_terminal.grad(x_terminal)/lp,1/lp,tag='x_{N}')
f_terminal(y_terminal)
V_terminal = f_terminal(y_terminal)-f_terminal(star_terminal)+g_terminal(y_terminal)-g_terminal(star_terminal)-lp/(4*np_)*anchor_terminal**2
LHS_terminal = V_terminal-interior_potential(x_terminal,anchor_terminal,star_terminal,f_terminal,g_terminal,np_-1,np_,lp)
RHS_terminal = ((1-alpha(np_,np_))*If(f_terminal,star_terminal,y_terminal,lp)
                + alpha(np_,np_)*If(f_terminal,x_terminal,y_terminal,lp)
                + (1-beta(np_-1,np_))*Ig(g_terminal,star_terminal,y_terminal)
                + beta(np_-1,np_)*Ig(g_terminal,x_terminal,y_terminal)
                - local_square_piece(x_terminal,y_terminal,star_terminal,f_terminal,g_terminal,np_,np_,lp))
diff_terminal = LHS_terminal-RHS_terminal
pm_terminal = pf.ExpressionManager(ctx_terminal,resolve_parameters=symbolic_mapping)
boundary_evidence = check_exact('Terminal: N >= 2',diff_terminal,pm_terminal,ctx_terminal)

```

```python
ctx_one = pf.PEPContext('symbolic_one').set_as_current()
f_one = pf.SmoothConvexFunction(is_basis=True,tags=['f_{one}'],L=lp)
g_one = pf.ConvexFunction(is_basis=True,tags=['g_{one}'])
star_one = (f_one+g_one).set_stationary_point('x_{star,one}')
x_one = pf.Vector(is_basis=True,tags=['x_{0,one}'])
y_one = g_one.prox(x_one-f_one.grad(x_one)/lp,1/lp,tag='x_{1,one}')
f_one(y_one)
one = sp.Integer(1)
LHS_one = f_one(y_one)-f_one(star_one)+g_one(y_one)-g_one(star_one)-lp/4*(x_one-star_one)**2
RHS_one = (sp.Rational(1,2)*If(f_one,star_one,x_one,lp)
           + sp.Rational(1,2)*If(f_one,star_one,y_one,lp)
           + sp.Rational(1,2)*If(f_one,x_one,y_one,lp)
           + Ig(g_one,star_one,y_one)
           - local_square_piece(x_one,y_one,star_one,f_one,g_one,one,one,lp))
diff_one = LHS_one-RHS_one
pm_one = pf.ExpressionManager(ctx_one,resolve_parameters=symbolic_mapping)
one_evidence = check_exact('Exceptional horizon N=1',diff_one,pm_one,ctx_one)
```

### Signs, normalization, and telescoping
For integers $1\leq i\leq N$ with $N\geq1$, the denominators $2N-i$, $2N+1-i$, $2N+1$, and $L$ are strictly positive. The factor $i-1$ in $c_{2,i}$ is nonnegative. Hence every square coefficient and every interpolation multiplier used above is nonnegative. The initial/terminal weights simplify as in the exact checks below.
The interior index range is empty when $N=2$; base and terminal identities still apply. The separate $N=1$ identity handles the only remaining case. Since every residual is nonpositive and each square is subtracted, the potential is nonincreasing. Its terminal definition then gives the theorem with the original squared-radius normalization.
Summing the checked local identities also establishes the analytic full-PEP identity
$$h_{N}-h_{\star}-\frac{L}{4N}\|d_{0}\|^{2}=\alpha_{1}I_{f}(\star,0)+\sum_{i=1}^{N-1}(\alpha_{i+1}-\alpha_{i})I_{f}(\star,i)+(1-\alpha_{N})I_{f}(\star,N)+\sum_{i=1}^N\alpha_{i}I_{f}(i-1,i)+\sum_{i=1}^N(\beta_{i}-\beta_{i-1})I_{g}(\star,i)+\sum_{i=2}^N\beta_{i-1}I_{g}(i-1,i)-\sum_{i=1}^NS_{i}.$$
Every active multiplier family is displayed, and every supplied square appears once. The finite sum telescopes by the verified base, interior, and terminal identities; the $N=1$ identity gives the same empty-sum convention. This is an all-horizon proof of the full candidate certificate, distinct from the earlier numerical solver checks.

```python
# Domain: integer N>=1, integer 1<=i<=N, and L>0.
i_sign, n_sign = sp.symbols('i N',integer=True,positive=True)
alpha_difference = sp.factor(alpha(i_sign+1,n_sign)-alpha(i_sign,n_sign))
beta_difference = sp.factor(beta(i_sign,n_sign)-beta(i_sign-1,n_sign))
assert sp.cancel(alpha_difference-(2*n_sign+1)/((2*n_sign-i_sign)*(2*n_sign+1-i_sign))) == 0
assert sp.cancel(beta_difference-2*n_sign/((2*n_sign-i_sign)*(2*n_sign+1-i_sign))) == 0
assert sp.cancel(1-alpha(n_sign,n_sign)-1/(n_sign+1)) == 0
assert sp.cancel(1-beta(n_sign-1,n_sign)-2/(n_sign+1)) == 0
print('Exact multiplier differences and terminal simplifications checked.')
print('For 1<=i<=N: 2N-i>=N>=1 and 2N+1-i>=N+1; all multipliers and all three SOS weights are nonnegative.')
print('Each local RHS is therefore <=0. V_0=0 and V_N=gap-L*||x_0-x_star||^2/(4N), giving gap<=L*R^2/(4N).')
# Sharpness for R>0: f=0, g(x)=c*max(x,0), c=L*R/(2N), x_0=R, x_star=0.
r_sharp = sp.Symbol('R',nonnegative=True)
c_sharp = ell_sym*r_sharp/(2*n_sym)
xn_sharp = r_sharp-n_sym*c_sharp/ell_sym
assert sp.simplify(xn_sharp-r_sharp/2) == 0
assert sp.simplify(c_sharp*xn_sharp-ell_sym*r_sharp**2/(4*n_sym)) == 0
print('Exact sharpness value for the one-dimensional hinge example:',sp.factor(c_sharp*xn_sharp))
```
