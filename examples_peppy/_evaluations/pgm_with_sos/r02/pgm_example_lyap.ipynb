# Proximal gradient method: convergence and Lyapunov discovery
Let $h=f+g$, where $f$ is convex and $L$-smooth and $g$ is closed, proper, and convex on a finite-dimensional real Euclidean space. Assume $L>0$, $R\geq0$, an integer $N\geq1$, a minimizer $x_\star$ of $h$, and $\|x_0-x_\star\|^2\leq R^2$. Proximal evaluations are well-defined.
$$x_{k+1}=\operatorname{prox}_{g/L}\left(x_k-\frac{1}{L}\nabla f(x_k)\right),\qquad k=0,\ldots,N-1.$$
The performance metric is $h(x_N)-h(x_\star)$. Stationarity is imposed on the sum: $\nabla f(x_\star)+s_\star=0$ with $s_\star\in\partial g(x_\star)$.

## Proof Statement

### Theorem
Under the assumptions above, for every integer $N\geq1$,
$$h(x_{N})-h(x_{\star})\leq\frac{L\|x_{0}-x_{\star}\|^2}{4N}\leq\frac{LR^2}{4N}.$$
Write $d_{k}=x_{k}-x_{\star}$, $a_{k}=\nabla f(x_{k})$, $a_{\star}=\nabla f(x_{\star})$, $F_{k}=f(x_{k})-f(x_{\star})$, and $G_{k}=g(x_{k})-g(x_{\star})$ for $k\geq1$. An explicit nonincreasing Lyapunov certificate is $V_{0}=0$,
$$V_{k}=-\frac{L}{4N}\|d_{0}\|^2+\frac{L(N-k)}{(2N-k)^2}\|d_{k}\|^2-\frac{\langle a_{k},d_{k}\rangle}{2N-k}+\frac{\|a_{k}-a_{\star}\|^2}{2L(2N-k)}+\frac{k+1}{2N-k}F_{k}+\frac{k}{2N-k}G_{k},\quad 1\leq k<N,$$
and
$$V_{N}=F_{N}+G_{N}-\frac{L}{4N}\|d_{0}\|^2.$$
For $N=1$ there are no interior indices, and the separately verified one-step identity gives the same conclusion. No value of $g(x_{0})$ is required. The constant is sharp: in one dimension, $f=0$, $g(x)=LR|x|/(2N)$, $x_{0}=R$, $x_{\star}=0$ give $x_{N}=R/2$ and equality (with the zero-radius case immediate).

### Proof outline
Set $s_{i}=L(x_{i-1}-x_{i})-a_{i-1}\in\partial g(x_{i})$ and $s_{\star}=-a_{\star}\in\partial g(x_{\star})$. Define the nonpositive interpolation residuals
$$I^f_{u,v}=f(v)-f(u)+\langle\nabla f(v),u-v\rangle+\frac{\|\nabla f(u)-\nabla f(v)\|^2}{2L},\qquad I^g_{u,v}=g(v)-g(u)+\langle s_v,u-v\rangle.$$
For the interior expression, write
$$\Phi_{k}=\frac{L(N-k)}{(2N-k)^2}\|d_{k}\|^2-\frac{\langle a_{k},d_{k}\rangle}{2N-k}+\frac{\|a_{k}-a_{\star}\|^2}{2L(2N-k)}+\frac{k+1}{2N-k}F_{k}+\frac{k}{2N-k}G_{k},\qquad V_{k}=\Phi_{k}-\frac{L}{4N}\|d_{0}\|^2.$$
Let $p_{0}=1/(2N)$. For $1\leq i<N$, set
$$p_{i}=\frac{2N+1}{(2N-i)(2N+1-i)},\qquad q_{i}=\frac{2N}{(2N-i)(2N+1-i)},$$
while $p_{N}=1/(N+1)$ and $q_{N}=2/(N+1)$. Also set $\alpha_{i}=i/(2N+1-i)$ and $\beta_{i}=(i-1)/(2N+1-i)$.
The supplied SOS aid, verified here by the exact local identities, is $S_{i}=c_{1,i}\|u_{1,i}\|^2+c_{2,i}\|u_{2,i}\|^2+c_{3,i}\|u_{3,i}\|^2$, with
$$c_{1,i}=\frac{N}{(2N+1-i)(2N-i)L},\qquad u_{1,i}=\frac{i}{2N}a_{i}+\frac{2N-i}{2N}a_{i-1}-a_{\star},$$
$$c_{2,i}=\frac{1}{2L}\left(\frac{i-1}{2N-i}\frac{2N+2}{2N+1}+\frac{2N}{(2N-i)^2(2N+1)}\right),\qquad u_{2,i}=a_{i-1}+s_{i}-\frac{L}{2N+1-i}d_{i-1},$$
$$c_{3,i}=\frac{i}{2N+1-i}\frac{2N}{2N+1}\frac{1}{2L},\qquad u_{3,i}=\frac{2N+1}{2N}(a_{i}+s_{i})-\frac{a_{i-1}+s_{i}}{2N}-\frac{L}{2N-i}d_{i}.$$
The exact base, interior and terminal calculations below establish the unified identity
$$V_{i}-V_{i-1}=p_{i}I^f_{x_{\star},x_{i}}+q_{i}I^g_{x_{\star},x_{i}}+\alpha_{i}I^f_{x_{i-1},x_{i}}+\beta_{i}I^g_{x_{i-1},x_{i}}-S_{i}+\mathbf1_{i=1}\,p_{0}I^f_{x_{\star},x_{0}},\quad 1\leq i\leq N.$$
The $g$ adjacent term is absent when $i=1$; it is not an evaluation of $g(x_{0})$ multiplied by zero. The $N=1$ identity is verified separately.
All interpolation multipliers and SOS coefficients are nonnegative: $2N-i\geq N\geq1$, $2N+1-i\geq N+1>0$, $L>0$, and $i-1\geq0$. Thus every right-hand side is nonpositive. Hence $V_{N}\leq V_{0}=0$, giving $h(x_{N})-h(x_{\star})\leq L\|d_{0}\|^2/(4N)\leq LR^2/(4N)$.
Summing the same exact identities also proves the global full-PEP identity with all displayed interpolation weights and $S=\sum_{i=1}^{N}S_i$. The normalization multiplier is $L/(4N)$ for $\|d_{0}\|^2-R^2\leq0$. This is an analytic argument for every horizon; the numerical factor blending used during discovery is not used in the proof.

## Functions, algorithm, and PEP setup

```python
"""PGM with step 1/L; composite objective gap and initial squared radius R**2."""

L = pf.Parameter("L")
R = pf.Parameter("R")
f = pf.SmoothConvexFunction(is_basis=True, tags=["f"], L=L)
g = pf.ConvexFunction(is_basis=True, tags=["g"])
h = f + g
h.add_tag("h")


def make_ctx_pgm(ctx_name: str, N, **kwargs):
    ctx = pf.PEPContext(ctx_name).set_as_current()
    x = pf.Vector(is_basis=True, tags=["x_0"])
    h.set_stationary_point("x_star")
    for k in range(int(N)):
        x = g.prox(x - f.grad(x) / L, 1 / L, tag=f"x_{k + 1}")
    f(x)
    return ctx


def get_pep_setup(N, params):
    ctx = make_ctx_pgm(f"ctx_{N}", N)
    pb = pf.PEPBuilder(ctx)
    pb.add_initial_constraint(
        ((ctx["x_0"] - ctx["x_star"]) ** 2).le(R**2, name="initial_condition")
    )
    pb.set_performance_metric(h(ctx[f"x_{N}"]) - h(ctx["x_star"]))
    return ctx, pb, f
```

## Lambda and S verification
Write $R_{ij}^f=f_j-f_i+\langle a_j,x_i-x_j\rangle+\|a_i-a_j\|^2/(2L)\leq0$ and $R_{ij}^g=g_j-g_i+\langle s_j,x_i-x_j\rangle\leq0$. For forward adjacent pairs, the candidate weights are
$$\lambda^f_{i,i+1}=\frac{i+1}{2N-i},\quad 0\leq i<N,\qquad \lambda^g_{i,i+1}=\frac{i}{2N-i},\quad 1\leq i<N.$$
The star weights equal successive increments of these weights, with the endpoint corrections displayed by the executable definitions below. These are inferred candidates, not yet an arbitrary-horizon certificate proof. The normalization multiplier is $L/(4N)$ for the squared-radius constraint.
The supplied three-family SOS is reproduced below in exact rational arithmetic. Its normalization factor relative to this certificate is 1. Its coefficients are nonnegative for $1\leq i\leq N$, $N\geq1$, $L>0$: every denominator is positive and every numerator is nonnegative. The first and third coefficients are strictly positive; the second also has a positive second summand.

```python
def lambda_candidate(owner, left, right, N):
    N = sp.sympify(N)
    if left == "x_star":
        j = int(right[2:])
        if owner == "f":
            if j == 0:
                return 1 / (2 * N)
            if j == N:
                return 1 / (N + 1)
            return (2 * N + 1) / ((2 * N - j) * (2 * N + 1 - j))
        if owner == "g":
            if j == N:
                return 2 / (N + 1)
            return 2 * N / ((2 * N - j) * (2 * N + 1 - j))
    elif right != "x_star":
        i, j = int(left[2:]), int(right[2:])
        if j == i + 1:
            return (
                (i + 1) / (2 * N - i) if owner == "f" else sp.Integer(i) / (2 * N - i)
            )
    return sp.S.Zero


def sos_pieces(ctx, N):
    N = sp.sympify(N)
    L = setup.L
    astar = setup.f.grad(ctx["x_star"])
    pieces = []
    for i in range(1, int(N) + 1):
        xprev, x = ctx[f"x_{i - 1}"], ctx[f"x_{i}"]
        aprev, a = setup.f.grad(xprev), setup.f.grad(x)
        s = setup.g.grad(x)
        c1 = N / ((2 * N + 1 - i) * (2 * N - i) * L)
        u1 = sp.Rational(i, 2) / N * a + (2 * N - i) / (2 * N) * aprev - astar
        c2 = (
            sp.Integer(i - 1) / (2 * N - i) * (2 * N + 2) / (2 * N + 1)
            + 2 * N / ((2 * N - i) ** 2 * (2 * N + 1))
        ) / (2 * L)
        u2 = aprev + s - L / (2 * N + 1 - i) * (xprev - ctx["x_star"])
        c3 = sp.Integer(i) / (2 * N + 1 - i) * 2 * N / (2 * N + 1) / (2 * L)
        u3 = (
            (2 * N + 1) / (2 * N) * (a + s)
            - (aprev + s) / (2 * N)
            - L / (2 * N - i) * (x - ctx["x_star"])
        )
        pieces.extend([(i, "u1", c1, u1), (i, "u2", c2, u2), (i, "u3", c3, u3)])
    return pieces


def sos_scalar(ctx, N):
    return sum((c * u**2 for _, _, c, u in sos_pieces(ctx, N)), pf.Scalar.zero())
```

```python
TOL = b2["numerical_checks"]["tolerances"]


def inventory(ctx, pb):
    entries = [(c, "initial", None, None) for c in pb.init_conditions]
    for owner in ctx.func_to_triplets:
        groups = ctx.get_constraint_data(owner)
        entries.extend(
            (c, "interpolation", str(owner), str(group))
            for group, constraints in groups.sc_dict.items()
            for c in constraints
        )
        entries.extend(
            (c, "interpolation", str(owner), str(group))
            for group, c in groups.psd_dict.items()
        )
    return entries


def residual_of(c):
    residual = c.lhs - c.rhs
    return -residual if c.cmp in (Comparator.GE, Comparator.SEQ) else residual


def verify(data, dense_value, decomposition=None):
    ctx, pb, _ = setup.get_pep_setup(data["N"], json.loads(data["params_json"]))
    if data["relaxed_constraints"]:
        pb.set_relaxed_constraints(data["relaxed_constraints"])
    for func in ctx.func_to_triplets:
        func.get_interpolation_constraints(ctx)
    entries = inventory(ctx, pb)
    assert set(data["constraint_metadata"]) == {c.name for c, *_ in entries}
    assert data["basis_vectors"] == [str(v) for v in ctx.basis_vectors()]
    assert data["basis_scalars"] == [str(v) for v in ctx.basis_scalars()]
    pm = pf.ExpressionManager(ctx, resolve_parameters={"L": sp.S.One, "R": sp.S.One})
    weighted = pf.Scalar.zero()
    vals = []
    for c, *_ in entries:
        value = data["named_dual_values"][c.name]
        if c.name in pb.relaxed_constraints:
            assert value is None
            continue
        assert isinstance(c, pf.ScalarConstraint), (
            "This PGM sample has only scalar constraints"
        )
        weighted += float(value) * (pf.Scalar.zero() + residual_of(c))
        if c.cmp != Comparator.EQ:
            vals.append(float(value))
    basis = ctx.basis_vectors()
    S = np.asarray(data["S_matrix"], dtype=float)
    S_scalar = sum(
        (float(S[i, j]) * basis[i] * basis[j] for i, j in np.ndindex(S.shape)),
        pf.Scalar.zero(),
    )
    checked = pm.eval_scalar(
        pb.performance_metric - data["opt_value"] - weighted + S_scalar
    )

    def ma(v):
        return float(np.max(np.abs(np.asarray(v, dtype=float)), initial=0.0))

    reconstruction = (
        S
        if decomposition is None
        else np.asarray(decomposition["ell_coords"]).T
        @ np.asarray(decomposition["D"])
        @ np.asarray(decomposition["ell_coords"])
    )
    residuals = dict(
        identity_gram_max_abs=ma(checked.inner_prod_coords),
        identity_func_max_abs=ma(checked.func_coords),
        identity_offset_abs=ma(checked.offset),
        S_reconstruction_max_abs=ma(reconstruction - S),
        dense_relaxed_gap=abs(data["opt_value"] - dense_value),
        minimum_scalar_inequality_dual=min(vals, default=0.0),
        minimum_S_eigenvalue=float(np.linalg.eigvalsh((S + S.T) / 2).min()),
        minimum_matrix_inequality_eigenvalue=0.0,
        candidate_rate_gap=abs(data["opt_value"] - 1 / (4 * data["N"])),
    )
    passed = (
        data["certificate_provenance"]["solver_status"] == "optimal"
        and np.all(np.isfinite(S))
        and np.allclose(S, S.T, atol=1e-6, rtol=0)
    )
    passed = passed and all(
        residuals[k] <= 1e-5
        for k in (
            "identity_gram_max_abs",
            "identity_func_max_abs",
            "identity_offset_abs",
            "dense_relaxed_gap",
            "candidate_rate_gap",
        )
    )
    passed = (
        passed
        and np.allclose(reconstruction, S, atol=1e-4, rtol=1e-5)
        and residuals["minimum_scalar_inequality_dual"] >= -1e-6
        and residuals["minimum_S_eigenvalue"] >= -1e-6
    )
    return (
        dict(tolerances=TOL, residuals=residuals, passed=bool(passed)),
        ctx,
        pb,
        checked,
    )
```

```python
checks, ctx, pb, numerical_residual = verify(
    b2, dense["opt_value"], b2["numerical_decomposition"]
)
assert checks["passed"]
print("Complete numerical certificate:", checks)
pf.pprint_labeled_matrix(
    np.asarray(numerical_residual.inner_prod_coords, dtype=float),
    b2["basis_vectors"],
    precision=9,
)
print("Function residual:", numerical_residual.func_coords)
print("Offset residual:", numerical_residual.offset)
print(
    "At the B2 handoff: numerical certificate verified; analytic proofs were still pending."
)
```

```python
N = b2["N"]
pm = pf.ExpressionManager(ctx, resolve_parameters={"L": sp.S.One, "R": sp.S.One})
weighted_exact = pf.Scalar.zero()
fit_errors = []
for constraint, origin, owner, group in inventory(ctx, pb):
    if origin == "initial":
        coefficient = sp.Rational(1, 4 * N)
    else:
        left, right = constraint.name.split(":")[1].split(",")
        coefficient = lambda_candidate(owner, left, right, N)
        fit_errors.append(
            abs(float(coefficient) - (b2["named_dual_values"][constraint.name] or 0.0))
        )
    weighted_exact += coefficient * residual_of(constraint)
for owner, labels in (
    ("f", ["x_star"] + [f"x_{i}" for i in range(N + 1)]),
    ("g", ["x_star"] + [f"x_{i}" for i in range(1, N + 1)]),
):
    matrix = np.array(
        [
            [
                sp.S.Zero
                if left == right or right == "x_star"
                else lambda_candidate(owner, left, right, N)
                for right in labels
            ]
            for left in labels
        ],
        dtype=object,
    )
    print(owner, "candidate weights:")
    pf.pprint_labeled_matrix(matrix, labels)
print("Maximum candidate multiplier discrepancy:", max(fit_errors))
assert max(fit_errors) <= 1e-4
pieces = sos_pieces(ctx, N)
D = np.diag(
    [
        float(sp.N(eval_parameter(coefficient, {"L": sp.S.One})))
        for _, _, coefficient, _ in pieces
    ]
)
E = np.asarray(
    [pm.eval_vector(vector).coords for _, _, _, vector in pieces], dtype=float
)
S_candidate_matrix = E.T @ D @ E
S_saved = np.asarray(b2["S_matrix"])
assert np.allclose(S_candidate_matrix, S_saved, atol=1e-4, rtol=1e-5)
print("SOS reconstruction residual:", np.max(abs(S_candidate_matrix - S_saved)))
pf.pprint_labeled_matrix(S_candidate_matrix, b2["basis_vectors"], precision=6)
exact_residual = pm.eval_scalar(
    pb.performance_metric - sp.Rational(1, 4 * N) - weighted_exact + sos_scalar(ctx, N),
    sympy_mode=True,
)
assert all(
    sp.simplify(value) == 0
    for arr in (
        exact_residual.inner_prod_coords,
        exact_residual.func_coords,
        np.asarray(exact_residual.offset),
    )
    for value in np.asarray(arr).flat
)
print(
    "Exact candidate identity at N=4: all Gram, function and offset coordinates vanish."
)
pf.pprint_labeled_matrix(exact_residual.inner_prod_coords, b2["basis_vectors"])
print("This finite-horizon exact check does not prove the all-horizon formula.")
```

## Partial-sum Lyapunov construction and rank profile
Let $I_{ij}^f$ and $I_{ij}^g$ denote the nonpositive interpolation residuals defined above, and let $S_i$ be the sum of the three blended numerical SOS pieces at step $i$. Start with $V_0=0$. Add $\lambda^f_{\star,0}I^f_{\star,0}$ at the first step. Thereafter, the canonical increments collect the new star and adjacent inequalities:
$$V_{k+1}-V_k=\lambda^f_{\star,k+1}I^f_{\star,k+1}+\lambda^g_{\star,k+1}I^g_{\star,k+1}+\lambda^f_{k,k+1}I^f_{k,k+1}+\lambda^g_{k,k+1}I^g_{k,k+1}-S_{k+1},$$
where the nonexistent $g$ pair at $k=0$ is absent and the stated initial term is included there. Every coefficient is nonnegative and every subtracted square is nonnegative, so $V_{k+1}\leq V_k$ under the exact candidate's domain conditions. Here the saved numerical multipliers are used to discover structure; numerical tolerances do not constitute the final analytic theorem.
The construction retains all interpolation terms. At the end, the full certificate gives $V_N\approx h(x_N)-h(x_\star)-\tau\|x_0-x_\star\|^2$; the small constant discrepancy between the numerical objective and $\tau R^2$ is retained by the certificate checks. No boundary interpolation or SOS term is omitted. The quadratic ranks are $0,4,\ldots,4,1$, confirmed at $N=4$ and $N=6$.

**Numerical decomposition recovery.** The supplied factors reconstruct the slack approximately; using them alone failed the terminal basis-span check. Assigning the discrepancy to the final piece was indefinite and rejected. A polar alignment $W=QS^{1/2}$, with $Q$ the polar factor of $US^{1/2}$ for supplied factors $U$, reconstructs the saved slack exactly, but using it alone failed two interior span checks. The selected recipe uses the fixed nonnegative blend
$$\widetilde S_i=\frac7{10}S_i^{\mathrm{supplied}}+\frac3{10}S_i^{\mathrm{aligned}}.$$
Each piece is a sum of squares. Its reconstruction and full-coordinate coverage errors are checked against the unchanged frozen tolerances at both horizons. The blend was chosen after measuring basis-span residuals, and this numerical tuning is recorded in supplementary diagnostics. The same value $3/10$ is used at both horizons. No certificate components or tolerances changed. The original supplied SOS remains the candidate for exact symbolic analysis; the finite-sample numerical blend is not an all-horizon formula.

```python
import hashlib

ROOT = notebook_dir
APPROVED = repo_root
CERTIFICATE_KEYS = (
    "N_verify",
    "params_json",
    "relaxed_constraints",
    "S_matrix",
    "S_row_names",
    "S_col_names",
    "basis_vectors",
    "basis_scalars",
    "opt_value",
    "tau_sol",
    "tau_name",
    "named_dual_values",
    "constraint_metadata",
    "normalization",
    "certificate_provenance",
    "numerical_checks",
    "numerical_decomposition",
)


def replay(certificate):
    digest = hashlib.sha256(
        json.dumps(
            {key: certificate[key] for key in CERTIFICATE_KEYS},
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode()
    ).hexdigest()
    assert digest == certificate["certificate_id"]
    assert certificate["closed_form_search"]["completed"]
    assert certificate["certificate_provenance"]["pep_type"] == "primal"
    assert Path(pf.__file__).resolve().is_relative_to(APPROVED)
    for name, expected in certificate["certificate_provenance"][
        "source_hashes"
    ].items():
        path = (ROOT / name).resolve()
        assert path.is_relative_to(ROOT)
        assert hashlib.sha256(path.read_bytes()).hexdigest() == expected
    checks, ctx, pb, _ = verify(
        certificate, certificate["opt_value"], certificate["numerical_decomposition"]
    )
    assert checks["passed"]
    assert certificate["N"] == certificate["N_verify"]
    assert (
        certificate["S_row_names"]
        == certificate["basis_vectors"]
        == certificate["S_col_names"]
    )
    entries = inventory(ctx, pb)
    assert len(entries) == len({c.name for c, *_ in entries})
    assert set(certificate["named_dual_values"]) == set(
        certificate["constraint_metadata"]
    )
    for c, origin, owner, group in entries:
        expected = dict(
            origin=origin,
            owner=owner,
            group=group,
            kind="scalar",
            comparator=c.cmp.value,
            shape=[],
            relaxed=c.name in pb.relaxed_constraints,
        )
        assert certificate["constraint_metadata"][c.name] == expected
    assert (
        certificate["normalization"]["tau_name"]
        == certificate["tau_name"]
        == "initial_condition"
    )
    assert (
        abs(
            certificate["tau_sol"]
            - certificate["named_dual_values"]["initial_condition"]
        )
        <= 1e-6
    )
    assert (
        abs(certificate["normalization"]["bound_value"] - certificate["opt_value"])
        <= 1e-5
    )
    return ctx, pb, checks
```

```python
from scipy.linalg import polar


def aligned_squares(certificate, ctx):
    S = np.asarray(certificate["S_matrix"], dtype=float)
    values, vectors = np.linalg.eigh((S + S.T) / 2)
    assert values.min() >= 0, "This exact numerical alignment requires a PSD saved S"
    sqrt_S = (vectors * np.sqrt(values)) @ vectors.T
    decomposition = certificate["numerical_decomposition"]
    D = np.asarray(decomposition["D"], dtype=float)
    E = np.asarray(decomposition["ell_coords"], dtype=float)
    assert np.array_equal(D, np.diag(np.diag(D))) and np.diag(D).min() >= 0
    supplied_factors = np.sqrt(np.diag(D))[:, None] * E
    Q, _ = polar(supplied_factors @ sqrt_S)
    aligned = Q @ sqrt_S
    assert np.allclose(aligned.T @ aligned, S, atol=1e-10, rtol=0)
    basis = ctx.basis_vectors()
    pieces = {
        label: sum((float(coef) * v for coef, v in zip(row, basis)), pf.Vector.zero())
        ** 2
        for label, row in zip(decomposition["piece_labels"], aligned)
    }
    return pieces, dict(
        method="polar alignment of supplied factors to saved S",
        factors=aligned.tolist(),
        factor_max_change=float(np.max(abs(aligned - supplied_factors))),
        reconstruction_max_abs=float(np.max(abs(aligned.T @ aligned - S))),
    )


def construct(certificate, variant="canonical"):
    ctx, pb, certificate_checks = replay(certificate)
    n = certificate["N"]
    pm = pf.ExpressionManager(ctx, resolve_parameters={"L": sp.S.One, "R": sp.S.One})
    group_terms = {}
    term_assignment = {}
    for constraint, origin, owner, group in inventory(ctx, pb):
        value = certificate["named_dual_values"][constraint.name]
        if origin == "initial" or value is None:
            continue
        group_terms[constraint.name] = float(value) * residual_of(constraint)
        left, right = constraint.name.split(":")[1].split(",")
        j = int(right[2:])
        if left == "x_star" and j == 0:
            term_assignment[constraint.name] = -1
        else:
            term_assignment[constraint.name] = j - 1
    S_piece_terms = {
        f"{family}_{i}": coefficient * vector**2
        for i, family, coefficient, vector in sos_pieces(ctx, n)
    }
    if variant == "polar-aligned":
        S_piece_terms, alignment = aligned_squares(certificate, ctx)
    elif variant == "blend-3/10":
        aligned_pieces, alignment = aligned_squares(certificate, ctx)
        S_piece_terms = {
            name: sp.Rational(7, 10) * piece + sp.Rational(3, 10) * aligned_pieces[name]
            for name, piece in S_piece_terms.items()
        }
    S_assignment = {name: int(name.split("_")[1]) - 1 for name in S_piece_terms}
    lyap = [pf.Scalar.zero()]
    partial_sum = sum(
        (term for name, term in group_terms.items() if term_assignment[name] == -1),
        pf.Scalar.zero(),
    )
    for k in range(n):
        partial_sum += sum(
            (term for name, term in group_terms.items() if term_assignment[name] == k),
            pf.Scalar.zero(),
        )
        partial_sum -= sum(
            (term for name, term in S_piece_terms.items() if S_assignment[name] == k),
            pf.Scalar.zero(),
        )
        lyap.append(partial_sum)
    evaluated = [pm.eval_scalar(v) for v in lyap]
    spectra = [
        np.linalg.svd(np.asarray(v.inner_prod_coords, dtype=float), compute_uv=False)
        for v in evaluated
    ]
    ranks = [int(np.sum(s > 1e-4)) for s in spectra]
    basis = ctx.basis_vectors()
    S = np.asarray(certificate["S_matrix"], dtype=float)
    S_scalar = sum(
        (float(S[i, j]) * basis[i] * basis[j] for i, j in np.ndindex(S.shape)),
        pf.Scalar.zero(),
    )
    checks = {}
    for label, expression in [
        (
            "coverage",
            lyap[n] - (sum(group_terms.values(), pf.Scalar.zero()) - S_scalar),
        ),
        ("pieces", sum(S_piece_terms.values(), pf.Scalar.zero()) - S_scalar),
    ]:
        value = pm.eval_scalar(expression)
        residual = max(
            float(np.max(np.abs(np.asarray(arr, dtype=float)), initial=0))
            for arr in (
                value.inner_prod_coords,
                value.func_coords,
                np.asarray(value.offset),
            )
        )
        checks[label] = residual
        assert residual <= 1e-5
    min_piece_eigen = 0.0
    for piece in S_piece_terms.values():
        value = pm.eval_scalar(piece)
        M = np.asarray(value.inner_prod_coords, dtype=float)
        min_piece_eigen = min(
            min_piece_eigen, float(np.linalg.eigvalsh((M + M.T) / 2).min())
        )
        assert (
            np.max(abs(value.func_coords), initial=0) <= 1e-5
            and abs(value.offset) <= 1e-5
        )
    assert min_piece_eigen >= -1e-6
    increment_errors = []
    for k in range(n):
        steps = (-1, 0) if k == 0 else (k,)
        inc = sum(
            (
                term
                for name, term in group_terms.items()
                if term_assignment[name] in steps
            ),
            pf.Scalar.zero(),
        )
        inc -= sum(
            (
                term
                for name, term in S_piece_terms.items()
                if S_assignment[name] in steps
            ),
            pf.Scalar.zero(),
        )
        value = pm.eval_scalar(lyap[k + 1] - lyap[k] - inc)
        error = max(
            float(np.max(np.abs(np.asarray(arr, dtype=float)), initial=0))
            for arr in (
                value.inner_prod_coords,
                value.func_coords,
                np.asarray(value.offset),
            )
        )
        increment_errors.append(error)
        assert error <= 1e-5
    checks.update(
        minimum_piece_eigenvalue=min_piece_eigen,
        maximum_increment_error=max(increment_errors),
        boundary_constraints=[],
        boundary_S=[],
        omitted_constraints=[],
    )
    observation = dict(
        N=n,
        certificate_id=certificate["certificate_id"],
        recipe_id=variant,
        rank_profile=ranks,
        rank_convention=dict(
            tolerance=1e-4,
            scaling="absolute, unscaled native Gram coordinates",
            coordinate_order="context basis creation order",
        ),
        singular_values=[s.tolist() for s in spectra],
        rank_credible=all(not np.any((s >= 5e-5) & (s <= 2e-4)) for s in spectra),
        certificate_checked=True,
        grouping_checked=True,
        checks=checks,
    )
    return dict(
        ctx=ctx,
        pb=pb,
        pm=pm,
        lyap=lyap,
        observation=observation,
        term_assignment=term_assignment,
        S_assignment=S_assignment,
        group_terms=group_terms,
        S_piece_terms=S_piece_terms,
        certificate_checks=certificate_checks,
    )


b3 = json.loads((notebook_dir / "state/pgm_b3.json").read_text())
selected_trial = construct(b3, variant="blend-3/10")
lyap = selected_trial["lyap"]
pm = selected_trial["pm"]
ctx = selected_trial["ctx"]
rank_left = rank_right = 1
N_int = b3["N"]
ranks = []
for k, Vk in enumerate(lyap):
    matrix = np.asarray(pm.eval_scalar(Vk).inner_prod_coords, dtype=float)
    rank = int(np.linalg.matrix_rank(matrix, tol=1e-4))
    ranks.append(rank)
    print(f"rank V_{k}: {rank}")
    if k == 0:
        print()
retained = ranks[rank_left : N_int + 1 - rank_right]
print(
    "Interior rank is constant at this N:",
    len(retained) >= 3 and len(set(retained)) == 1,
    "boundary exclusions:",
    (rank_left, rank_right),
)
print(
    "Full-coordinate coverage residual:",
    selected_trial["observation"]["checks"]["coverage"],
)
print(
    "Maximum increment residual:",
    selected_trial["observation"]["checks"]["maximum_increment_error"],
)
print("Boundary interpolation/S leftovers: none; omitted terms: none.")
```

### Candidate-vector scan
Candidates include tagged iterates, basis vectors, all available smooth gradients and proximal subgradients, their pairwise differences, and the meaningful point and gradient gaps. Zero and coordinate-duplicate vectors are removed, with preferred descriptive labels first.

```python
from itertools import combinations
from pepflow.lyapunov_utils import (
    vectors_in_column_space,
    find_basis_with_sparsest_coefficients,
    find_symmetric_coefficient_matrix,
)

candidate_pairs = {}
for k in range(1, N_int + 1):
    x = ctx[f"x_{k}"]
    a = setup.f.grad(x)
    pairs = [("x_0-x_star", ctx["x_0"] - ctx["x_star"])]
    if k < N_int:
        pairs += [
            (f"x_{k}-x_star", x - ctx["x_star"]),
            (f"a_{k}", a),
            (f"a_{k}-a_star", a - setup.f.grad(ctx["x_star"])),
        ]
    base = [(str(v), v) for v in ctx.basis_vectors()]
    base += [(f"x_{i}", ctx[f"x_{i}"]) for i in range(N_int + 1)] + [
        ("x_star", ctx["x_star"])
    ]
    base += [(f"a_{i}", setup.f.grad(ctx[f"x_{i}"])) for i in range(N_int + 1)]
    base += [(f"s_{i}", setup.g.grad(ctx[f"x_{i}"])) for i in range(1, N_int + 1)]
    base += [("a_star", setup.f.grad(ctx["x_star"]))]
    pairs += base
    pairs += [
        (f"({name1})-({name2})", v1 - v2)
        for (name1, v1), (name2, v2) in combinations(base, 2)
    ]
    unique = []
    coordinates = []
    for label, vector in pairs:
        row = np.asarray(pm.eval_vector(vector).coords, dtype=float)
        if np.linalg.norm(row) <= 1e-12 or any(
            np.allclose(row, old, atol=1e-12, rtol=0) for old in coordinates
        ):
            continue
        unique.append((label, vector))
        coordinates.append(row)
    candidate_pairs[k] = unique
```

```python
spanning_by_k = {}
labels_by_k = {}
for k, pairs in candidate_pairs.items():
    spanning_by_k[k] = vectors_in_column_space(
        lyap[k],
        [v for _, v in pairs],
        pep_context=ctx,
        resolve_parameters={"L": sp.S.One, "R": sp.S.One},
        rtol=1e-4,
        atol=1e-4,
    )
    labels_by_k[k] = {id(v): label for label, v in pairs}
    print(f"V_{k} column-space candidates:")
    for vector in spanning_by_k[k]:
        print("   " + labels_by_k[k][id(vector)])
```

### Selected basis pattern
The sparse search below selects a consistent interior pattern
$$[x_0-x_\star,\ x_k-x_\star,\ a_k,\ a_k-a_\star].$$
At $k=0$ the quadratic part is zero and has an empty basis. At $k=N$ the basis is $[x_0-x_\star]$. The same templates are then independently checked against each partial sum.

```python
sparse_basis_by_k = {}
sparse_coefficients = {}
for k in range(1, N_int + 1):
    basis, C = find_basis_with_sparsest_coefficients(
        lyap[k],
        spanning_by_k[k],
        pep_context=ctx,
        resolve_parameters={"L": sp.S.One, "R": sp.S.One},
        zero_tol=1e-6,
        indep_tol=1e-7,
        span_tol=1e-5,
    )
    assert len(basis) == b4["rank_profile"][k]
    sparse_basis_by_k[k] = basis
    sparse_coefficients[k] = C
    print(
        f"k={k}: rank {len(basis)}; sparse basis: {[labels_by_k[k][id(v)] for v in basis]}"
    )
    print(f"zero coefficients: {int(np.sum(abs(C) < 1e-6))}/{C.size}")


def V_k_basis(k, ctx, N):
    if k == 0:
        return []
    x0gap = ctx["x_0"] - ctx["x_star"]
    if k == N:
        return [x0gap]
    x = ctx[f"x_{k}"]
    a = setup.f.grad(x)
    return [x0gap, x - ctx["x_star"], a, a - setup.f.grad(ctx["x_star"])]


def V_k_basis_labels(k, N):
    if k == 0:
        return []
    if k == N:
        return ["x_0-x_star"]
    return ["x_0-x_star", f"x_{k}-x_star", f"a_{k}", f"a_{k}-a_star"]


for k in range(N_int + 1):
    basis = V_k_basis(k, ctx, N_int)
    B = (
        np.column_stack([pm.eval_vector(v).coords for v in basis]).astype(float)
        if basis
        else np.zeros((len(ctx.basis_vectors()), 0))
    )
    rank = int(np.linalg.matrix_rank(B, tol=1e-7)) if basis else 0
    assert rank == len(basis) == b4["rank_profile"][k]
    print(f"k={k}: rank {rank} basis {V_k_basis_labels(k, N_int)}")
```

### Coefficient matrices
The basis order is exactly the displayed template. Symmetric off-diagonal matrix entries contribute twice to the quadratic form. Function-value gaps and the scalar offset are reconstructed separately. All formulas below are candidates supported by this selected sample; no arbitrary-horizon conclusion follows from a fit.

```python
def coeff_pattern(k, N, L=sp.S.One):
    k, N, L = map(sp.sympify, (k, N, L))
    if k == 0:
        return sp.zeros(0)
    if k == N:
        return sp.Matrix([[-L / (4 * N)]])
    return sp.Matrix(
        [
            [-L / (4 * N), 0, 0, 0],
            [0, L * (N - k) / (2 * N - k) ** 2, -1 / (2 * (2 * N - k)), 0],
            [0, -1 / (2 * (2 * N - k)), 0, 0],
            [0, 0, 0, 1 / (2 * L * (2 * N - k))],
        ]
    )


def function_coeff_pattern(k, N):
    k, N = map(sp.sympify, (k, N))
    assert isinstance(k, sp.Expr) and isinstance(N, sp.Expr)
    if k == 0:
        return sp.S.Zero, sp.S.Zero
    if k == N:
        return sp.S.One, sp.S.One
    return (k + 1) / (2 * N - k), k / (2 * N - k)


def V_k_structure(k, ctx, N, coeff_by_k, function_coefficients, offset_coefficients):
    basis = V_k_basis(k, ctx, N)
    C = coeff_by_k[k]
    value = sum(
        (float(C[i, j]) * basis[i] * basis[j] for i, j in np.ndindex(C.shape)),
        pf.Scalar.zero(),
    )
    if k > 0:
        fcoef, gcoef = function_coefficients[k]
        value += float(fcoef) * (setup.f(ctx[f"x_{k}"]) - setup.f(ctx["x_star"]))
        value += float(gcoef) * (setup.g(ctx[f"x_{k}"]) - setup.g(ctx["x_star"]))
    return value + float(offset_coefficients[k])


def extract_structure(trial, N):
    ctx, pm, lyap = trial["ctx"], trial["pm"], trial["lyap"]
    coeffs, funcs, offsets, errors = {}, {}, {}, {}
    for k in range(N + 1):
        basis = V_k_basis(k, ctx, N)
        target = pm.eval_scalar(lyap[k])
        M = np.asarray(target.inner_prod_coords, dtype=float)
        if not basis:
            C = np.zeros((0, 0))
            B = np.zeros((len(M), 0))
        else:
            B = np.column_stack([pm.eval_vector(v).coords for v in basis]).astype(float)
            assert (
                np.linalg.matrix_rank(B, tol=1e-7)
                == len(basis)
                == trial["observation"]["rank_profile"][k]
            )
            C = find_symmetric_coefficient_matrix(
                lyap[k],
                basis,
                pep_context=ctx,
                resolve_parameters={"L": sp.S.One, "R": sp.S.One},
                indep_tol=1e-7,
                span_tol=1e-5,
            )
        residual = float(np.linalg.norm(B @ C @ B.T - M, "fro"))
        assert residual <= 1e-5 * max(1.0, np.linalg.norm(M, "fro"))
        coeffs[k] = C
        if k == 0:
            funcs[k] = [0.0, 0.0]
        else:
            columns = np.column_stack(
                [
                    pm.eval_scalar(fun(ctx[f"x_{k}"]) - fun(ctx["x_star"])).func_coords
                    for fun in (setup.f, setup.g)
                ]
            ).astype(float)
            funcs[k] = np.linalg.lstsq(
                columns, np.asarray(target.func_coords, dtype=float), rcond=None
            )[0].tolist()
        offsets[k] = float(target.offset)
        diff = pm.eval_scalar(
            V_k_structure(k, ctx, N, coeffs, funcs, offsets) - lyap[k]
        )
        components = {
            name: float(np.max(abs(np.asarray(arr, dtype=float)), initial=0))
            for name, arr in [
                ("gram", diff.inner_prod_coords),
                ("function", diff.func_coords),
                ("offset", np.asarray(diff.offset)),
            ]
        }
        assert all(value <= 1e-5 for value in components.values())
        Cguess = np.asarray(coeff_pattern(k, N), dtype=float).reshape(C.shape)
        fit = float(np.max(abs(Cguess - C), initial=0))
        assert fit <= 1e-4
        assert (
            np.max(
                abs(
                    np.asarray(function_coeff_pattern(k, N), dtype=float)
                    - np.asarray(funcs[k])
                ),
                initial=0,
            )
            <= 1e-5
        )
        errors[k] = dict(
            quadratic_frobenius=residual,
            components=components,
            candidate_coefficient_max_abs=fit,
        )
    return coeffs, funcs, offsets, errors


coeff_by_k, function_coefficients, offset_coefficients, structure_errors = (
    extract_structure(trial, N_int)
)
for k, C in coeff_by_k.items():
    labels = V_k_basis_labels(k, N_int)
    print(
        f"k={k}: quadratic Frobenius residual={structure_errors[k]['quadratic_frobenius']:.8g}"
    )
    if C.size:
        pf.pprint_labeled_matrix(C, labels, precision=8)
        print("Candidate coefficient matrix:")
        pf.pprint_labeled_matrix(
            np.asarray(coeff_pattern(k, N_int), dtype=object), labels
        )
    else:
        print("Empty 0-by-0 quadratic matrix.")
    print(
        "Coefficients of f-gap and g-gap:",
        function_coefficients[k],
        "offset:",
        offset_coefficients[k],
    )
    print("Full-component residuals:", structure_errors[k]["components"])
    print(
        "Candidate matrix maximum discrepancy:",
        structure_errors[k]["candidate_coefficient_max_abs"],
    )
```

### Block 4 conclusion
For $1\leq k<N$, put $d_k=x_k-x_\star$ and $a_k=\nabla f(x_k)$. The reconstructed structure suggests
$$V_k=-\frac{L}{4N}\|d_0\|^2+\frac{L(N-k)}{(2N-k)^2}\|d_k\|^2-\frac{\langle a_k,d_k\rangle}{2N-k}+\frac{\|a_k-a_\star\|^2}{2L(2N-k)}+\frac{k+1}{2N-k}(f_k-f_\star)+\frac{k}{2N-k}(g_k-g_\star).$$
The endpoint candidates are $V_0=0$ and $V_N=h(x_N)-h(x_\star)-L\|d_0\|^2/(4N)$. There are no scalar offset terms. The actual numerical matrix coefficients and gap coefficients reconstruct all selected partial sums within the frozen tolerances, including both endpoints. These numerical checks use the blended pieces; B5 must derive and verify the exact formulas and all step/base/boundary identities. Analytic full-PEP and Lyapunov proof statuses remain false.

## Closed-form derivation and exact verification

The sparse coefficient pattern from B4 fixes the candidate $\Phi_{k}$ displayed in the theorem. The adjacent interpolation weights are $\alpha_i=i/(2N+1-i)$ and $\beta_i=(i-1)/(2N+1-i)$; their differences generate the star weights in the interior. The base and terminal coefficients are fixed by $V_{0}=0$ and $V_{N}=h(x_N)-h(x_\star)-L\|d_0\|^2/(4N)$. We now substitute these candidates into the actual proximal-gradient recurrence and verify all coordinates exactly, with independent base and terminal contexts. No coefficient fitting or numerical blending is used in these symbolic identities.

```python
assert b4["numerical_reconstruction"]["certificate_id"] == b4["certificate_id"]
assert all(error["quadratic_frobenius"] <= 1e-5 for error in structure_errors.values())
Kpar, Npar, Lpar = pf.Parameter("k"), pf.Parameter("N"), pf.Parameter("L")
k_symbolic, N_symbolic = sp.symbols("k N", integer=True, positive=True)
L_symbolic = sp.Symbol("L", positive=True)
symbolic_parameters = {"k": k_symbolic, "N": N_symbolic, "L": L_symbolic}
symbolic_checks = {}


def smooth_residual(F, u, v, L):
    return F(v) - F(u) + F.grad(v) * (u - v) + (F.grad(u) - F.grad(v)) ** 2 / (2 * L)


def convex_residual(G, u, v):
    return G(v) - G(u) + G.grad(v) * (u - v)


def phi(k, N, L, x, star, F, G):
    d = x - star
    a, astar = F.grad(x), F.grad(star)
    return (
        L * (N - k) / (2 * N - k) ** 2 * d**2
        - a * d / (2 * N - k)
        + (a - astar) ** 2 / (2 * L * (2 * N - k))
        + (k + 1) / (2 * N - k) * (F(x) - F(star))
        + k / (2 * N - k) * (G(x) - G(star))
    )


def step_square(i, N, L, x, xplus, star, F, G):
    a0, a1, astar = F.grad(x), F.grad(xplus), F.grad(star)
    s = G.grad(xplus)
    c1 = N / ((2 * N + 1 - i) * (2 * N - i) * L)
    u1 = i / (2 * N) * a1 + (2 * N - i) / (2 * N) * a0 - astar
    c2 = (
        (i - 1) / (2 * N - i) * (2 * N + 2) / (2 * N + 1)
        + 2 * N / ((2 * N - i) ** 2 * (2 * N + 1))
    ) / (2 * L)
    u2 = a0 + s - L / (2 * N + 1 - i) * (x - star)
    c3 = i / (2 * N + 1 - i) * 2 * N / (2 * N + 1) / (2 * L)
    u3 = (
        (2 * N + 1) / (2 * N) * (a1 + s)
        - (a0 + s) / (2 * N)
        - L / (2 * N - i) * (xplus - star)
    )
    return c1 * u1**2 + c2 * u2**2 + c3 * u3**2


def exact_components(expression, ctx, parameters):
    manager = pf.ExpressionManager(ctx, resolve_parameters=parameters)
    value = manager.eval_scalar(expression, sympy_mode=True)
    gram = sp.Matrix(value.inner_prod_coords).applyfunc(
        lambda z: sp.factor(sp.cancel(z))
    )
    funcs = sp.Matrix(list(value.func_coords)).applyfunc(
        lambda z: sp.factor(sp.cancel(z))
    )
    offset = sp.factor(sp.cancel(value.offset))
    return gram, funcs, offset


def assert_identity(name, LHS, RHS, ctx, parameters, display=False):
    gram, funcs, offset = exact_components(LHS - RHS, ctx, parameters)
    passed = all(v == 0 for v in [*gram, *funcs, offset])
    print(
        name, "exact zero:", passed, "coordinates:", gram.shape, funcs.shape, flush=True
    )
    if display:
        pf.pprint_labeled_matrix(
            __import__("numpy").array(gram.tolist(), dtype=object),
            [str(v) for v in ctx.basis_vectors()],
        )
        print("Function-coordinate residual:", funcs)
        print("Offset residual:", offset)
    if not passed:
        print(
            "Nonzero entries:",
            [v for v in [*gram, *funcs, offset] if v != 0],
            flush=True,
        )
    assert passed
    return dict(
        passed=passed,
        gram_shape=list(gram.shape),
        function_shape=list(funcs.shape),
        offset=str(offset),
    )
```

```python
# Exact recurrence relations for the candidate interpolation coefficients.
i_symbolic = sp.Symbol("i", integer=True, positive=True)
p_increment = (i_symbolic + 1) / (2 * N_symbolic - i_symbolic) - i_symbolic / (
    2 * N_symbolic + 1 - i_symbolic
)
q_increment = i_symbolic / (2 * N_symbolic - i_symbolic) - (i_symbolic - 1) / (
    2 * N_symbolic + 1 - i_symbolic
)
assert (
    sp.cancel(
        p_increment
        - (2 * N_symbolic + 1)
        / ((2 * N_symbolic - i_symbolic) * (2 * N_symbolic + 1 - i_symbolic))
    )
    == 0
)
assert (
    sp.cancel(
        q_increment
        - 2
        * N_symbolic
        / ((2 * N_symbolic - i_symbolic) * (2 * N_symbolic + 1 - i_symbolic))
    )
    == 0
)
assert sp.cancel(1 - N_symbolic / (N_symbolic + 1) - 1 / (N_symbolic + 1)) == 0
assert sp.cancel(1 - (N_symbolic - 1) / (N_symbolic + 1) - 2 / (N_symbolic + 1)) == 0
print("Interior star increments and terminal coefficient corrections verified exactly.")
```

## Symbolic Step Recursion Verification

For $1\leq k\leq N-2$, the equation checked is
$$\Phi_{k+1}-\Phi_k=\frac{2N+1}{(2N-k-1)(2N-k)}I^f_{x_\star,x_{k+1}}+\frac{2N}{(2N-k-1)(2N-k)}I^g_{x_\star,x_{k+1}}+\frac{k+1}{2N-k}I^f_{x_k,x_{k+1}}+\frac{k}{2N-k}I^g_{x_k,x_{k+1}}-S_{k+1}.$$
The anchored initial-distance term cancels. The residual $\mathrm{LHS}-\mathrm{RHS}$ must vanish in every Gram, function-value and constant coordinate.

```python
ctx_symbolic = pf.PEPContext("step").set_as_current()
F_symbolic = pf.SmoothConvexFunction(is_basis=True, tags=["f_{step}"], L=Lpar)
G_symbolic = pf.ConvexFunction(is_basis=True, tags=["g_{step}"])
H_symbolic = F_symbolic + G_symbolic
star_symbolic = H_symbolic.set_stationary_point("x_{star,step}")
x_symbolic = pf.Vector(is_basis=True, tags=["x_{step}"])
xplus_symbolic = G_symbolic.prox(
    x_symbolic - F_symbolic.grad(x_symbolic) / Lpar, 1 / Lpar, tag="x_{plus,step}"
)
F_symbolic(xplus_symbolic)
LHS = phi(
    Kpar + 1, Npar, Lpar, xplus_symbolic, star_symbolic, F_symbolic, G_symbolic
) - phi(Kpar, Npar, Lpar, x_symbolic, star_symbolic, F_symbolic, G_symbolic)
RHS = (
    (2 * Npar + 1)
    / ((2 * Npar - Kpar - 1) * (2 * Npar - Kpar))
    * smooth_residual(F_symbolic, star_symbolic, xplus_symbolic, Lpar)
)
RHS += (
    2
    * Npar
    / ((2 * Npar - Kpar - 1) * (2 * Npar - Kpar))
    * convex_residual(G_symbolic, star_symbolic, xplus_symbolic)
)
RHS += (
    (Kpar + 1)
    / (2 * Npar - Kpar)
    * smooth_residual(F_symbolic, x_symbolic, xplus_symbolic, Lpar)
)
RHS += (
    Kpar / (2 * Npar - Kpar) * convex_residual(G_symbolic, x_symbolic, xplus_symbolic)
)
RHS -= step_square(
    Kpar + 1,
    Npar,
    Lpar,
    x_symbolic,
    xplus_symbolic,
    star_symbolic,
    F_symbolic,
    G_symbolic,
)
diff_step = LHS - RHS
symbolic_checks["step"] = assert_identity(
    "Interior step, 1 <= k <= N-2",
    LHS,
    RHS,
    ctx_symbolic,
    symbolic_parameters,
    display=True,
)
```

## Base Case and Boundary Symbolic Verification

For $N\geq2$, the initial equation is
$$V_1=\frac{1}{2N}I^f_{x_\star,x_0}+\frac{2N+1}{(2N-1)(2N)}I^f_{x_\star,x_1}+\frac{1}{2N-1}I^g_{x_\star,x_1}+\frac{1}{2N}I^f_{x_0,x_1}-S_1.$$
Here $V_1=\Phi_1-L\|d_0\|^2/(4N)$ and $V_0=0$. No $g(x_0)$ is introduced. The residual $\mathrm{LHS}-\mathrm{RHS}$ must simplify to zero.

```python
ctx_symbolic = pf.PEPContext("base").set_as_current()
F_symbolic = pf.SmoothConvexFunction(is_basis=True, tags=["f_{base}"], L=Lpar)
G_symbolic = pf.ConvexFunction(is_basis=True, tags=["g_{base}"])
H_symbolic = F_symbolic + G_symbolic
star_symbolic = H_symbolic.set_stationary_point("x_{star,base}")
x_symbolic = pf.Vector(is_basis=True, tags=["x_{base}"])
xplus_symbolic = G_symbolic.prox(
    x_symbolic - F_symbolic.grad(x_symbolic) / Lpar, 1 / Lpar, tag="x_{plus,base}"
)
F_symbolic(xplus_symbolic)
LHS = (
    phi(sp.S.One, Npar, Lpar, xplus_symbolic, star_symbolic, F_symbolic, G_symbolic)
    - Lpar / (4 * Npar) * (x_symbolic - star_symbolic) ** 2
)
RHS = smooth_residual(F_symbolic, star_symbolic, x_symbolic, Lpar) / (2 * Npar)
RHS += (
    (2 * Npar + 1)
    / ((2 * Npar - 1) * (2 * Npar))
    * smooth_residual(F_symbolic, star_symbolic, xplus_symbolic, Lpar)
)
RHS += (
    sp.S.One
    / (2 * Npar - 1)
    * convex_residual(G_symbolic, star_symbolic, xplus_symbolic)
)
RHS += smooth_residual(F_symbolic, x_symbolic, xplus_symbolic, Lpar) / (2 * Npar)
RHS -= step_square(
    sp.S.One,
    Npar,
    Lpar,
    x_symbolic,
    xplus_symbolic,
    star_symbolic,
    F_symbolic,
    G_symbolic,
)
diff_base = LHS - RHS
symbolic_checks["base"] = assert_identity(
    "Base, N >= 2", LHS, RHS, ctx_symbolic, symbolic_parameters, display=True
)
```

### Boundary Identity Symbolic Verification

For $N\geq2$, the terminal equation is
$$V_N-V_{N-1}=\frac{1}{N+1}I^f_{x_\star,x_N}+\frac{2}{N+1}I^g_{x_\star,x_N}+\frac{N}{N+1}I^f_{x_{N-1},x_N}+\frac{N-1}{N+1}I^g_{x_{N-1},x_N}-S_N.$$
Equivalently, the left side is $h(x_N)-h(x_\star)-\Phi_{N-1}$. The residual $\mathrm{LHS}-\mathrm{RHS}$ must simplify to zero.

```python
ctx_symbolic = pf.PEPContext("boundary").set_as_current()
F_symbolic = pf.SmoothConvexFunction(is_basis=True, tags=["f_{boundary}"], L=Lpar)
G_symbolic = pf.ConvexFunction(is_basis=True, tags=["g_{boundary}"])
H_symbolic = F_symbolic + G_symbolic
star_symbolic = H_symbolic.set_stationary_point("x_{star,boundary}")
x_symbolic = pf.Vector(is_basis=True, tags=["x_{boundary}"])
xplus_symbolic = G_symbolic.prox(
    x_symbolic - F_symbolic.grad(x_symbolic) / Lpar, 1 / Lpar, tag="x_{plus,boundary}"
)
F_symbolic(xplus_symbolic)
LHS = (
    H_symbolic(xplus_symbolic)
    - H_symbolic(star_symbolic)
    - phi(Npar - 1, Npar, Lpar, x_symbolic, star_symbolic, F_symbolic, G_symbolic)
)
RHS = smooth_residual(F_symbolic, star_symbolic, xplus_symbolic, Lpar) / (
    Npar + 1
) + 2 / (Npar + 1) * convex_residual(G_symbolic, star_symbolic, xplus_symbolic)
RHS += Npar / (Npar + 1) * smooth_residual(F_symbolic, x_symbolic, xplus_symbolic, Lpar)
RHS += (Npar - 1) / (Npar + 1) * convex_residual(G_symbolic, x_symbolic, xplus_symbolic)
RHS -= step_square(
    Npar, Npar, Lpar, x_symbolic, xplus_symbolic, star_symbolic, F_symbolic, G_symbolic
)
diff_boundary = LHS - RHS
symbolic_checks["boundary"] = assert_identity(
    "Final step, N >= 2", LHS, RHS, ctx_symbolic, symbolic_parameters, display=True
)
```

### Exceptional one-step horizon

At $N=1$, verify the separate identity
$$h(x_1)-h(x_\star)-\frac{L}{4}\|d_0\|^2=\frac12 I^f_{x_\star,x_0}+\frac12 I^f_{x_\star,x_1}+I^g_{x_\star,x_1}+\frac12 I^f_{x_0,x_1}-S_1\big|_{N=1}.$$
The residual $\mathrm{LHS}-\mathrm{RHS}$ must vanish exactly; this also avoids extending the interior formula outside its domain.

```python
ctx_symbolic = pf.PEPContext("one").set_as_current()
F_symbolic = pf.SmoothConvexFunction(is_basis=True, tags=["f_{one}"], L=Lpar)
G_symbolic = pf.ConvexFunction(is_basis=True, tags=["g_{one}"])
H_symbolic = F_symbolic + G_symbolic
star_symbolic = H_symbolic.set_stationary_point("x_{star,one}")
x_symbolic = pf.Vector(is_basis=True, tags=["x_{one}"])
xplus_symbolic = G_symbolic.prox(
    x_symbolic - F_symbolic.grad(x_symbolic) / Lpar, 1 / Lpar, tag="x_{plus,one}"
)
F_symbolic(xplus_symbolic)
LHS = (
    H_symbolic(xplus_symbolic)
    - H_symbolic(star_symbolic)
    - Lpar / 4 * (x_symbolic - star_symbolic) ** 2
)
RHS = sp.Rational(1, 2) * smooth_residual(
    F_symbolic, star_symbolic, x_symbolic, Lpar
) + sp.Rational(1, 2) * smooth_residual(F_symbolic, star_symbolic, xplus_symbolic, Lpar)
RHS += convex_residual(G_symbolic, star_symbolic, xplus_symbolic) + sp.Rational(
    1, 2
) * smooth_residual(F_symbolic, x_symbolic, xplus_symbolic, Lpar)
RHS -= step_square(
    sp.S.One,
    sp.S.One,
    Lpar,
    x_symbolic,
    xplus_symbolic,
    star_symbolic,
    F_symbolic,
    G_symbolic,
)
diff_one = LHS - RHS
symbolic_checks["N1"] = assert_identity(
    "Exceptional N=1", LHS, RHS, ctx_symbolic, symbolic_parameters, display=True
)
```

### Signs, domains, normalization, and full-PEP identity

For $1\leq i\leq N$, every denominator in the residual multipliers and SOS coefficients is positive because $L>0$, $2N-i\geq N\geq1$, and $2N+1-i\geq N+1$. The only possibly zero numerator is $i-1$, which is nonnegative. Thus all residual multipliers are nonnegative and $S_i\geq0$. Smooth-convex and convex interpolation make their residuals nonpositive, so the exact identities imply $V_i\leq V_{i-1}$. No positivity of $V_i$ is assumed or required.
The exact telescoping identity for every $N\geq1$ is
$$h(x_N)-h(x_\star)-\frac{L}{4N}\|d_0\|^2=p_0I^f_{x_\star,x_0}+\sum_{i=1}^{N}\left(p_iI^f_{x_\star,x_i}+q_iI^g_{x_\star,x_i}+\alpha_iI^f_{x_{i-1},x_i}-S_i\right)+\sum_{i=2}^{N}\beta_iI^g_{x_{i-1},x_i}.$$
There is exactly one copy of each supplied $S_i$, and these weights coincide with the B2 candidate formulas. Combining the initial residual $I_0=\|d_0\|^2-R^2\leq0$ with multiplier $L/(4N)$ therefore yields the global full-PEP certificate. The exact normalization identity below retains the radius term explicitly.

```python
Rpar = pf.Parameter("R")
R_symbolic = sp.Symbol("R", nonnegative=True)
x_initial = pf.Vector(is_basis=True, tags=["x_{initial,normalization}"])
metric_symbolic = H_symbolic(xplus_symbolic) - H_symbolic(star_symbolic)
terminal_potential = (
    metric_symbolic - Lpar / (4 * Npar) * (x_initial - star_symbolic) ** 2
)
initial_residual = (x_initial - star_symbolic) ** 2 - Rpar**2
LHS = metric_symbolic - Lpar * Rpar**2 / (4 * Npar)
RHS = terminal_potential + Lpar / (4 * Npar) * initial_residual
normalization_parameters = dict(symbolic_parameters)
normalization_parameters["R"] = R_symbolic
symbolic_checks["normalization"] = assert_identity(
    "Radius normalization",
    LHS,
    RHS,
    ctx_symbolic,
    normalization_parameters,
    display=True,
)
assert all(check["passed"] for check in symbolic_checks.values())
print("Exact base, step, terminal, N=1 and radius identities all pass.")
print(
    "Sign argument: nonpositive interpolation residuals, nonnegative multipliers and subtracted SOS imply V_i <= V_(i-1)."
)
print("Therefore h(x_N)-h(x_star) <= L ||x_0-x_star||^2/(4N) <= L R^2/(4N).")
print(
    "Summing the local identities gives the complete global candidate certificate for arbitrary N."
)
```

### Sharpness

For $R>0$, take the one-dimensional instance $f(x)=0$ and $g(x)=c|x|$, with $c=LR/(2N)$ and $x_0=R$. The proximal step subtracts $c/L=R/(2N)$ while the iterate is positive, so $x_k=R(1-k/(2N))$ for $0\leq k\leq N$. Therefore $x_N=R/2$, $x_\star=0$, and $h(x_N)-h(x_\star)=LR^2/(4N)$. This admissible instance attains the bound. At $R=0$, the initial point is a minimizer and the bound is zero.

```python
sharp_c = L_symbolic * R_symbolic / (2 * N_symbolic)
sharp_xN = R_symbolic / 2
assert (
    sp.cancel(sharp_c * sharp_xN - L_symbolic * R_symbolic**2 / (4 * N_symbolic)) == 0
)
print("Sharp example objective matches the theorem constant exactly.")
```
