# Proximal gradient: composite convex objective gap
Let $f$ be convex and $L$-smooth, $L>0$, and let $g$ be closed, proper, and convex in a real finite-dimensional Euclidean space. Write $h=f+g$. Assume $x_\star$ minimizes $h$, $R\ge0$, $N\ge1$ is an integer, and $\|x_0-x_\star\|^2\le R^2$. Proximal steps and evaluations are well-defined.
$$x_{k+1}=\operatorname{prox}_{g/L}\left(x_k-\frac{1}{L}\nabla f(x_k)\right),\qquad k=0,\ldots,N-1.$$
The performance metric is $h(x_N)-h(x_\star)$.
The input supplies a candidate SOS decomposition, to be verified in subsequent certificate work; it is not imposed in this numerical model.

## Proof Statement

### Theorem
Let $h=f+g$, with $f$ convex and differentiable with $L$-Lipschitz gradient, $L>0$, and $g$ closed, proper and convex. Suppose $x_{\star}$ minimizes $h$ and $\|x_{0}-x_{\star}\|^{2}\le R^{2}$, $R\ge0$. For any integer $N\ge1$, proximal gradient with step $1/L$ satisfies
$$h(x_{N})-h(x_{\star})\le\frac{L\|x_{0}-x_{\star}\|^{2}}{4N}\le\frac{LR^{2}}{4N}.$$
Here is a nonincreasing Lyapunov certificate. Write $a_{k}=\nabla f(x_{k})$, $a_{\star}=\nabla f(x_{\star})$, $d_{k}=x_{k}-x_{\star}$ and $D_{k}=2N-k$. Define $V_{0}=0$ and
$$V_{k}=\frac{k+1}{D_{k}}[f(x_{k})-f(x_{\star})]+\frac{k}{D_{k}}[g(x_{k})-g(x_{\star})]-\frac{\langle a_{k},d_{k}\rangle}{D_{k}}+\frac{\|a_{k}-a_{\star}\|^{2}}{2LD_{k}}+\frac{L(N-k)}{D_{k}^{2}}\|d_{k}\|^{2}-\frac{L}{4N}\|d_{0}\|^{2},\qquad 1\le k<N.$$
The terminal case is defined separately:
$$V_{N}=h(x_{N})-h(x_{\star})-\frac{L}{4N}\|d_{0}\|^{2}.$$
Then $V_{N}\le V_{N-1}\le\cdots\le V_{0}=0$. The statement also covers $N=1$ and $R=0$; it requires no finite value of $g(x_{0})$.


### Proof outline
Use $a_{k}=\nabla f(x_{k})$, $s_{i}=L(x_{i-1}-x_{i})-a_{i-1}\in\partial g(x_{i})$, $s_{\star}=-a_{\star}$, and $d_{k}=x_{k}-x_{\star}$. The interpolation residuals are
$$I_{f}(u,v)=f(v)-f(u)+\langle\nabla f(v),u-v\rangle+\frac{\|\nabla f(u)-\nabla f(v)\|^{2}}{2L}\le0,$$
$$I_{g}(u,v)=g(v)-g(u)+\langle s_{v},u-v\rangle\le0.$$
Define $D_{k}=2N-k$, $V_{0}=0$, $V_{N}=h(x_{N})-h(x_{\star})-L\|d_{0}\|^{2}/(4N)$, and, for $1\le k<N$,
$$V_{k}=\frac{(k+1)[f(x_{k})-f(x_{\star})]+k[g(x_{k})-g(x_{\star})]-\langle a_{k},d_{k}\rangle}{D_{k}}+\frac{\|a_{k}-a_{\star}\|^{2}}{2LD_{k}}+\frac{L(N-k)\|d_{k}\|^{2}}{D_{k}^{2}}-\frac{L\|d_{0}\|^{2}}{4N}.$$
For $1\le i\le N$, set $Q_{i}=\sum_{r=1}^{3}c_{i,r}\|u_{i,r}\|^{2}$, where
$$c_{i,1}=\frac{N}{(2N+1-i)(2N-i)L},\qquad u_{i,1}=\frac{i}{2N}a_{i}+\frac{2N-i}{2N}a_{i-1}-a_{\star},$$
$$c_{i,2}=\frac{1}{2L}\left[\frac{i-1}{2N-i}\frac{2N+2}{2N+1}+\frac{2N}{(2N-i)^{2}(2N+1)}\right],\qquad u_{i,2}=a_{i-1}+s_{i}-\frac{L}{2N+1-i}d_{i-1},$$
$$c_{i,3}=\frac{i}{2N+1-i}\frac{2N}{2N+1}\frac{1}{2L},\qquad u_{i,3}=\frac{2N+1}{2N}(a_{i}+s_{i})-\frac{a_{i-1}+s_{i}}{2N}-\frac{L}{2N-i}d_{i}.$$
Let
$$\alpha_{i}=\begin{cases}\dfrac{2N+1}{(2N-i)(2N+1-i)}&i<N,\\ \dfrac{1}{N+1}&i=N,\end{cases}\quad
\beta_{i}=\begin{cases}\dfrac{2N}{(2N-i)(2N+1-i)}&i<N,\\ \dfrac{2}{N+1}&i=N.\end{cases}$$
The exact identity for every step is
$$V_{i}-V_{i-1}=\mathbf{1}_{\{i=1\}}\frac{I_{f}(x_{\star},x_{0})}{2N}+\alpha_{i}I_{f}(x_{\star},x_{i})+\frac{i}{2N+1-i}I_{f}(x_{i-1},x_{i})+\beta_{i}I_{g}(x_{\star},x_{i})+\frac{i-1}{2N+1-i}I_{g}(x_{i-1},x_{i})-Q_{i}.$$
The last $I_{g}$ term is **omitted** for $i=1$, avoiding an evaluation of $g(x_{0})$. For $N=1$, use the terminal multipliers and the initial residual together. The notebook verifies the interior, base, terminal, and $N=1$ identities separately, in all scalar components with symbolic parameters.

All multipliers and square coefficients are nonnegative: $2N-i\ge N>0$, $2N+1-i\ge N+1>0$, $L>0$, and $i-1\ge0$. Thus each displayed right-hand side is nonpositive. Telescoping yields $V_{N}\le V_{0}=0$, so $h(x_{N})-h(x_{\star})\le L\|d_{0}\|^{2}/(4N)\le LR^{2}/(4N)$. The use of a supplied SOS candidate is justified by these exact identities, not by numerical agreement.

```python
L = pf.Parameter("L")
R = pf.Parameter("R")
f = pf.SmoothConvexFunction(is_basis=True, tags=["f"], L=L)
g = pf.ConvexFunction(is_basis=True, tags=["g"])
h = (f + g).add_tag("h")
```

```python
def make_ctx_pgm(ctx_name: str, N, **kwargs) -> pf.PEPContext:
    ctx = pf.PEPContext(ctx_name).set_as_current()
    x = pf.Vector(is_basis=True, tags=["x_0"])
    h.set_stationary_point("x_star")
    for i in range(int(N)):
        x = g.prox(x - f.grad(x) / L, 1 / L, tag=f"x_{i + 1}")
    f.grad(x)
    return ctx


def get_pep_setup(N, params):
    ctx = make_ctx_pgm(f"ctx_{N}", N)
    pb = pf.PEPBuilder(ctx)
    pb.add_initial_constraint(
        ((ctx["x_0"] - ctx["x_star"]) ** 2).le(R * R, name="initial_condition")
    )
    pb.set_performance_metric(h(ctx[f"x_{N}"]) - h(ctx["x_star"]))
    return ctx, pb, h

ctx, pb, objective = get_pep_setup(1, {"L": 1, "R": 1})
print("Setup built:", len(ctx.basis_vectors()), "basis vectors")
```

## Lambda and S verification
Define all interpolation residuals with sign $R\le0$. The numerical certificate is
$$h(x_N)-h(x_\star)-\mathrm{opt}-\sum_q\lambda_qR_q+S=0.$$
The sum includes the radius constraint. Inequality multipliers and the Gram slack must be nonnegative and positive semidefinite, respectively.

The sparse fractions suggest $f$ successor multipliers $(i+1)/(2N-i)$ and $g$ successor multipliers $i/(2N-i)$. Optimum-to-iterate multipliers are their consecutive differences, with endpoint adjustments. The following executable definitions include all endpoints. They are candidates for arbitrary $N$, not yet an all-horizon proof. The supplied SOS aid has normalization factor 1 in this objective-gap certificate. Its positive coefficients use exact rational expressions; the notebook verifies reconstruction against the same selected numerical slack.

```python
def aid_pieces(ctx, f, g, L, N):
    n = sp.S(N)
    pieces = []
    for j in range(1, int(N) + 1):
        i = sp.S(j)
        xp, x, xs = ctx[f"x_{j-1}"], ctx[f"x_{j}"], ctx["x_star"]
        ap, a, astar, s = f.grad(xp), f.grad(x), f.grad(xs), g.grad(x)
        c1 = n / ((2*n+1-i)*(2*n-i)) / L
        u1 = i/(2*n)*a + (2*n-i)/(2*n)*ap - astar
        c2 = ((i-1)/(2*n-i)*(2*n+2)/(2*n+1) + 2*n/((2*n-i)**2*(2*n+1))) / (2*L)
        u2 = ap+s-L/(2*n+1-i)*(xp-xs)
        c3 = i/(2*n+1-i)*(2*n)/(2*n+1)/(2*L)
        u3 = (2*n+1)/(2*n)*(a+s)-(ap+s)/(2*n)-L/(2*n-i)*(x-xs)
        pieces.extend([(j, 'u1', c1, u1), (j, 'u2', c2, u2), (j, 'u3', c3, u3)])
    return pieces


def multiplier_candidate(name, N):
    n = sp.S(N)
    if name == 'initial_condition':
        return sp.Rational(1, 4)/n
    owner, pair = name.split(':')
    row, col = pair.split(',')
    if col == 'x_star':
        return sp.S.Zero
    j = sp.S(int(col[2:]))
    if row == 'x_star':
        if owner == 'f':
            if j == 0:
                return 1/(2*n)
            if j == n:
                return 1/(n+1)
            return (2*n+1)/((2*n-j)*(2*n+1-j))
        if j == n:
            return 2/(n+1)
        return 2*n/((2*n-j)*(2*n+1-j))
    i = sp.S(int(row[2:]))
    if j != i+1:
        return sp.S.Zero
    return (i+1)/(2*n-i) if owner == 'f' else i/(2*n-i)


def square_sum(pieces):
    return sum((c*u**2 for _, _, c, u in pieces), pf.Scalar.zero())
```

```python
from pepflow.parameter import eval_parameter

ctx_b2, pb_b2, obj_b2 = get_pep_setup(N, {"L": 1, "R": 1})
entries = [(c, "initial", None, None) for c in pb_b2.init_conditions]
for owner in ctx_b2.func_to_triplets:
    owner.get_interpolation_constraints(ctx_b2)
    groups = ctx_b2.get_constraint_data(owner)
    entries.extend((c, "interpolation", str(owner), str(group))
                   for group, constraints in groups.sc_dict.items() for c in constraints)
    assert not groups.psd_dict, "Unexpected PSD interpolation extras"
assert len({c.name for c, *_ in entries}) == len(entries)
assert set(b2["named_dual_values"]) == {c.name for c, *_ in entries}
assert [str(v) for v in ctx_b2.basis_vectors()] == b2["basis_vectors"]
assert [str(v) for v in ctx_b2.basis_scalars()] == b2["basis_scalars"]
for filename, expected in b2["certificate_provenance"]["source_hashes"].items():
    assert hashlib.sha256(Path(filename).read_bytes()).hexdigest() == expected
certificate_keys = ("N_verify", "params_json", "relaxed_constraints", "S_matrix", "S_row_names",
    "S_col_names", "basis_vectors", "basis_scalars", "opt_value", "tau_sol", "tau_name",
    "named_dual_values", "constraint_metadata", "normalization", "certificate_provenance",
    "numerical_checks", "numerical_decomposition")
assert hashlib.sha256(json.dumps({k: b2[k] for k in certificate_keys}, sort_keys=True,
    separators=(",", ":"), allow_nan=False).encode()).hexdigest() == b2["certificate_id"]
expected_dropped = []
for c, *_ in entries:
    if c.name == "initial_condition":
        continue
    ri, ci = c.name.split(":")[1].split(",")
    keep = ri == "x_star" or (ci != "x_star" and int(ci[2:]) == int(ri[2:]) + 1)
    if not keep:
        expected_dropped.append(c.name)
assert set(expected_dropped) == set(b2["relaxed_constraints"])
pb_b2.set_relaxed_constraints(expected_dropped)
pm = pf.ExpressionManager(ctx_b2, resolve_parameters={"L": sp.S.One, "R": sp.S.One})
pieces = aid_pieces(ctx_b2, f, g, L, N)
S_candidate = square_sum(pieces)
E = np.array([pm.eval_vector(u).coords for _, _, _, u in pieces], dtype=float)
D = np.diag(np.asarray([eval_parameter(c, {"L": sp.S.One}) for _, _, c, _ in pieces], dtype=float))
S_saved = np.array(b2["S_matrix"])
assert np.allclose(E.T @ D @ E, S_saved, atol=1e-4, rtol=1e-5)
assert np.min(np.diag(D)) >= 0
assert np.allclose(S_saved, S_saved.T, atol=1e-6, rtol=0)
assert np.min(np.linalg.eigvalsh(S_saved)) >= -1e-6
assert all(value is None or value >= -1e-6 for value in b2["named_dual_values"].values())
weighted = pf.Scalar.zero()
weighted_exact = pf.Scalar.zero()
for c, origin, owner, group in entries:
    meta = b2["constraint_metadata"][c.name]
    assert (meta["origin"], meta["owner"], meta["group"], meta["comparator"]) == (
        origin, owner, group, c.cmp.value)
    residual = c.lhs - c.rhs
    if c.cmp == pf.Comparator.GE:
        residual = -residual
    value = b2["named_dual_values"][c.name]
    if value is not None:
        weighted += value * residual
    else:
        assert c.name in expected_dropped
    weighted_exact += multiplier_candidate(c.name, N) * residual
residual_numeric = pm.eval_scalar(pb_b2.performance_metric - b2["opt_value"] - weighted + S_candidate)
def maxabs(values):
    return float(np.max(np.abs(np.asarray(values, dtype=float)), initial=0))

print("Full numerical identity residuals:", maxabs(residual_numeric.inner_prod_coords),
      maxabs(residual_numeric.func_coords), maxabs(residual_numeric.offset))
assert all(maxabs(v) <= 1e-5 for v in (residual_numeric.inner_prod_coords,
    residual_numeric.func_coords, residual_numeric.offset))
print("SOS reconstruction error:", maxabs(E.T @ D @ E - S_saved))
print("Minimum S eigenvalue:", np.min(np.linalg.eigvalsh(S_saved)))
pf.pprint_labeled_matrix(E.T @ D @ E, b2["basis_vectors"], b2["basis_vectors"], precision=4)
pf.pprint_labeled_matrix(np.asarray(residual_numeric.inner_prod_coords, dtype=float),
                        b2["basis_vectors"], b2["basis_vectors"], precision=8)
# Exact-mode fixed-N check; this is separate from the numerical certificate.
exact_manager = pf.ExpressionManager(ctx_b2, resolve_parameters={"L": sp.S.One, "R": sp.S.One})
exact = exact_manager.eval_scalar(pb_b2.performance_metric - sp.Rational(1, 4*N)
    - weighted_exact + S_candidate, sympy_mode=True)
assert all(sp.simplify(v) == 0 for v in np.asarray(exact.inner_prod_coords).flat)
assert all(sp.simplify(v) == 0 for v in np.asarray(exact.func_coords).flat)
assert sp.simplify(exact.offset) == 0
print("Exact candidate fixed-N identity: all coordinates are zero")
for owner in ("f", "g"):
    tags = ["x_star"] + [f"x_{j}" for j in range(0 if owner == "f" else 1, N+1)]
    candidate_matrix = np.array([[sp.S.Zero if ri == ci else multiplier_candidate(
        f"{owner}:{ri},{ci}", N) for ci in tags] for ri in tags], dtype=object)
    pf.pprint_labeled_matrix(candidate_matrix, tags, tags)
candidate_error = max(abs(value - float(multiplier_candidate(name, N)))
    for name, value in b2["named_dual_values"].items() if value is not None)
assert candidate_error < 1e-4
print("Candidate multiplier error:", candidate_error)
print("Numerical certificate verified; arbitrary-horizon analytic proof remains pending.")
```

### Initial closed-form search status
Sparse consecutive multipliers were rationalized and their differences identified. Both function classes have candidate formulas, including endpoints. The supplied SOS reconstructs the same selected slack within the fixed tolerance; the candidate multipliers and SOS satisfy an exact fixed-$N$ identity above. This does not establish the identity for arbitrary horizons. Global analytic verification and the Lyapunov theorem remain pending.

## Lyapunov partial sums and ranks
For interpolation residuals $I_q\le0$, set $V_0=0$ and
$$V_k=\sum_{q\text{ assigned through }k}\lambda_q I_q-\sum_{i=1}^{k}Q_i.$$
The initial optimum-to-$x_0$ smooth residual enters the first increment. Each step adds both functions' optimum-to-current and predecessor-to-current residuals and subtracts a numerically checked PSD block $Q_i$. The original supplied squares remain the analytic candidate. To repair accumulated numerical reconstruction error, the saved $S$ is reallocated into blocks using the observed support $[a_\star,x_0-x_\star,x_k-x_\star,a_k-a_\star]$. The allocation minimizes distance from the original partial-sum coefficient matrices while constraining step blocks toward PSD; all blocks are then independently checked at the frozen PSD tolerance $10^{-6}$. The auxiliary allocation solves report optimal_inaccurate, which is retained; their labels do not replace the checks. Nonnegative multipliers and PSD step blocks give the intended nonincreasing direction. Block PSD is checked numerically here; exact signs remain a B5 obligation. The numerical construction retains every selected multiplier and all Gram, function, and constant coordinates.

The same construction has rank 4 at every retained index for $N=4$ and $N=6$, excluding only $V_0,V_N$. Rank uses absolute tolerance $10^{-4}$ on unscaled Gram coordinates at $L=R=1$. No interpolation term or part of the saved slack is left out. The terminal rank is 1; the radius constraint is reserved for the boundary argument. These observations do not establish an arbitrary-horizon theorem.

```python
certificate = json.loads((notebook_dir/"state/pgm_b3.json").read_text())
basis = ctx_b2.basis_vectors()
S_scalar = sum((float(S_saved[i,j])*basis[i]*basis[j]
                for i,j in np.ndindex(S_saved.shape)),pf.Scalar.zero())
saved_identity = pm.eval_scalar(pb_b2.performance_metric-b2["opt_value"]-weighted+S_scalar)
assert all(maxabs(v)<=1e-5 for v in (saved_identity.inner_prod_coords,saved_identity.func_coords,saved_identity.offset))
# Reallocated per-step PSD blocks reproduce the same saved S.
# These are finite-sample numerical blocks, distinct from the supplied analytic SOS candidate.
allocation = certificate['reallocated_S_pieces']
assert allocation['certificate_id'] == certificate['certificate_id']
assert allocation['N'] == N
S_piece_terms = {}
for j, saved_matrix in enumerate(allocation['Q'],1):
    matrix = np.asarray(saved_matrix,dtype=float)
    assert matrix.shape == S_saved.shape
    assert np.all(np.isfinite(matrix))
    assert np.allclose(matrix,matrix.T,atol=1e-6,rtol=0)
    assert np.linalg.eigvalsh((matrix+matrix.T)/2).min() >= -1e-6
    S_piece_terms[f'{j}:block'] = sum((float(matrix[i,col_idx])*basis[i]*basis[col_idx]
        for i,col_idx in np.ndindex(matrix.shape)),pf.Scalar.zero())
S_assignment={name:int(name.split(':')[0])-1 for name in S_piece_terms}
piece_diff = pm.eval_scalar(sum(S_piece_terms.values(),pf.Scalar.zero())-S_scalar)
assert all(maxabs(v)<=1e-5 for v in (piece_diff.inner_prod_coords,piece_diff.func_coords,piece_diff.offset))
for piece in S_piece_terms.values():
    ev=pm.eval_scalar(piece)
    assert maxabs(ev.func_coords)<=1e-5 and maxabs(ev.offset)<=1e-5

# Normalized interpolation residuals are <=0; subtracting each PSD square gives a nonincreasing sequence.
group_terms = {}
term_assignment = {}
for c, origin, _, _ in entries:
    if c.name == 'initial_condition' or b2['named_dual_values'][c.name] is None:
        continue
    assert origin == 'interpolation'
    residual = c.lhs-c.rhs
    if c.cmp == pf.Comparator.GE:
        residual = -residual
    group_terms[c.name] = b2['named_dual_values'][c.name]*residual
    row,col = c.name.split(':')[1].split(',')
    j = int(col[2:])
    term_assignment[c.name] = -1 if j==0 else j-1
assert set(term_assignment) == set(group_terms)
assert set(S_assignment) == set(S_piece_terms)
lyap = [pf.Scalar.zero()]
partial_sum = sum((group_terms[name] for name,target in term_assignment.items() if target == -1),pf.Scalar.zero())
for step in range(N):
    partial_sum += sum((group_terms[name] for name,target in term_assignment.items() if target == step),pf.Scalar.zero())
    partial_sum -= sum((S_piece_terms[name] for name,target in S_assignment.items() if target == step),pf.Scalar.zero())
    lyap.append(partial_sum)
rank_tolerance = 1e-4
rank_left,rank_right = 1,1
ranks=[]
singular_values=[]
for k,Vk in enumerate(lyap):
    matrix=np.asarray(pm.eval_scalar(Vk).inner_prod_coords,dtype=float)
    sv=np.linalg.svd(matrix,compute_uv=False)
    singular_values.append(sv.tolist())
    rank=int(np.linalg.matrix_rank(matrix,tol=rank_tolerance))
    ranks.append(rank)
    print(f'rank V_{k}: {rank}')
    if k==0:
        print()
retained=ranks[rank_left:N+1-rank_right]
print('Interior rank is constant at this N:',len(retained)>=3 and len(set(retained))==1,'boundary exclusions:',(rank_left,rank_right))
coverage = lyap[N] - (sum(group_terms.values(),pf.Scalar.zero())-S_scalar)
coverage_eval=pm.eval_scalar(coverage)
coverage_residual=max(maxabs(v) for v in (coverage_eval.inner_prod_coords,coverage_eval.func_coords,coverage_eval.offset))
assert coverage_residual <= 1e-5
increment_residuals=[]
for k in range(N):
    assigned={-1,0} if k==0 else {k}
    increment=sum((group_terms[name] for name,step in term_assignment.items() if step in assigned),pf.Scalar.zero())
    increment-=sum((S_piece_terms[name] for name,step in S_assignment.items() if step in assigned),pf.Scalar.zero())
    ev=pm.eval_scalar(lyap[k+1]-lyap[k]-increment)
    error=max(maxabs(v) for v in (ev.inner_prod_coords,ev.func_coords,ev.offset))
    increment_residuals.append(error)
    assert error <= 1e-5
grouping_checked=True
print('Full-coordinate coverage residual:',coverage_residual)

rank_evidence=json.loads((notebook_dir/"supplementary/rank_search.json").read_text())
for sample in rank_evidence["samples"]:
    retained_sample=sample["rank_profile"][1:sample["N"]]
    assert len(retained_sample)>=3 and set(retained_sample)=={4}
    print("Confirming horizon",sample["N"],"retained indices",list(range(1,sample["N"])),
          "common rank",4,"coverage residual",sample["coverage_residual"])
assert rank_evidence["summary"]["selected_certificate_id"]==b2["certificate_id"]
assert ranks==json.loads((notebook_dir/"state/pgm_b3.json").read_text())["rank_profile"]
```

For $1\le k<N$, adopt the ordered basis
$$[a_\star,\ x_0-x_\star,\ x_k-x_\star,\ a_k-a_\star].$$
This spans the same space as the search-selected bases and has a simpler uniform template. The initial basis is empty; the terminal basis is $[x_0-x_\star]$. The next cell verifies independence and the adopted decomposition at all indices.

```python
def V_k_basis(k):
    if k == 0:
        return []
    d0=ctx_b2['x_0']-ctx_b2['x_star']
    if k == N:
        return [d0]
    astar=f.grad(ctx_b2['x_star'])
    return [astar,d0,ctx_b2[f'x_{k}']-ctx_b2['x_star'],f.grad(ctx_b2[f'x_{k}'])-astar]


def V_k_basis_labels(k):
    if k == 0:
        return []
    if k == N:
        return ['x_0-x_star']
    return ['a_star','x_0-x_star',f'x_{k}-x_star',f'a_{k}-a_star']
```

### Coefficient matrices
For the chosen basis, the quadratic part is $\sum_{i,j}C_{ij}\langle v_i,v_j\rangle$, with off-diagonal coefficients counted twice. Function values are represented separately by $A_k(f(x_k)-f_\star)+B_k(g(x_k)-g_\star)$. Scalar offsets are checked to vanish. The displayed matrices and gap coefficients are numerical samples.

```python
from pepflow.lyapunov_utils import find_symmetric_coefficient_matrix

coeff_by_k={}
function_coefficients={}
coefficient_residuals={}
required_indices=list(range(N+1))
for k in required_indices:
    selected=V_k_basis(k)
    C=find_symmetric_coefficient_matrix(lyap[k],selected,pep_context=ctx_b2,
        resolve_parameters={'L':sp.S.One,'R':sp.S.One},indep_tol=1e-7,span_tol=1e-5)
    coeff_by_k[k]=C
    M=np.asarray(pm.eval_scalar(lyap[k]).inner_prod_coords,dtype=float)
    if selected:
        B=np.column_stack([pm.eval_vector(v).coords for v in selected]).astype(float)
        assert np.linalg.matrix_rank(B,tol=1e-7)==len(selected)==ranks[k]
        error=float(np.linalg.norm(M-B@C@B.T,ord='fro'))
    else:
        assert ranks[k]==0
        error=float(np.linalg.norm(M,ord='fro'))
    assert error<=1e-5*max(1,np.linalg.norm(M,ord='fro'))
    coefficient_residuals[k]=error
    print(f'k={k}: rank {ranks[k]} basis {V_k_basis_labels(k)}; residual {error:.3g}')
    if C.size:
        pf.pprint_labeled_matrix(C,V_k_basis_labels(k),V_k_basis_labels(k),precision=7)
    if k==0:
        function_coefficients[k]=[0.0,0.0]
    else:
        x=ctx_b2[f'x_{k}']
        xs=ctx_b2['x_star']
        F=np.column_stack([pm.eval_scalar(term).func_coords for term in (f(x)-f(xs),g(x)-g(xs))]).astype(float)
        target=np.asarray(pm.eval_scalar(lyap[k]).func_coords,dtype=float)
        coeff=np.linalg.lstsq(F,target,rcond=None)[0]
        assert np.max(np.abs(F@coeff-target),initial=0)<=1e-4
        function_coefficients[k]=coeff.tolist()
    assert abs(float(pm.eval_scalar(lyap[k]).offset))<=1e-5
    print('Function gap coefficients [f,g]:',function_coefficients[k],'; scalar offset: 0')


def V_k_structure(k):
    selected=V_k_basis(k)
    C=coeff_by_k[k]
    quadratic=sum((float(C[i,j])*selected[i]*selected[j]
        for i,j in np.ndindex(C.shape)),pf.Scalar.zero())
    if k==0:
        return quadratic
    x,xs=ctx_b2[f'x_{k}'],ctx_b2['x_star']
    A,B=function_coefficients[k]
    return quadratic+A*(f(x)-f(xs))+B*(g(x)-g(xs))


component_residuals={}
for k in required_indices:
    difference=pm.eval_scalar(V_k_structure(k)-lyap[k])
    target=pm.eval_scalar(lyap[k])
    component_residuals[str(k)]={}
    for label,actual,scale in (('gram',difference.inner_prod_coords,target.inner_prod_coords),
        ('function',difference.func_coords,target.func_coords),('offset',difference.offset,target.offset)):
        error=maxabs(actual)
        assert error<=1e-4+1e-5*maxabs(scale)
        component_residuals[str(k)][label]=error
print('All-index full-component reconstruction passed:',component_residuals)
```

## Closed-form coefficient conclusion
The numerical pattern suggests the following normalized solution; the generic coefficient equations below derive its current-step coefficients from the preceding potential and the candidate local residual weights. Base and terminal equations then verify its normalization without assuming endpoint extrapolation.

In the ordered basis $[a_{\star},d_{0},d_{k},a_{k}-a_{\star}]$, the only nonzero entries of the symmetric coefficient matrix are
$$C_{11}=-\frac{L}{4N},\quad C_{22}=\frac{L(N-k)}{(2N-k)^{2}},\quad C_{33}=\frac{1}{2L(2N-k)},\quad C_{02}=C_{20}=C_{23}=C_{32}=-\frac{1}{2(2N-k)}.$$
Indices here are zero based. The function coefficients are $(k+1)/(2N-k)$ and $k/(2N-k)$, with zero constant offset. Hence
$$V_{k}=\frac{k+1}{2N-k}(f_{k}-f_{\star})+\frac{k}{2N-k}(g_{k}-g_{\star})-\frac{\langle a_{k},d_{k}\rangle}{2N-k}+\frac{\|a_{k}-a_{\star}\|^{2}}{2L(2N-k)}+\frac{L(N-k)}{(2N-k)^{2}}\|d_{k}\|^{2}-\frac{L}{4N}\|d_{0}\|^{2},\quad1\le k<N.$$
Take $V_{0}=0$ and $V_{N}=h_{N}-h_{\star}-L\|d_{0}\|^{2}/(4N)$. The following sections derive and verify the one-step identity and both boundaries. Numerical matrix comparisons are diagnostics only; arbitrary-horizon validity follows from exact symbolic residuals.


```python
import sympy as sp
import numpy as np
import pepflow as pf

n_symbol, i_symbol = sp.symbols('N i', integer=True, positive=True)
ell_symbol = sp.symbols('L', positive=True)


def analytic_potential(point, fs, gs, xs, x0, index, horizon, ell):
    k,n = index,horizon
    d = 2*n-k
    gap_f=fs(point)-fs(xs)
    gap_g=gs(point)-gs(xs)
    a,astar=fs.grad(point),fs.grad(xs)
    dx=point-xs
    return ((k+1)/d*gap_f+k/d*gap_g-a*dx/d+(a-astar)**2/(2*d*ell)
            +ell*(n-k)/d**2*dx**2-ell/(4*n)*(x0-xs)**2)


def smooth_residual(fs, first, second, ell):
    return fs(second)-fs(first)+fs.grad(second)*(first-second)+(fs.grad(first)-fs.grad(second))**2/(2*ell)


def convex_residual(gs, first, second):
    return gs(second)-gs(first)+gs.grad(second)*(first-second)


def local_squares(fs, gs, xp, xn, xs, index, horizon, ell):
    i,n=index,horizon
    ap,a,astar,s=fs.grad(xp),fs.grad(xn),fs.grad(xs),gs.grad(xn)
    c1=n/((2*n+1-i)*(2*n-i))/ell
    u1=i/(2*n)*a+(2*n-i)/(2*n)*ap-astar
    c2=((i-1)/(2*n-i)*(2*n+2)/(2*n+1)+2*n/((2*n-i)**2*(2*n+1)))/(2*ell)
    u2=ap+s-ell/(2*n+1-i)*(xp-xs)
    c3=i/(2*n+1-i)*(2*n)/(2*n+1)/(2*ell)
    u3=(2*n+1)/(2*n)*(a+s)-(ap+s)/(2*n)-ell/(2*n-i)*(xn-xs)
    return c1*u1**2+c2*u2**2+c3*u3**2


def local_contribution(fs,gs,xp,xn,xs,index,horizon,ell,terminal=False,include_previous_g=True):
    i,n=index,horizon
    wf=(1/(n+1) if terminal else (2*n+1)/((2*n-i)*(2*n+1-i)))
    wg=(2/(n+1) if terminal else 2*n/((2*n-i)*(2*n+1-i)))
    total=wf*smooth_residual(fs,xs,xn,ell)+i/(2*n+1-i)*smooth_residual(fs,xp,xn,ell)
    total+=wg*convex_residual(gs,xs,xn)
    if include_previous_g:
        total+=(i-1)/(2*n+1-i)*convex_residual(gs,xp,xn)
    return total-local_squares(fs,gs,xp,xn,xs,i,n,ell)


def symbolic_components(expression,manager):
    ev=manager.eval_scalar(expression,sympy_mode=True)
    gram=sp.Matrix(ev.inner_prod_coords).applyfunc(sp.factor)
    funcs=sp.Matrix(list(ev.func_coords)).applyfunc(sp.factor)
    offset=sp.factor(ev.offset)
    return gram,funcs,offset


def verify_identity(label,expression,manager):
    gram,funcs,offset=symbolic_components(expression,manager)
    assert all(v==0 for v in [*gram,*funcs,offset]),(label,gram,funcs,offset)
    labels=[str(v) for v in manager.context.basis_vectors()]
    pf.pprint_labeled_matrix(np.array(gram,dtype=object),labels,labels)
    print(label,'Gram residual zero; function residual:',funcs.T,'offset:',offset)
    return True
```

```python
# Compare the derived analytic matrices with the B4 numerical extraction.
analytic_comparison={}
for k in range(1,N+1):
    if k==N:
        C_exact=sp.Matrix([[-sp.Rational(1,4*N)]])
        gaps_exact=sp.Matrix([1,1])
    else:
        denominator=sp.Integer(2*N-k)
        C_exact=sp.zeros(4)
        C_exact[1,1]=-sp.Rational(1,4*N)
        C_exact[2,2]=sp.Rational(N-k,1)/denominator**2
        C_exact[3,3]=1/(2*denominator)
        C_exact[0,2]=C_exact[2,0]=C_exact[2,3]=C_exact[3,2]=-1/(2*denominator)
        gaps_exact=sp.Matrix([(k+1)/denominator,k/denominator])
    matrix_error=float(np.max(np.abs(np.array(C_exact,dtype=float)-coeff_by_k[k])))
    gap_error=float(np.max(np.abs(np.array(gaps_exact,dtype=float).reshape(-1)-function_coefficients[k])))
    analytic_comparison[str(k)]={'matrix':matrix_error,'function':gap_error}
    assert max(matrix_error,gap_error)<=1e-4
    print('Derived matrix at k=',k,'numerical matrix/function errors=',matrix_error,gap_error)
    pf.pprint_labeled_matrix(np.array(C_exact,dtype=object),V_k_basis_labels(k),V_k_basis_labels(k))
```

### Generic coefficient equations and solution
The unknowns are the ten symmetric entries $C_{rs}(i,N,L)$ and the two function-gap coefficients. We equate every component of $V_{i}^{\mathrm{unknown}}-V_{i-1}-T_{i}$ to zero. The preceding candidate fixes the radius normalization $-L/(4N)$; the subsequent base and terminal equations verify that choice. The supplied square block and rational local weights suggested by the numerical certificate remain candidates until all identities and signs pass.

```python
ctx_step=pf.PEPContext('symbolic_step').set_as_current()
L_step=pf.Parameter('L_symbolic')
N_step=pf.Parameter('N_symbolic')
i_step=pf.Parameter('i_symbolic')
f_step=pf.SmoothConvexFunction(is_basis=True,tags=['f_{step}'],L=L_step)
g_step=pf.ConvexFunction(is_basis=True,tags=['g_{step}'])
h_step=f_step+g_step
xs_step=h_step.set_stationary_point('x_{star}')
x0_step=pf.Vector(is_basis=True,tags=['x_{0}'])
xp_step=pf.Vector(is_basis=True,tags=['x_{i-1}'])
xn_step=g_step.prox(xp_step-f_step.grad(xp_step)/L_step,1/L_step,tag='x_{i}')
V_previous=analytic_potential(xp_step,f_step,g_step,xs_step,x0_step,i_step-1,N_step,L_step)
V_current=analytic_potential(xn_step,f_step,g_step,xs_step,x0_step,i_step,N_step,L_step)
RHS_step=local_contribution(f_step,g_step,xp_step,xn_step,xs_step,i_step,N_step,L_step)

unknown_names=[f'c{r}{s}' for r in range(4) for s in range(r,4)]+['Af','Bg']
unknown_symbols=sp.symbols(' '.join(unknown_names),real=True)
unknown_parameters={name:pf.Parameter(name) for name in unknown_names}
resolve_unknown={'L_symbolic':ell_symbol,'N_symbolic':n_symbol,'i_symbolic':i_symbol,**dict(zip(unknown_names,unknown_symbols))}
vectors=[f_step.grad(xs_step),x0_step-xs_step,xn_step-xs_step,f_step.grad(xn_step)-f_step.grad(xs_step)]
V_unknown=pf.Scalar.zero()
for r in range(4):
    for s in range(r,4):
        V_unknown+=(1 if r==s else 2)*unknown_parameters[f'c{r}{s}']*vectors[r]*vectors[s]
V_unknown+=unknown_parameters['Af']*(f_step(xn_step)-f_step(xs_step))
V_unknown+=unknown_parameters['Bg']*(g_step(xn_step)-g_step(xs_step))
unknown_manager=pf.ExpressionManager(ctx_step,resolve_parameters=resolve_unknown)
unknown_gram,unknown_func,unknown_offset=symbolic_components(V_unknown-V_previous-RHS_step,unknown_manager)
coefficient_equations=[sp.Eq(value,0) for value in [*unknown_gram,*unknown_func,unknown_offset] if value!=0]
print('All nonzero coefficient equations:',coefficient_equations)
coefficient_solution=sp.solve(coefficient_equations,unknown_symbols,dict=True)
assert len(coefficient_solution)==1
print('Derived current coefficient functions:')
for variable,value in coefficient_solution[0].items():
    print(variable,sp.factor(value))
```

## Symbolic Step Recursion Verification

For $2\le i<N$, verify
$$V_{i}-V_{i-1}=\frac{2N+1}{(2N-i)(2N+1-i)}I_{f}(x_{\star},x_{i})+\frac{i}{2N+1-i}I_{f}(x_{i-1},x_{i})+\frac{2N}{(2N-i)(2N+1-i)}I_{g}(x_{\star},x_{i})+\frac{i-1}{2N+1-i}I_{g}(x_{i-1},x_{i})-Q_{i}.$$
The residual $\mathrm{LHS}-\mathrm{RHS}$ must simplify to zero in every scalar component. All three squares in $Q_{i}$ are defined in the proof outline and constructed above.

```python
ctx_step=pf.PEPContext('symbolic_step').set_as_current()
L_step=pf.Parameter('L_symbolic')
N_step=pf.Parameter('N_symbolic')
i_step=pf.Parameter('i_symbolic')
f_step=pf.SmoothConvexFunction(is_basis=True,tags=['f_{step}'],L=L_step)
g_step=pf.ConvexFunction(is_basis=True,tags=['g_{step}'])
h_step=f_step+g_step
xs_step=h_step.set_stationary_point('x_{star}')
x0_step=pf.Vector(is_basis=True,tags=['x_{0}'])
xp_step=pf.Vector(is_basis=True,tags=['x_{i-1}'])
xn_step=g_step.prox(xp_step-f_step.grad(xp_step)/L_step,1/L_step,tag='x_{i}')
V_previous=analytic_potential(xp_step,f_step,g_step,xs_step,x0_step,i_step-1,N_step,L_step)
V_current=analytic_potential(xn_step,f_step,g_step,xs_step,x0_step,i_step,N_step,L_step)
RHS_step=local_contribution(f_step,g_step,xp_step,xn_step,xs_step,i_step,N_step,L_step)
LHS_step=V_current-V_previous
diff_step=LHS_step-RHS_step
pm_step=pf.ExpressionManager(ctx_step,resolve_parameters={'L_symbolic':ell_symbol,'N_symbolic':n_symbol,'i_symbolic':i_symbol})
step_verified=verify_identity('Interior step',diff_step,pm_step)
```

## Base Case and Boundary Symbolic Verification

For $N\ge2$, verify
$$V_{1}-V_{0}=\frac{I_{f}(x_{\star},x_{0})}{2N}+\frac{2N+1}{(2N-1)2N}I_{f}(x_{\star},x_{1})+\frac{I_{f}(x_{0},x_{1})}{2N}+\frac{I_{g}(x_{\star},x_{1})}{2N-1}-Q_{1},\qquad V_{0}=0.$$
The residual $\mathrm{LHS}-\mathrm{RHS}$ must simplify to zero. No $g(x_{0})$ term is formed.

```python
ctx_base=pf.PEPContext('symbolic_base').set_as_current()
L_base=pf.Parameter('L_symbolic')
N_base=pf.Parameter('N_symbolic')
f_base=pf.SmoothConvexFunction(is_basis=True,tags=['f_{base}'],L=L_base)
g_base=pf.ConvexFunction(is_basis=True,tags=['g_{base}'])
xs_base=(f_base+g_base).set_stationary_point('x_{star}')
x0_base=pf.Vector(is_basis=True,tags=['x_{0}'])
x1_base=g_base.prox(x0_base-f_base.grad(x0_base)/L_base,1/L_base,tag='x_{1}')
LHS_base=analytic_potential(x1_base,f_base,g_base,xs_base,x0_base,sp.S.One,N_base,L_base)
RHS_base=smooth_residual(f_base,xs_base,x0_base,L_base)/(2*N_base)
RHS_base+=local_contribution(f_base,g_base,x0_base,x1_base,xs_base,sp.S.One,N_base,L_base,include_previous_g=False)
diff_base=LHS_base-RHS_base
pm_base=pf.ExpressionManager(ctx_base,resolve_parameters={'L_symbolic':ell_symbol,'N_symbolic':n_symbol})
base_verified=verify_identity('Base, N>=2',diff_base,pm_base)
```

### Boundary Identity Symbolic Verification

For $N\ge2$, verify
$$V_{N}-V_{N-1}=\frac{I_{f}(x_{\star},x_{N})}{N+1}+\frac{N}{N+1}I_{f}(x_{N-1},x_{N})+\frac{2}{N+1}I_{g}(x_{\star},x_{N})+\frac{N-1}{N+1}I_{g}(x_{N-1},x_{N})-Q_{N},$$
where $V_{N}=h(x_{N})-h(x_{\star})-L\|d_{0}\|^{2}/(4N)$. The residual $\mathrm{LHS}-\mathrm{RHS}$ must simplify to zero.

```python
ctx_boundary=pf.PEPContext('symbolic_boundary').set_as_current()
L_boundary=pf.Parameter('L_symbolic')
N_boundary=pf.Parameter('N_symbolic')
f_boundary=pf.SmoothConvexFunction(is_basis=True,tags=['f_{boundary}'],L=L_boundary)
g_boundary=pf.ConvexFunction(is_basis=True,tags=['g_{boundary}'])
xs_boundary=(f_boundary+g_boundary).set_stationary_point('x_{star}')
x0_boundary=pf.Vector(is_basis=True,tags=['x_{0}'])
xp_boundary=pf.Vector(is_basis=True,tags=['x_{N-1}'])
xn_boundary=g_boundary.prox(xp_boundary-f_boundary.grad(xp_boundary)/L_boundary,1/L_boundary,tag='x_{N}')
V_terminal=f_boundary(xn_boundary)-f_boundary(xs_boundary)+g_boundary(xn_boundary)-g_boundary(xs_boundary)-L_boundary/(4*N_boundary)*(x0_boundary-xs_boundary)**2
V_penultimate=analytic_potential(xp_boundary,f_boundary,g_boundary,xs_boundary,x0_boundary,N_boundary-1,N_boundary,L_boundary)
LHS_boundary=V_terminal-V_penultimate
RHS_boundary=local_contribution(f_boundary,g_boundary,xp_boundary,xn_boundary,xs_boundary,N_boundary,N_boundary,L_boundary,terminal=True)
diff_boundary=LHS_boundary-RHS_boundary
pm_boundary=pf.ExpressionManager(ctx_boundary,resolve_parameters={'L_symbolic':ell_symbol,'N_symbolic':n_symbol})
boundary_verified=verify_identity('Terminal, N>=2',diff_boundary,pm_boundary)
```

### Exceptional horizon symbolic verification

$$h(x_{1})-h(x_{\star})-\frac{L}{4}\|d_{0}\|^{2}=\frac12 I_{f}(x_{\star},x_{0})+\frac12 I_{f}(x_{\star},x_{1})+\frac12 I_{f}(x_{0},x_{1})+I_{g}(x_{\star},x_{1})-Q_{1}.$$
Here $Q_{1}$ uses $N=1$. The residual $\mathrm{LHS}-\mathrm{RHS}$ must simplify to zero.

```python
ctx_small=pf.PEPContext('symbolic_small').set_as_current()
L_small=pf.Parameter('L_symbolic')
f_small=pf.SmoothConvexFunction(is_basis=True,tags=['f_{small}'],L=L_small)
g_small=pf.ConvexFunction(is_basis=True,tags=['g_{small}'])
xs_small=(f_small+g_small).set_stationary_point('x_{star}')
x0_small=pf.Vector(is_basis=True,tags=['x_{0}'])
x1_small=g_small.prox(x0_small-f_small.grad(x0_small)/L_small,1/L_small,tag='x_{1}')
LHS_small=f_small(x1_small)-f_small(xs_small)+g_small(x1_small)-g_small(xs_small)-L_small/4*(x0_small-xs_small)**2
RHS_small=smooth_residual(f_small,xs_small,x0_small,L_small)/2
RHS_small+=local_contribution(f_small,g_small,x0_small,x1_small,xs_small,sp.S.One,sp.S.One,L_small,terminal=True,include_previous_g=False)
diff_small=LHS_small-RHS_small
pm_small=pf.ExpressionManager(ctx_small,resolve_parameters={'L_symbolic':ell_symbol})
small_verified=verify_identity('Exceptional horizon N=1',diff_small,pm_small)
```
