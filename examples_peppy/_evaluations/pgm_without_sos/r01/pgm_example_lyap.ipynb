# Proximal gradient: convergence and Lyapunov discovery

Let $f$ be convex and $L$-smooth with $L>0$, and let $g$ be closed, proper and convex in a real finite-dimensional Euclidean space. Let $h=f+g$, let $x_\star$ minimize $h$, and suppose $\|x_{0}-x_\star\|^2\le R^2$, $R\ge0$. Updates and evaluations are well-defined. For an integer $N\ge1$,
$$x_{k+1}=\operatorname{prox}_{g/L}\left(x_{k}-\frac{1}{L}\nabla f(x_{k})\right),\qquad 0\le k<N.$$
We study $h(x_{N})-h(x_\star)$. Composite stationarity imposes $\nabla f(x_\star)+s_\star=0$ with $s_\star\in\partial g(x_\star)$.

## Proof Statement

### Theorem

For the stated PGM recurrence with convex $L$-smooth $f$, closed proper convex $g$, $L>0$, $R\ge0$, integer $N\ge1$, a minimizer $x_{\star}$ and $\|x_{0}-x_{\star}\|^{2}\le R^{2}$, the following potential is nonincreasing. All stated evaluations and proximal steps are well-defined.

Put $d_{k}=x_{k}-x_{\star}$ and $p_{k}=(\nabla f(x_{k})-\nabla f(x_{\star}))/L$.
For $1\le k<N$, define
$$V_{k}=-\frac{L}{4N}\|d_{0}\|^{2}
+\frac{L(N-k)}{(2N-k)^{2}}\|d_{k}\|^{2}
-\frac{\langle d_{k},\nabla f(x_{k})\rangle}{2N-k}
-\frac{Lk}{2(2N-k)^{2}}\|p_{k}\|^{2}
+\frac{k+1}{2N-k}(f(x_{k})-f(x_{\star}))
+\frac{k}{2N-k}(g(x_{k})-g(x_{\star})).$$
The endpoints are $V_{0}=0$ and
$$V_{N}=h(x_{N})-h(x_{\star})-\frac{L}{4N}\|d_{0}\|^{2}.$$
Consequently, $$h(x_{N})-h(x_{\star})\le\frac{LR^{2}}{4N}.$$

### Proof outline

Put $d_{k}=x_{k}-x_{\star}$ and $p_{k}=(\nabla f(x_{k})-\nabla f(x_{\star}))/L$.
For $1\le k<N$, define
$$V_{k}=-\frac{L}{4N}\|d_{0}\|^{2}
+\frac{L(N-k)}{(2N-k)^{2}}\|d_{k}\|^{2}
-\frac{\langle d_{k},\nabla f(x_{k})\rangle}{2N-k}
-\frac{Lk}{2(2N-k)^{2}}\|p_{k}\|^{2}
+\frac{k+1}{2N-k}(f(x_{k})-f(x_{\star}))
+\frac{k}{2N-k}(g(x_{k})-g(x_{\star})).$$
The endpoints are $V_{0}=0$ and
$$V_{N}=h(x_{N})-h(x_{\star})-\frac{L}{4N}\|d_{0}\|^{2}.$$

Define the nonpositive interpolation residuals
$$I_{f}(u,v)=f(v)-f(u)+\langle\nabla f(v),u-v\rangle+\frac{\|\nabla f(u)-\nabla f(v)\|^{2}}{2L},\quad
I_{g}(u,v)=g(v)-g(u)+\langle s_{v},u-v\rangle,$$
where $s_{v}\in\partial g(v)$ is the selected oracle subgradient. At the minimizer choose $s_{\star}=-\nabla f(x_{\star})$; at a new iterate use the proximal subgradient.
Write $a_{k}=(k+1)/(2N-k)$ and $b_{k}=k/(2N-k)$.
For a step, put $m=2N-k$, $p=p_{k}$, $q=p_{k+1}$,
$s=(s_{x_{k+1}}+\nabla f(x_{\star}))/L$, $z=s-d_{k}/m$.
The recurrence gives $d_{k+1}=d_{k}-p-s$.
Define
$$A_{k}=\frac{k+1}{2(m-1)^{2}}+\frac{k+2}{2(m-1)},\quad
B_{k}=\frac{k+1}{2(m-1)},\quad C_{k}=\frac{km+N-k}{(m-1)^{2}},$$
$$D_{k}=\frac{k+1}{2m(m-1)},\quad E_{k}=\frac{km+1}{2(m-1)^{2}},\quad
F_{k}=\frac{a_{k}}2-\frac{N-k-1}{(m-1)^{2}}-\frac{k}{2m^{2}}.$$
Set $\alpha_{k}=A_{k}$, $\beta_{k}=C_{k}-B_{k}^{2}/A_{k}$,
$\eta_{k}=E_{k}-B_{k}D_{k}/A_{k}$,
$\gamma_{k}=F_{k}-D_{k}^{2}/A_{k}-\eta_{k}^{2}/\beta_{k}$ and
$$Q_{k}=L\alpha_{k}\|q+(B_{k}/A_{k})z+(D_{k}/A_{k})p\|^{2}
+L\beta_{k}\|z+(\eta_{k}/\beta_{k})p\|^{2}+L\gamma_{k}\|p\|^{2}.$$
The sign lemma below proves $\alpha_{k},\beta_{k}>0$ and $\gamma_{k}\ge0$ for $0\le k\le N-2$.
For $N\ge2$, the exact base identity is
$$V_{1}=a_{0}I_{f}(x_{\star},x_{0})+(a_{1}-a_{0})I_{f}(x_{\star},x_{1})
+b_{1}I_{g}(x_{\star},x_{1})+a_{0}I_{f}(x_{0},x_{1})
-\frac{L}{4N}\|p_{0}\|^{2}-Q_{0}.$$
For $1\le k\le N-2$, the exact recursion is
$$V_{k+1}-V_{k}=(a_{k+1}-a_{k})I_{f}(x_{\star},x_{k+1})
+(b_{k+1}-b_{k})I_{g}(x_{\star},x_{k+1})
+a_{k}I_{f}(x_{k},x_{k+1})+b_{k}I_{g}(x_{k},x_{k+1})-Q_{k}.$$
At the final step put $p=p_{N-1}$, $q=p_{N}$, $s=(s_{x_{N}}+\nabla f(x_{\star}))/L$, $d=d_{N-1}$ and
$$T_{N}=\frac L2\left\|q+s+\frac{p-d}{N+1}\right\|^{2}
+\frac L2\left\|s+\frac{Np-d}{N+1}\right\|^{2}.$$
Then, for $N\ge2$,
$$V_{N}-V_{N-1}=(1-a_{N-1})I_{f}(x_{\star},x_{N})+(1-b_{N-1})I_{g}(x_{\star},x_{N})
+a_{N-1}I_{f}(x_{N-1},x_{N})+b_{N-1}I_{g}(x_{N-1},x_{N})-T_{N}.$$
For $N=1$, use the separate identity
$$V_{1}=\tfrac12 I_{f}(x_{\star},x_{0})+\tfrac12 I_{f}(x_{\star},x_{1})+I_{g}(x_{\star},x_{1})
+\tfrac12 I_{f}(x_{0},x_{1})-\tfrac L4\|p_{0}\|^{2}-T_{1}.$$
All residual multipliers are nonnegative: $a_{k+1}-a_{k}=(2N+1)/[(2N-k-1)(2N-k)]$,
$b_{k+1}-b_{k}=2N/[(2N-k-1)(2N-k)]$,
$1-a_{N-1}=1/(N+1)$ and $1-b_{N-1}=2/(N+1)$.
Thus $V_{N}\le\cdots\le V_{1}\le V_{0}=0$, and
$$h(x_{N})-h(x_{\star})\le\frac{L}{4N}\|d_{0}\|^{2}\le\frac{LR^{2}}{4N}.$$

## Function and parameter definitions

```python
L = pf.Parameter("L")
R = pf.Parameter("R")
f = pf.SmoothConvexFunction(is_basis=True, tags=["f"], L=L)
g = pf.ConvexFunction(is_basis=True, tags=["g"])
h = f + g
```

## Algorithm and performance estimation setup

```python
def make_ctx_pgm(ctx_name: str, N, **kwargs) -> pf.PEPContext:
    ctx = pf.PEPContext(ctx_name).set_as_current()
    x = pf.Vector(is_basis=True, tags=["x_0"])
    h.set_stationary_point("x_star")
    for k in range(int(N)):
        x = g.prox(x - f.grad(x) / L, 1 / L, tag=f"x_{k + 1}")
    f(x)
    return ctx


def get_pep_setup(N, params):
    ctx = make_ctx_pgm(f"pgm_{N}", N)
    pb = pf.PEPBuilder(ctx)
    pb.add_initial_constraint(
        ((ctx["x_0"] - ctx["x_star"]) ** 2).le(R**2, name="initial_condition")
    )
    pb.set_performance_metric(h(ctx[f"x_{N}"]) - h(ctx["x_star"]))
    return ctx, pb, f
```

```python
import hashlib
import json
import uuid

import numpy as np
import pepflow as pf
from pepflow.lyapunov_utils import ldl_decompose_with_reversed_basis

from runtime import solve_setup  # ty: ignore[unresolved-import]  # Notebook bootstrap adds the helper directory to sys.path.


def inventory(ctx, pb):
    entries = [(c, "initial", None, None) for c in pb.init_conditions]
    for owner in ctx.func_to_triplets:
        groups = ctx.get_constraint_data(owner)
        entries.extend(
            (c, "interpolation", str(owner), str(group))
            for group, constraints in groups.sc_dict.items()
            for c in constraints
        )
        entries.extend(
            (c, "interpolation", str(owner), str(group))
            for group, c in groups.psd_dict.items()
        )
    assert len({c.name for c, *_ in entries}) == len(entries)
    return entries


def capture(ctx, pb, obj, params, dense_value=None):
    result, solver_record = solve_setup(ctx, pb, params)
    ctx.set_as_current()
    pm = pf.ExpressionManager(ctx, resolve_parameters=params)
    S = result.get_gram_dual_matrix()
    metadata, duals = {}, {}
    for c, origin, owner, group in inventory(ctx, pb):
        assert isinstance(c, pf.ScalarConstraint), "Unexpected matrix constraint"
        dropped = c.name in pb.relaxed_constraints
        value = result.get_dual_value(c.name)
        assert dropped or value is not None, c.name
        metadata[c.name] = dict(
            origin=origin,
            owner=owner,
            group=group,
            kind="scalar",
            comparator=c.cmp.value,
            shape=[],
            relaxed=dropped,
        )
        duals[c.name] = None if dropped else float(value)
    D, ell = ldl_decompose_with_reversed_basis(
        S, ctx.basis_vectors(), print_output=False
    )
    E = np.asarray([pm.eval_vector(v).coords for v in ell], dtype=float)
    state = dict(
        opt_value=float(result.opt_value),
        tau_sol=float(duals["initial_condition"]),
        tau_name="initial_condition",
        relaxed_constraints=list(pb.relaxed_constraints),
        S_matrix=np.asarray(S.matrix, dtype=float).tolist(),
        S_row_names=list(S.row_names),
        S_col_names=list(S.col_names),
        basis_vectors=[str(v) for v in ctx.basis_vectors()],
        basis_scalars=[str(v) for v in ctx.basis_scalars()],
        named_dual_values=duals,
        constraint_metadata=metadata,
        certificate_provenance=dict(
            solve_id=str(uuid.uuid4()),
            pep_type="primal",
            solver_status=str(result.solver_status),
            solver=solver_record,
        ),
        numerical_decomposition=dict(
            method="reverse_ldl",
            D=np.asarray(D, dtype=float).tolist(),
            ell_coords=E.tolist(),
            basis_labels=[str(v) for v in ctx.basis_vectors()],
        ),
    )
    displays = {}
    for owner in (obj, *[o for o in ctx.func_to_triplets if o is not obj]):
        value = result.get_scalar_constraint_dual_value_in_numpy(owner)
        groups = value if isinstance(value, dict) else {"default": value}
        for group, matrix in groups.items():
            displays[f"{owner}:{group}"] = dict(
                matrix=np.asarray(matrix.matrix, dtype=float).tolist(),
                row_names=list(matrix.row_names),
                col_names=list(matrix.col_names),
            )
    state["all_lambda_groups"] = displays
    state["lambda_groups"] = {
        key: value for key, value in displays.items() if key.startswith(str(obj) + ":")
    }
    state["dense_opt_value"] = (
        state["opt_value"] if dense_value is None else dense_value
    )
    state["numerical_checks"], _, _ = verify(ctx, pb, params, state)
    return state


def verify(ctx, pb, params, state):
    ctx.set_as_current()
    pm = pf.ExpressionManager(ctx, resolve_parameters=params)
    basis = ctx.basis_vectors()
    assert [str(v) for v in basis] == state["basis_vectors"]
    assert [str(v) for v in ctx.basis_scalars()] == state["basis_scalars"]
    entries = inventory(ctx, pb)
    assert {c.name for c, *_ in entries} == set(state["named_dual_values"])
    assert set(state["constraint_metadata"]) == set(state["named_dual_values"])
    weighted = pf.Scalar.zero()
    nonnegative = []
    for c, origin, owner, group in entries:
        meta = state["constraint_metadata"][c.name]
        assert meta == dict(
            origin=origin,
            owner=owner,
            group=group,
            kind="scalar",
            comparator=c.cmp.value,
            shape=[],
            relaxed=c.name in pb.relaxed_constraints,
        )
        value = state["named_dual_values"][c.name]
        if meta["relaxed"]:
            assert value is None
            continue
        assert value is not None and np.isfinite(value)
        residual = c.lhs - c.rhs
        if c.cmp == pf.Comparator.GE:
            residual = -residual
        weighted += value * residual
        if c.cmp != pf.Comparator.EQ:
            nonnegative.append(value)
    D = np.asarray(state["numerical_decomposition"]["D"])
    E = np.asarray(state["numerical_decomposition"]["ell_coords"])
    ell = [
        sum((float(a) * v for a, v in zip(row, basis)), pf.Vector.zero()) for row in E
    ]
    S_numeric = sum(
        (float(D[i, j]) * ell[i] * ell[j] for i, j in np.ndindex(D.shape)),
        pf.Scalar.zero(),
    )
    checked = pm.eval_scalar(
        pb.performance_metric - state["opt_value"] - weighted + S_numeric
    )
    S = np.asarray(state["S_matrix"])

    def maxabs(value):
        return float(np.max(np.abs(np.asarray(value, dtype=float)), initial=0))

    residuals = dict(
        identity_gram_max_abs=maxabs(checked.inner_prod_coords),
        identity_func_max_abs=maxabs(checked.func_coords),
        identity_offset_abs=maxabs(checked.offset),
        S_reconstruction_max_abs=maxabs(E.T @ D @ E - S),
        dense_relaxed_gap=abs(state["dense_opt_value"] - state["opt_value"]),
        minimum_scalar_inequality_dual=min(nonnegative, default=0),
        minimum_S_eigenvalue=float(np.linalg.eigvalsh((S + S.T) / 2).min()),
        minimum_matrix_inequality_eigenvalue=0.0,
    )
    tolerances = dict(
        identity_atol=1e-5,
        identity_rtol=0,
        reconstruction_atol=1e-4,
        reconstruction_rtol=1e-5,
        dual_tol=1e-6,
        psd_tol=1e-6,
        dense_opt_atol=1e-5,
    )
    passed = (
        state["certificate_provenance"]["solver_status"] == "optimal"
        and np.all(np.isfinite(S))
        and np.allclose(S, S.T, atol=1e-6, rtol=0)
        and all(
            residuals[key] <= 1e-5
            for key in (
                "identity_gram_max_abs",
                "identity_func_max_abs",
                "identity_offset_abs",
            )
        )
        and np.allclose(E.T @ D @ E, S, atol=1e-4, rtol=1e-5)
        and residuals["dense_relaxed_gap"] < 1e-5
        and residuals["minimum_scalar_inequality_dual"] >= -1e-6
        and residuals["minimum_S_eigenvalue"] >= -1e-6
    )
    return (
        dict(tolerances=tolerances, residuals=residuals, passed=bool(passed)),
        checked,
        S_numeric,
    )


def certificate_id(state):
    keys = (
        "N_verify",
        "params_json",
        "relaxed_constraints",
        "S_matrix",
        "S_row_names",
        "S_col_names",
        "basis_vectors",
        "basis_scalars",
        "opt_value",
        "tau_sol",
        "tau_name",
        "named_dual_values",
        "constraint_metadata",
        "normalization",
        "certificate_provenance",
        "numerical_checks",
        "numerical_decomposition",
    )
    return hashlib.sha256(
        json.dumps(
            {key: state[key] for key in keys},
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode()
    ).hexdigest()
```

### Initial closed-form search

The consecutive coefficients suggest $a_{k}=(k+1)/(2N-k)$ for the smooth function and $b_{k}=k/(2N-k)$ for the nonsmooth function. Optimum-to-iterate coefficients are their successive differences, with terminal masses chosen to sum to one. This hypothesis came from the fractions of this attempt's sparse solve. We compare every active multiplier below.

For $S$, reverse LDL gives a complete numerical decomposition. Its factors are not simple enough to infer a useful general square formula directly. We also construct the candidate remainder from the rational multiplier expressions and compare all its scalar coordinates with this same numerical $S$. PSD for arbitrary horizons and an interpretable square decomposition remain unresolved.

```python
def candidate_multiplier(name, N):
    if name == "initial_condition":
        return sp.Rational(1, 4 * N)
    owner, pair = name.split(":")
    first, second = pair.split(",")

    def cumulative(k):
        return sp.Rational(k + (1 if owner == "f" else 0), 2 * N - k)

    if first == "x_star":
        j = int(second[2:])
        if j == N:
            return 1 - cumulative(N - 1)
        if j == 0:
            return cumulative(0)
        return cumulative(j) - cumulative(j - 1)
    if second != "x_star" and int(second[2:]) == int(first[2:]) + 1:
        return cumulative(int(first[2:]))
    return sp.S.Zero


lambda_error = max(
    abs(float(candidate_multiplier(name, N)) - value)
    for name, value in selected["named_dual_values"].items()
    if value is not None
)
print("Candidate lambda maximum error:", lambda_error)
assert lambda_error < 1e-4
for group, display in selected["all_lambda_groups"].items():
    owner = group.split(":")[0]
    rows, cols = display["row_names"], display["col_names"]
    candidate = np.array(
        [
            [
                candidate_multiplier(f"{owner}:{ri},{ci}", N) if ri != ci else 0
                for ci in cols
            ]
            for ri in rows
        ],
        dtype=object,
    )
    print("Candidate", group)
    pf.pprint_labeled_matrix(candidate, rows, cols)

pm = pf.ExpressionManager(ctx, resolve_parameters=params)
candidate_weighted = pf.Scalar.zero()
for c, *_ in inventory(ctx, pb):
    residual = c.lhs - c.rhs
    if c.cmp == pf.Comparator.GE:
        residual = -residual
    candidate_weighted += candidate_multiplier(c.name, N) * residual
S_guess = candidate_weighted - pb.performance_metric + sp.Rational(1, 4 * N)
S_guess_coords = pm.eval_scalar(S_guess)
S_candidate_error = float(
    np.max(
        np.abs(
            np.asarray(S_guess_coords.inner_prod_coords, dtype=float)
            - np.asarray(selected["S_matrix"])
        )
    )
)
print("Candidate S Gram error:", S_candidate_error)
print("Candidate S function/offset:", S_guess_coords.func_coords, S_guess_coords.offset)
assert S_candidate_error < 1e-4
assert (
    np.max(np.abs(np.asarray(S_guess_coords.func_coords, dtype=float)), initial=0)
    < 1e-5
)
assert abs(float(S_guess_coords.offset)) < 1e-5
pf.pprint_labeled_matrix(
    np.asarray(S_guess_coords.inner_prod_coords), selected["basis_vectors"], precision=5
)
```

## Lyapunov partial sums

With residuals $r_{i}\le0$, assign each interpolation inequality to the step ending at its destination iterate (the initial optimum-to-$x_{0}$ term is included in the first step). Start with the explicit analytic PSD squares below. To match the saved numerical matrix $S$, apply $T=S^{1/2}S_{\mathrm{analytic}}^{\dagger/2}$ to every square matrix, and add the PSD remainder $S^{1/2}(I-P)S^{1/2}$ at the terminal step, where $P$ projects onto the analytic range. This preserves each piece's numerical sign and reconstructs the same saved $S$ to roundoff. No saved multiplier is changed. The original reverse-LDL discovery is archived in supplementary/ldl-discovery.
$$V_{0}=0,\qquad V_{k+1}-V_{k}=\sum_{i\in\mathcal I_{k}}\lambda_{i} r_{i}-\sum_{j\in\mathcal S_{k}}S_{j}\le0.$$
The final sum covers every interpolation contribution and every $S$ piece. The initial distance constraint supplies the normalization boundary:
$$V_{N}=h(x_{N})-h(x_\star)-\tau\|x_{0}-x_\star\|^2$$
up to the reported numerical certificate residual. Interior Gram rank is 4 at $N=4,6$, using fixed exclusions of one endpoint on each side. This is numerical structural evidence, not an arbitrary-horizon proof.

```python
import sympy as sp
import pepflow as pf


def a(k, N):
    return (k + 1) / (2 * N - k)


def b(k, N):
    return k / (2 * N - k)


def distance_weight(k, N):
    return (N - k) / (2 * N - k) ** 2


def gradient_weight(k, N):
    return k / (2 * (2 * N - k) ** 2)


def potential_interior(x, x0, star, f, g, L, k, N):
    d = x - star
    p = f.grad(x) - f.grad(star)
    return (
        -L / (4 * N) * (x0 - star) ** 2
        + L * distance_weight(k, N) * d**2
        - 1 / (2 * N - k) * d * f.grad(x)
        - gradient_weight(k, N) / L * p**2
        + a(k, N) * (f(x) - f(star))
        + b(k, N) * (g(x) - g(star))
    )


def potential_terminal(x, x0, star, f, g, L, N):
    return f(x) + g(x) - f(star) - g(star) - L / (4 * N) * (x0 - star) ** 2


def sos_coefficients(k, N):
    m = 2 * N - k
    aa = gradient_weight(k + 1, N) + (k + 2) / (2 * (m - 1))
    bb = (k + 1) / (2 * (m - 1))
    cc = (k * m + N - k) / (m - 1) ** 2
    dd = (k + 1) / (2 * m * (m - 1))
    ee = (k * m + 1) / (2 * (m - 1) ** 2)
    ff = a(k, N) / 2 - distance_weight(k + 1, N) - gradient_weight(k, N)
    beta = cc - bb**2 / aa
    eta = ee - bb * dd / aa
    gamma = ff - dd**2 / aa - eta**2 / beta
    return aa, beta, gamma, bb / aa, dd / aa, eta / beta


def step_squares(x, xn, star, f, g, L, k, N):
    p = (f.grad(x) - f.grad(star)) / L
    q = (f.grad(xn) - f.grad(star)) / L
    s = (g.grad(xn) + f.grad(star)) / L
    z = s - (x - star) / (2 * N - k)
    alpha, beta, gamma, u, v, w = sos_coefficients(k, N)
    return [
        L * alpha * (q + u * z + v * p) ** 2,
        L * beta * (z + w * p) ** 2,
        L * gamma * p**2,
    ]


def terminal_squares(x, xn, star, f, g, L, N):
    p = (f.grad(x) - f.grad(star)) / L
    q = (f.grad(xn) - f.grad(star)) / L
    s = (g.grad(xn) + f.grad(star)) / L
    d = x - star
    return [
        L / 2 * (q + s + (p - d) / (N + 1)) ** 2,
        L / 2 * (s + (N * p - d) / (N + 1)) ** 2,
    ]


def interpolation(f, g, L, u, v):
    If = f(v) - f(u) + f.grad(v) * (u - v) + (f.grad(u) - f.grad(v)) ** 2 / (2 * L)
    Ig = g(v) - g(u) + g.grad(v) * (u - v)
    return If, Ig


def finite_pieces(ctx, f, g, L, N):
    """PSD per-step pieces for exact rational indices."""
    N = sp.S(N)
    star = ctx["x_star"]
    pieces, assignment = {}, {}
    p0 = (f.grad(ctx["x_0"]) - f.grad(star)) / L
    pieces["initial_smooth_square"] = L / (4 * N) * p0**2
    assignment["initial_smooth_square"] = 0
    for index in range(int(N)):
        k = sp.S(index)
        x, xn = ctx[f"x_{index}"], ctx[f"x_{index + 1}"]
        local = (
            terminal_squares(x, xn, star, f, g, L, N)
            if k == N - 1
            else step_squares(x, xn, star, f, g, L, k, N)
        )
        for j, piece in enumerate(local):
            name = f"step_{index}_square_{j}"
            pieces[name] = piece
            assignment[name] = index
    return pieces, assignment


def exact_components(expression, manager):
    evaluated = manager.eval_scalar(expression, sympy_mode=True)
    gram = sp.Matrix(evaluated.inner_prod_coords).applyfunc(sp.factor)
    functions = sp.Matrix(list(evaluated.func_coords)).applyfunc(sp.factor)
    offset = sp.factor(evaluated.offset)
    return gram, functions, offset


def local_context(tag):
    ctx = pf.PEPContext(tag).set_as_current()
    L = pf.Parameter("L")
    f = pf.SmoothConvexFunction(is_basis=True, tags=[f"f_{{{tag}}}"], L=L)
    g = pf.ConvexFunction(is_basis=True, tags=[f"g_{{{tag}}}"])
    star = (f + g).set_stationary_point("x_star")
    x = pf.Vector(is_basis=True, tags=["x"])
    xn = g.prox(x - f.grad(x) / L, 1 / L, tag="x_next")
    f(xn)
    g(x)
    return ctx, L, f, g, star, x, xn


def rebuild_check(root, state, get_pep_setup):
    assert certificate_id(state) == state["certificate_id"]
    for relative, expected in state["certificate_provenance"]["source_hashes"].items():
        assert hashlib.sha256((root / relative).read_bytes()).hexdigest() == expected
    assert state["closed_form_search"]["completed"]
    params = json.loads(state["params_json"])
    ctx, pb, obj = get_pep_setup(state["N_verify"], params)
    pb.set_relaxed_constraints(state["relaxed_constraints"])
    checks, _, _ = verify(ctx, pb, params, state)
    assert checks["passed"]
    assert state["S_row_names"] == state["basis_vectors"] == state["S_col_names"]
    assert state["normalization"]["bound_value"] == state["opt_value"]
    assert (
        state["normalization"]["tau_name"] == state["tau_name"] == "initial_condition"
    )
    assert state["named_dual_values"]["initial_condition"] == state["tau_sol"]
    assert state["normalization"]["initial_condition"] == "||x_0-x_star||^2 <= R^2"
    assert state["normalization"]["performance_metric"] == "h(x_N)-h(x_star)"
    return ctx, pb, obj, checks


def build_pieces(ctx, state, pm):
    f = next(owner for owner in ctx.func_to_triplets if str(owner) == "f")
    g = next(owner for owner in ctx.func_to_triplets if str(owner) == "g")
    pieces, assignment = finite_pieces(ctx, f, g, f.L, state["N_verify"])
    basis = ctx.basis_vectors()
    raw = {
        name: pm.eval_scalar(piece).inner_prod_coords.astype(float)
        for name, piece in pieces.items()
    }
    analytic_S = sum(raw.values())
    numerical_S = np.asarray(state["S_matrix"])
    eigen_a, U = np.linalg.eigh((analytic_S + analytic_S.T) / 2)
    active = eigen_a > 1e-10
    projector = U[:, active] @ U[:, active].T
    inverse_root = (U[:, active] / np.sqrt(eigen_a[active])) @ U[:, active].T
    assert (
        np.max(np.abs((U[:, active] * eigen_a[active]) @ U[:, active].T - analytic_S))
        < 1e-10
    )
    eigen_s, W = np.linalg.eigh((numerical_S + numerical_S.T) / 2)
    assert eigen_s.min() >= -1e-6
    root_s = (W * np.sqrt(np.maximum(eigen_s, 0))) @ W.T
    transform = root_s @ inverse_root
    corrected = {name: transform @ matrix @ transform.T for name, matrix in raw.items()}
    corrected["numerical_nullspace_remainder"] = (
        root_s @ (np.eye(len(basis)) - projector) @ root_s.T
    )
    assignment["numerical_nullspace_remainder"] = state["N_verify"] - 1
    pieces = {
        name: sum(
            (
                float(matrix[i, j]) * basis[i] * basis[j]
                for i, j in np.ndindex(matrix.shape)
            ),
            pf.Scalar.zero(),
        )
        for name, matrix in corrected.items()
    }
    for piece in pieces.values():
        evaluated = pm.eval_scalar(piece)
        matrix = np.asarray(evaluated.inner_prod_coords, dtype=float)
        assert np.linalg.eigvalsh((matrix + matrix.T) / 2).min() >= -1e-6
        assert (
            np.max(np.abs(np.asarray(evaluated.func_coords, dtype=float)), initial=0)
            < 1e-5
        )
        assert abs(float(evaluated.offset)) < 1e-5
    reconstructed = pm.eval_scalar(sum(pieces.values(), pf.Scalar.zero()))
    assert np.allclose(
        reconstructed.inner_prod_coords.astype(float),
        state["S_matrix"],
        atol=1e-5,
        rtol=0,
    )
    return pieces, assignment


def construct(ctx, pb, state, verbose=False):
    ctx.set_as_current()
    params = json.loads(state["params_json"])
    pm = pf.ExpressionManager(ctx, resolve_parameters=params)
    N = state["N_verify"]
    pieces, S_assignment = build_pieces(ctx, state, pm)
    terms, assignment = {}, {}
    for c, origin, _, _ in inventory(ctx, pb):
        value = state["named_dual_values"][c.name]
        if origin == "initial":
            assert c.name == "initial_condition"
            continue
        if value is None:
            continue
        assert value >= -1e-6
        assert c.cmp in (pf.Comparator.LE, pf.Comparator.GE)
        residual = c.lhs - c.rhs
        if c.cmp == pf.Comparator.GE:
            residual = -residual
        terms[c.name] = value * residual
        destination = int(c.name.split(",")[1][2:])
        assignment[c.name] = max(0, min(N - 1, destination - 1))
    lyap = [pf.Scalar.zero()]
    partial_sum = pf.Scalar.zero()
    for k in range(N):
        increment = sum(
            (term for name, term in terms.items() if assignment[name] == k),
            pf.Scalar.zero(),
        )
        increment -= sum(
            (piece for name, piece in pieces.items() if S_assignment[name] == k),
            pf.Scalar.zero(),
        )
        partial_sum += increment
        lyap.append(partial_sum)
    values = [pm.eval_scalar(value) for value in lyap]
    matrices = [np.asarray(value.inner_prod_coords, dtype=float) for value in values]
    singular = [np.linalg.svd(matrix, compute_uv=False) for matrix in matrices]
    ranks = [int(np.sum(sv > 1e-4)) for sv in singular]
    if verbose:
        for k, rank in enumerate(ranks):
            print(f"rank V_{k}: {rank}")
            if k == 0:
                print()
        print(
            "Interior rank is constant at this N:",
            len(ranks[1:-1]) >= 3 and len(set(ranks[1:-1])) == 1,
            "boundary exclusions:",
            (1, 1),
        )

    def full_error(expression):
        v = pm.eval_scalar(expression)
        return max(
            float(
                np.max(np.abs(np.asarray(v.inner_prod_coords, dtype=float)), initial=0)
            ),
            float(np.max(np.abs(np.asarray(v.func_coords, dtype=float)), initial=0)),
            abs(float(v.offset)),
        )

    S_scalar = sum(
        (
            float(a) * u * v
            for row, u in zip(state["S_matrix"], ctx.basis_vectors())
            for a, v in zip(row, ctx.basis_vectors())
        ),
        pf.Scalar.zero(),
    )
    coverage = full_error(lyap[-1] - (sum(terms.values(), pf.Scalar.zero()) - S_scalar))
    increments = []
    for k in range(N):
        increment = sum(
            (term for name, term in terms.items() if assignment[name] == k),
            pf.Scalar.zero(),
        )
        increment -= sum(
            (piece for name, piece in pieces.items() if S_assignment[name] == k),
            pf.Scalar.zero(),
        )
        increments.append(full_error(lyap[k + 1] - lyap[k] - increment))
    boundary = full_error(
        lyap[-1]
        - (pb.performance_metric - state["tau_sol"] * (ctx["x_0"] - ctx["x_star"]) ** 2)
    )
    assert max(coverage, boundary, *increments) < 1e-5
    record = dict(
        N=N,
        certificate_id=state["certificate_id"],
        recipe_id="forward_destination_analytic_psd_congruence_v3",
        rank_profile=ranks,
        singular_values=[sv.tolist() for sv in singular],
        rank_convention={
            "tolerance": 1e-4,
            "scaling": "absolute_unscaled",
            "coordinate_order": "original_framework_basis_earlier_iterates_first",
        },
        rank_credible=all(not np.any((sv > 5e-5) & (sv < 2e-4)) for sv in singular),
        certificate_checked=True,
        grouping_checked=True,
        coverage_max_abs=coverage,
        boundary_max_abs=boundary,
        increment_max_abs=max(increments),
        constraint_assignment=assignment,
        S_assignment=S_assignment,
        boundary_constraints=[],
        boundary_S=[],
        omitted_constraints=[],
        sign_convention="V(k+1)-V(k)=sum(nonnegative_dual * residual_le_0)-sum(PSD_pieces), hence nonincreasing",
        normalization_constraint="initial_condition",
        lyap_inner_prod_coords=[
            v.inner_prod_coords.astype(float).tolist() for v in values
        ],
        lyap_func_coords=[v.func_coords.astype(float).tolist() for v in values],
        lyap_offsets=[float(v.offset) for v in values],
    )
    return lyap, pieces, terms, record
```

### Candidate-vector scan

Candidates include tagged iterates, existing basis vectors, point-to-optimum differences, smooth-gradient differences, and composite gradient sums. We add pairwise differences, remove coordinate duplicates and zero vectors, and filter to each quadratic column space.

```python
candidates_by_k = {}
for k in range(1, N_int + 1):
    labeled = [
        ("x0-star", ctx["x_0"] - ctx["x_star"]),
        ("xk-star", ctx[f"x_{k}"] - ctx["x_star"]),
        ("grad_f_k-star", f.grad(ctx[f"x_{k}"]) - f.grad(ctx["x_star"])),
        ("grad_f_k", f.grad(ctx[f"x_{k}"])),
        ("grad_g_k+grad_f_star", g.grad(ctx[f"x_{k}"]) + f.grad(ctx["x_star"])),
        ("grad_f_k+grad_g_k", f.grad(ctx[f"x_{k}"]) + g.grad(ctx[f"x_{k}"])),
    ]
    base = list(ctx.basis_vectors()) + [ctx[f"x_{j}"] for j in range(N_int + 1)]
    labeled += [(str(v), v) for v in base]
    labeled += [(f"({u})-({v})", u - v) for u, v in combinations(base, 2)]
    unique, seen = [], []
    for label, vector in labeled:
        coords = pm.eval_vector(vector).coords.astype(float)
        if np.linalg.norm(coords) < 1e-6 or any(
            np.allclose(coords, old, atol=1e-7, rtol=0) for old in seen
        ):
            continue
        unique.append((label, vector))
        seen.append(coords)
    candidates_by_k[k] = unique
```

```python
filtered_by_k = {}
labels_by_k = {}
for k, labeled in candidates_by_k.items():
    found = vectors_in_column_space(
        lyap[k],
        [v for _, v in labeled],
        pep_context=ctx,
        resolve_parameters=params_sp,
        rtol=1e-4,
        atol=1e-4,
    )
    labels = {id(v): label for label, v in labeled}
    filtered_by_k[k], labels_by_k[k] = found, labels
    print(f"V_{k} column-space candidates:")
    for vector in found:
        print("  ", labels[id(vector)])
```

### Selected basis pattern

The sparse search is shown for every nonzero index. We then use a consistent basis across interior indices to make the recurrence interpretable; the chosen basis is checked independently.

```python
search_bases, search_coefficients = {}, {}
for k in range(1, N_int + 1):
    basis_found, matrix_found = find_basis_with_sparsest_coefficients(
        lyap[k],
        filtered_by_k[k],
        pep_context=ctx,
        resolve_parameters=params_sp,
        zero_tol=1e-6,
        indep_tol=1e-7,
        span_tol=1e-5,
    )
    search_bases[k], search_coefficients[k] = basis_found, matrix_found
    assert len(basis_found) == b3["rank_profile"][k]
    print(f"k={k}, rank={len(basis_found)}")
    print("sparse basis:", [labels_by_k[k][id(v)] for v in basis_found])
    print(
        "zero coefficients:",
        int(np.sum(np.abs(matrix_found) < 1e-6)),
        "/",
        matrix_found.size,
    )
```

For $1\le k<N$, use the ordered template
$$[\,x_{0}-x_\star,\ x_{k}-x_\star,\ \nabla f(x_{k})-\nabla f(x_\star),\ \nabla f(x_{k})\,].$$
The first two vectors record distances and the last two separate the smooth-gradient difference from the affine-gradient term. At $k=0$ the basis is empty; at $k=N$ it is $[x_{0}-x_\star]$.

```python
def V_k_basis(k):
    if k == 0:
        return []
    anchor = ctx["x_0"] - ctx["x_star"]
    if k == N_int:
        return [anchor]
    return [
        anchor,
        ctx[f"x_{k}"] - ctx["x_star"],
        f.grad(ctx[f"x_{k}"]) - f.grad(ctx["x_star"]),
        f.grad(ctx[f"x_{k}"]),
    ]


def V_k_basis_labels(k):
    if k == 0:
        return []
    if k == N_int:
        return ["x0-star"]
    return ["x0-star", "xk-star", "grad_f_k-star", "grad_f_k"]


for k in range(N_int + 1):
    vectors = V_k_basis(k)
    matrix = (
        np.column_stack([pm.eval_vector(v).coords for v in vectors]).astype(float)
        if vectors
        else np.zeros((len(ctx.basis_vectors()), 0))
    )
    rank = int(np.linalg.matrix_rank(matrix, tol=1e-7)) if vectors else 0
    assert rank == len(vectors) == b3["rank_profile"][k]
    print(f"k={k}: rank {rank} basis {V_k_basis_labels(k)}")
```

### Coefficient matrices

The quadratic part is $\sum_{i,j}C_{ij}\langle v_{i},v_{j}\rangle$ with symmetric $C$; off-diagonal entries count twice. Function terms are represented by the two gaps $f(x_{k})-f(x_\star)$ and $g(x_{k})-g(x_\star)$. Their numerical weights are retained, and all scalar components are checked.

```python
coeff_by_k, coefficient_residuals, function_weights = {}, {}, {}
for k in range(N_int + 1):
    basis_k = V_k_basis(k)
    target = pm.eval_scalar(lyap[k])
    target_matrix = target.inner_prod_coords.astype(float)
    if basis_k:
        C = find_symmetric_coefficient_matrix(
            lyap[k],
            basis_k,
            pep_context=ctx,
            resolve_parameters=params_sp,
            indep_tol=1e-7,
            span_tol=1e-5,
        )
        B = np.column_stack([pm.eval_vector(v).coords for v in basis_k]).astype(float)
        quadratic_error = float(np.linalg.norm(B @ C @ B.T - target_matrix, ord="fro"))
    else:
        C = np.zeros((0, 0))
        quadratic_error = float(np.linalg.norm(target_matrix, ord="fro"))
    assert quadratic_error <= 1e-5 * max(1.0, np.linalg.norm(target_matrix, ord="fro"))
    coeff_by_k[k], coefficient_residuals[k] = C, quadratic_error
    if k == 0:
        weights = np.zeros(2)
    else:
        gaps = [
            f(ctx[f"x_{k}"]) - f(ctx["x_star"]),
            g(ctx[f"x_{k}"]) - g(ctx["x_star"]),
        ]
        F = np.column_stack([pm.eval_scalar(gap).func_coords for gap in gaps]).astype(
            float
        )
        weights = np.linalg.lstsq(F, target.func_coords.astype(float), rcond=None)[0]
        assert np.max(np.abs(F @ weights - target.func_coords.astype(float))) < 1e-5
    function_weights[k] = weights.tolist()
    assert abs(float(target.offset)) < 1e-5
    print(f"k={k}: quadratic reconstruction residual {quadratic_error:.3e}")
    if basis_k:
        pf.pprint_labeled_matrix(C, V_k_basis_labels(k), precision=7)
    print("Function gap weights [f_k-f_star, g_k-g_star]:", weights)
    print("Constant offset:", target.offset)
```

```python
def V_k_structure(k):
    vectors = V_k_basis(k)
    C = coeff_by_k[k]
    expression = sum(
        (float(C[i, j]) * vectors[i] * vectors[j] for i, j in np.ndindex(C.shape)),
        pf.Scalar.zero(),
    )
    if k > 0:
        af, ag = function_weights[k]
        expression += af * (f(ctx[f"x_{k}"]) - f(ctx["x_star"]))
        expression += ag * (g(ctx[f"x_{k}"]) - g(ctx["x_star"]))
    return expression


component_residuals = {}
for k in range(N_int + 1):
    evaluated = pm.eval_scalar(V_k_structure(k) - lyap[k])
    target = pm.eval_scalar(lyap[k])
    errors = {}
    for label, residual, reference in (
        ("gram", evaluated.inner_prod_coords, target.inner_prod_coords),
        ("function", evaluated.func_coords, target.func_coords),
        ("offset", evaluated.offset, target.offset),
    ):
        error = float(np.max(np.abs(np.asarray(residual, dtype=float)), initial=0))
        scale = float(np.max(np.abs(np.asarray(reference, dtype=float)), initial=0))
        assert error <= 1e-4 + 1e-5 * scale
        errors[label] = error
    component_residuals[str(k)] = errors
    print("Complete reconstruction", k, errors)
```

### Block 4 conclusion

Every index, including both endpoints, passes independence and full-expression reconstruction. For $1\le k<N$ the observed structure is
$$V_{k}=\sum_{i,j=0}^{3}c_{ij}(k,N,L)\langle v_{i},v_{j}\rangle
+a_{f}(k,N,L)\bigl(f(x_{k})-f(x_\star)\bigr)
+a_{g}(k,N,L)\bigl(g(x_{k})-g(x_\star)\bigr).$$
The symmetric coefficient functions remain unresolved. At $k=0$, $V_{0}=0$ numerically. At $k=N$, the structure has one quadratic coefficient and two function-gap weights, with no constant term. Block 5 must derive the coefficients and establish the exact step, initial and terminal identities; this numerical reconstruction does not prove them.

## Closed-form coefficient derivation

```python
from typing import Any, cast

b4 = json.loads((notebook_dir / "state/pgm_b4.json").read_text())
assert b4["certificate_id"] == certificate["certificate_id"]
assert b4["numerical_reconstruction"]["passed"]
print("B4 unknowns resolved below:", [v["name"] for v in b4["unresolved_coefficients"]])
proof_checks = {}
```

The numerical structure suggests the displayed $a_{k},b_{k}$ and distance/cross coefficients. The gradient coefficient has freedom. Write its negative normalized value as $t_{k}$, retain $t_{k+1}$ independently in the difference, and form the residual quadratic below. We select $t_{k}=k/[2(2N-k)^{2}]$; exact identities and positive pivots certify this choice. This is an analytically chosen decomposition, not a formula claim for the original reverse LDL factors.

```python
k_sym, N_sym = sp.symbols("k N", integer=True)
t_prev, t_next = sp.symbols("t_k t_next", real=True)
d_sym, p_sym, q_sym, s_sym = sp.symbols("d p q s", real=True)
dn_sym = d_sym - p_sym - s_sym
If_star_quad = -q_sym * dn_sym + q_sym**2 / 2
Ig_star_quad = -s_sym * dn_sym
If_step_quad = q_sym * (d_sym - dn_sym) + (p_sym - q_sym) ** 2 / 2
Ig_step_quad = s_sym * (d_sym - dn_sym)
V_quad = (
    distance_weight(k_sym, N_sym) * d_sym**2
    - d_sym * p_sym / (2 * N_sym - k_sym)
    - t_prev * p_sym**2
)
W_quad = (
    distance_weight(k_sym + 1, N_sym) * dn_sym**2
    - dn_sym * q_sym / (2 * N_sym - k_sym - 1)
    - t_next * q_sym**2
)
S_unknown = (
    (a(k_sym + 1, N_sym) - a(k_sym, N_sym)) * If_star_quad
    + (b(k_sym + 1, N_sym) - b(k_sym, N_sym)) * Ig_star_quad
    + a(k_sym, N_sym) * If_step_quad
    + b(k_sym, N_sym) * Ig_step_quad
    - W_quad
    + V_quad
)
M_unknown = (sp.hessian(S_unknown, [q_sym, s_sym, d_sym, p_sym]) / 2).applyfunc(
    sp.factor
)
print("Residual quadratic with unresolved gradient coefficients:")
pf.pprint_labeled_matrix(
    np.asarray(M_unknown, dtype=object), ["q", "s", "d", "p"], precision=0
)
chosen_S = S_unknown.subs(
    {t_prev: gradient_weight(k_sym, N_sym), t_next: gradient_weight(k_sym + 1, N_sym)}
)
alpha_s, beta_s, gamma_s, u_s, v_s, w_s = sos_coefficients(k_sym, N_sym)
z_sym = s_sym - d_sym / (2 * N_sym - k_sym)
chosen_squares = (
    alpha_s * (q_sym + u_s * z_sym + v_s * p_sym) ** 2
    + beta_s * (z_sym + w_s * p_sym) ** 2
    + gamma_s * p_sym**2
)
assert sp.factor(chosen_S - chosen_squares) == 0
print("Chosen gradient coefficients solve the exact residual-square equations.")
(
    ctx_derivation,
    L_derivation,
    f_derivation,
    g_derivation,
    star_derivation,
    x_derivation,
    xn_derivation,
) = local_context("derivation")
k_derivation, N_derivation = pf.Parameter("k"), pf.Parameter("N")
t_derivation, tn_derivation = pf.Parameter("t_prev"), pf.Parameter("t_next")
anchor_derivation = pf.Vector(is_basis=True, tags=["x_0"])
p_derivation = (
    f_derivation.grad(x_derivation) - f_derivation.grad(star_derivation)
) / L_derivation
q_derivation = (
    f_derivation.grad(xn_derivation) - f_derivation.grad(star_derivation)
) / L_derivation
V_unknown = (
    potential_interior(
        x_derivation,
        anchor_derivation,
        star_derivation,
        f_derivation,
        g_derivation,
        L_derivation,
        k_derivation,
        N_derivation,
    )
    + L_derivation
    * (gradient_weight(k_derivation, N_derivation) - t_derivation)
    * p_derivation**2
)
W_unknown = (
    potential_interior(
        xn_derivation,
        anchor_derivation,
        star_derivation,
        f_derivation,
        g_derivation,
        L_derivation,
        k_derivation + 1,
        N_derivation,
    )
    + L_derivation
    * (gradient_weight(k_derivation + 1, N_derivation) - tn_derivation)
    * q_derivation**2
)
If_d, Ig_d = interpolation(
    f_derivation, g_derivation, L_derivation, star_derivation, xn_derivation
)
If_s, Ig_s = interpolation(
    f_derivation, g_derivation, L_derivation, x_derivation, xn_derivation
)
R_unknown = (
    (a(k_derivation + 1, N_derivation) - a(k_derivation, N_derivation)) * If_d
    + (b(k_derivation + 1, N_derivation) - b(k_derivation, N_derivation)) * Ig_d
    + a(k_derivation, N_derivation) * If_s
    + b(k_derivation, N_derivation) * Ig_s
)
R_unknown -= sum(
    step_squares(
        x_derivation,
        xn_derivation,
        star_derivation,
        f_derivation,
        g_derivation,
        L_derivation,
        k_derivation,
        N_derivation,
    ),
    pf.Scalar.zero(),
)
t_function: Any = sp.Function("t")
derivation_parameters: dict[str, Any] = {
    "L": sp.Symbol("L", positive=True),
    "k": k_sym,
    "N": N_sym,
    "t_prev": t_function(k_sym, N_sym),
    "t_next": t_function(k_sym + 1, N_sym),
}
pm_derivation = pf.ExpressionManager(
    ctx_derivation, resolve_parameters=derivation_parameters
)
unknown_components = exact_components(W_unknown - V_unknown - R_unknown, pm_derivation)
equations = [
    sp.Eq(term, 0)
    for component in unknown_components[:2]
    for term in component
    if term != 0
]
print("Native coefficient equations:", equations)
chosen_t = {
    t_function(k_sym, N_sym): gradient_weight(k_sym, N_sym),
    t_function(k_sym + 1, N_sym): gradient_weight(k_sym + 1, N_sym),
}
assert all(
    sp.factor(term.subs(chosen_t)) == 0
    for component in unknown_components[:2]
    for term in component
)
assert sp.factor(unknown_components[2]) == 0
ctx.set_as_current()
```

```python
def closed_coefficient_matrix(index, horizon, smoothness):
    index, horizon, smoothness = [
        cast(sp.Expr, sp.sympify(value)) for value in (index, horizon, smoothness)
    ]
    if index == 0:
        return sp.zeros(0, 0)
    if index == horizon:
        return sp.Matrix([[-smoothness / (4 * horizon)]])
    M = sp.zeros(4, 4)
    M[0, 0] = -smoothness / (4 * horizon)
    M[1, 1] = smoothness * distance_weight(index, horizon)
    M[2, 2] = -gradient_weight(index, horizon) / smoothness
    M[1, 3] = M[3, 1] = -1 / (2 * (2 * horizon - index))
    return M


coefficient_comparisons = {}
for index in range(N_int + 1):
    closed_C = closed_coefficient_matrix(index, N_int, params_sp["L"])
    if index:
        print("Closed coefficient matrix", index)
        pf.pprint_labeled_matrix(
            np.asarray(closed_C, dtype=object), V_k_basis_labels(index), precision=0
        )
    err = float(
        np.max(
            np.abs(
                np.asarray(closed_C, dtype=float).reshape(coeff_by_k[index].shape)
                - coeff_by_k[index]
            ),
            initial=0,
        )
    )
    target_weights = (
        [0, 0]
        if index == 0
        else (
            [1, 1]
            if index == N_int
            else [a(sp.S(index), sp.S(N_int)), b(sp.S(index), sp.S(N_int))]
        )
    )
    ferr = float(
        np.max(
            np.abs(np.asarray(target_weights, dtype=float) - function_weights[index]),
            initial=0,
        )
    )
    assert err < 1e-4 and ferr < 1e-4
    coefficient_comparisons[str(index)] = {"quadratic": err, "function": ferr}
    print("Coefficient comparison", index, coefficient_comparisons[str(index)])
```

### Closed-form conclusion

Put $d_{k}=x_{k}-x_{\star}$ and $p_{k}=(\nabla f(x_{k})-\nabla f(x_{\star}))/L$.
For $1\le k<N$, define
$$V_{k}=-\frac{L}{4N}\|d_{0}\|^{2}
+\frac{L(N-k)}{(2N-k)^{2}}\|d_{k}\|^{2}
-\frac{\langle d_{k},\nabla f(x_{k})\rangle}{2N-k}
-\frac{Lk}{2(2N-k)^{2}}\|p_{k}\|^{2}
+\frac{k+1}{2N-k}(f(x_{k})-f(x_{\star}))
+\frac{k}{2N-k}(g(x_{k})-g(x_{\star})).$$
The endpoints are $V_{0}=0$ and
$$V_{N}=h(x_{N})-h(x_{\star})-\frac{L}{4N}\|d_{0}\|^{2}.$$

The following native symbolic checks verify this candidate for arbitrary admissible horizons.

## Sign and domain verification

Set $n=N-k-1\ge0$. The following exact positive-polynomial representations prove all three square signs. Every coefficient of $A,B,P$ is positive and $A,B$ have positive constants. Therefore all denominators and division pivots are strictly positive. At $k=0$, the third square has zero weight, which is allowed.

```python
sign_k, sign_n = sp.symbols("k n", nonnegative=True)
sign_N = sign_k + 1 + sign_n
alpha_sign, beta_sign, gamma_sign, *_ = sos_coefficients(sign_k, sign_N)
A = sign_k**2 + 2 * sign_k * sign_n + 4 * sign_k + 4 * sign_n + 3
B = (
    sign_k**4
    + 4 * sign_k**3 * sign_n
    + 8 * sign_k**3
    + 4 * sign_k**2 * sign_n**2
    + 22 * sign_k**2 * sign_n
    + 18 * sign_k**2
    + 12 * sign_k * sign_n**2
    + 28 * sign_k * sign_n
    + 16 * sign_k
    + 4 * sign_n**2
    + 10 * sign_n
    + 5
)
P = (
    sign_k**4
    + 5 * sign_k**3 * sign_n
    + 3 * sign_k**3
    + 8 * sign_k**2 * sign_n**2
    + 13 * sign_k**2 * sign_n
    + 3 * sign_k**2
    + 4 * sign_k * sign_n**3
    + 16 * sign_k * sign_n**2
    + 13 * sign_k * sign_n
    + sign_k
    + 4 * sign_n**3
    + 10 * sign_n**2
    + 5 * sign_n
)
sign_forms = [
    A / (2 * (sign_k + 2 * sign_n + 1) ** 2),
    B / (2 * (sign_k + 2 * sign_n + 1) ** 2 * A),
    sign_k * P / ((sign_k + 2 * sign_n + 2) ** 2 * B),
]
assert [
    sp.factor(x - y) for x, y in zip([alpha_sign, beta_sign, gamma_sign], sign_forms)
] == [0, 0, 0]
assert all(c > 0 for poly in (A, B, P) for c in sp.Poly(poly, sign_k, sign_n).coeffs())
assert A.subs({sign_k: 0, sign_n: 0}) > 0 and B.subs({sign_k: 0, sign_n: 0}) > 0
for name, expr in zip(["alpha", "beta", "gamma"], sign_forms):
    print(name, "=", expr)
    print(sp.latex(expr))
assert (
    sp.factor(
        a(k_sym + 1, N_sym)
        - a(k_sym, N_sym)
        - (2 * N_sym + 1) / ((2 * N_sym - k_sym - 1) * (2 * N_sym - k_sym))
    )
    == 0
)
assert (
    sp.factor(
        b(k_sym + 1, N_sym)
        - b(k_sym, N_sym)
        - 2 * N_sym / ((2 * N_sym - k_sym - 1) * (2 * N_sym - k_sym))
    )
    == 0
)
assert sp.factor(1 - a(N_sym - 1, N_sym) - 1 / (N_sym + 1)) == 0
assert sp.factor(1 - b(N_sym - 1, N_sym) - 2 / (N_sym + 1)) == 0
proof_checks["signs"] = True
proof_checks["domain"] = True
print(
    "For L>0, integer N>=1, all used residual multipliers and square weights are nonnegative; denominators are nonzero."
)
```

## Symbolic Step Recursion Verification

$$V_{k+1}-V_{k}=(a_{k+1}-a_{k})I_{f}(x_{\star},x_{k+1})+(b_{k+1}-b_{k})I_{g}(x_{\star},x_{k+1})+a_{k}I_{f}(x_{k},x_{k+1})+b_{k}I_{g}(x_{k},x_{k+1})-Q_{k}.$$ For $1\le k\le N-2$. The residual $\mathrm{LHS}-\mathrm{RHS}$ must be exactly zero.

```python
# Independent native symbolic step context and the actual proximal recurrence.
ctx_proof = pf.PEPContext("proof_step").set_as_current()
L_proof = pf.Parameter("L")
N_proof = pf.Parameter("N")
k_proof = pf.Parameter("k")
f_proof = pf.SmoothConvexFunction(is_basis=True, tags=["f_{step}"], L=L_proof)
g_proof = pf.ConvexFunction(is_basis=True, tags=["g_{step}"])
star_proof = (f_proof + g_proof).set_stationary_point("x_star")
x_proof = pf.Vector(is_basis=True, tags=["x"])
xn_proof = g_proof.prox(
    x_proof - f_proof.grad(x_proof) / L_proof, 1 / L_proof, tag="x_next"
)
f_proof(xn_proof)
g_proof(x_proof)
x0_proof = pf.Vector(is_basis=True, tags=["x_0"])
Ifstar, Igstar = interpolation(f_proof, g_proof, L_proof, star_proof, xn_proof)
Ifstep, Igstep = interpolation(f_proof, g_proof, L_proof, x_proof, xn_proof)
LHS = potential_interior(
    xn_proof, x0_proof, star_proof, f_proof, g_proof, L_proof, k_proof + 1, N_proof
) - potential_interior(
    x_proof, x0_proof, star_proof, f_proof, g_proof, L_proof, k_proof, N_proof
)
RHS = (a(k_proof + 1, N_proof) - a(k_proof, N_proof)) * Ifstar + (
    b(k_proof + 1, N_proof) - b(k_proof, N_proof)
) * Igstar
RHS += (
    a(k_proof, N_proof) * Ifstep
    + b(k_proof, N_proof) * Igstep
    - sum(
        step_squares(
            x_proof, xn_proof, star_proof, f_proof, g_proof, L_proof, k_proof, N_proof
        ),
        pf.Scalar.zero(),
    )
)
diff = LHS - RHS
L_symbol = sp.Symbol("L", positive=True)
# The framework accepts symbolic substitutions in sympy_mode; its annotation is narrower.
symbolic_parameters: dict[str, Any] = {"L": L_symbol, "N": N_sym, "k": k_sym}
pm_proof = pf.ExpressionManager(ctx_proof, resolve_parameters=symbolic_parameters)
gram_residual, function_residual, constant_residual = exact_components(diff, pm_proof)
pf.pprint_labeled_matrix(
    np.asarray(gram_residual, dtype=object),
    [f"v_{i}" for i in range(gram_residual.rows)],
    precision=0,
)
print("step function residual:", function_residual)
print("step constant residual:", constant_residual)
assert all(v == 0 for v in [*gram_residual, *function_residual, constant_residual])
proof_checks["step"] = True
print("step: exact identity passed in all scalar components.")
```

## Base Case and Boundary Symbolic Verification

$$V_{1}=a_{0}I_{f}(x_{\star},x_{0})+(a_{1}-a_{0})I_{f}(x_{\star},x_{1})+b_{1}I_{g}(x_{\star},x_{1})+a_{0}I_{f}(x_{0},x_{1})-\frac{L}{4N}\|p_{0}\|^2-Q_{0}.$$ For $N\ge2$. The residual $\mathrm{LHS}-\mathrm{RHS}$ must be exactly zero.

```python
# Independent native symbolic base context and the actual proximal recurrence.
ctx_proof = pf.PEPContext("proof_base").set_as_current()
L_proof = pf.Parameter("L")
N_proof = pf.Parameter("N")
k_proof = pf.Parameter("k")
f_proof = pf.SmoothConvexFunction(is_basis=True, tags=["f_{base}"], L=L_proof)
g_proof = pf.ConvexFunction(is_basis=True, tags=["g_{base}"])
star_proof = (f_proof + g_proof).set_stationary_point("x_star")
x_proof = pf.Vector(is_basis=True, tags=["x"])
xn_proof = g_proof.prox(
    x_proof - f_proof.grad(x_proof) / L_proof, 1 / L_proof, tag="x_next"
)
f_proof(xn_proof)
g_proof(x_proof)
x0_proof = x_proof
Ifstar, Igstar = interpolation(f_proof, g_proof, L_proof, star_proof, xn_proof)
Ifstep, Igstep = interpolation(f_proof, g_proof, L_proof, x_proof, xn_proof)
k_proof = sp.S.Zero
LHS = potential_interior(
    xn_proof, x_proof, star_proof, f_proof, g_proof, L_proof, sp.S.One, N_proof
)
Ifinitial, _ = interpolation(f_proof, g_proof, L_proof, star_proof, x_proof)
RHS = (
    a(k_proof, N_proof) * Ifinitial
    + (a(k_proof + 1, N_proof) - a(k_proof, N_proof)) * Ifstar
)
RHS += (b(k_proof + 1, N_proof) - b(k_proof, N_proof)) * Igstar + a(
    k_proof, N_proof
) * Ifstep
RHS -= (
    L_proof
    / (4 * N_proof)
    * ((f_proof.grad(x_proof) - f_proof.grad(star_proof)) / L_proof) ** 2
)
RHS -= sum(
    step_squares(
        x_proof, xn_proof, star_proof, f_proof, g_proof, L_proof, k_proof, N_proof
    ),
    pf.Scalar.zero(),
)
diff = LHS - RHS
L_symbol = sp.Symbol("L", positive=True)
# The framework accepts symbolic substitutions in sympy_mode; its annotation is narrower.
symbolic_parameters: dict[str, Any] = {"L": L_symbol, "N": N_sym, "k": k_sym}
pm_proof = pf.ExpressionManager(ctx_proof, resolve_parameters=symbolic_parameters)
gram_residual, function_residual, constant_residual = exact_components(diff, pm_proof)
pf.pprint_labeled_matrix(
    np.asarray(gram_residual, dtype=object),
    [f"v_{i}" for i in range(gram_residual.rows)],
    precision=0,
)
print("base function residual:", function_residual)
print("base constant residual:", constant_residual)
assert all(v == 0 for v in [*gram_residual, *function_residual, constant_residual])
proof_checks["base"] = True
print("base: exact identity passed in all scalar components.")
```

### Boundary Identity Symbolic Verification

$$V_{N}-V_{N-1}=(1-a_{N-1})I_{f}(x_{\star},x_{N})+(1-b_{N-1})I_{g}(x_{\star},x_{N})+a_{N-1}I_{f}(x_{N-1},x_{N})+b_{N-1}I_{g}(x_{N-1},x_{N})-T_{N}.$$ For $N\ge2$. The residual $\mathrm{LHS}-\mathrm{RHS}$ must be exactly zero.

```python
# Independent native symbolic terminal context and the actual proximal recurrence.
ctx_proof = pf.PEPContext("proof_terminal").set_as_current()
L_proof = pf.Parameter("L")
N_proof = pf.Parameter("N")
k_proof = pf.Parameter("k")
f_proof = pf.SmoothConvexFunction(is_basis=True, tags=["f_{terminal}"], L=L_proof)
g_proof = pf.ConvexFunction(is_basis=True, tags=["g_{terminal}"])
star_proof = (f_proof + g_proof).set_stationary_point("x_star")
x_proof = pf.Vector(is_basis=True, tags=["x"])
xn_proof = g_proof.prox(
    x_proof - f_proof.grad(x_proof) / L_proof, 1 / L_proof, tag="x_next"
)
f_proof(xn_proof)
g_proof(x_proof)
x0_proof = pf.Vector(is_basis=True, tags=["x_0"])
Ifstar, Igstar = interpolation(f_proof, g_proof, L_proof, star_proof, xn_proof)
Ifstep, Igstep = interpolation(f_proof, g_proof, L_proof, x_proof, xn_proof)
k_proof = N_proof - 1
LHS = potential_terminal(
    xn_proof, x0_proof, star_proof, f_proof, g_proof, L_proof, N_proof
) - potential_interior(
    x_proof, x0_proof, star_proof, f_proof, g_proof, L_proof, k_proof, N_proof
)
RHS = (
    (1 - a(k_proof, N_proof)) * Ifstar
    + (1 - b(k_proof, N_proof)) * Igstar
    + a(k_proof, N_proof) * Ifstep
    + b(k_proof, N_proof) * Igstep
)
RHS -= sum(
    terminal_squares(x_proof, xn_proof, star_proof, f_proof, g_proof, L_proof, N_proof),
    pf.Scalar.zero(),
)
diff = LHS - RHS
L_symbol = sp.Symbol("L", positive=True)
# The framework accepts symbolic substitutions in sympy_mode; its annotation is narrower.
symbolic_parameters: dict[str, Any] = {"L": L_symbol, "N": N_sym, "k": k_sym}
pm_proof = pf.ExpressionManager(ctx_proof, resolve_parameters=symbolic_parameters)
gram_residual, function_residual, constant_residual = exact_components(diff, pm_proof)
pf.pprint_labeled_matrix(
    np.asarray(gram_residual, dtype=object),
    [f"v_{i}" for i in range(gram_residual.rows)],
    precision=0,
)
print("terminal function residual:", function_residual)
print("terminal constant residual:", constant_residual)
assert all(v == 0 for v in [*gram_residual, *function_residual, constant_residual])
proof_checks["boundary"] = True
print("terminal: exact identity passed in all scalar components.")
```

### Exceptional horizon N=1

$$V_{1}=\tfrac12 I_{f}(x_{\star},x_{0})+\tfrac12 I_{f}(x_{\star},x_{1})+I_{g}(x_{\star},x_{1})+\tfrac12 I_{f}(x_{0},x_{1})-\tfrac L4\|p_{0}\|^2-T_{1}.$$ For $N=1$. The residual $\mathrm{LHS}-\mathrm{RHS}$ must be exactly zero.

```python
# Independent native symbolic N1 context and the actual proximal recurrence.
ctx_proof = pf.PEPContext("proof_N1").set_as_current()
L_proof = pf.Parameter("L")
N_proof = pf.Parameter("N")
k_proof = pf.Parameter("k")
f_proof = pf.SmoothConvexFunction(is_basis=True, tags=["f_{N1}"], L=L_proof)
g_proof = pf.ConvexFunction(is_basis=True, tags=["g_{N1}"])
star_proof = (f_proof + g_proof).set_stationary_point("x_star")
x_proof = pf.Vector(is_basis=True, tags=["x"])
xn_proof = g_proof.prox(
    x_proof - f_proof.grad(x_proof) / L_proof, 1 / L_proof, tag="x_next"
)
f_proof(xn_proof)
g_proof(x_proof)
x0_proof = x_proof
Ifstar, Igstar = interpolation(f_proof, g_proof, L_proof, star_proof, xn_proof)
Ifstep, Igstep = interpolation(f_proof, g_proof, L_proof, x_proof, xn_proof)
N_proof = sp.S.One
k_proof = N_proof - 1
LHS = potential_terminal(
    xn_proof, x0_proof, star_proof, f_proof, g_proof, L_proof, N_proof
) - potential_interior(
    x_proof, x0_proof, star_proof, f_proof, g_proof, L_proof, k_proof, N_proof
)
RHS = (
    (1 - a(k_proof, N_proof)) * Ifstar
    + (1 - b(k_proof, N_proof)) * Igstar
    + a(k_proof, N_proof) * Ifstep
    + b(k_proof, N_proof) * Igstep
)
RHS -= sum(
    terminal_squares(x_proof, xn_proof, star_proof, f_proof, g_proof, L_proof, N_proof),
    pf.Scalar.zero(),
)
LHS = potential_terminal(
    xn_proof, x_proof, star_proof, f_proof, g_proof, L_proof, N_proof
)
Ifinitial, _ = interpolation(f_proof, g_proof, L_proof, star_proof, x_proof)
RHS += (
    a(sp.S.Zero, N_proof) * Ifinitial
    - L_proof
    / (4 * N_proof)
    * ((f_proof.grad(x_proof) - f_proof.grad(star_proof)) / L_proof) ** 2
)
diff = LHS - RHS
L_symbol = sp.Symbol("L", positive=True)
# The framework accepts symbolic substitutions in sympy_mode; its annotation is narrower.
symbolic_parameters: dict[str, Any] = {"L": L_symbol, "N": N_sym, "k": k_sym}
pm_proof = pf.ExpressionManager(ctx_proof, resolve_parameters=symbolic_parameters)
gram_residual, function_residual, constant_residual = exact_components(diff, pm_proof)
pf.pprint_labeled_matrix(
    np.asarray(gram_residual, dtype=object),
    [f"v_{i}" for i in range(gram_residual.rows)],
    precision=0,
)
print("N1 function residual:", function_residual)
print("N1 constant residual:", constant_residual)
assert all(v == 0 for v in [*gram_residual, *function_residual, constant_residual])
proof_checks["N1"] = True
print("N1: exact identity passed in all scalar components.")
```

## Normalization and theorem conclusion

Put $d_{k}=x_{k}-x_{\star}$ and $p_{k}=(\nabla f(x_{k})-\nabla f(x_{\star}))/L$.
For $1\le k<N$, define
$$V_{k}=-\frac{L}{4N}\|d_{0}\|^{2}
+\frac{L(N-k)}{(2N-k)^{2}}\|d_{k}\|^{2}
-\frac{\langle d_{k},\nabla f(x_{k})\rangle}{2N-k}
-\frac{Lk}{2(2N-k)^{2}}\|p_{k}\|^{2}
+\frac{k+1}{2N-k}(f(x_{k})-f(x_{\star}))
+\frac{k}{2N-k}(g(x_{k})-g(x_{\star})).$$
The endpoints are $V_{0}=0$ and
$$V_{N}=h(x_{N})-h(x_{\star})-\frac{L}{4N}\|d_{0}\|^{2}.$$
The exact identities and sign lemma give $V_{N}\le0$. Therefore $h(x_{N})-h(x_{\star})\le L\|d_{0}\|^{2}/(4N)\le LR^{2}/(4N)$, including $N=1$.

```python
normalization_L, normalization_N = sp.symbols("L N", positive=True)
normalization_R, initial_distance = sp.symbols("R D", nonnegative=True)
terminal_gap = sp.Symbol("H", real=True)
terminal_value = terminal_gap - normalization_L * initial_distance / (
    4 * normalization_N
)
assert (
    sp.expand(
        terminal_value
        + normalization_L * initial_distance / (4 * normalization_N)
        - terminal_gap
    )
    == 0
)
assert (
    sp.factor(
        normalization_L * normalization_R**2 / (4 * normalization_N)
        - normalization_L * initial_distance / (4 * normalization_N)
        - normalization_L
        * (-initial_distance + normalization_R**2)
        / (4 * normalization_N)
    )
    == 0
)
proof_checks["normalization"] = True
assert all(
    proof_checks[name]
    for name in ("step", "base", "boundary", "N1", "signs", "domain", "normalization")
)
print("V_N <= V_0 = 0; D=||x0-star||^2 <= R^2 gives h(x_N)-h(x_star) <= L R^2/(4N).")
print(r"\forall N\in\mathbb{N}_{\ge1},\quad h(x_N)-h(x_\star)\le\frac{LR^2}{4N}.")
print(
    "Numerical full certificate verified; global analytic lambda/S status remains unresolved; analytic Lyapunov proof verified."
)
```
