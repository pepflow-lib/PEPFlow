# Proximal gradient: composite convex minimization
Let $f$ be convex and $L$-smooth, $L>0$, and let $g$ be closed, proper and convex on a finite-dimensional real Euclidean space. Write $h=f+g$. Assume $x_{\star}$ minimizes $h$, $R\geq0$, and $\|x_{0}-x_{\star}\|^2\leq R^2$. For integer $N\geq1$, perform
$$x_{k+1}=\operatorname{prox}_{g/L}\left(x_{k}-\frac1L\nabla f(x_{k})\right),\qquad k=0,\ldots,N-1.$$
The performance metric is $h(x_{N})-h(x_{\star})$. Proximal optimality is encoded as $L(x_{k}-x_{k+1})-\nabla f(x_{k})\in\partial g(x_{k+1})$. Stationarity is imposed on the sum, not separately on its components.

## Proof Statement

### Theorem
Under the assumptions above, for every integer $N\geq1$,
$$h(x_{N})-h(x_{\star})\leq\frac{LR^{2}}{4N}.$$
Here is an explicit Lyapunov certificate. Put $e_{k}=x_{k}-x_{\star}$, $p_{k}=(\nabla f(x_{k})-\nabla f(x_{\star}))/L$, and $t_{k}=2N-k$. Set $V_{0}=0$, and for $1\leq k<N$ set
$$V_{k}=-\frac{\|e_{0}\|^{2}}{4N}+\frac{k+1}{Lt_{k}}(f(x_{k})-f(x_{\star}))+\frac{k}{Lt_{k}}(g(x_{k})-g(x_{\star}))+\frac{N-k}{t_{k}^{2}}\|e_{k}\|^{2}-\frac{\langle e_{k},\nabla f(x_{k})\rangle}{Lt_{k}}-\beta_{k}\|p_{k}\|^{2}.$$
The terminal value is
$$V_{N}=\frac{h(x_{N})-h(x_{\star})}{L}-\frac{\|e_{0}\|^{2}}{4N}.$$
The coefficient has the following closed finite-product expression. For $1\leq j\leq N-2$, with $t=2N-j$, define
$$M_{j}=\begin{pmatrix}2j(j+1)t(t-1)^{2}&j(j+1)(t-1)(j+t+1)\\4t^{2}(t-1)((2j+1)t-j)&2t^{2}((j^{2}+3j+1)t+1)\end{pmatrix}.$$
For $1\leq k<N$, form
$$\binom{u_{k}}{v_{k}}=\left(M_{k}M_{k+1}\cdots M_{N-2}\right)\binom{(N-1)/(2(N+1)^{2})}{1},\qquad\beta_{k}=u_{k}/v_{k}.$$
An empty product is the identity; thus $\beta_{N-1}=(N-1)/(2(N+1)^{2})$. Set $\beta_{0}=\beta_{N}=0$. No product or interior potential is needed when $N=1$. Every denominator is positive on the stated domain. These formulas give $V_{k+1}\leq V_{k}$, including the exceptional first and last steps.


### Proof outline
Use the potential $V_{0}=0$,
$$V_{k}=-\frac{\|e_{0}\|^{2}}{4N}+a_{k}\frac{f(x_{k})-f(x_{\star})}{L}+b_{k}\frac{g(x_{k})-g(x_{\star})}{L}+A_{k}\|e_{k}\|^{2}-d_{k}\langle e_{k},\nabla f(x_{k})/L\rangle-\beta_{k}\|p_{k}\|^{2}$$
for $1\leq k<N$, with $a_{k}=(k+1)/t_{k}$, $b_{k}=k/t_{k}$, $A_{k}=(N-k)/t_{k}^{2}$, $d_{k}=1/t_{k}$ and the finite-product $\beta_{k}$ defined in the theorem. Its terminal value is $V_{N}=(h(x_{N})-h(x_{\star}))/L-\|e_{0}\|^{2}/(4N)$.

For a selected subgradient $s_{g}(v)\in\partial g(v)$, define the nonpositive interpolation residuals
$$J_{f}(u,v)=\frac{f(v)-f(u)+\langle\nabla f(v),u-v\rangle}{L}+\frac{\|\nabla f(u)-\nabla f(v)\|^{2}}{2L^{2}},\qquad J_{g}(u,v)=\frac{g(v)-g(u)+\langle s_{g}(v),u-v\rangle}{L}.$$
Take $s_{g}(x_{\star})=-\nabla f(x_{\star})$ and $s_{g}(x_{k+1})=L(x_{k}-x_{k+1})-\nabla f(x_{k})$. Put $q_{k+1}=(s_{g}(x_{k+1})+\nabla f(x_{\star}))/L$; then $e_{k+1}=e_{k}-p_{k}-q_{k+1}$.

For $0\leq k\leq N-2$, set $t=2N-k$, $c=\beta_{k+1}$, and
$$H_{k}=\begin{pmatrix}c+\frac{k+2}{2(t-1)}&\frac{k+1}{2(t-1)}\\\frac{k+1}{2(t-1)}&\frac{(2k+1)t-k}{2(t-1)^{2}}\end{pmatrix},\quad K_{k}=\begin{pmatrix}-\frac{k+1}{2t(t-1)}&\frac{k+1}{2t(t-1)}\\-\frac{(2k+1)t-k}{2t(t-1)^{2}}&\frac{kt+1}{2(t-1)^{2}}\end{pmatrix}.$$
Writing $z=(e_{k},p_{k})^{T}$ and $w=(p_{k+1},q_{k+1})^{T}$, let
$$Q_{k}=\|w+H_{k}^{-1}K_{k}z\|_{H_{k}}^{2},\qquad\|y\|_{H}^{2}=\sum_{i,j=1}^{2}H_{ij}\langle y_{i},y_{j}\rangle.$$
The exact interior identity, for $1\leq k\leq N-2$, is
$$V_{k+1}-V_{k}=(a_{k+1}-a_{k})J_{f}(x_{\star},x_{k+1})+a_{k}J_{f}(x_{k},x_{k+1})+(b_{k+1}-b_{k})J_{g}(x_{\star},x_{k+1})+b_{k}J_{g}(x_{k},x_{k+1})-Q_{k}.$$
For $N\geq2$, the base identity is
$$V_{1}=a_{0}J_{f}(x_{\star},x_{0})+(a_{1}-a_{0})J_{f}(x_{\star},x_{1})+a_{0}J_{f}(x_{0},x_{1})+b_{1}J_{g}(x_{\star},x_{1})-Q_{0}-\frac{\|p_{0}\|^{2}}{4N}.$$
No value or subgradient of $g$ at $x_{0}$ is required.

For the final step define $z=(e_{N-1},p_{N-1})^{T}$, $w=(p_{N},q_{N})^{T}$, and
$$H_{T}=\begin{pmatrix}1/2&1/2\\1/2&1\end{pmatrix},\quad K_{T}=\begin{pmatrix}-\frac1{2(N+1)}&\frac1{2(N+1)}\\-\frac1{N+1}&1/2\end{pmatrix},\quad Q_{T}=\|w+H_{T}^{-1}K_{T}z\|_{H_{T}}^{2}.$$
For $N\geq2$,
$$V_{N}-V_{N-1}=\frac1{N+1}J_{f}(x_{\star},x_{N})+\frac{N}{N+1}J_{f}(x_{N-1},x_{N})+\frac2{N+1}J_{g}(x_{\star},x_{N})+\frac{N-1}{N+1}J_{g}(x_{N-1},x_{N})-Q_{T}.$$
For $N=1$, the separate identity is
$$V_{1}=\tfrac12J_{f}(x_{\star},x_{0})+\tfrac12J_{f}(x_{\star},x_{1})+\tfrac12J_{f}(x_{0},x_{1})+J_{g}(x_{\star},x_{1})-Q_{T}-\tfrac14\|p_{0}\|^{2}.$$
Every multiplier is nonnegative. The positive-entry transfer matrices imply $\beta_{k}\geq0$ and $v_{k}>0$. For $t>1$, $k\geq0$, $c\geq0$, the first diagonal entry of $H_{k}$ is positive, and
$$\det H_{k}=\frac{2c(t-1)((2k+1)t-k)+(k^{2}+3k+1)t+1}{4(t-1)^{3}}>0.$$
Thus $Q_{k}\geq0$; also $H_{T}\succ0$ because its leading principal minors are $1/2$ and $1/4$. The identities imply $V_{N}\leq V_{0}=0$. Consequently
$$h(x_{N})-h(x_{\star})=L\left(V_{N}+\frac{\|e_{0}\|^{2}}{4N}\right)\leq\frac{L\|e_{0}\|^{2}}{4N}\leq\frac{LR^{2}}{4N}.$$
The symbolic cells below verify every Gram, function-value and constant coordinate, including the exceptional $N=1$ case. The local analytic proof does not assert a separate global closed-form full-PEP certificate.

## Function and parameter definitions

```python
L = pf.Parameter("L")
R = pf.Parameter("R")
f = pf.SmoothConvexFunction(is_basis=True, tags=["f"], L=L)
g = pf.ConvexFunction(is_basis=True, tags=["g"])
h = (f + g).add_tag("h")
```

## Algorithm and PEP setup

```python
def make_ctx_pgm(ctx_name: str, N, **kwargs) -> pf.PEPContext:
    ctx = pf.PEPContext(ctx_name).set_as_current()
    x = pf.Vector(is_basis=True, tags=["x_0"])
    h.set_stationary_point("x_star")
    for k in range(int(N)):
        x = g.prox(x - f.grad(x) / L, 1 / L, tag=f"x_{k + 1}")
    return ctx


def get_pep_setup(N, params):
    ctx = make_ctx_pgm(f"pgm_{N}", N)
    pb = pf.PEPBuilder(ctx)
    pb.add_initial_constraint(
        ((ctx["x_0"] - ctx["x_star"]) ** 2).le(R**2, name="initial_condition")
    )
    pb.set_performance_metric(h(ctx[f"x_{N}"]) - h(ctx["x_star"]))
    return ctx, pb, f

ctx_check, pb_check, obj_check = get_pep_setup(4, {"L": 1, "R": 1})
print("Setup built for N=4; metric:", pb_check.performance_metric)
```

## Lambda and S verification
For $0\leq i<N$, the candidate cumulative multipliers are $a_{i}=(i+1)/(2N-i)$ for $f$ and $b_{i}=i/(2N-i)$ for $g$. Set $a_{-1}=0$, $a_{N}=b_{N}=1$, and $b_{0}=0$. Consecutive-pair weights are $a_{i},b_{i}$; optimizer-pair weights are their successive differences. These are candidates inferred from this sample, not an all-horizon proof.
The candidate quadratic remainder is $S_{\rm guess}=\|x_{0}-x_{\star}\|^2/(4N)-(h_{N}-h_{\star})+\sum\lambda I$, at $L=R=1$. Its agreement and the saved numerical reverse-LDL decomposition are checked below. No fixed SOS aid or pre-existing answer was consulted.

```python

# Initial formula search inferred from the two groups' rational sequences.
def candidate_lambda(owner, ri, ci, horizon):
    def cumulative(i):
        if owner == 'f':
            if i < 0:
                return sp.S.Zero
            if i == horizon:
                return sp.S.One
            return sp.Rational(i + 1, 2 * horizon - i)
        if i <= 0:
            return sp.S.Zero
        if i == horizon:
            return sp.S.One
        return sp.Rational(i, 2 * horizon - i)
    if ci == 'x_star':
        return sp.S.Zero
    j = int(ci[2:])
    if ri == 'x_star':
        return cumulative(j) - cumulative(j - 1)
    i = int(ri[2:])
    return cumulative(i) if j == i + 1 else sp.S.Zero

candidate_errors = []
weighted_guess = pf.Scalar.zero()
for c, origin, owner, group in entries:
    if origin != 'interpolation':
        continue
    ri, ci = c.name.split(':')[1].split(',')
    guess = candidate_lambda(owner, ri, ci, N)
    actual = sample['named_dual_values'][c.name]
    actual = 0.0 if actual is None else actual
    candidate_errors.append(abs(float(guess) - actual))
    weighted_guess += guess * (c.lhs - c.rhs)
print('Maximum lambda candidate error:', max(candidate_errors))
assert max(candidate_errors) <= 1e-4
for owner in ('f', 'g'):
    tags = ['x_star'] + [f'x_{j}' for j in range(0 if owner == 'f' else 1, N + 1)]
    candidate_matrix = np.array([[float(candidate_lambda(owner, ri, ci, N)) for ci in tags] for ri in tags])
    print('Candidate lambda:', owner)
    pf.pprint_labeled_matrix(candidate_matrix, tags, precision=5)
assert pb.performance_metric is not None
S_guess = sp.Rational(1, 4 * N) * (ctx['x_0'] - ctx['x_star'])**2 - pb.performance_metric + weighted_guess
S_array = np.asarray(sample['S_matrix'], dtype=float)
S_guess_eval = pm.eval_scalar(S_guess)
S_candidate_error = float(np.max(np.abs(np.asarray(S_guess_eval.inner_prod_coords, float) - S_array)))
print('S candidate coordinate discrepancy:', S_candidate_error)
assert np.allclose(S_guess_eval.inner_prod_coords, S_array, atol=1e-4, rtol=1e-5)
assert np.allclose(S_guess_eval.func_coords, 0, atol=1e-4, rtol=0)
assert abs(float(S_guess_eval.offset)) <= 1e-4
print('S eigenvalues:', np.linalg.eigvalsh(S_array))
# The candidate S is a residual formula, not yet a proof of all-horizon positivity.
# No pre-existing decomposition is permitted. Apply approved reverse LDL to this sample.
S_named = pf.MatrixWithNames(S_array, sample['S_row_names'], sample['S_col_names'])
pf.pprint_labeled_matrix(S_named, precision=5)
D, ell = ldl_decompose_with_reversed_basis(S_named, basis, print_output=False)
D = np.asarray(D, dtype=float)
E = np.asarray([pm.eval_vector(v).coords for v in ell], dtype=float)
S_numeric = sum((float(D[i,j]) * ell[i] * ell[j] for i,j in np.ndindex(D.shape)), pf.Scalar.zero())
print('Reverse-LDL diagonal:', np.diag(D))
print('Significant factors (full factors retained in state):')
for i, coefficient in enumerate(np.diag(D)):
    if abs(coefficient) > 1e-6:
        print(i, coefficient, str(ell[i]))
pf.pprint_labeled_matrix(D, [f'ell_{i}' for i in range(len(D))], precision=5)
```

### Complete numerical certificate identity
All inequalities use nonpositive residuals. Nonnegative multipliers and a positive semidefinite remainder support the upper-bound direction. The finite-sample checks include Gram, function-value and constant coordinates; no analytic full-PEP claim is made.

```python

# Complete numerical identity: metric - bound - sum(dual * residual) + S = 0.
weighted_constraints = pf.Scalar.zero()
scalar_duals = []
for c, origin, owner, group in entries:
    meta = sample['constraint_metadata'][c.name]
    value = sample['named_dual_values'][c.name]
    assert meta['origin'] == origin and meta['owner'] == owner and meta['group'] == group
    assert meta['comparator'] == c.cmp.value
    if meta['relaxed']:
        assert value is None and c.name in pb.relaxed_constraints
        continue
    assert value is not None and np.isfinite(value)
    residual = c.lhs - c.rhs
    if c.cmp in (pf.Comparator.GE, pf.Comparator.SEQ):
        residual = -residual
    weighted_constraints += float(value) * residual
    if c.cmp != pf.Comparator.EQ:
        scalar_duals.append(float(value))
checked = pm.eval_scalar(pb.performance_metric - sample['opt_value'] - weighted_constraints + S_numeric)
def max_abs(values):
    return float(np.max(np.abs(np.asarray(values, dtype=float)), initial=0))
residuals = dict(identity_gram_max_abs=max_abs(checked.inner_prod_coords),
                 identity_func_max_abs=max_abs(checked.func_coords),
                 identity_offset_abs=max_abs(checked.offset),
                 S_reconstruction_max_abs=max_abs(E.T @ D @ E - S_array),
                 dense_relaxed_gap=abs(dense['opt_value'] - sample['opt_value']),
                 minimum_scalar_inequality_dual=min(scalar_duals),
                 minimum_S_eigenvalue=float(np.linalg.eigvalsh(S_array).min()),
                 minimum_matrix_inequality_eigenvalue=0.0)
tolerances = dict(identity_atol=1e-5, identity_rtol=0.0, reconstruction_atol=1e-4,
                  reconstruction_rtol=1e-5, dual_tol=1e-6, psd_tol=1e-6, dense_opt_atol=1e-5)
assert sample['solver_status'] == 'optimal'
assert all(residuals[key] <= 1e-5 for key in ('identity_gram_max_abs', 'identity_func_max_abs', 'identity_offset_abs'))
assert np.all(np.isfinite(S_array)) and np.allclose(S_array, S_array.T, atol=1e-6, rtol=0)
assert np.allclose(E.T @ D @ E, S_array, atol=1e-4, rtol=1e-5)
assert residuals['dense_relaxed_gap'] < 1e-5
assert residuals['minimum_scalar_inequality_dual'] >= -1e-6
assert residuals['minimum_S_eigenvalue'] >= -1e-6
assert abs(sample['opt_value'] - 1/(4*N)) < 1e-5
print('Complete numerical certificate checks:', residuals)
pf.pprint_labeled_matrix(np.asarray(checked.inner_prod_coords, float), sample['basis_vectors'], precision=8)
print('Function residuals:', checked.func_coords, 'constant residual:', checked.offset)
print('Numerical certificate verified; analytic full-PEP and Lyapunov proof remain unverified.')

```

## Partial-sum Lyapunov construction and ranks
Write each normalized interpolation residual as $I_{c}\leq0$ and each reverse-LDL remainder piece as $Q_{i}\geq0$. Starting at $V_{0}=0$, assign the optimizer-to-initial interpolation and the initial LDL pieces to the first increment. At each later iterate, assign the new optimizer and consecutive interpolation terms, and the two LDL pieces whose latest gradient has that iterate index:
$$V_{k+1}-V_{k}=\sum_{c\text{ assigned at }k}\lambda_{c} I_{c}-\sum_{i\text{ assigned at }k}Q_{i}\leq0.$$
The construction includes every interpolation contribution and every remainder piece. The normalization constraint is reserved for the final boundary argument. Both $N=4$ and $N=6$ have interior rank four when one index at each end is excluded; this is finite-sample evidence, not an analytic theorem. Small numerical pieces are retained.

```python
import hashlib
import json
from pathlib import Path
import numpy as np
import sympy as sp
import pepflow as pf
from pepflow.constraint import ScalarConstraint
from pepflow.operator import LinearOperatorTranspose
from pepflow.pep_result import MatrixWithNames
from pepflow.utils import name_to_vector_tuple

certificate = json.loads((notebook_dir / 'state/pgm_b2.json').read_text())
REPO_ROOT = repo_root
N_int = certificate['N_verify']
params_sp: dict[str, pf.NUMERICAL_TYPE] = {'L': sp.S.One, 'R': sp.S.One}
ctx, pb, obj = saved_get_pep_setup(N_int, params_sp)
pb.set_relaxed_constraints(certificate['relaxed_constraints'])
certificate_checked = False
if not certificate.get("numerical_certificate_verified", False):
    raise ValueError("Recover or verify the complete numerical certificate in Block 2")
if not certificate.get("closed_form_search", {}).get("completed", False):
    raise ValueError("Complete or recover Block 2's initial closed-form attempt")
certificate_keys = (
    "N_verify", "params_json", "relaxed_constraints",
    "S_matrix", "S_row_names", "S_col_names", "basis_vectors", "basis_scalars",
    "opt_value", "tau_sol", "tau_name", "named_dual_values", "constraint_metadata",
    "normalization", "certificate_provenance", "numerical_checks",
    "numerical_decomposition",
)
payload = {key: certificate[key] for key in certificate_keys}
digest = hashlib.sha256(json.dumps(
    payload, sort_keys=True, separators=(",", ":"), allow_nan=False
).encode("utf-8")).hexdigest()
if digest != certificate["certificate_id"]:
    raise ValueError("Certificate components changed; recover one coherent certificate")
provenance = certificate["certificate_provenance"]
if provenance["pep_type"] != "primal":
    raise ValueError("Recover and verify the certificate's actual dual sign convention")
for relative_path, expected_hash in provenance["source_hashes"].items():
    source_path = (notebook_dir / relative_path).resolve()
    if not source_path.is_relative_to(notebook_dir):
        raise ValueError("Certificate source path must stay in the selected repository")
    if hashlib.sha256(source_path.read_bytes()).hexdigest() != expected_hash:
        raise ValueError(f"Certificate source changed: {relative_path}")
if not Path(pf.__file__).resolve().is_relative_to(REPO_ROOT):
    raise ValueError("The imported PEPFlow source is not the selected repository")
if N_int != certificate["N_verify"]:
    raise ValueError("Certificate horizon differs from the rebuilt setup")
saved_params = json.loads(certificate["params_json"])
expected_params = {
    key: (sp.Rational(value) if isinstance(value, str) else
          sp.nsimplify(value, rational=True, tolerance=1e-9)
          if isinstance(value, float) else sp.S(value))
    for key, value in saved_params.items()
}
if params_sp != expected_params:
    raise ValueError("Certificate parameters differ from the rebuilt setup")
if not certificate["numerical_checks"]["passed"]:
    raise ValueError("Block 2 numerical checks did not pass")
tol = certificate["numerical_checks"]["tolerances"]

records = [(c, "initial", None, None) for c in pb.init_conditions]
owners = list(ctx.func_to_triplets) + [
    owner for owner in ctx.oper_to_duplets
    if not isinstance(owner, LinearOperatorTranspose)
]
for owner in owners:
    data = ctx.get_constraint_data(owner)
    for group, family in data.sc_dict.items():
        records.extend((c, "interpolation", str(owner), str(group)) for c in family)
    records.extend((c, "interpolation", str(owner), str(group))
                   for group, c in data.psd_dict.items())
constraints_by_name = {c.name: c for c, _, _, _ in records}
if len(constraints_by_name) != len(records):
    raise ValueError("Constraint names must identify one unique contribution")
saved_meta = certificate["constraint_metadata"]
saved_duals = certificate["named_dual_values"]
if set(constraints_by_name) != set(saved_meta) or set(saved_duals) != set(saved_meta):
    raise ValueError("The saved certificate does not cover the rebuilt constraints")
relaxed = set(certificate["relaxed_constraints"])
if set(pb.relaxed_constraints) != relaxed or not relaxed.issubset(constraints_by_name):
    raise ValueError("The selected relaxation differs from the saved certificate")
if pb.ctx is not ctx:
    raise ValueError("The builder and expression context differ")
basis = list(ctx.basis_vectors())
basis_labels = [repr(v) for v in basis]
scalar_labels = [repr(v) for v in ctx.basis_scalars()]
if basis_labels != certificate["basis_vectors"] or scalar_labels != certificate["basis_scalars"]:
    raise ValueError("Certificate coordinate order differs from the rebuilt context")
if certificate["S_row_names"] != basis_labels or certificate["S_col_names"] != basis_labels:
    raise ValueError("S labels do not match the rebuilt basis")
S_array = np.asarray(certificate["S_matrix"], dtype=float)
if S_array.shape != (len(basis), len(basis)) or not np.all(np.isfinite(S_array)):
    raise ValueError("Invalid S shape or entries")
if not np.allclose(S_array, S_array.T, atol=tol["psd_tol"], rtol=0):
    raise ValueError("S is not numerically symmetric")
if len(basis) and np.linalg.eigvalsh((S_array + S_array.T) / 2).min() < -tol["psd_tol"]:
    raise ValueError("S failed the recorded PSD tolerance")
S_sol = MatrixWithNames(S_array, basis_labels, basis_labels)
pm = pf.ExpressionManager(ctx, resolve_parameters=params_sp)
S_scalar = sum((float(S_array[i, j]) * basis[i] * basis[j]
                for i, j in np.ndindex(S_array.shape)), pf.Scalar.zero())

weighted_terms = {}
interpolation_terms = {}
interpolation_pairs = {}
extra_terms = {}
extra_duals = {}
for constraint, origin, owner, group in records:
    name = constraint.name
    kind = "scalar" if isinstance(constraint, ScalarConstraint) else "psd"
    residual = constraint.lhs - constraint.rhs
    shape = [] if kind == "scalar" else list(np.shape(residual))
    expected_meta = dict(origin=origin, owner=owner, group=group, kind=kind,
                         comparator=constraint.cmp.value, shape=shape,
                         relaxed=name in relaxed)
    if any(saved_meta[name].get(key) != value for key, value in expected_meta.items()):
        raise ValueError(f"Constraint metadata changed: {name}")
    value = saved_duals[name]
    if name in relaxed:
        if value is not None:
            raise ValueError(f"A relaxed constraint must have no retained multiplier: {name}")
        continue
    if value is None:
        raise ValueError(f"Missing unrelaxed dual: {name}")
    dual = np.asarray(value, dtype=float)
    if list(dual.shape) != shape or not np.all(np.isfinite(dual)):
        raise ValueError(f"Invalid saved dual: {name}")
    # Normalize inequalities to residual <= 0; primal equality duals keep their sign.
    if constraint.cmp.name in ("GE", "SEQ"):
        residual = -residual
    if kind == "scalar":
        if constraint.cmp.name != "EQ" and float(dual) < -tol["dual_tol"]:
            raise ValueError(f"Negative inequality multiplier: {name}")
        contribution = pf.Scalar.zero() + float(dual) * residual
    else:
        if constraint.cmp.name != "EQ" and not np.allclose(dual, dual.T, atol=tol["psd_tol"], rtol=0):
            raise ValueError(f"Nonsymmetric matrix dual: {name}")
        if constraint.cmp.name != "EQ" and dual.size and np.linalg.eigvalsh((dual + dual.T) / 2).min() < -tol["psd_tol"]:
            raise ValueError(f"Matrix dual is not PSD: {name}")
        contribution = sum((float(dual[i, j]) * residual[j, i]
                            for i, j in np.ndindex(dual.shape)), pf.Scalar.zero())
    weighted_terms[name] = contribution
    if origin == "interpolation" and kind == "scalar":
        interpolation_terms[name] = contribution
        try:
            interpolation_pairs[name] = name_to_vector_tuple(name)
        except (ValueError, IndexError):
            interpolation_pairs[name] = None  # Assign this term explicitly when grouping.
    elif name != certificate["tau_name"]:
        extra_terms[name] = contribution
        extra_duals[name] = value

tau_name = certificate["tau_name"]
if (tau_name not in weighted_terms or saved_meta[tau_name]["origin"] != "initial"
        or saved_meta[tau_name]["kind"] != "scalar" or tau_name in relaxed
        or certificate["normalization"]["tau_name"] != tau_name):
    raise ValueError("Missing or inconsistent bound normalization constraint")
if not np.allclose(saved_duals[tau_name], certificate["tau_sol"], atol=tol["dual_tol"], rtol=0):
    raise ValueError("The saved normalization dual is inconsistent")
if not np.isfinite(certificate["opt_value"]) or not np.allclose(
    certificate["normalization"]["bound_value"], certificate["opt_value"],
    atol=tol["identity_atol"], rtol=tol["identity_rtol"]
):
    raise ValueError("The saved bound/normalization is inconsistent")
assert pb.performance_metric is not None
assert pb.performance_metric is not None
certificate_diff = (pb.performance_metric - float(certificate["opt_value"])
                    - sum(weighted_terms.values(), pf.Scalar.zero()) + S_scalar)
evaluated_diff = pm.eval_scalar(certificate_diff)
for label, coords in (
    ("Gram", evaluated_diff.inner_prod_coords),
    ("function", evaluated_diff.func_coords),
    ("offset", np.asarray(evaluated_diff.offset)),
):
    if not np.allclose(np.asarray(coords, dtype=float), 0,
                       atol=tol["identity_atol"], rtol=tol["identity_rtol"]):
        raise ValueError(f"Numerical certificate {label} identity failed")
decomposition = certificate["numerical_decomposition"]
D_saved = np.asarray(decomposition["D"], dtype=float)
E_saved = np.asarray(decomposition["ell_coords"], dtype=float)
if decomposition["basis_labels"] != basis_labels or E_saved.ndim != 2:
    raise ValueError("Numerical S decomposition basis differs")
if E_saved.shape[1] != len(basis) or D_saved.shape != (E_saved.shape[0], E_saved.shape[0]):
    raise ValueError("Numerical S decomposition shape differs")
if not np.all(np.isfinite(D_saved)) or not np.all(np.isfinite(E_saved)):
    raise ValueError("Numerical S decomposition contains nonfinite values")
if not np.allclose(E_saved.T @ D_saved @ E_saved, S_array,
                   atol=tol["reconstruction_atol"], rtol=tol["reconstruction_rtol"]):
    raise ValueError("Numerical S decomposition does not reconstruct this certificate")
certificate_checked = True
print("Numerical certificate checked for this N and parameter sample.")
```

### Numerical remainder pieces
Reuse the selected complete LDL decomposition. Its reconstruction and piece signs are checked before grouping.

```python
# Reuse the saved complete reverse-LDL factors; never truncate small pieces.
D = np.asarray(certificate['numerical_decomposition']['D'], dtype=float)
E = np.asarray(certificate['numerical_decomposition']['ell_coords'], dtype=float)
assert np.array_equal(D, np.diag(np.diag(D)))
ell = [sum((float(coef)*vector for coef,vector in zip(row,basis)), pf.Vector.zero()) for row in E]
S_piece_terms = {i: float(D[i,i])*ell[i]**2 for i in range(len(D))}
S_assignment = {i: N_int-i//2-1 if i < 2*N_int else -1 for i in S_piece_terms}
assert np.allclose(E.T@D@E, S_array, atol=1e-4, rtol=1e-5)
for piece in S_piece_terms.values():
    evaluated_piece = pm.eval_scalar(piece)
    assert np.allclose(evaluated_piece.func_coords, 0, atol=1e-5, rtol=0)
    assert abs(float(evaluated_piece.offset)) <= 1e-5
    assert np.linalg.eigvalsh(np.asarray(evaluated_piece.inner_prod_coords,float)).min() >= -1e-6
```

```python

# Nonpositive interpolation residuals minus nonnegative S pieces give V_(k+1)<=V_k.
grouping_checked = False
group_terms = {name:term for name,term in weighted_terms.items() if name != tau_name}
term_assignment = {}
for name in group_terms:
    pair = interpolation_pairs[name]
    assert pair is not None
    ri, ci = pair
    j = int(ci[2:])
    assert ri == 'x_star' or int(ri[2:])+1 == j
    term_assignment[name] = j-1
assert set(term_assignment) == set(group_terms)
assert set(S_assignment) == set(S_piece_terms)
rank_tolerance = 1e-4
rank_left, rank_right = 1, 1
lyap = [pf.Scalar.zero()]
partial_sum = sum((term for name,term in group_terms.items() if term_assignment[name] == -1), pf.Scalar.zero())
partial_sum -= sum((piece for i,piece in S_piece_terms.items() if S_assignment[i] == -1), pf.Scalar.zero())
for k in range(N_int):
    partial_sum += sum((term for name,term in group_terms.items() if term_assignment[name] == k), pf.Scalar.zero())
    partial_sum -= sum((piece for i,piece in S_piece_terms.items() if S_assignment[i] == k), pf.Scalar.zero())
    lyap.append(partial_sum)
ranks = []
singular_values = []
for k,Vk in enumerate(lyap):
    matrix = np.asarray(pm.eval_scalar(Vk).inner_prod_coords, float)
    singular = np.linalg.svd(matrix, compute_uv=False)
    singular_values.append(singular.tolist())
    rank = int(np.sum(singular > rank_tolerance))
    ranks.append(rank)
    print(f'rank V_{k}: {rank}')
    if k == 0:
        print()
retained_ranks = ranks[rank_left:N_int+1-rank_right]
print('Interior rank is constant at this N:',len(retained_ranks)>=3 and len(set(retained_ranks))==1,'boundary exclusions:',(rank_left,rank_right))
# No interpolation or S piece is omitted or held at the terminal boundary.
coverage_diff = lyap[N_int] - (sum(group_terms.values(), pf.Scalar.zero()) - S_scalar)
piece_diff = sum(S_piece_terms.values(), pf.Scalar.zero()) - S_scalar
checks = [('coverage',coverage_diff), ('S pieces',piece_diff)]
for k in range(N_int):
    targets = {-1,0} if k == 0 else {k}
    increment = sum((term for name,term in group_terms.items() if term_assignment[name] in targets),pf.Scalar.zero())
    increment -= sum((piece for i,piece in S_piece_terms.items() if S_assignment[i] in targets),pf.Scalar.zero())
    checks.append((f'increment_{k}',lyap[k+1]-lyap[k]-increment))
coverage_residuals = {}
for label, expression in checks:
    evaluated = pm.eval_scalar(expression)
    residual = max(float(np.max(np.abs(np.asarray(coords,float)),initial=0)) for coords in
                   (evaluated.inner_prod_coords,evaluated.func_coords,np.asarray(evaluated.offset)))
    coverage_residuals[label] = residual
    assert residual <= 1e-5
assert all(value is not None and value >= -1e-6 for name,value in saved_duals.items() if name in group_terms)
grouping_checked = True
print('Full-coordinate coverage and increment checks:',coverage_residuals)

```

### Cross-horizon check
Both observations were recomputed from their own complete certificates using the identical saved construction. The N=6 selected sparse solve was optimal; its dense comparison solve was optimal_inaccurate, with the status retained. Selected-certificate identities and signs passed the unchanged tolerances at both horizons. Singular values are saved with each observation, including values below the rank threshold.

```python
def assess_rank_pattern(samples, selected_certificate_id, left=1, right=1):
    if type(left) is not int or type(right) is not int or not (0 <= left <= 2 and 0 <= right <= 2):
        raise ValueError("Exclude at most two indices at each end")
    selected = [s for s in samples if s["certificate_id"] == selected_certificate_id]
    if len(selected) != 1:
        raise ValueError("Select exactly one current certificate observation")
    selected = selected[0]
    family = [s for s in samples if s["recipe_id"] == selected["recipe_id"]]
    if any(s["rank_convention"] != selected["rank_convention"] for s in family):
        raise ValueError("Recompute the candidate with comparable rank conventions")
    if len({s["N"] for s in family}) != len(family):
        raise ValueError("Compare one selected trial per horizon for this recipe")
    eligible, unresolved, inconsistent = [], [], False
    for sample in family:
        n, profile = sample["N"], sample["rank_profile"]
        if type(n) is not int or n < 1 or len(profile) != n + 1:
            raise ValueError("Rank profile must contain V_0 through V_N")
        if any(type(r) is not int or r < 0 for r in profile):
            raise ValueError("Ranks must be nonnegative integers")
        retained = profile[left:n + 1 - right]
        if not (sample["certificate_checked"] is True and sample["grouping_checked"] is True):
            unresolved.append({"N": n, "reason": "unverified_certificate_or_grouping"})
        elif sample["rank_credible"] is not True:
            unresolved.append({"N": n, "reason": "ambiguous_rank"})
        elif len(retained) < 3:
            unresolved.append({"N": n, "reason": "insufficient_interior_indices"})
        elif len(set(retained)) != 1:
            inconsistent = True
            unresolved.append({"N": n, "reason": "nonconstant_interior_rank"})
        else:
            eligible.append((n, retained[0], sample["certificate_id"]))
    common_ranks = {r for _, r, _ in eligible}
    inconsistent = inconsistent or len(common_ranks) > 1
    selected_eligible = any(cid == selected_certificate_id for _, _, cid in eligible)
    confirmed = not inconsistent and len(eligible) >= 2 and selected_eligible
    return {
        "status": "confirmed" if confirmed else "inconsistent" if inconsistent else "insufficient_evidence",
        "boundary_exclusions": {"left": left, "right": right},
        "rank": next(iter(common_ranks)) if len(common_ranks) == 1 else None,
        "confirmed_horizons": sorted(n for n, _, _ in eligible) if confirmed else [],
        "selected_certificate_id": selected_certificate_id,
        "recipe_id": selected["recipe_id"],
        "rank_convention": selected["rank_convention"],
        "unresolved_samples": unresolved,
    }

rank_samples = [json.loads((notebook_dir / f'supplementary/sample_n{n}/observation.json').read_text()) for n in (4, 6)]
rank_search = assess_rank_pattern(rank_samples, certificate['certificate_id'], left=1, right=1)
assert rank_search['status'] == 'confirmed'
assert rank_samples[0]['rank_profile'] == ranks
for observation in rank_samples:
    print('N=', observation['N'], 'ranks=', observation['rank_profile'], 'credible=', observation['rank_credible'])
print('Confirmed interior rank:', rank_search['rank'], 'horizons:', rank_search['confirmed_horizons'])
print('Absolute threshold 1e-4 in the setup coordinate order; fixed exclusions (1,1).')
```

## Identify the vectors composing the Lyapunov function
Reconstruct the selected Block 3 partial sums and seek interpretable spanning vectors. All endpoint cases remain included.

```python
from itertools import combinations
from pepflow.lyapunov_utils import vectors_in_column_space, find_basis_with_sparsest_coefficients, find_symmetric_coefficient_matrix
from pgm_setup import f as setup_f, g as setup_g

b3 = json.loads((notebook_dir / 'state/pgm_b3.json').read_text())
assert b3['certificate_id'] == certificate['certificate_id']
assert ranks == b3['rank_profile']
rank_evidence = json.loads((notebook_dir / b3['rank_search']['evidence_path']).read_text())
assert rank_evidence['summary'] == b3['rank_search']
assert b3['rank_search']['boundary_exclusions'] == {'left': 1, 'right': 1}
for k, value in enumerate(lyap):
    evaluated = pm.eval_scalar(value)
    assert np.allclose(evaluated.inner_prod_coords, b3['lyap_inner_prod_coords'][k], atol=1e-4, rtol=1e-5)
    assert np.allclose(np.asarray(evaluated.func_coords,float), b3['lyap_func_coords'][k], atol=1e-4, rtol=1e-5)
    assert abs(float(evaluated.offset)-b3['lyap_offsets'][k]) <= 1e-4
print('Selected ranks:', ranks, '; interior 1 <= k <= N-1; endpoints checked separately.')
```

### Candidate-vector scan
Candidates include initial/current point gaps, gradient differences, the optimizer gradient, all basis vectors and tagged iterates, and pairwise differences. Duplicate coordinates and zero vectors are removed.

```python

candidate_pairs = [('e_0', ctx['x_0']-ctx['x_star']), ('s_star', setup_f.grad(ctx['x_star']))]
for j in range(1,N_int+1):
    candidate_pairs += [(f'e_{j}',ctx[f'x_{j}']-ctx['x_star']),
                        (f'p_{j}',setup_f.grad(ctx[f'x_{j}'])-setup_f.grad(ctx['x_star']))]
candidate_pairs += [(str(v),v) for v in ctx.basis_vectors()]
candidate_pairs += [(f'x_{j}',ctx[f'x_{j}']) for j in range(N_int+1)]
base_pairs = list(candidate_pairs)
for (label1,v1),(label2,v2) in combinations(base_pairs,2):
    candidate_pairs.append((f'({label1})-({label2})',v1-v2))
unique_pairs = []
coordinate_keys = set()
for label, vector in candidate_pairs:
    coords = np.asarray(pm.eval_vector(vector).coords,float)
    if np.linalg.norm(coords) <= 1e-12:
        continue
    key = tuple(np.round(coords,12))
    if key not in coordinate_keys:
        unique_pairs.append((label,vector))
        coordinate_keys.add(key)
print('Distinct nonzero candidates:',len(unique_pairs))
```

```python

in_column = {}
for k in range(1,N_int+1):
    in_column[k] = vectors_in_column_space(lyap[k],[v for _,v in unique_pairs],pep_context=ctx,
                                          resolve_parameters=params_sp,rtol=1e-4,atol=1e-4)
    print(f'V_{k} column-space candidates:')
    for label,vector in unique_pairs:
        if any(vector is selected for selected in in_column[k]):
            print('   ',label)
```

### Selected basis pattern
Run the sparsity search on the actual column-space candidates at each nonzero-rank index. Then adopt a uniform order using the optimizer gradient explicitly, because it makes the expression consistent across indices.

```python

sparse_bases = {}
sparse_coeffs = {}
for k in range(1,N_int+1):
    sparse_basis, sparse_C = find_basis_with_sparsest_coefficients(lyap[k],in_column[k],pep_context=ctx,
        resolve_parameters=params_sp,zero_tol=1e-6,indep_tol=1e-7,span_tol=1e-5)
    assert len(sparse_basis) == ranks[k]
    sparse_bases[k],sparse_coeffs[k] = sparse_basis,sparse_C
    labels = [next(label for label,v in unique_pairs if v is selected) for selected in sparse_basis]
    print(f'k={k}: rank {ranks[k]}; sparse basis: {labels}')
    print(f'zero coefficients: {int(np.sum(np.abs(sparse_C)<1e-6))}/{sparse_C.size}')
```

Define $e_{k}=x_{k}-x_{\star}$, $p_{k}=\nabla f(x_{k})-\nabla f(x_{\star})$, and $s_{\star}=\nabla f(x_{\star})$. The adopted ordered basis is
$$\mathcal B_{k}=[e_{0},e_{k},p_{k},s_{\star}]\quad(1\leq k<N),\qquad \mathcal B_{0}=[],\quad\mathcal B_{N}=[e_{0}].$$
This choice is independently checked below, including the initial and terminal ranks.

```python

# Adopt one uniform interpretable order even if a sparse search returns differences.
def V_k_basis(k):
    if k == 0:
        return []
    if k == N_int:
        return [ctx['x_0']-ctx['x_star']]
    return [ctx['x_0']-ctx['x_star'],ctx[f'x_{k}']-ctx['x_star'],
            setup_f.grad(ctx[f'x_{k}'])-setup_f.grad(ctx['x_star']),setup_f.grad(ctx['x_star'])]

def V_k_basis_labels(k):
    if k == 0:
        return []
    if k == N_int:
        return ['e_0']
    return ['e_0',f'e_{k}',f'p_{k}','s_star']

for k in range(N_int+1):
    vectors = V_k_basis(k)
    coordinate_matrix = np.column_stack([np.asarray(pm.eval_vector(v).coords,float) for v in vectors]) if vectors else np.zeros((len(ctx.basis_vectors()),0))
    basis_rank = int(np.linalg.matrix_rank(coordinate_matrix,tol=1e-7))
    assert basis_rank == len(vectors) == ranks[k]
    print(f'k={k}: rank {basis_rank} basis {V_k_basis_labels(k)}')
```

### Coefficient matrices
The symmetric matrix uses the basis order above; off-diagonal terms occur twice. Function gaps and offsets are checked separately. Coefficient fitting is deferred: c_ij(k,N,L), a(k,N,L), b(k,N,L), and terminal t00(N,L), tF(N,L), tG(N,L) are analytic unknowns.

```python

coeff_by_k = {}
coefficient_residuals = {}
function_sample_coefficients = {}
for k in range(N_int+1):
    vectors = V_k_basis(k)
    target = pm.eval_scalar(lyap[k])
    matrix = np.asarray(target.inner_prod_coords,float)
    if not vectors:
        C = np.zeros((0,0))
        reconstructed = np.zeros_like(matrix)
    else:
        C = find_symmetric_coefficient_matrix(lyap[k],vectors,pep_context=ctx,
                                            resolve_parameters=params_sp,indep_tol=1e-7,span_tol=1e-5)
        B = np.column_stack([np.asarray(pm.eval_vector(v).coords,float) for v in vectors])
        reconstructed = B@C@B.T
    residual = float(np.linalg.norm(reconstructed-matrix,ord='fro'))
    assert residual <= 1e-5*max(1.0,float(np.linalg.norm(matrix,ord='fro')))
    assert np.array_equal(C,C.T)
    coeff_by_k[k] = C
    coefficient_residuals[k] = residual
    print(f'k={k}: quadratic reconstruction residual {residual:.3e}')
    if vectors:
        pf.pprint_labeled_matrix(C,V_k_basis_labels(k),precision=7)
    if k == 0:
        function_sample_coefficients[k] = [0.0,0.0]
        assert np.allclose(np.asarray(target.func_coords,float),0,atol=1e-5,rtol=0)
    else:
        Fgap = setup_f(ctx[f'x_{k}'])-setup_f(ctx['x_star'])
        Ggap = setup_g(ctx[f'x_{k}'])-setup_g(ctx['x_star'])
        Fbasis = np.column_stack([np.asarray(pm.eval_scalar(term).func_coords,float) for term in (Fgap,Ggap)])
        weights = np.linalg.lstsq(Fbasis,np.asarray(target.func_coords,float),rcond=None)[0]
        assert np.allclose(Fbasis@weights,np.asarray(target.func_coords,float),atol=1e-5,rtol=0)
        function_sample_coefficients[k] = weights.tolist()
        print('Function coefficients [f_k-f_star,g_k-g_star]:',weights)
    assert abs(float(target.offset)) <= 1e-5
    print('Constant offset:',float(target.offset))
```

```python

# Full expression assembly: quadratic form plus identified function gaps.
def V_k_structure(k):
    if k == 0:
        return pf.Scalar.zero()
    vectors = V_k_basis(k)
    C = coeff_by_k[k]
    quadratic = sum((float(C[i,j])*vectors[i]*vectors[j] for i,j in np.ndindex(C.shape)),pf.Scalar.zero())
    f_weight,g_weight = function_sample_coefficients[k]
    return quadratic+f_weight*(setup_f(ctx[f'x_{k}'])-setup_f(ctx['x_star']))+g_weight*(setup_g(ctx[f'x_{k}'])-setup_g(ctx['x_star']))

component_residuals = {}
for k in range(N_int+1):
    residual = pm.eval_scalar(V_k_structure(k)-lyap[k])
    target = pm.eval_scalar(lyap[k])
    component_residuals[k] = {}
    for name,coords,target_coords in [('gram',residual.inner_prod_coords,target.inner_prod_coords),
                                      ('function',residual.func_coords,target.func_coords),
                                      ('offset',residual.offset,target.offset)]:
        error = float(np.max(np.abs(np.asarray(coords,float)),initial=0))
        scale = float(np.max(np.abs(np.asarray(target_coords,float)),initial=0))
        assert error <= 1e-4+1e-5*scale
        component_residuals[k][name] = error
    print(f'k={k}: complete reconstruction',component_residuals[k])
print('All indices reconstructed. Coefficient functions remain analytically unresolved.')
```

### Block 4 conclusion
The full interior structure is
$$V_{k}=\sum_{i,j=0}^3c_{ij}(k,N,L)\langle (\mathcal B_{k})_{i},(\mathcal B_{k})_{j}\rangle+a(k,N,L)(f(x_{k})-f(x_{\star}))+b(k,N,L)(g(x_{k})-g(x_{\star})).$$
At $k=0$, $V_{0}=0$. At $k=N$, the observed structure is $t_{00}\|e_{0}\|^2+t_{F}(f_{N}-f_{\star})+t_{G}(g_{N}-g_{\star})$. No scalar offset is present. Every component is numerically reconstructed at every selected index. These sample coefficients are not an analytic formula; Block 5 must derive them and prove the exact step, base, boundary and sign conditions.

## Closed-form coefficient derivation
The B4 coefficients are now resolved. The finite product is an explicit solution, with no unknown coefficient functions or recursive evaluation required. Compare it with all selected numerical matrices and function-gap coefficients.

```python
# Closed-form coefficient solution as an explicit finite matrix product.
b4 = json.loads((notebook_dir / 'state/pgm_b4.json').read_text())
assert b4['certificate_id'] == certificate['certificate_id']
assert b4['numerical_reconstruction']['certificate_id'] == certificate['certificate_id']

def beta_closed(index, horizon):
    if index in (0,horizon):
        return sp.S.Zero
    assert 1 <= index <= horizon-1
    terminal = sp.Matrix([sp.Rational(horizon-1,2*(horizon+1)**2),1])
    product = sp.eye(2)
    for j in range(index,horizon-1):
        t = 2*horizon-j
        product *= sp.Matrix([[2*j*(j+1)*t*(t-1)**2,j*(j+1)*(t-1)*(j+t+1)],
                              [4*t*t*(t-1)*((2*j+1)*t-j),2*t*t*((j*j+3*j+1)*t+1)]])
    value = product*terminal
    return sp.cancel(value[0]/value[1])

def analytic_C(index,horizon,smoothness=1):
    ell = sp.sympify(smoothness)
    if index == 0:
        return sp.zeros(0)
    if index == horizon:
        return sp.Matrix([[-sp.Rational(1,4*horizon)]])
    t = 2*horizon-index
    Cmat = sp.zeros(4)
    Cmat[0,0] = -sp.Rational(1,4*horizon)
    Cmat[1,1] = sp.Rational(horizon-index,t*t)
    Cmat[1,2] = Cmat[2,1] = -1/(2*ell*t)
    Cmat[1,3] = Cmat[3,1] = -1/(2*ell*t)
    Cmat[2,2] = -beta_closed(index,horizon)/ell**2
    return Cmat

for j in range(N_int+1):
    closed = np.asarray(analytic_C(j,N_int),float).reshape(coeff_by_k[j].shape)
    error = float(np.max(np.abs(closed-coeff_by_k[j]),initial=0))
    assert error <= 1e-4
    print(f'k={j}: analytic/sample matrix discrepancy {error:.3e}; beta={beta_closed(j,N_int)}')
    if closed.size:
        pf.pprint_labeled_matrix(np.asarray(analytic_C(j,N_int),dtype=object),V_k_basis_labels(j),precision=7)
    weights = [0,0] if j==0 else [1,1] if j==N_int else [(j+1)/(2*N_int-j),j/(2*N_int-j)]
    assert np.allclose(weights,function_sample_coefficients[j],atol=1e-4,rtol=1e-5)
```

### Transfer formula and coefficient equation
The terminal condition is $\beta_{N-1}=(N-1)/(2(N+1)^{2})$. Eliminating the future-gradient block gives
$$\beta_{k}=\frac{k(k+1)(t-1)\bigl(2t(t-1)\beta_{k+1}+k+t+1\bigr)}{2t^{2}\bigl(2(t-1)((2k+1)t-k)\beta_{k+1}+(k^{2}+3k+1)t+1\bigr)},\qquad t=2N-k.$$
The matrix product in the theorem solves this relation: splitting its first factor gives exactly the ratio identity checked below. Its last empty product is the terminal condition. The recurrence evaluates to zero at $k=0$ as required for the initial cancellation. The next cell derives the relation from the actual local quadratic remainder rather than fitting samples.

```python

k_sym,N_sym = sp.symbols('k N',integer=True,nonnegative=True)
t_sym = sp.symbols('t',positive=True)
c_next = sp.symbols('c_next',nonnegative=True)
u_sym,v_sym = sp.symbols('u v',positive=True)

def beta_recursion(k,t,c):
    return sp.cancel(k*(k+1)*(t-1)*(2*c*t*(t-1)+k+t+1) /
                     (2*t*t*(2*c*(t-1)*((2*k+1)*t-k)+(k*k+3*k+1)*t+1)))

def local_blocks(k,t,c):
    H = sp.Matrix([[c+(k+2)/(2*(t-1)),(k+1)/(2*(t-1))],
                   [(k+1)/(2*(t-1)),((2*k+1)*t-k)/(2*(t-1)**2)]])
    K = sp.Matrix([[-(k+1)/(2*t*(t-1)),(k+1)/(2*t*(t-1))],
                   [-((2*k+1)*t-k)/(2*t*(t-1)**2),(k*t+1)/(2*(t-1)**2)]])
    return H,K

M_transfer = sp.Matrix([[2*k_sym*(k_sym+1)*t_sym*(t_sym-1)**2,k_sym*(k_sym+1)*(t_sym-1)*(k_sym+t_sym+1)],
                       [4*t_sym*t_sym*(t_sym-1)*((2*k_sym+1)*t_sym-k_sym),2*t_sym*t_sym*((k_sym*k_sym+3*k_sym+1)*t_sym+1)]])
transferred = M_transfer*sp.Matrix([u_sym,v_sym])
assert sp.cancel(transferred[0]/transferred[1]-beta_recursion(k_sym,t_sym,u_sym/v_sym)) == 0
H_symbolic,K_symbolic = local_blocks(k_sym,t_sym,c_next)
det_expected = (2*c_next*(t_sym-1)*((2*k_sym+1)*t_sym-k_sym)+(k_sym*k_sym+3*k_sym+1)*t_sym+1)/(4*(t_sym-1)**3)
assert sp.cancel(H_symbolic.det()-det_expected) == 0
print('Exact transfer-ratio identity: zero; positive determinant formula:',sp.factor(det_expected))
print('beta_0 recurrence:',sp.cancel(beta_recursion(0,t_sym,c_next)))
```

```python
# Derivation: eliminate the two future-gradient directions from the local quadratic remainder.
beta_unknown = sp.symbols('beta_unknown',real=True)
e_alg,p_alg,q_alg,r_alg = sp.symbols('e p q r',real=True)
t_alg = t_sym
k_alg = k_sym
next_e = e_alg-p_alg-q_alg
old_quad = (t_alg-k_alg)*e_alg**2/(2*t_alg**2)-e_alg*p_alg/t_alg-beta_unknown*p_alg**2
new_quad = (t_alg-k_alg-2)*next_e**2/(2*(t_alg-1)**2)-next_e*r_alg/(t_alg-1)-c_next*r_alg**2
a_alg=(k_alg+1)/t_alg
b_alg=k_alg/t_alg
an_alg=(k_alg+2)/(t_alg-1)
bn_alg=(k_alg+1)/(t_alg-1)
weighted_quad = ((an_alg-a_alg)*(-r_alg*next_e+r_alg**2/2)
 +a_alg*(r_alg*(e_alg-next_e)+(p_alg-r_alg)**2/2)
 +(bn_alg-b_alg)*(-q_alg*next_e)+b_alg*q_alg*(e_alg-next_e))
remainder_matrix = (sp.hessian(weighted_quad-new_quad+old_quad,[r_alg,q_alg,e_alg,p_alg])/2).applyfunc(sp.cancel)
assert all(sp.cancel(v)==0 for v in remainder_matrix[:2,:2]-H_symbolic)
assert all(sp.cancel(v)==0 for v in remainder_matrix[:2,2:]-K_symbolic)
schur_residual = (remainder_matrix[2:,2:]-K_symbolic.T*H_symbolic.inv()*K_symbolic).applyfunc(sp.cancel)
assert schur_residual[0,0]==schur_residual[0,1]==schur_residual[1,0]==0
beta_solution = sp.solve(schur_residual[1,1],beta_unknown)[0]
assert sp.cancel(beta_solution-beta_recursion(k_sym,t_sym,c_next))==0
print('Solved coefficient equation:',sp.Eq(beta_unknown,beta_recursion(k_sym,t_sym,c_next)))
print('The finite product above solves this recurrence with beta_(N-1)=(N-1)/(2(N+1)^2).')
```

### Closed-form Lyapunov conclusion
For $1\leq k<N$ the resulting expression is
$$V_{k}=-\frac{\|e_{0}\|^{2}}{4N}+\frac{k+1}{L(2N-k)}(f_{k}-f_{\star})+\frac{k}{L(2N-k)}(g_{k}-g_{\star})+\frac{N-k}{(2N-k)^{2}}\|e_{k}\|^{2}-\frac{\langle e_{k},\nabla f(x_{k})\rangle}{L(2N-k)}-\beta_{k}\|p_{k}\|^{2},$$
with the explicit transfer-product coefficient from the theorem, $V_{0}=0$, and $V_{N}=(h_{N}-h_{\star})/L-\|e_{0}\|^{2}/(4N)$. The next sections verify the exact local identities over symbolic index and parameter domains.

## Symbolic recursion contexts
The following definitions create the actual proximal-gradient recursion, sum stationarity, normalized gradients and quadratic block remainders in the approved PEPFlow framework.

```python

# All following identities are actual PEPFlow Scalar expressions.
L_sym = sp.symbols('L',positive=True)

def local_context(label,with_current_g=True):
    context = pf.PEPContext(label).set_as_current()
    Lpar = pf.Parameter('L_exact')
    fs = pf.SmoothConvexFunction(is_basis=True,tags=['f_{'+label+'}'],L=Lpar)
    gs = pf.ConvexFunction(is_basis=True,tags=['g_{'+label+'}'])
    hs = fs+gs
    xs = hs.set_stationary_point('x_{star}')
    x = pf.Vector(is_basis=True,tags=['x_{k}'])
    anchor = pf.Vector(is_basis=True,tags=['e_{0}'])
    xp = gs.prox(x-fs.grad(x)/Lpar,1/Lpar,tag='x_{k+1}')
    fs(xp)
    if with_current_g:
        gs(x)
    mapping = {'L_exact':L_sym}
    return context,fs,gs,xs,x,xp,anchor,Lpar,mapping

def scalar_parameter(mapping,name,value):
    mapping[name] = sp.cancel(value)
    return pf.Parameter(name)

def local_potential(fs,gs,xs,x,anchor,Lpar,mapping,a,b,A,d,beta,N,name):
    e = x-xs
    p = (fs.grad(x)-fs.grad(xs))/Lpar
    return (scalar_parameter(mapping,name+'_a',a)*(fs(x)-fs(xs))/Lpar
            +scalar_parameter(mapping,name+'_b',b)*(gs(x)-gs(xs))/Lpar
            +scalar_parameter(mapping,name+'_A',A)*e**2
            -scalar_parameter(mapping,name+'_d',d)*e*(fs.grad(x)/Lpar)
            -scalar_parameter(mapping,name+'_beta',beta)*p**2
            -scalar_parameter(mapping,name+'_tau',1/(4*N))*anchor**2)

def block_square(H,K,e,p,r,q,mapping,name):
    # Q = [w+H^{-1}Kz]^T H [w+H^{-1}Kz], w=(r,q), z=(e,p).
    lower = (K.T*H.inv()*K).applyfunc(sp.cancel)
    block = H.row_join(K).col_join(K.T.row_join(lower))
    vectors = [r,q,e,p]
    return sum((scalar_parameter(mapping,f'{name}_{i}_{j}',block[i,j])*vectors[i]*vectors[j]
                for i,j in np.ndindex((4,4))),pf.Scalar.zero())

def verify_exact(label,expression,context,mapping):
    manager = pf.ExpressionManager(context,resolve_parameters=mapping)
    evaluated = manager.eval_scalar(expression,sympy_mode=True)
    gram = sp.Matrix(evaluated.inner_prod_coords).applyfunc(sp.cancel)
    funcs = sp.Matrix(list(evaluated.func_coords)).applyfunc(sp.cancel)
    offset = sp.cancel(evaluated.offset)
    assert all(value==0 for value in list(gram)+list(funcs)+[offset]),label
    print(label,': exact zero in Gram, function and offset coordinates')
    pf.pprint_labeled_matrix(np.asarray(gram,dtype=object),[str(v) for v in context.basis_vectors()])
    print('Function residual:',funcs.T,'offset:',offset)
    return True
```

## Symbolic Step Recursion Verification

For $1\leq k\leq N-2$, check
$$V_{k+1}-V_{k}=(a_{k+1}-a_{k})J_{f}(x_{\star},x_{k+1})+a_{k}J_{f}(x_{k},x_{k+1})+(b_{k+1}-b_{k})J_{g}(x_{\star},x_{k+1})+b_{k}J_{g}(x_{k},x_{k+1})-Q_{k}.$$
Here $\beta_{k+1}$ is a nonnegative symbolic coefficient and $\beta_{k}$ is its proved transfer formula. The residual $\mathrm{LHS}-\mathrm{RHS}$ must vanish in every component.

```python

ctx_step,f_step,g_step,xs_step,x_step,xp_step,e0_step,Lp_step,map_step = local_context('step')
t_step = 2*N_sym-k_sym
beta_k = beta_recursion(k_sym,t_step,c_next)
H_step,K_step = local_blocks(k_sym,t_step,c_next)
V_step = local_potential(f_step,g_step,xs_step,x_step,e0_step,Lp_step,map_step,
    (k_sym+1)/t_step,k_sym/t_step,(N_sym-k_sym)/t_step**2,1/t_step,beta_k,N_sym,'old')
V_next = local_potential(f_step,g_step,xs_step,xp_step,e0_step,Lp_step,map_step,
    (k_sym+2)/(t_step-1),(k_sym+1)/(t_step-1),(N_sym-k_sym-1)/(t_step-1)**2,1/(t_step-1),c_next,N_sym,'new')
a_step = (k_sym+1)/t_step
b_step = k_sym/t_step
mu_f = (k_sym+2)/(t_step-1)-a_step
mu_g = (k_sym+1)/(t_step-1)-b_step
I_step = (scalar_parameter(map_step,'mu_f',mu_f)*f_step.interp_ineq(xs_step,xp_step)
          +scalar_parameter(map_step,'a',a_step)*f_step.interp_ineq(x_step,xp_step)
          +scalar_parameter(map_step,'mu_g',mu_g)*g_step.interp_ineq(xs_step,xp_step)
          +scalar_parameter(map_step,'b',b_step)*g_step.interp_ineq(x_step,xp_step))/Lp_step
p_step = (f_step.grad(x_step)-f_step.grad(xs_step))/Lp_step
r_step = (f_step.grad(xp_step)-f_step.grad(xs_step))/Lp_step
q_step = (g_step.grad(xp_step)+f_step.grad(xs_step))/Lp_step
Q_step = block_square(H_step,K_step,x_step-xs_step,p_step,r_step,q_step,map_step,'Q')
LHS_step = V_next-V_step
RHS_step = I_step-Q_step
step_verified = verify_exact('Interior step',LHS_step-RHS_step,ctx_step,map_step)
```

## Base Case and Boundary Symbolic Verification

For $N\geq2$, check
$$V_{1}=a_{0}J_{f}(x_{\star},x_{0})+(a_{1}-a_{0})J_{f}(x_{\star},x_{1})+a_{0}J_{f}(x_{0},x_{1})+b_{1}J_{g}(x_{\star},x_{1})-Q_{0}-\frac{\|p_{0}\|^{2}}{4N}.$$
This uses a separate initial context and no evaluation of $g(x_{0})$. The residual $\mathrm{LHS}-\mathrm{RHS}$ must vanish.

```python

ctx_base,f_base,g_base,xs_base,x_base,xp_base,e0_unused,Lp_base,map_base = local_context('base',with_current_g=False)
t_base = 2*N_sym
H_base,K_base = local_blocks(sp.S.Zero,t_base,c_next)
V_base = local_potential(f_base,g_base,xs_base,xp_base,x_base-xs_base,Lp_base,map_base,
    2/(2*N_sym-1),1/(2*N_sym-1),(N_sym-1)/(2*N_sym-1)**2,1/(2*N_sym-1),c_next,N_sym,'base')
a0 = 1/(2*N_sym)
base_terms = (scalar_parameter(map_base,'a0',a0)*f_base.interp_ineq(xs_base,x_base)
    +scalar_parameter(map_base,'af1',2/(2*N_sym-1)-a0)*f_base.interp_ineq(xs_base,xp_base)
    +scalar_parameter(map_base,'a01',a0)*f_base.interp_ineq(x_base,xp_base)
    +scalar_parameter(map_base,'bg1',1/(2*N_sym-1))*g_base.interp_ineq(xs_base,xp_base))/Lp_base
p_base = (f_base.grad(x_base)-f_base.grad(xs_base))/Lp_base
r_base = (f_base.grad(xp_base)-f_base.grad(xs_base))/Lp_base
q_base = (g_base.grad(xp_base)+f_base.grad(xs_base))/Lp_base
Q_base = block_square(H_base,K_base,x_base-xs_base,p_base,r_base,q_base,map_base,'Qb')
LHS_base = V_base
RHS_base = base_terms-Q_base-scalar_parameter(map_base,'initial_square',1/(4*N_sym))*p_base**2
base_verified = verify_exact('Base N>=2',LHS_base-RHS_base,ctx_base,map_base)
```

### Boundary Identity Symbolic Verification

For $N\geq2$, check
$$V_{N}-V_{N-1}=\frac1{N+1}J_{f}(x_{\star},x_{N})+\frac{N}{N+1}J_{f}(x_{N-1},x_{N})+\frac2{N+1}J_{g}(x_{\star},x_{N})+\frac{N-1}{N+1}J_{g}(x_{N-1},x_{N})-Q_{T}.$$
Also check $h(x_{N})-h(x_{\star})=L(V_{N}+\|e_{0}\|^{2}/(4N))$. Both residuals must vanish in all coordinates.

```python

ctx_end,f_end,g_end,xs_end,x_end,xp_end,e0_end,Lp_end,map_end = local_context('boundary')
a_end = N_sym/(N_sym+1)
b_end = (N_sym-1)/(N_sym+1)
V_before = local_potential(f_end,g_end,xs_end,x_end,e0_end,Lp_end,map_end,
    a_end,b_end,1/(N_sym+1)**2,1/(N_sym+1),(N_sym-1)/(2*(N_sym+1)**2),N_sym,'last')
V_terminal = (f_end(xp_end)+g_end(xp_end)-f_end(xs_end)-g_end(xs_end))/Lp_end-scalar_parameter(map_end,'tau_end',1/(4*N_sym))*e0_end**2
H_end = sp.Matrix([[sp.Rational(1,2),sp.Rational(1,2)],[sp.Rational(1,2),1]])
K_end = sp.Matrix([[-1/(2*(N_sym+1)),1/(2*(N_sym+1))],[-1/(N_sym+1),sp.Rational(1,2)]])
p_end = (f_end.grad(x_end)-f_end.grad(xs_end))/Lp_end
r_end = (f_end.grad(xp_end)-f_end.grad(xs_end))/Lp_end
q_end = (g_end.grad(xp_end)+f_end.grad(xs_end))/Lp_end
Q_end = block_square(H_end,K_end,x_end-xs_end,p_end,r_end,q_end,map_end,'Qe')
end_terms = (scalar_parameter(map_end,'afstar',1-a_end)*f_end.interp_ineq(xs_end,xp_end)
    +scalar_parameter(map_end,'afcon',a_end)*f_end.interp_ineq(x_end,xp_end)
    +scalar_parameter(map_end,'bgstar',1-b_end)*g_end.interp_ineq(xs_end,xp_end)
    +scalar_parameter(map_end,'bgcon',b_end)*g_end.interp_ineq(x_end,xp_end))/Lp_end
LHS_boundary = V_terminal-V_before
RHS_boundary = end_terms-Q_end
boundary_verified = verify_exact('Terminal step N>=2',LHS_boundary-RHS_boundary,ctx_end,map_end)
normalization_diff = (f_end(xp_end)+g_end(xp_end)-f_end(xs_end)-g_end(xs_end))-Lp_end*(V_terminal+scalar_parameter(map_end,'tau_normalization',1/(4*N_sym))*e0_end**2)
normalization_verified = verify_exact('Terminal normalization',normalization_diff,ctx_end,map_end)
```

### Exceptional one-step horizon

For $N=1$, check separately
$$V_{1}=\tfrac12J_{f}(x_{\star},x_{0})+\tfrac12J_{f}(x_{\star},x_{1})+\tfrac12J_{f}(x_{0},x_{1})+J_{g}(x_{\star},x_{1})-Q_{T}-\tfrac14\|p_{0}\|^{2}.$$
The residual $\mathrm{LHS}-\mathrm{RHS}$ must vanish without using an interior coefficient formula.
