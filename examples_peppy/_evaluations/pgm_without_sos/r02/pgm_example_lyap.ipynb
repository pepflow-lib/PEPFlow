# Proximal gradient method
Let $f$ be convex and $L$-smooth, $g$ closed, proper and convex, and $h=f+g$.
We work in a real finite-dimensional Euclidean space, with $L>0$, $R\geq0$, and integer $N\geq1$.
Let $x_\star$ minimize $h$ and assume $\|x_0-x_\star\|^2\leq R^2$.
The well-defined updates are
$$x_{k+1}=\operatorname{prox}_{g/L}\left(x_k-\frac{1}{L}\nabla f(x_k)\right),\qquad 0\leq k<N.$$
The metric is $h(x_N)-h(x_\star)$. Composite stationarity is encoded as $0\in\nabla f(x_\star)+\partial g(x_\star)$.


## Proof Statement



### Theorem
Let $f$ be convex and differentiable with $L$-Lipschitz gradient, $L>0$, and let $g$ be closed proper convex. Suppose $x_\star$ minimizes $h=f+g$ and $\|x_0-x_\star\|\le R$, $R\ge0$. For integer $N\ge1$, proximal gradient with step $1/L$ satisfies
$$h(x_N)-h(x_\star)\le\frac{LR^2}{4N}.$$
Write $u_k=x_k-x_\star$, $p_k=\nabla f(x_k)-\nabla f(x_\star)$, $F_k=f(x_k)-f(x_\star)$, $G_k=g(x_k)-g(x_\star)$, and $t_k=2N-k$. The following explicit Lyapunov sequence is nonincreasing:
$$V_0=0,\qquad V_N=F_N+G_N-\frac{L}{4N}\|u_0\|^2,$$
$$V_k=-\frac{L}{4N}\|u_0\|^2+\frac{L(N-k)}{t_k^2}\|u_k\|^2-\frac{\langle u_k,\nabla f(x_k)\rangle}{t_k}-\frac{\beta_k}{L}\|p_k\|^2+\frac{k+1}{t_k}F_k+\frac{k}{t_k}G_k,\quad1\le k<N.$$
Here $\beta_0=0$ and, for $1\le k<N$, the closed form is the finite matrix product
$$\beta_k=\frac{P_k}{Q_k},\qquad\binom{P_k}{Q_k}=T_kT_{k+1}\cdots T_{N-2}\binom{(N-1)/(2(N+1)^2)}{1},$$
$$T_j=\begin{pmatrix}2j(j+1)t_j(t_j-1)^2&j(j+1)(t_j-1)(j+t_j+1)\\4t_j^2(t_j-1)((2j+1)t_j-j)&2t_j^2((j^2+3j+1)t_j+1)\end{pmatrix}.$$
The empty product is the identity. No recurrence remains to be solved to evaluate this formula. For $N=1$ there are no interior coefficients or interior potentials. No value of $g(x_0)$ is required.



### Proof outline
The potential used throughout is explicitly
$$V_k=-\frac L{4N}\|u_0\|^2+\frac{L(N-k)}{(2N-k)^2}\|u_k\|^2-\frac{\langle u_k,\nabla f(x_k)\rangle}{2N-k}-\frac{\beta_k}L\|p_k\|^2+\frac{k+1}{2N-k}F_k+\frac{k}{2N-k}G_k\quad(1\le k<N),$$
with $V_0=0$, $V_N=F_N+G_N-L\|u_0\|^2/(4N)$ and the finite-product coefficients above.
For evaluated points define the nonpositive interpolation residuals
$$I_f(u,v)=f(v)-f(u)+\langle\nabla f(v),u-v\rangle+\frac{\|\nabla f(u)-\nabla f(v)\|^2}{2L},$$
$$I_g(u,v)=g(v)-g(u)+\langle g'_v,u-v\rangle.$$
Here $g'_{k+1}=L(x_k-x_{k+1})-\nabla f(x_k)\in\partial g(x_{k+1})$, and $g'_\star=-\nabla f(x_\star)$. An index in a residual denotes its iterate, and $\star$ denotes $x_\star$.
For $0\le k\le N-2$, put $t=2N-k$, $c=\beta_{k+1}$, $p=p_k$, $q=g'_{k+1}+\nabla f(x_\star)$, $r=p_{k+1}$, $z=q-Lu_k/t$, and
$$a=c+\frac{k+2}{2(t-1)},\quad d=\frac{(2k+1)t-k}{2(t-1)^2},\quad e=\frac{k+1}{2(t-1)},\quad v=\frac{k+1}{2t(t-1)},\quad w=\frac{kt+1}{2(t-1)^2},\quad\delta=d-e^2/a.$$
Define the complete nonnegative slack
$$\mathcal Q_k=\frac1L\left[a\left\|r+\frac ea z+\frac va p\right\|^2+\delta\left\|z+\frac{w-ev/a}{\delta}p\right\|^2\right].$$
For $1\le k\le N-2$ the exact step identity is
$$V_{k+1}-V_k=\frac{2N+1}{t(t-1)}I_f(\star,k+1)+\frac{k+1}{t}I_f(k,k+1)+\frac{2N}{t(t-1)}I_g(\star,k+1)+\frac{k}{t}I_g(k,k+1)-\mathcal Q_k.$$
The separate base identity, for $N\ge2$, is
$$V_1=\frac{I_f(\star,0)}{2N}+\frac{2N+1}{2N(2N-1)}I_f(\star,1)+\frac{I_f(0,1)}{2N}+\frac{I_g(\star,1)}{2N-1}-\mathcal Q_0-\frac{\|p_0\|^2}{4NL}.$$
For the terminal step let $p=p_{N-1}$, $q=g'_N+\nabla f(x_\star)$, $r=p_N$, and $u=u_{N-1}$. Then
$$\mathcal Q_{\rm end}=\frac1{2L}\left[\left\|r+q+\frac{p-Lu}{N+1}\right\|^2+\left\|q+\frac{Np-Lu}{N+1}\right\|^2\right],$$
$$V_N-V_{N-1}=\frac{I_f(\star,N)+N I_f(N-1,N)+2I_g(\star,N)+(N-1)I_g(N-1,N)}{N+1}-\mathcal Q_{\rm end}.$$
For $N=1$, use $p=p_0$, $q=g'_1+\nabla f(x_\star)$, $r=p_1$, $u=u_0$ and the same displayed $\mathcal Q_{\rm end}$ with $N=1$:
$$V_1=\tfrac12 I_f(\star,0)+\tfrac12 I_f(\star,1)+\tfrac12 I_f(0,1)+I_g(\star,1)-\mathcal Q_{\rm end}-\frac{\|p_0\|^2}{4L}.$$
All active interpolation multipliers are nonnegative. The product defining $\beta_k$ has positive entries for $1\le k<N-1$ and a positive denominator seed; hence $Q_k>0$ and $\beta_k\ge0$. For local squares $k\ge0$, $t>1$, $c\ge0$ give $a>0$ and
$$\delta=\frac{2c(t-1)((2k+1)t-k)+(k^2+3k+1)t+1}{4(t-1)^3a}>0.$$
Every square is subtracted and every residual is nonpositive, so $V_N\le\cdots\le V_1\le V_0=0$. Therefore
$$h(x_N)-h(x_\star)\le\frac{L}{4N}\|u_0\|^2\le\frac{LR^2}{4N}.$$
The exact identities, including all Gram, function-value and constant coordinates, are checked below. In one dimension $f=0$, $g(x)=LR|x|/(2N)$, $x_0=R$, and $x_\star=0$ attain equality when $R>0$; $R=0$ gives the zero bound.

## Function definitions

```python
L = pf.Parameter("L")
R = pf.Parameter("R")
f = pf.SmoothConvexFunction(is_basis=True, tags=["f"], L=L)
g = pf.ConvexFunction(is_basis=True, tags=["g"])
h = f + g
h.add_tag("h")
```

## Algorithm and PEP setup

```python
def make_ctx_pgm(ctx_name: str, N, **kwargs) -> pf.PEPContext:
    ctx = pf.PEPContext(ctx_name).set_as_current()
    x = pf.Vector(is_basis=True, tags=["x_0"])
    h.set_stationary_point("x_star")
    for k in range(N):
        x = g.prox(x - f.grad(x) / L, 1 / L, tag=f"x_{k + 1}")
    f(x)
    return ctx


def get_pep_setup(N, params):
    ctx = make_ctx_pgm(f"pgm_{N}", N)
    pb = pf.PEPBuilder(ctx)
    pb.add_initial_constraint(
        ((ctx["x_0"] - ctx["x_star"]) ** 2).le(R**2, name="initial_condition")
    )
    pb.set_performance_metric(h(ctx[f"x_{N}"]) - h(ctx["x_star"]))
    return ctx, pb, f
```

### Load the coherent selected certificate and rebuild its context without solving.

```python
import hashlib
import json
from fractions import Fraction
import numpy as np
import sympy as sp
import pepflow as pf
from pepflow.constraint import ScalarConstraint
from pepflow.lyapunov_utils import ldl_decompose_with_reversed_basis
from pepflow.utils import Comparator

b1 = json.loads((notebook_dir / "state/pgm_b1.json").read_text())
dense = json.loads((notebook_dir / "state/pgm_dense.json").read_text())
selected = json.loads(
    (notebook_dir / "supplementary/b2-relaxation-001.json").read_text()
)
N = selected["N"]
ctx, pb, obj = get_pep_setup(N, {"L": 1, "R": 1})
pb.set_relaxed_constraints(selected["relaxed_constraints"])
for owner in ctx.func_to_triplets:
    owner.get_interpolation_constraints(ctx)
entries = [(c, "initial", None, None) for c in pb.init_conditions]
for owner in ctx.func_to_triplets:
    groups = ctx.get_constraint_data(owner)
    entries.extend(
        (c, "interpolation", str(owner), str(group))
        for group, cs in groups.sc_dict.items()
        for c in cs
    )
    entries.extend(
        (c, "interpolation", str(owner), str(group))
        for group, c in groups.psd_dict.items()
    )
metadata = selected["constraint_metadata"]
duals = selected["named_dual_values"]
assert set(metadata) == set(duals) == {c.name for c, *_ in entries}
for c, origin, owner, group in entries:
    assert isinstance(c, ScalarConstraint), "Unexpected matrix constraint"
    assert metadata[c.name] == dict(
        origin=origin,
        owner=owner,
        group=group,
        kind="scalar",
        comparator=c.cmp.value,
        shape=[],
        relaxed=c.name in pb.relaxed_constraints,
    )
for source, expected in selected["certificate_provenance"]["source_hashes"].items():
    assert hashlib.sha256(Path(source).read_bytes()).hexdigest() == expected
pm = pf.ExpressionManager(ctx, resolve_parameters={"L": sp.S.One, "R": sp.S.One})
basis = ctx.basis_vectors()
assert (
    [str(v) for v in basis]
    == selected["basis_vectors"]
    == selected["S_row_names"]
    == selected["S_col_names"]
)
assert [str(v) for v in ctx.basis_scalars()] == selected["basis_scalars"]
print("Dense bound:", dense["opt_value"], "Selected bound:", selected["opt_value"])
print("Basis:", selected["basis_vectors"])
for owner_name, group in sorted(
    {(m["owner"], m["group"]) for m in metadata.values() if m["owner"] is not None}
):
    vals = [
        (name, v)
        for name, v in dense["named_dual_values"].items()
        if v is not None
        and metadata[name]["owner"] == owner_name
        and metadata[name]["group"] == group
    ]
    active = [(name, round(v, 6)) for name, v in vals if abs(v) > 1e-6]
    print(
        owner_name,
        group,
        "dense nonzero count at 1e-6:",
        len(active),
        "largest:",
        sorted(active, key=lambda z: -abs(z[1]))[:6],
    )
print("Dense S eigenvalues:", np.linalg.eigvalsh(np.asarray(dense["S_matrix"])))
print(
    "Relaxation: retain star-to-iterate and consecutive forward pairs for both owners."
)
print("Dropped constraints:", len(selected["relaxed_constraints"]))
for name, value in duals.items():
    if value is not None:
        print(name, f"{value:.9f}", Fraction(value).limit_denominator(1000))
```

### Candidate multipliers inferred from fractions and scalar conservation.

```python
def candidate_multiplier(name, horizon):
    n = sp.S(horizon)
    if name == "initial_condition":
        return 1 / (4 * n)
    owner, pair = name.split(":")
    first, second = pair.split(",")
    if second == "x_star":
        return sp.S.Zero
    j = int(second[2:])

    def flow(index):
        if owner == "f":
            return (index + 1) / (2 * n - index) if index >= 0 else sp.S.Zero
        return index / (2 * n - index) if index >= 1 else sp.S.Zero

    if first == "x_star":
        return 1 - flow(horizon - 1) if j == horizon else flow(j) - flow(j - 1)
    i = int(first[2:])
    return flow(i) if j == i + 1 else sp.S.Zero


lambda_error = max(
    abs(float(candidate_multiplier(name, N)) - (value or 0))
    for name, value in duals.items()
)
print("Candidate lambda max error:", lambda_error)
assert lambda_error <= 1e-4
for owner_tag, tags in [
    ("f", ["x_star"] + [f"x_{i}" for i in range(N + 1)]),
    ("g", ["x_star"] + [f"x_{i}" for i in range(1, N + 1)]),
]:
    matrix = np.array(
        [
            [
                float(candidate_multiplier(f"{owner_tag}:{i},{j}", N)) if i != j else 0
                for j in tags
            ]
            for i in tags
        ]
    )
    print("Candidate", owner_tag, "multipliers, numerically supported only:")
    pf.pprint_labeled_matrix(matrix, tags, precision=5)
```

### Selected numerical reverse-LDL decomposition; retain every factor.

```python
S_array = np.asarray(selected["S_matrix"], dtype=float)
S_labeled = pf.MatrixWithNames(
    matrix=S_array, row_names=selected["S_row_names"], col_names=selected["S_col_names"]
)
D, ell = ldl_decompose_with_reversed_basis(S_labeled, basis, print_output=False)
D = np.asarray(D, dtype=float)
E = np.asarray([pm.eval_vector(v).coords for v in ell], dtype=float)
S_numeric = sum(
    (float(D[i, j]) * ell[i] * ell[j] for i, j in np.ndindex(D.shape)), pf.Scalar.zero()
)
print("Reverse-LDL diagonal:", np.diag(D))
pf.pprint_labeled_matrix(
    E, [f"ell_{i}" for i in range(len(E))], selected["basis_vectors"], precision=4
)
print(
    "Full D retained, including tiny pivots; off-diagonal maximum:",
    np.max(np.abs(D - np.diag(np.diag(D)))),
)
S_reconstructed = E.T @ D @ E
```

### All-coordinate numerical identity and cone checks.

```python
weighted = pf.Scalar.zero()
weighted_candidate = pf.Scalar.zero()
scalar_inequalities = []
for c, *_ in entries:
    residual = c.lhs - c.rhs
    if c.cmp in (Comparator.GE, Comparator.SEQ):
        residual = -residual
    weighted_candidate += candidate_multiplier(c.name, N) * (
        pf.Scalar.zero() + residual
    )
    value = duals[c.name]
    if metadata[c.name]["relaxed"]:
        assert value is None
        continue
    assert value is not None and np.isfinite(value)
    weighted += value * (pf.Scalar.zero() + residual)
    if c.cmp != Comparator.EQ:
        scalar_inequalities.append(value)
diff = pb.performance_metric - selected["opt_value"] - weighted + S_numeric
checked = pm.eval_scalar(diff)


def max_abs(array):
    return float(np.max(np.abs(np.asarray(array, dtype=float)), initial=0))


S_eigs = np.linalg.eigvalsh((S_array + S_array.T) / 2)
residuals = dict(
    identity_gram_max_abs=max_abs(checked.inner_prod_coords),
    identity_func_max_abs=max_abs(checked.func_coords),
    identity_offset_abs=max_abs(checked.offset),
    S_reconstruction_max_abs=max_abs(S_reconstructed - S_array),
    dense_relaxed_gap=abs(dense["opt_value"] - selected["opt_value"]),
    conjecture_gap=abs(selected["opt_value"] - 1 / (4 * N)),
    minimum_scalar_inequality_dual=min(scalar_inequalities, default=0),
    minimum_S_eigenvalue=float(S_eigs.min()),
    minimum_matrix_inequality_eigenvalue=0.0,
)
tolerances = dict(
    identity_atol=1e-5,
    identity_rtol=0,
    reconstruction_atol=1e-4,
    reconstruction_rtol=1e-5,
    dual_tol=1e-6,
    psd_tol=1e-6,
    dense_opt_atol=1e-5,
)
passed = (
    selected["certificate_provenance"]["solver_status"] == "optimal"
    and np.isfinite(S_array).all()
    and np.allclose(S_array, S_array.T, atol=1e-6, rtol=0)
    and all(
        residuals[key] <= 1e-5
        for key in [
            "identity_gram_max_abs",
            "identity_func_max_abs",
            "identity_offset_abs",
        ]
    )
    and np.allclose(S_reconstructed, S_array, atol=1e-4, rtol=1e-5)
    and residuals["dense_relaxed_gap"] < 1e-5
    and residuals["conjecture_gap"] < 1e-5
    and residuals["minimum_scalar_inequality_dual"] >= -1e-6
    and residuals["minimum_S_eigenvalue"] >= -1e-6
)
print("Numerical certificate verified:", bool(passed))
print(json.dumps(residuals, indent=2))
pf.pprint_labeled_matrix(
    np.asarray(checked.inner_prod_coords, dtype=float),
    selected["basis_vectors"],
    precision=9,
)
assert passed
```

### Candidate multipliers inferred from fractions and scalar conservation.

```python
def candidate_multiplier(name, horizon):
    n = sp.S(horizon)
    if name == "initial_condition":
        return 1 / (4 * n)
    owner, pair = name.split(":")
    first, second = pair.split(",")
    if second == "x_star":
        return sp.S.Zero
    j = int(second[2:])

    def flow(index):
        if owner == "f":
            return (index + 1) / (2 * n - index) if index >= 0 else sp.S.Zero
        return index / (2 * n - index) if index >= 1 else sp.S.Zero

    if first == "x_star":
        return 1 - flow(horizon - 1) if j == horizon else flow(j) - flow(j - 1)
    i = int(first[2:])
    return flow(i) if j == i + 1 else sp.S.Zero


lambda_error = max(
    abs(float(candidate_multiplier(name, N)) - (value or 0))
    for name, value in duals.items()
)
print("Candidate lambda max error:", lambda_error)
assert lambda_error <= 1e-4
for owner_tag, tags in [
    ("f", ["x_star"] + [f"x_{i}" for i in range(N + 1)]),
    ("g", ["x_star"] + [f"x_{i}" for i in range(1, N + 1)]),
]:
    matrix = np.array(
        [
            [
                float(candidate_multiplier(f"{owner_tag}:{i},{j}", N)) if i != j else 0
                for j in tags
            ]
            for i in tags
        ]
    )
    print("Candidate", owner_tag, "multipliers, numerically supported only:")
    pf.pprint_labeled_matrix(matrix, tags, precision=5)
```

### Selected numerical reverse-LDL decomposition; retain every factor.

```python
S_array = np.asarray(selected["S_matrix"], dtype=float)
S_labeled = pf.MatrixWithNames(
    matrix=S_array, row_names=selected["S_row_names"], col_names=selected["S_col_names"]
)
D, ell = ldl_decompose_with_reversed_basis(S_labeled, basis, print_output=False)
D = np.asarray(D, dtype=float)
E = np.asarray([pm.eval_vector(v).coords for v in ell], dtype=float)
S_numeric = sum(
    (float(D[i, j]) * ell[i] * ell[j] for i, j in np.ndindex(D.shape)), pf.Scalar.zero()
)
print("Reverse-LDL diagonal:", np.diag(D))
pf.pprint_labeled_matrix(
    E, [f"ell_{i}" for i in range(len(E))], selected["basis_vectors"], precision=4
)
print(
    "Full D retained, including tiny pivots; off-diagonal maximum:",
    np.max(np.abs(D - np.diag(np.diag(D)))),
)
S_reconstructed = E.T @ D @ E
```

### All-coordinate numerical identity and cone checks.

```python
weighted = pf.Scalar.zero()
weighted_candidate = pf.Scalar.zero()
scalar_inequalities = []
for c, *_ in entries:
    residual = c.lhs - c.rhs
    if c.cmp in (Comparator.GE, Comparator.SEQ):
        residual = -residual
    weighted_candidate += candidate_multiplier(c.name, N) * (
        pf.Scalar.zero() + residual
    )
    value = duals[c.name]
    if metadata[c.name]["relaxed"]:
        assert value is None
        continue
    assert value is not None and np.isfinite(value)
    weighted += value * (pf.Scalar.zero() + residual)
    if c.cmp != Comparator.EQ:
        scalar_inequalities.append(value)
diff = pb.performance_metric - selected["opt_value"] - weighted + S_numeric
checked = pm.eval_scalar(diff)


def max_abs(array):
    return float(np.max(np.abs(np.asarray(array, dtype=float)), initial=0))


S_eigs = np.linalg.eigvalsh((S_array + S_array.T) / 2)
residuals = dict(
    identity_gram_max_abs=max_abs(checked.inner_prod_coords),
    identity_func_max_abs=max_abs(checked.func_coords),
    identity_offset_abs=max_abs(checked.offset),
    S_reconstruction_max_abs=max_abs(S_reconstructed - S_array),
    dense_relaxed_gap=abs(dense["opt_value"] - selected["opt_value"]),
    conjecture_gap=abs(selected["opt_value"] - 1 / (4 * N)),
    minimum_scalar_inequality_dual=min(scalar_inequalities, default=0),
    minimum_S_eigenvalue=float(S_eigs.min()),
    minimum_matrix_inequality_eigenvalue=0.0,
)
tolerances = dict(
    identity_atol=1e-5,
    identity_rtol=0,
    reconstruction_atol=1e-4,
    reconstruction_rtol=1e-5,
    dual_tol=1e-6,
    psd_tol=1e-6,
    dense_opt_atol=1e-5,
)
passed = (
    selected["certificate_provenance"]["solver_status"] == "optimal"
    and np.isfinite(S_array).all()
    and np.allclose(S_array, S_array.T, atol=1e-6, rtol=0)
    and all(
        residuals[key] <= 1e-5
        for key in [
            "identity_gram_max_abs",
            "identity_func_max_abs",
            "identity_offset_abs",
        ]
    )
    and np.allclose(S_reconstructed, S_array, atol=1e-4, rtol=1e-5)
    and residuals["dense_relaxed_gap"] < 1e-5
    and residuals["conjecture_gap"] < 1e-5
    and residuals["minimum_scalar_inequality_dual"] >= -1e-6
    and residuals["minimum_S_eigenvalue"] >= -1e-6
)
print("Numerical certificate verified:", bool(passed))
print(json.dumps(residuals, indent=2))
pf.pprint_labeled_matrix(
    np.asarray(checked.inner_prod_coords, dtype=float),
    selected["basis_vectors"],
    precision=9,
)
assert passed
```

## Partial-sum Lyapunov construction and rank profile
Let $r_c\leq0$ be the normalized interpolation residuals and $P_i\geq0$ the retained LDL squares of each horizon's own certificate. Starting from $V_0=0$, we assign interpolation terms by their destination iterate and squares by their latest supported iterate (support cutoff $10^{-5}$ is used only for assignment; no coefficients are discarded). The initial $(*,0)$ term enters the first transition:
$$V_{k+1}-V_k=\sum_{c\text{ assigned at }k}\lambda_c r_c-\sum_{i\text{ assigned at }k}P_i\leq0.$$
All interpolation contributions and all squares enter by $V_N$; the initial-distance normalization remains for the boundary proof. Thus $V_N$ has a distance-square component, which need not have rank zero. The following visible construction rebuilds and rechecks each complete saved certificate without solving.


```python
def reconstruct_partial_sums(certificate):
    # Rebuild and revalidate a complete saved primal certificate without solving.
    import hashlib
    import json
    import numpy as np
    import pepflow as pf
    from pathlib import Path
    from pepflow.constraint import ScalarConstraint
    from pepflow.utils import Comparator

    certificate_checked = False
    certificate_keys = (
        "N_verify",
        "params_json",
        "relaxed_constraints",
        "S_matrix",
        "S_row_names",
        "S_col_names",
        "basis_vectors",
        "basis_scalars",
        "opt_value",
        "tau_sol",
        "tau_name",
        "named_dual_values",
        "constraint_metadata",
        "normalization",
        "certificate_provenance",
        "numerical_checks",
        "numerical_decomposition",
    )
    payload = {key: certificate[key] for key in certificate_keys}
    assert (
        hashlib.sha256(
            json.dumps(
                payload, sort_keys=True, separators=(",", ":"), allow_nan=False
            ).encode()
        ).hexdigest()
        == certificate["certificate_id"]
    )
    assert certificate["certificate_provenance"]["pep_type"] == "primal"
    assert certificate["certificate_provenance"]["solver_status"] == "optimal"
    assert certificate["closed_form_search"]["completed"]
    for source, expected in certificate["certificate_provenance"][
        "source_hashes"
    ].items():
        source_path = Path(source).resolve()
        assert source_path.is_relative_to(notebook_dir)
        assert hashlib.sha256(source_path.read_bytes()).hexdigest() == expected
    assert (
        Path(pf.__file__).resolve()
        == notebook_dir / "supplementary/approved/pepflow/__init__.py"
    )
    N_int = int(certificate["N_verify"])
    N = N_int
    params = json.loads(certificate["params_json"])
    assert params == {"L": 1, "R": 1}
    ctx, pb, obj = get_pep_setup(N, params)
    pb.set_relaxed_constraints(certificate["relaxed_constraints"])
    ctx.set_as_current()
    assert pb.ctx is ctx
    entries = [(c, "initial", None, None) for c in pb.init_conditions]
    assert not ctx.oper_to_duplets
    for owner in ctx.func_to_triplets:
        owner.get_interpolation_constraints(ctx)
        data = ctx.get_constraint_data(owner)
        entries.extend(
            (c, "interpolation", str(owner), str(group))
            for group, cs in data.sc_dict.items()
            for c in cs
        )
        assert not data.psd_dict
    metadata = certificate["constraint_metadata"]
    saved_duals = certificate["named_dual_values"]
    assert len(entries) == len({c.name for c, *_ in entries})
    assert set(metadata) == set(saved_duals) == {c.name for c, *_ in entries}
    assert set(pb.relaxed_constraints).issubset(metadata)
    basis = ctx.basis_vectors()
    basis_labels = [str(v) for v in basis]
    assert (
        basis_labels
        == certificate["basis_vectors"]
        == certificate["S_row_names"]
        == certificate["S_col_names"]
    )
    assert [str(v) for v in ctx.basis_scalars()] == certificate["basis_scalars"]
    pm = pf.ExpressionManager(ctx, resolve_parameters=params)
    tol = certificate["numerical_checks"]["tolerances"]
    assert (
        tol["identity_atol"] == 1e-5
        and tol["psd_tol"] == 1e-6
        and tol["dual_tol"] == 1e-6
    )
    S_array = np.asarray(certificate["S_matrix"], dtype=float)
    assert S_array.shape == (len(basis), len(basis)) and np.isfinite(S_array).all()
    assert np.allclose(S_array, S_array.T, atol=1e-6, rtol=0)
    assert np.linalg.eigvalsh((S_array + S_array.T) / 2).min() >= -1e-6
    S_scalar = sum(
        (
            float(S_array[i, j]) * basis[i] * basis[j]
            for i, j in np.ndindex(S_array.shape)
        ),
        pf.Scalar.zero(),
    )
    weighted_terms = {}
    interpolation_terms = {}
    interpolation_pairs = {}
    extra_terms = {}
    extra_duals = {}
    for c, origin, owner, group in entries:
        assert isinstance(c, ScalarConstraint)
        assert metadata[c.name] == dict(
            origin=origin,
            owner=owner,
            group=group,
            kind="scalar",
            comparator=c.cmp.value,
            shape=[],
            relaxed=c.name in pb.relaxed_constraints,
        )
        value = saved_duals[c.name]
        if c.name in pb.relaxed_constraints:
            assert value is None
            continue
        assert value is not None and np.isfinite(value)
        assert c.cmp == Comparator.EQ or value >= -1e-6
        residual = c.lhs - c.rhs
        if c.cmp in (Comparator.GE, Comparator.SEQ):
            residual = -residual
        weighted_terms[c.name] = value * (pf.Scalar.zero() + residual)
        if origin == "interpolation":
            interpolation_terms[c.name] = weighted_terms[c.name]
            interpolation_pairs[c.name] = tuple(c.name.split(":")[1].split(","))
        elif c.name != certificate["tau_name"]:
            extra_terms[c.name] = weighted_terms[c.name]
            extra_duals[c.name] = value
    tau_name = certificate["tau_name"]
    assert tau_name == "initial_condition" and metadata[tau_name]["origin"] == "initial"
    assert (
        metadata[tau_name]["comparator"] == "LE" and not metadata[tau_name]["relaxed"]
    )
    assert saved_duals[tau_name] == certificate["tau_sol"]
    assert certificate["normalization"]["tau_name"] == tau_name
    assert certificate["normalization"]["bound_value"] == certificate["opt_value"]
    assert (
        certificate["normalization"]["initial_condition"] == "||x_0-x_star||^2 <= R^2"
    )
    assert certificate["normalization"]["performance_metric"] == "h(x_N) - h(x_star)"
    assert abs(certificate["opt_value"] - 1 / (4 * N)) < 1e-5
    certificate_diff = (
        pb.performance_metric
        - certificate["opt_value"]
        - sum(weighted_terms.values(), pf.Scalar.zero())
        + S_scalar
    )
    value = pm.eval_scalar(certificate_diff)

    def max_abs(array):
        return float(np.max(np.abs(np.asarray(array, dtype=float)), initial=0))

    certificate_residuals = [
        max_abs(value.inner_prod_coords),
        max_abs(value.func_coords),
        max_abs(value.offset),
    ]
    assert max(certificate_residuals) < 1e-5
    D_saved = np.asarray(certificate["numerical_decomposition"]["D"], dtype=float)
    E_saved = np.asarray(
        certificate["numerical_decomposition"]["ell_coords"], dtype=float
    )
    assert certificate["numerical_decomposition"]["basis_labels"] == basis_labels
    assert E_saved.shape[1] == len(basis) and D_saved.shape == (
        len(E_saved),
        len(E_saved),
    )
    assert np.isfinite(D_saved).all() and np.isfinite(E_saved).all()
    assert np.allclose(E_saved.T @ D_saved @ E_saved, S_array, atol=1e-4, rtol=1e-5)
    certificate_checked = True
    print(
        "Certificate rechecked at N=",
        N,
        "Gram/function/offset residuals:",
        certificate_residuals,
    )

    # Reuse all factors of this same certificate. No candidate formula replaces numerical data.
    assert np.allclose(D_saved, np.diag(np.diag(D_saved)), atol=0, rtol=0)
    ell_saved = [
        sum((float(c) * v for c, v in zip(row, basis)), pf.Vector.zero())
        for row in E_saved
    ]
    S_piece_terms = {
        i: float(D_saved[i, i]) * ell_saved[i] ** 2 for i in range(len(D_saved))
    }
    piece_residual = pm.eval_scalar(
        sum(S_piece_terms.values(), pf.Scalar.zero()) - S_scalar
    )
    assert (
        max(
            max_abs(piece_residual.inner_prod_coords),
            max_abs(piece_residual.func_coords),
            max_abs(piece_residual.offset),
        )
        < 1e-5
    )
    for piece in S_piece_terms.values():
        value = pm.eval_scalar(piece)
        assert max_abs(value.func_coords) == 0 and max_abs(value.offset) == 0
        assert (
            np.linalg.eigvalsh(np.asarray(value.inner_prod_coords, dtype=float)).min()
            >= -1e-6
        )

    # Every increment is a sum of nonpositive weighted residuals minus PSD pieces.
    grouping_checked = False
    rank_tolerance = 1e-4
    rank_left, rank_right = 1, 1
    group_terms = {
        name: term for name, term in weighted_terms.items() if name != tau_name
    }
    term_assignment = {
        name: int(interpolation_pairs[name][1][2:]) - 1 for name in group_terms
    }
    basis_times = [
        0
        if label in ["x_0", "x_star", "grad_f(x_star)"]
        else int(label.split("x_")[1].split(")")[0])
        for label in basis_labels
    ]
    S_assignment = {
        i: max(
            0,
            max((basis_times[j] for j, v in enumerate(row) if abs(v) > 1e-5), default=0)
            - 1,
        )
        for i, row in enumerate(E_saved)
    }
    assert set(term_assignment) == set(group_terms) and set(S_assignment) == set(
        S_piece_terms
    )
    assert set(term_assignment.values()).issubset(set(range(-1, N)))
    assert set(S_assignment.values()).issubset(set(range(N)))
    lyap = [pf.Scalar.zero()]
    partial_sum = sum(
        (group_terms[name] for name, target in term_assignment.items() if target == -1),
        pf.Scalar.zero(),
    )
    for step in range(N):
        partial_sum += sum(
            (
                group_terms[name]
                for name, target in term_assignment.items()
                if target == step
            ),
            pf.Scalar.zero(),
        )
        partial_sum -= sum(
            (S_piece_terms[i] for i, target in S_assignment.items() if target == step),
            pf.Scalar.zero(),
        )
        lyap.append(partial_sum)
    ranks = []
    singular_values = []
    for k, Vk in enumerate(lyap):
        sv = np.linalg.svd(
            np.asarray(pm.eval_scalar(Vk).inner_prod_coords, dtype=float),
            compute_uv=False,
        )
        singular_values.append(sv.tolist())
        ranks.append(int(np.sum(sv > rank_tolerance)))
        print(f"rank V_{k}: {ranks[-1]}")
        if k == 0:
            print()
    retained_ranks = ranks[rank_left : N + 1 - rank_right]
    print(
        "Interior rank is constant at this N:",
        len(retained_ranks) >= 3 and len(set(retained_ranks)) == 1,
        "boundary exclusions:",
        (rank_left, rank_right),
    )
    coverage = lyap[N] - (sum(group_terms.values(), pf.Scalar.zero()) - S_scalar)
    coverage_value = pm.eval_scalar(coverage)
    coverage_residuals = [
        max_abs(coverage_value.inner_prod_coords),
        max_abs(coverage_value.func_coords),
        max_abs(coverage_value.offset),
    ]
    assert max(coverage_residuals) < 1e-5
    increment_residuals = []
    for k in range(N):
        steps = {-1, 0} if k == 0 else {k}
        increment = sum(
            (
                group_terms[name]
                for name, target in term_assignment.items()
                if target in steps
            ),
            pf.Scalar.zero(),
        )
        increment -= sum(
            (S_piece_terms[i] for i, target in S_assignment.items() if target in steps),
            pf.Scalar.zero(),
        )
        value = pm.eval_scalar(lyap[k + 1] - lyap[k] - increment)
        residual = max(
            max_abs(value.inner_prod_coords),
            max_abs(value.func_coords),
            max_abs(value.offset),
        )
        assert residual < 1e-5
        increment_residuals.append(residual)
    # Inspect a fixed margin about the frozen rank threshold; ambiguity triggers recovery.
    rank_credible = all(
        not any(1e-5 <= v <= 1e-3 for v in sv) for sv in singular_values
    )
    assert rank_credible
    boundary_names = []
    boundary_S = []
    grouping_checked = True
    print(
        "Full-coordinate coverage residuals:",
        coverage_residuals,
        "maximum increment residual:",
        max(increment_residuals),
    )
    return {
        "N": N,
        "certificate_id": certificate["certificate_id"],
        "rank_profile": ranks,
        "coverage": coverage_residuals,
        "credible": rank_credible,
        "certificate_checked": certificate_checked,
        "grouping_checked": grouping_checked,
        "boundary_names": boundary_names,
        "boundary_S": boundary_S,
        "ctx": ctx,
        "pm": pm,
        "lyap": lyap,
        "params": params,
    }
```

### Candidate-vector scan
Tagged iterates, point-to-solution gaps, smooth-gradient differences, proximal subgradients, composite residuals, and pairwise differences are scanned. Duplicate and zero coordinates are removed.

```python
from itertools import combinations
from pepflow.lyapunov_utils import (
    vectors_in_column_space,
    find_basis_with_sparsest_coefficients,
    find_symmetric_coefficient_matrix,
)

raw_candidates = []
for i in range(N + 1):
    raw_candidates.append((f"x_{i}-x_star", ctx[f"x_{i}"] - ctx["x_star"]))
    raw_candidates.append(
        (f"grad_f(x_{i})-grad_f(x_star)", f.grad(ctx[f"x_{i}"]) - f.grad(ctx["x_star"]))
    )
    if i > 0:
        raw_candidates.append(
            (
                f"grad_g(x_{i})+grad_f(x_star)",
                g.grad(ctx[f"x_{i}"]) + f.grad(ctx["x_star"]),
            )
        )
        raw_candidates.append((f"q_{i}", f.grad(ctx[f"x_{i}"]) + g.grad(ctx[f"x_{i}"])))
raw_candidates.extend((str(v), v) for v in ctx.basis_vectors())
raw_candidates.extend((f"x_{i}", ctx[f"x_{i}"]) for i in range(N + 1))
base_candidates = list(raw_candidates)
for (a, u), (b, v) in combinations(base_candidates, 2):
    raw_candidates.append((f"({a})-({b})", u - v))
candidates = []
seen = set()
for label, vector in raw_candidates:
    coordinates = np.asarray(pm.eval_vector(vector).coords, dtype=float)
    if np.linalg.norm(coordinates) < 1e-10:
        continue
    key = tuple(np.round(coordinates, 10))
    neg = tuple(np.round(-coordinates, 10))
    if key in seen or neg in seen:
        continue
    seen.add(key)
    candidates.append((label, vector))
labels_by_id = {id(v): label for label, v in candidates}
print("Deduplicated nonzero candidates:", len(candidates))
```

### Adopt a uniform interior basis; every endpoint has its own definition.
The first sparse search uses $\nabla f(x_\star)$, while later searches use $\nabla f(x_k)$. We adopt the uniform equivalent interior basis
$$[x_0-x_\star,\ x_k-x_\star,\ \nabla f(x_k)-\nabla f(x_\star),\ \nabla f(x_k)].$$
The following independent reconstruction checks justify this choice. At $k=0$ the basis is empty; at $k=N$ it is $[x_0-x_\star]$.

```python
def V_k_basis(k):
    if k == 0:
        return []
    u0 = ctx["x_0"] - ctx["x_star"]
    if k == N_int:
        return [u0]
    return [
        u0,
        ctx[f"x_{k}"] - ctx["x_star"],
        f.grad(ctx[f"x_{k}"]) - f.grad(ctx["x_star"]),
        f.grad(ctx[f"x_{k}"]),
    ]


def V_k_basis_labels(k):
    if k == 0:
        return []
    if k == N_int:
        return ["x_0-x_star"]
    return [
        "x_0-x_star",
        f"x_{k}-x_star",
        f"grad_f(x_{k})-grad_f(x_star)",
        f"grad_f(x_{k})",
    ]


for k in range(N + 1):
    selected_basis = V_k_basis(k)
    matrix = (
        np.column_stack([pm.eval_vector(v).coords for v in selected_basis]).astype(
            float
        )
        if selected_basis
        else np.zeros((len(ctx.basis_vectors()), 0))
    )
    assert np.linalg.matrix_rank(matrix, tol=1e-7) == len(selected_basis) == ranks[k]
    print(f"k={k}: rank {ranks[k]} basis {V_k_basis_labels(k)}")
```

### Coefficient matrices
The displayed symmetric matrices use the adopted ordered basis; cross terms are counted twice. Function gaps and offsets are checked separately.

```python
coeff_by_k = {}
coefficient_residuals = {}
function_coefficients = {}
function_residuals = {}
for k in range(N + 1):
    chosen = V_k_basis(k)
    target = pm.eval_scalar(lyap[k])
    if chosen:
        C = find_symmetric_coefficient_matrix(
            lyap[k],
            chosen,
            pep_context=ctx,
            resolve_parameters=params,
            indep_tol=1e-7,
            span_tol=1e-5,
        )
        matrix = np.column_stack([pm.eval_vector(v).coords for v in chosen]).astype(
            float
        )
        residual = np.linalg.norm(
            matrix @ C @ matrix.T - np.asarray(target.inner_prod_coords, dtype=float),
            ord="fro",
        )
    else:
        C = np.zeros((0, 0))
        residual = np.linalg.norm(
            np.asarray(target.inner_prod_coords, dtype=float), ord="fro"
        )
    assert residual <= 1e-5 * max(
        1, np.linalg.norm(np.asarray(target.inner_prod_coords, dtype=float), ord="fro")
    )
    coeff_by_k[k] = C
    coefficient_residuals[k] = float(residual)
    if k == 0:
        alpha, beta = 0.0, 0.0
        func_error = max_abs(target.func_coords)
    else:
        fgap = f(ctx[f"x_{k}"]) - f(ctx["x_star"])
        ggap = g(ctx[f"x_{k}"]) - g(ctx["x_star"])
        F = np.column_stack(
            [pm.eval_scalar(fgap).func_coords, pm.eval_scalar(ggap).func_coords]
        ).astype(float)
        alpha, beta = np.linalg.lstsq(
            F, np.asarray(target.func_coords, dtype=float), rcond=None
        )[0]
        func_error = max_abs(F @ np.array([alpha, beta]) - target.func_coords)
    assert func_error < 1e-5 and max_abs(target.offset) == 0
    function_coefficients[k] = [float(alpha), float(beta)]
    function_residuals[k] = float(func_error)
    print(
        f"k={k}: quadratic reconstruction residual {residual:.3e}; function residual {func_error:.3e}; offset {target.offset}"
    )
    pf.pprint_labeled_matrix(C, V_k_basis_labels(k), precision=6)
    print(
        "Function coefficients for f(x_k)-f_star, g(x_k)-g_star:",
        function_coefficients[k],
    )
```

### Reconstruct the complete potential from basis, C, and the identified gaps.


```python
def V_k_structure(k):
    chosen = V_k_basis(k)
    C = coeff_by_k[k]
    quadratic = sum(
        (float(C[i, j]) * chosen[i] * chosen[j] for i, j in np.ndindex(C.shape)),
        pf.Scalar.zero(),
    )
    if k == 0:
        return quadratic
    alpha, beta = function_coefficients[k]
    return (
        quadratic
        + alpha * (f(ctx[f"x_{k}"]) - f(ctx["x_star"]))
        + beta * (g(ctx[f"x_{k}"]) - g(ctx["x_star"]))
    )


component_residuals = {}
for k in range(N + 1):
    value = pm.eval_scalar(V_k_structure(k) - lyap[k])
    target = pm.eval_scalar(lyap[k])
    component_residuals[str(k)] = {}
    for name, actual, reference in [
        ("gram", value.inner_prod_coords, target.inner_prod_coords),
        ("function", value.func_coords, target.func_coords),
        ("offset", value.offset, target.offset),
    ]:
        error = max_abs(actual)
        assert error <= 1e-4 + 1e-5 * max_abs(reference)
        component_residuals[str(k)][name] = error
print("Full-component reconstruction passed for all indices 0 through", N)
```

## Explicit coefficients and their finite-product solution.
Eliminating the next-gradient coordinate and then the proximal-subgradient coordinate gives a rank-two slack matrix with entries $H_{pp}=m-\beta_k$, $H_{zz}=d$, $H_{rr}=a$, $H_{pz}=w$, $H_{pr}=v$, $H_{zr}=e$, where $m=(kt^2-kt+k+1)/(2t(t-1)^2)$. The final pivot equation $m-\beta_k-v^2/a-(w-ev/a)^2/\delta=0$ gives the fractional-linear recursion encoded by $T_k$. The terminal identity imposes $\beta_{N-1}=(N-1)/(2(N+1)^2)$. Multiplication of the transfer matrices solves this recursion exactly, as the first-factor identity verifies. The separate square-completion cell below verifies the elimination equation exactly.

```python
import sympy as sp
import numpy as np
import pepflow as pf

k_sp, N_sp = sp.symbols("k N", integer=True)
L_sp = sp.symbols("L", positive=True)
c_sp = sp.symbols("c", nonnegative=True)


def transfer_matrix(j, n):
    t = 2 * n - j
    return sp.Matrix(
        [
            [2 * j * (j + 1) * t * (t - 1) ** 2, j * (j + 1) * (t - 1) * (j + t + 1)],
            [
                4 * t**2 * (t - 1) * ((2 * j + 1) * t - j),
                2 * t**2 * ((j**2 + 3 * j + 1) * t + 1),
            ],
        ]
    )


def beta_closed(index, horizon):
    n = sp.Integer(horizon)
    if index == 0:
        return sp.S.Zero
    product = sp.eye(2)
    for j in range(index, horizon - 1):
        product = product * transfer_matrix(sp.Integer(j), n)
    seed = sp.Matrix([(n - 1) / (2 * (n + 1) ** 2), 1])
    result = product * seed
    return sp.cancel(result[0] / result[1])


def beta_recurrence(k, n, c):
    T = transfer_matrix(k, n)
    return sp.cancel((T[0, 0] * c + T[0, 1]) / (T[1, 0] * c + T[1, 1]))


a_num, a_den = sp.symbols("a_num a_den", positive=True)
T_generic = transfer_matrix(k_sp, N_sp)
product_identity = sp.cancel(
    (T_generic[0, 0] * a_num + T_generic[0, 1] * a_den)
    / (T_generic[1, 0] * a_num + T_generic[1, 1] * a_den)
    - beta_recurrence(k_sp, N_sp, a_num / a_den)
)
assert product_identity == 0
assert beta_recurrence(0, N_sp, c_sp) == 0
print("Finite-product first-factor identity:", product_identity, "; beta_0=0")
print("Exact discovered coefficients at N=4:", [beta_closed(j, 4) for j in range(1, 4)])
```

### Closed-form coefficient reconstruction
Every initial, interior and terminal quadratic, function-value and constant component is compared against the coherent numerical partial sums. The exact proof below is independent of the numerical tolerance.

```python
# %% Compare the exact closed coefficients with the reconstructed numerical potential.
def closed_coefficient_matrix(index, horizon, smoothness):
    n = sp.Integer(horizon)
    ell = sp.sympify(smoothness)
    if index == 0:
        return sp.zeros(0)
    if index == horizon:
        return sp.Matrix([[-ell / (4 * n)]])
    t = 2 * n - index
    C = sp.zeros(4)
    C[0, 0] = -ell / (4 * n)
    C[1, 1] = ell * (n - index) / t**2
    C[1, 3] = C[3, 1] = -1 / (2 * t)
    C[2, 2] = -beta_closed(index, horizon) / ell
    return C


def closed_gap_coefficients(index, horizon):
    if index == 0:
        return sp.S.Zero, sp.S.Zero
    if index == horizon:
        return sp.S.One, sp.S.One
    t = sp.Integer(2 * horizon - index)
    return (index + 1) / t, index / t


closed_comparison = {}
for index in range(N_int + 1):
    C_exact = closed_coefficient_matrix(index, N_int, sp.S.One)
    C_float = np.array(C_exact, dtype=float).reshape(coeff_by_k[index].shape)
    assert np.allclose(C_float, coeff_by_k[index], atol=1e-4, rtol=1e-5)
    basis_exact = V_k_basis(index)
    exact_sample = sum(
        (
            float(C_exact[i, j]) * basis_exact[i] * basis_exact[j]
            for i, j in np.ndindex(C_float.shape)
        ),
        pf.Scalar.zero(),
    )
    alpha_exact, gamma_exact = closed_gap_coefficients(index, N_int)
    if index:
        exact_sample += float(alpha_exact) * (
            f(ctx[f"x_{index}"]) - f(ctx["x_star"])
        ) + float(gamma_exact) * (g(ctx[f"x_{index}"]) - g(ctx["x_star"]))
    difference = pm.eval_scalar(exact_sample - lyap[index])
    errors = {
        key: max_abs(value)
        for key, value in [
            ("gram", difference.inner_prod_coords),
            ("function", difference.func_coords),
            ("offset", difference.offset),
        ]
    }
    assert max(errors.values()) < 1e-4
    closed_comparison[str(index)] = errors
    print(
        "Closed coefficient matrix at k=", index, "; complete scalar residuals:", errors
    )
    pf.pprint_labeled_matrix(C_float, V_k_basis_labels(index), precision=7)
    print("Exact function coefficients:", alpha_exact, gamma_exact)
print("True selected certificate S:")
pf.pprint_labeled_matrix(
    np.asarray(b4["S_matrix"], dtype=float),
    [str(v) for v in ctx.basis_vectors()],
    precision=6,
)
```

## Local square factors obtained by eliminating the next gradient and proximal subgradient.


```python
def local_coefficients(k, n, c):
    t = 2 * n - k
    a = c + (k + 2) / (2 * (t - 1))
    d = ((2 * k + 1) * t - k) / (2 * (t - 1) ** 2)
    e = (k + 1) / (2 * (t - 1))
    v = (k + 1) / (2 * t * (t - 1))
    w = (k * t + 1) / (2 * (t - 1) ** 2)
    delta = sp.factor(d - e**2 / a)
    return a, delta, e / a, v / a, sp.factor((w - e * v / a) / delta)


t_sp = sp.symbols("t", positive=True)
a, delta, ez, vp, wp = local_coefficients(k_sp, (t_sp + k_sp) / 2, c_sp)
positive_numerator = (
    2 * c_sp * (t_sp - 1) * ((2 * k_sp + 1) * t_sp - k_sp)
    + (k_sp**2 + 3 * k_sp + 1) * t_sp
    + 1
)
delta_identity = sp.factor(delta - positive_numerator / (4 * (t_sp - 1) ** 3 * a))
assert delta_identity == 0
print("Schur complement positive-factor identity:", delta_identity)
```

## Native PEPFlow symbolic helper definitions, shared by the separate identity cells.


```python
proof_parameters = {}
weight_counter = 0


def proof_weight(value):
    global weight_counter
    weight_counter += 1
    name = f"proof_weight_{weight_counter}"
    proof_parameters[name] = sp.factor(sp.cancel(sp.sympify(value)))
    return pf.Parameter(name)


def analytic_potential(x, anchor, xstar, fun, regularizer, k, n, beta, Lvalue):
    u = x - xstar
    grad = fun.grad(x)
    t = 2 * n - k
    return (
        -proof_weight(Lvalue / (4 * n)) * (anchor - xstar) ** 2
        + proof_weight(Lvalue * (n - k) / t**2) * u**2
        - proof_weight(1 / t) * (u * grad)
        - proof_weight(beta / Lvalue) * (grad - fun.grad(xstar)) ** 2
        + proof_weight((k + 1) / t) * (fun(x) - fun(xstar))
        + (
            proof_weight(k / t) * (regularizer(x) - regularizer(xstar))
            if k != 0
            else pf.Scalar.zero()
        )
    )


def local_square_sum(x, xnext, xstar, fun, regularizer, k, n, c, Lvalue):
    p = fun.grad(x) - fun.grad(xstar)
    q = regularizer.grad(xnext) + fun.grad(xstar)
    r = fun.grad(xnext) - fun.grad(xstar)
    z = q - proof_weight(Lvalue / (2 * n - k)) * (x - xstar)
    a, delta, ez, vp, wp = local_coefficients(k, n, c)
    first = r + proof_weight(ez) * z + proof_weight(vp) * p
    second = z + proof_weight(wp) * p
    return (
        proof_weight(a / Lvalue) * first**2 + proof_weight(delta / Lvalue) * second**2
    )


def local_matrix_sum(x, xnext, xstar, fun, regularizer, k, n, c, Lvalue):
    t = 2 * n - k
    p = fun.grad(x) - fun.grad(xstar)
    q = regularizer.grad(xnext) + fun.grad(xstar)
    r = fun.grad(xnext) - fun.grad(xstar)
    z = q - proof_weight(Lvalue / t) * (x - xstar)
    a = c + (k + 2) / (2 * (t - 1))
    d = ((2 * k + 1) * t - k) / (2 * (t - 1) ** 2)
    e = (k + 1) / (2 * (t - 1))
    v = (k + 1) / (2 * t * (t - 1))
    w = (k * t + 1) / (2 * (t - 1) ** 2)
    m = (k * t * t - k * t + k + 1) / (2 * t * (t - 1) ** 2)
    return (
        proof_weight(m / Lvalue) * p**2
        - proof_weight(beta_recurrence(k, n, c) / Lvalue) * p**2
        + proof_weight(d / Lvalue) * z**2
        + proof_weight(a / Lvalue) * r**2
        + proof_weight(2 * w / Lvalue) * (p * z)
        + proof_weight(2 * v / Lvalue) * (p * r)
        + proof_weight(2 * e / Lvalue) * (z * r)
    )


def check_exact(name, expression, manager):
    value = manager.eval_scalar(expression, sympy_mode=True)
    print(name, "coordinates extracted; simplifying", flush=True)

    def reduction(z):
        return (
            sp.factor(sp.cancel(z.subs(N_sp, (t_sp + k_sp) / 2)))
            if name == "Step"
            else sp.factor(sp.cancel(z))
        )

    gram = sp.Matrix(value.inner_prod_coords).applyfunc(reduction)
    funcs = sp.Matrix(list(value.func_coords)).applyfunc(reduction)
    offset = sp.factor(sp.cancel(sp.sympify(value.offset)))
    labels = [str(v) for v in manager.context.basis_vectors()]
    print(name, "Gram residual:")
    pf.pprint_labeled_matrix(np.array(gram, dtype=object), labels, precision=0)
    print(name, "function residual:", list(funcs), "constant residual:", offset)
    assert all(v == 0 for v in [*gram, *funcs, offset]), name
    return True
```

## Symbolic step identity for 1 <= k <= N-2.
For $1\le k\le N-2$, the checked identity is
$$V_{k+1}-V_k=\frac{2N+1}{t(t-1)}I_f(\star,k+1)+\frac{k+1}{t}I_f(k,k+1)+\frac{2N}{t(t-1)}I_g(\star,k+1)+\frac{k}{t}I_g(k,k+1)-\mathcal Q_k.$$
The native identity uses the equivalent quadratic matrix for $\mathcal Q_k$; the final exact Gram square-completion check proves equality to the two displayed squares. This avoids unnecessary nested rational expansion without changing the identity.

```python
ctx_step = pf.PEPContext("pgm_symbolic_step").set_as_current()
L_proof = pf.Parameter("L_proof")
f_step = pf.SmoothConvexFunction(is_basis=True, tags=["f_{step}"], L=L_proof)
g_step = pf.ConvexFunction(is_basis=True, tags=["g_{step}"])
h_step = f_step + g_step
xstar_step = h_step.set_stationary_point("x_{star}")
x_step = pf.Vector(is_basis=True, tags=["x_{k}"])
anchor_step = pf.Vector(is_basis=True, tags=["x_{0}"])
xnext_step = g_step.prox(
    x_step - f_step.grad(x_step) / L_proof, 1 / L_proof, tag="x_{k+1}"
)
f_step(xnext_step)
g_step(x_step)
proof_parameters = {"L_proof": L_sp}
weight_counter = 0
beta_k = beta_recurrence(k_sp, N_sp, c_sp)
V_step = analytic_potential(
    x_step, anchor_step, xstar_step, f_step, g_step, k_sp, N_sp, beta_k, L_sp
)
V_next = analytic_potential(
    xnext_step, anchor_step, xstar_step, f_step, g_step, k_sp + 1, N_sp, c_sp, L_sp
)
t = 2 * N_sp - k_sp
step_residuals = (
    proof_weight((2 * N_sp + 1) / (t * (t - 1)))
    * f_step.interp_ineq(xstar_step, xnext_step)
    + proof_weight((k_sp + 1) / t) * f_step.interp_ineq(x_step, xnext_step)
    + proof_weight(2 * N_sp / (t * (t - 1)))
    * g_step.interp_ineq(xstar_step, xnext_step)
    + proof_weight(k_sp / t) * g_step.interp_ineq(x_step, xnext_step)
)
Q_step = local_matrix_sum(
    x_step, xnext_step, xstar_step, f_step, g_step, k_sp, N_sp, c_sp, L_sp
)
step_LHS = V_next - V_step
step_RHS = step_residuals - Q_step
diff_step = step_LHS - step_RHS
pm_step = pf.ExpressionManager(ctx_step, resolve_parameters=proof_parameters)
step_verified = check_exact("Step", diff_step, pm_step)
```

## Base identity: the first transition, for N >= 2.
For $N\ge2$, the separately constructed initial context checks
$$V_1=\frac{I_f(\star,0)}{2N}+\frac{2N+1}{2N(2N-1)}I_f(\star,1)+\frac{I_f(0,1)}{2N}+\frac{I_g(\star,1)}{2N-1}-\mathcal Q_0-\frac{\|p_0\|^2}{4NL}.$$
No call evaluates $g(x_0)$.

```python
ctx_base = pf.PEPContext("pgm_symbolic_base").set_as_current()
f_base = pf.SmoothConvexFunction(is_basis=True, tags=["f_{base}"], L=L_proof)
g_base = pf.ConvexFunction(is_basis=True, tags=["g_{base}"])
h_base = f_base + g_base
xstar_base = h_base.set_stationary_point("x_{star}")
x0_base = pf.Vector(is_basis=True, tags=["x_{0}"])
x1_base = g_base.prox(
    x0_base - f_base.grad(x0_base) / L_proof, 1 / L_proof, tag="x_{1}"
)
f_base(x1_base)
proof_parameters = {"L_proof": L_sp}
weight_counter = 0
base_LHS = analytic_potential(
    x1_base, x0_base, xstar_base, f_base, g_base, sp.S.One, N_sp, c_sp, L_sp
)
t0 = 2 * N_sp
base_RHS = (
    proof_weight(1 / (2 * N_sp)) * f_base.interp_ineq(xstar_base, x0_base)
    + proof_weight((2 * N_sp + 1) / (t0 * (t0 - 1)))
    * f_base.interp_ineq(xstar_base, x1_base)
    + proof_weight(1 / t0) * f_base.interp_ineq(x0_base, x1_base)
    + proof_weight(2 * N_sp / (t0 * (t0 - 1))) * g_base.interp_ineq(xstar_base, x1_base)
    - local_matrix_sum(
        x0_base, x1_base, xstar_base, f_base, g_base, sp.S.Zero, N_sp, c_sp, L_sp
    )
    - proof_weight(1 / (4 * N_sp * L_sp))
    * (f_base.grad(x0_base) - f_base.grad(xstar_base)) ** 2
)
diff_base = base_LHS - base_RHS
pm_base = pf.ExpressionManager(ctx_base, resolve_parameters=proof_parameters)
base_verified = check_exact("Base", diff_base, pm_base)
```

## Boundary identity: last transition, for N >= 2, and distance normalization.
For $N\ge2$, the last transition and normalization are
$$V_N-V_{N-1}=\frac{I_f(\star,N)+NI_f(N-1,N)+2I_g(\star,N)+(N-1)I_g(N-1,N)}{N+1}-\mathcal Q_{\rm end},$$
$$h(x_N)-h(x_\star)-\frac{LR^2}{4N}=V_N+\frac{L}{4N}(\|u_0\|^2-R^2).$$

```python
ctx_boundary = pf.PEPContext("pgm_symbolic_boundary").set_as_current()
f_boundary = pf.SmoothConvexFunction(is_basis=True, tags=["f_{boundary}"], L=L_proof)
g_boundary = pf.ConvexFunction(is_basis=True, tags=["g_{boundary}"])
h_boundary = f_boundary + g_boundary
xstar_boundary = h_boundary.set_stationary_point("x_{star}")
xprev = pf.Vector(is_basis=True, tags=["x_{N-1}"])
anchor_boundary = pf.Vector(is_basis=True, tags=["x_{0}"])
xlast = g_boundary.prox(
    xprev - f_boundary.grad(xprev) / L_proof, 1 / L_proof, tag="x_{N}"
)
f_boundary(xlast)
g_boundary(xprev)
proof_parameters = {"L_proof": L_sp}
weight_counter = 0
beta_last = (N_sp - 1) / (2 * (N_sp + 1) ** 2)
V_previous = analytic_potential(
    xprev,
    anchor_boundary,
    xstar_boundary,
    f_boundary,
    g_boundary,
    N_sp - 1,
    N_sp,
    beta_last,
    L_sp,
)
last_gap = h_boundary(xlast) - h_boundary(xstar_boundary)
V_terminal = (
    last_gap - proof_weight(L_sp / (4 * N_sp)) * (anchor_boundary - xstar_boundary) ** 2
)
p_end = f_boundary.grad(xprev) - f_boundary.grad(xstar_boundary)
q_end = g_boundary.grad(xlast) + f_boundary.grad(xstar_boundary)
r_end = f_boundary.grad(xlast) - f_boundary.grad(xstar_boundary)
u_end = xprev - xstar_boundary
terminal_square_1 = (
    r_end
    + q_end
    + proof_weight(1 / (N_sp + 1)) * p_end
    - proof_weight(L_sp / (N_sp + 1)) * u_end
)
terminal_square_2 = (
    q_end
    + proof_weight(N_sp / (N_sp + 1)) * p_end
    - proof_weight(L_sp / (N_sp + 1)) * u_end
)
Q_terminal = proof_weight(1 / (2 * L_sp)) * (
    terminal_square_1**2 + terminal_square_2**2
)
boundary_LHS = V_terminal - V_previous
boundary_RHS = (
    proof_weight(1 / (N_sp + 1)) * f_boundary.interp_ineq(xstar_boundary, xlast)
    + proof_weight(N_sp / (N_sp + 1)) * f_boundary.interp_ineq(xprev, xlast)
    + proof_weight(2 / (N_sp + 1)) * g_boundary.interp_ineq(xstar_boundary, xlast)
    + proof_weight((N_sp - 1) / (N_sp + 1)) * g_boundary.interp_ineq(xprev, xlast)
    - Q_terminal
)
diff_boundary = boundary_LHS - boundary_RHS
pm_boundary = pf.ExpressionManager(ctx_boundary, resolve_parameters=proof_parameters)
boundary_verified = check_exact("Boundary", diff_boundary, pm_boundary)
R_sp = sp.symbols("R", nonnegative=True)
R_proof = pf.Parameter("R_proof")
proof_parameters["R_proof"] = R_sp
normalization_LHS = last_gap - proof_weight(L_sp / (4 * N_sp)) * R_proof**2
normalization_RHS = V_terminal + proof_weight(L_sp / (4 * N_sp)) * (
    (anchor_boundary - xstar_boundary) ** 2 - R_proof**2
)
diff_normalization = normalization_LHS - normalization_RHS
pm_normalization = pf.ExpressionManager(
    ctx_boundary, resolve_parameters=proof_parameters
)
normalization_verified = check_exact(
    "Normalization", diff_normalization, pm_normalization
)
```

## The exceptional horizon N=1, without evaluating g at x_0.
With the explicitly constructed single proximal step,
$$V_1=\tfrac12I_f(\star,0)+\tfrac12I_f(\star,1)+\tfrac12I_f(0,1)+I_g(\star,1)-\mathcal Q_{\rm end}-\frac{\|p_0\|^2}{4L}.$$
This independently checks the exceptional horizon. Both squares are displayed explicitly in executable form.

```python
ctx_one = pf.PEPContext("pgm_symbolic_one").set_as_current()
f_one = pf.SmoothConvexFunction(is_basis=True, tags=["f_{one}"], L=L_proof)
g_one = pf.ConvexFunction(is_basis=True, tags=["g_{one}"])
h_one = f_one + g_one
xs_one = h_one.set_stationary_point("x_{star}")
x0_one = pf.Vector(is_basis=True, tags=["x_{0}"])
x1_one = g_one.prox(x0_one - f_one.grad(x0_one) / L_proof, 1 / L_proof, tag="x_{1}")
f_one(x1_one)
proof_parameters = {"L_proof": L_sp}
weight_counter = 0
p_one = f_one.grad(x0_one) - f_one.grad(xs_one)
q_one = g_one.grad(x1_one) + f_one.grad(xs_one)
r_one = f_one.grad(x1_one) - f_one.grad(xs_one)
u_one = x0_one - xs_one
square_one_1 = (
    r_one + q_one + sp.Rational(1, 2) * p_one - proof_weight(L_sp / 2) * u_one
)
square_one_2 = q_one + sp.Rational(1, 2) * p_one - proof_weight(L_sp / 2) * u_one
one_LHS = h_one(x1_one) - h_one(xs_one) - proof_weight(L_sp / 4) * u_one**2
one_RHS = (
    sp.Rational(1, 2) * f_one.interp_ineq(xs_one, x0_one)
    + sp.Rational(1, 2) * f_one.interp_ineq(xs_one, x1_one)
    + sp.Rational(1, 2) * f_one.interp_ineq(x0_one, x1_one)
    + g_one.interp_ineq(xs_one, x1_one)
    - proof_weight(1 / (2 * L_sp)) * (square_one_1**2 + square_one_2**2)
    - proof_weight(1 / (4 * L_sp)) * p_one**2
)
diff_one = one_LHS - one_RHS
pm_one = pf.ExpressionManager(ctx_one, resolve_parameters=proof_parameters)
one_verified = check_exact("N=1", diff_one, pm_one)
```

## Exact quadratic square completion, valid for arbitrary vector Gram entries.


```python
import sympy as sp

k, t, c, b = sp.symbols("k t c b")
a = c + (k + 2) / (2 * (t - 1))
d = ((2 * k + 1) * t - k) / (2 * (t - 1) ** 2)
e = (k + 1) / (2 * (t - 1))
v = (k + 1) / (2 * t * (t - 1))
w = (k * t + 1) / (2 * (t - 1) ** 2)
m = (k * t * t - k * t + k + 1) / (2 * t * (t - 1) ** 2)
delta = sp.factor(d - e * e / a)
rec = (
    k
    * (k + 1)
    * (t - 1)
    * (2 * c * t * (t - 1) + k + t + 1)
    / (
        2
        * t
        * t
        * (2 * c * (t - 1) * ((2 * k + 1) * t - k) + (k * k + 3 * k + 1) * t + 1)
    )
)
derived_beta = sp.factor(m - v * v / a - (w - e * v / a) ** 2 / delta)
assert sp.cancel(rec - beta_recurrence(k, (t + k) / 2, c)) == 0
print("Solved final-pivot equation for beta_k:", derived_beta, flush=True)
print("checking coefficient", flush=True)
coefficient_identity = sp.factor(m - rec - v * v / a - (w - e * v / a) ** 2 / delta)
assert coefficient_identity == 0
print(coefficient_identity, flush=True)
A, D, E, V, W = sp.symbols("A D E V W", nonzero=True)
p, z, r = sp.symbols("p z r")
rho = V * V / A + (W - E * V / A) ** 2 / (D - E * E / A)
Q = rho * p * p + D * z * z + A * r * r + 2 * W * p * z + 2 * V * p * r + 2 * E * z * r
squares = (
    A * (r + E / A * z + V / A * p) ** 2
    + (D - E * E / A) * (z + (W - E * V / A) / (D - E * E / A) * p) ** 2
)
square_identity = sp.hessian(Q - squares, (p, z, r)).applyfunc(sp.factor)
assert square_identity == sp.zeros(3)
print("generic Gram square identity", square_identity, flush=True)
squares_verified = True
```

## Sign/domain verification and telescoping
All active interpolation multipliers are nonnegative. The product defining $\beta_k$ has positive entries for $1\le k<N-1$ and a positive denominator seed; hence $Q_k>0$ and $\beta_k\ge0$. For local squares $k\ge0$, $t>1$, $c\ge0$ give $a>0$ and
$$\delta=\frac{2c(t-1)((2k+1)t-k)+(k^2+3k+1)t+1}{4(t-1)^3a}>0.$$
Every square is subtracted and every residual is nonpositive, so $V_N\le\cdots\le V_1\le V_0=0$. Therefore
$$h(x_N)-h(x_\star)\le\frac{L}{4N}\|u_0\|^2\le\frac{LR^2}{4N}.$$
The exact identities, including all Gram, function-value and constant coordinates, are checked below. In one dimension $f=0$, $g(x)=LR|x|/(2N)$, $x_0=R$, and $x_\star=0$ attain equality when $R>0$; $R=0$ gives the zero bound.
